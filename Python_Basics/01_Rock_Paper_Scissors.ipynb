## 1. Syntax Reference: Strings, Operators, & Conditionals
* **String Formatting**: `f-strings` (introduced in Python 3.6) allow variables to be injected directly into strings using curly braces `{}`. They are significantly cleaner and faster than using the `+` concatenation operator.
* **Comparison Operators**: `==` (equal to), `!=` (not equal to), `>` (greater than).
* **Control Flow**: Python evaluates `if`, `elif`, and `else` top-to-bottom. The first condition to evaluate as `True` executes, bypassing the rest.

```python
# --- String Formatting ---
name = "XiaoHei"
age = 25

# Old Concatenation Method
greetings = "My name is " + name + "."
print(greetings)

# Modern f-string Method
print(f"{name} is {age} years old.")

# --- Comparison Operators & Control Flow ---
a = 5
b = 10

if a > b:
    print("a is greater than b")
elif a == b:
    print("a and b are equal")
elif a != b:
    print("a and b are not equal")

# --- Elif Chains ---
age_check = 20
if age_check >= 18:
    print("You are an adult.")
elif age_check > 12:
    print("You are a teenager.")
elif age_check > 1:
    print("You are a child.")
else:
    print("You are a baby.")
```

## 2. RPS Project: Data Collections & Input
* **Modules**: Using `import` allows you to bring in Python's Standard Library tools, like the `random` module.
* **Lists**: Ordered, mutable arrays (e.g., `["rock", "paper", "scissors"]`).
* **Dictionaries**: Unordered data collections that store information in `key: value` pairs. Data retrieval from a dictionary has a Time Complexity of **O(1)**.
* **Functions (`def`)**: Reusable logic blocks that can gather data and pass it back to the main program using `return`.

```python
import random

def get_choices():
    # input() pauses execution to accept a string from the user in the console
    player_choice = input("Enter a choice (rock, paper, scissors): ")
    
    # Lists and random choice
    options = ["rock", "paper", "scissors"]
    computer_choice = random.choice(options)
    
    # Storing the results in a Dictionary
    choices = {"player": player_choice, "computer": computer_choice}
    return choices
```

## 3. RPS Project: Nested Conditionals & Function Arguments
* **Arguments**: Variables passed into a function when it is called (e.g., `player` and `computer`).
* **Nested Conditionals**: Placing `if` statements inside other `if` statements. This is useful for evaluating secondary logic only if a primary condition is met.

```python
def check_win(player, computer):
    print(f"You chose {player}, computer chose {computer}.")
    
    if player == computer:
        return "It's a tie!"
    elif player == "rock":
        if computer == "scissors":
            return "Rock beats scissors! You win!"
        else:
            return "Paper beats rock! You lose!"
    elif player == "paper":
        if computer == "rock":
            return "Paper beats rock! You win!"
        else:
            return "Scissors beats paper! You lose!"
    elif player == "scissors":
        if computer == "paper":
            return "Scissors beats paper! You win!"
        else:
            return "Rock beats scissors! You lose!"
```

## 4. RPS Project: Execution & Dictionary Access
* **Dictionary Access**: Values are retrieved from a dictionary by referencing their key inside square brackets: `dictionary_name["key_name"]`.

```python
# 1. Call the function and save the returned dictionary
choices = get_choices()

# 2. Access the values using their keys and pass them as arguments
result = check_win(choices["player"], choices["computer"])

# 3. Print the final returned string
print(result)
```

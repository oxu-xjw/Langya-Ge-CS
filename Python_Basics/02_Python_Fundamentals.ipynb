```python
#Variables
# No numbers at the beginning of variable names, no special characters except for underscores, and no Python reserved keywords.

cat1 = "XiaoHei"
age = 20

name1 = 1
HEIGHT = "one"
my_name = 2
_name = "two"



#Expressions and Statements 
1 + 1
"XiaoBai"
print(cat1)
cat2 = "XiaoBai"; print(cat2)



#Comments
# this is a single line comment
name = "XiaoHei"  # this is an inline comment
"""This is a multi-line comment
that spans multiple lines."""



#Data Types
"string" # str for string
123 # int for integer
3.14 # float for floating-point number
True # bool for boolean
[1, 2, 3] # list for lists
(1, 2, 3) # tuple for tuples
{"key": "value"} # dict for dictionaries
{1, 2, 3} # set for sets
None # NoneType for None values

print(type(name) == str)
print(isinstance(name, str))

print(type(age) == int)
print(isinstance(age, int)) 

length = 2.9
print(type(length) == float)
print(isinstance(length, float))

width = float(3)
print(type(width) == float)
print(isinstance(width, float))

number = "20"
number = int(number) 
print(isinstance(number, int))



#Arithmetic Operators
age = 20
1  + 1 # Addition
5 - 2 # Subtraction
3 * 4 # Multiplication
10 / 2 # Division
10 // 3 # Floor Division
2 ** 3 # Exponentiation
10 % 3 # Modulo 

print("XiaoBai" + "is a good cat.") # String concatenation

age += 1 # age = age + 1
print(age)
age *= 2 # age = age * 2
print(age)



#Comparison Operators
a = 5
b = 10

a == b # Equal to
a != b # Not equal to
a > b # Greater than
a < b # Less than
a >= b # Greater than or equal to
a <= b # Less than or equal to



#Boolean Operators
condition1 = True
condition2 = False

not condition1 # Logical NOT
condition1 and condition2 # Logical AND
condition1 or condition2 # Logical OR

print(0 or 1) # Logical OR with integers (0 is False, 1 is True)
print(False or 'hey') # Logical OR with a boolean and a non-empty string (False is False, 'hey' is True, returns the first True value)
print('hi' or 'hey') # Logical OR with non-empty strings (both are True, returns the first one)
print([] or False) # Logical OR with an empty list and False (both are False, returns the last value)
print(False or []) # Logical OR with False and an empty list (both are False, returns the last value)

print(0 and 1) # Logical AND with integers (0 is False, 1 is True, returns the first False value)
print(1 and 0) # Logical AND with integers (1 is True, 0 is False, returns the first False value
print(False and 'hey') # Logical AND with a boolean and a non-empty string (False is False, 'hey' is True, returns the first False value#)
print('hi' and 'hey') # Logical AND with non-empty strings (both are True, returns the last one)
print([] and False) # Logical AND with an empty list and False (both are False, returns the first False value)
print(False and []) # Logical AND with False and an empty list (both are False, returns the first False value)



#Bitwise Operators
"""
& performs a binary AND
| performs a binary OR
^ performs a binary XOR operation 
~ performs a binary NOT operation
<< performs a left bitwise shift
>> performs a right bitwise shift
"""


#is & in Operators
"""
The 'is' operator checks if two variables point to the same object in memory
The 'in' operator checks if a value exists within an iterable (like a list, tuple, or string)
"""


#Ternary Operator
def is_adult(age):
    if age >= 18:
        return True
    else:
        return False
    
def is_adult_ternary(age):
    return True if age >= 18 else False



#Strings
"XiaoHei"
'XiaoBai'
name = "XiaoHei"
phrase = 'XiaoBai' + 'is a good cat.'
phrase2 = name + " is a good cat."
name += " is my cat."
age = str(20)
print(name)

print("""" XiaoHei
is

a good cat. "
""")



#String Methods
print("XiaoHei".upper()) # Convert to uppercase
print("XiaoHei".lower()) # Convert to lowercase
print("xIaOhEi".title()) # Convert to title case
print("XiaoHei".islower()) # Check if all characters are lowercase

"""
isalpha() to check if a string contains only characters and is not empty
isalnum() to check if a string contains characters or digits and is not empty
isdecimal() to check if a string contains digits and is not empty
lower() to get a lowercase version of the string
islower() to check if a string is lowercase
upper() to get an uppercase version of the string
isupper() to check if a string is uppercase
title() to get a capitalized version of the string
startsswith() to check if the string starts with a specific substring
endswith() to check if the string ends with a specific substring
replace() to replace a part of a string
split() to split a string on a specific character separator
strip() to trim the whitespace from a string
join() to append new letters to a string
find() to find the position of a substring
"""

cat = "XiaoBai"
print(cat.lower())
print(cat)
print(len(cat))
print("Bai" in cat)



#Escaping Characters
print("She said, \"Hello!\"") # Using backslash to escape double quotes



#String Characters & Slicing
name = "XiaoBai"
print(name[0])
print(name[-1])
print(name[1:4]) # Slicing from index 1 to 3 (4 is exclusive)
print(name[:4]) # Slicing from the beginning to index 3



#Booleans
done = True

print(type(done) == bool)

if done:
    print("Task is completed.")
else:
    print("Task is not completed.")
    
book_1_read = True
book_2_read = False
read_any_book = any([book_1_read, book_2_read]) # True if at least one book is read

ingredients_purchased = True
meal_cooked = False
ready_to_eat = all([ingredients_purchased, meal_cooked]) # True if all conditions are True



#Number Data Types
num1 = 2 + 3j
num2 = complex(2,3)

print(num2.real, num2.imag)



#Built-in Functions
print(abs(-5.5)) # Absolute value
print(round(3.14159, 2)) # Rounding to 2 decimal places



#Enums
from enum import Enum

class state(Enum):
    ON = 1
    OFF = 0
    
print(state.ON)
print(state.ON.value)
print(state['ON'].value)
print(state(0))
print(state['OFF'])
print(list(state))



#User Input
age = input("Please enter your age: ")
print("You age is " + age)



#Control Statements
condition = True

if condition == True:
    print("The condition")
    print("was true")

if condition == True:
    print("The condition was true")
else:
    print("The condition was false")
    
if condition == True:
    print("The condition was true")
elif condition == False:
    print("The condition was false")
else:
    print("The condition was neither true nor false")



#Lists
cats = ["XiaoHei", "XiaoBai", 1, True]
dogs = []

print("XiaoHei" in cats) # Check if "XiaoHei" is in the list
print(cats[0]) # Access the first element of the list
print(cats[-1]) # Access the last element of the list
print(cats[1:3]) # Slice the list from index 1 to 2

cats[2] = 2
print(cats)
print(len(cats)) # Get the length of the list

cats.append("new cat") # Add an element to the end of the list
dogs.extend(["Fido", "Spot"]) # Extend the list with another list
dogs += ["Buddy"] # Concatenate another list to the existing list
dogs.remove("Fido") # Remove a specific element from the list
cats.pop() # Remove and return the last element of the list
dogs.insert(0, "Rover") # Insert an element at a specific index
cats[3:3] = ["Cat 1", "Cat 2"] # Insert multiple elements at a specific index

#Sorting Lists
items = ["apple", "banana", "cherry", "Cherry", "banana"]
items.sort() # Sort the list in place
print(items)
items.sort(key=str.lower) # Sort the list in place, ignoring case

items_copy = items[:] # Create a copy of the list
print(sorted(items, key=str.lower)) # Return a new sorted list, ignoring case



#Tuples
names = ("XiaoHei", "XiaoBai", "Oscar")

names[0] # Access the first element of the tuple#
names[0:2] # Slice the tuple from index 0 to 1 (2 is exclusive)
names.index("XiaoBai") # Get the index of a specific element
len(names) # Get the length of the tuple

print("XiaoHei" in names) # Check if "XiaoHei" is in the tuple
print(sorted(names)) # Return a sorted list of the tuple elements

new_names = names + ("New Cat",) # Concatenate a new tuple to the existing tuple



#Dictionaries
cat = {"name": "XiaoHei", "age": 1, "color": "white"}

cat["name"] = "XiaoBai" # Update the value associated with the key "name"

print(cat["name"]) # Access the value associated with the key "name"
print(cat.get("age")) # Access the value associated with the key "age" using get() method
print(cat.get("breed", "Unknown")) # Access a non-existent key with a default value

print(cat.pop("color")) # Remove and return the value associated with the key "color"
print(cat)
print(cat.popitem()) # Removes the last inserted key-value pair and returns it as a tuple

cats = {"cat1": "XiaoHei", "cat2": "XiaoBai"}
print("XiaoHei" in cats) # Check if "XiaoHei" is a value in the dictionary
print(cats.keys()) # Get a view of the dictionary's keys
print(list(cats.keys())) # Get a list of the dictionary's keys
print(cats.values()) # Get a view of the dictionary's values
print(list(cats.values())) # Get a list of the dictionary's values
print(cats.items()) # Get a view of the dictionary's key-value pairs
print(len(cats)) # Get the number of key-value pairs in the dictionary

cats["favorite_food"] = "fish" # Add a new key-value pair to the dictionary
del cats["favorite_food"]
catsCopy = cats.copy() # Create a shallow copy of the dictionary



#Sets
# Sets do not allow duplicate elements and are unordered collections of unique items.
set1 = {"red", "green", "blue", "red"}
set2 = {"red"}
set3 = {"yellow", "green", "blue"}

intersection = set1 & set2 # Get the intersection of two sets
union = set2 | set3 # Get the union of two sets
difference = set1 - set3 # Get the difference of two sets
superset = set1 > set2 # Check if set1 is a superset of set2
subset = set2 < set1 # Check if set2 is a subset of set1
print(list(set1)) # Get a list of the unique elements in the set
print("red" in set1) # Check if "red" is in the set



#Functions
"""
Parameters are the variables defined in the function definition
Arguments are the values passed to the function when it is called
"""
def hello():
    print("Hello!")
    
hello() # Call the function to execute its code


def hello_name(name):
    print("Hello, " + name + "!")
    
hello_name("XiaoHei") # Call the function with an argument


def hello_name_empty(name="stranger"):
    print("Hello, " + name + "!")

hello_name_empty() # Call the function without an argument
hello_name_empty("XiaoBai") # Call the function with an argument


def hello_name_and_age(name, age):
    print("Hello, " + name + "! You are " + str(age) + " years old.")
    
hello_name_and_age("XiaoHei", 20) # Call the function with multiple arguments


def change(value):
    value = 2
    
val = 1
change(val)
print(val) # The value of val remains unchanged because integers are immutable and passed by value


def change_(value):
    value["name"] = "XiaoBai"

val = {"name": "XiaoHei"}
change_(val)
print(val) # The value of val is changed because dictionaries are mutable and passed by reference


def hello_return_1(name):
    if not name:
        return "No name provided"
    return "Hello, " + name + "!"

hello_return_1(False) # Call the function with an empty string
hello_return_1("XiaoHei") # Call the function with a valid name


def hello_return_2(name):
    print("Hello, " + name + "!")
    return name, "cat", 1

print(hello_return_2("XiaoHei")) # Call the function and print the returned tuple



#Variable Scope
age = 20 # Global variable

def age_global():
    print(age) # Accessing the global variable
    
print(age) # Accessing the global variable outside the function
age_global() # Call the function to print the global variable

def age_local():
    age = 30 # Local variable
    print(age) # Accessing the local variable

print(age) # Accessing the global variable again to show it is unchanged
age_local() # Call the function to print the local variable



#Nested Functions 
def talk(phrase):
    def say(word):
        print(word)
        
    words = phrase.split(' ')
    for word in words:
        say(word)
        
talk("Hello, how are you?") # Call the function to execute the nested function

def count():
    count = 0
    
    def increment():
        nonlocal count # Declare count as nonlocal to modify the variable in the enclosing scope
        count += 1
        return count
    
    increment() # Call the nested function to increment the count
    
count() # Call the outer function to execute the nested function and modify the count variable



#Closures
def count_closure():
    count = 0
    
    def increment():
        nonlocal count # Declare count as nonlocal to modify the variable in the enclosing scope
        count += 1
        return count
    
    return increment # Return the nested function as a closure

increment = count_closure() # Get the closure function

print(increment()) # Call the closure function to increment and return the count
print(increment()) # Call the closure function again to see the updated count
print(increment()) # Call the closure function again to see the updated count



#Objects
age = 8
print(age.real) # Access the real part of the number (for complex numbers)
print(age.imag) # Access the imaginary part of the number (for complex numbers)
print(age.bit_length()) # Get the number of bits necessary to represent the number in binary

items = [1,2]
items.append(3) # Call the append method of the list object to add an element to the list
items.pop() # Call the pop method of the list object to remove and return the last element of the list
print(id(items)) # Get the memory address of the list object



#Loops
condition = True
while condition == True:
    print("The condition is true.")
    condition = False # Update the condition to break the loop
    
count = 0
while count < 8:
    print("Count is: " + str(count))
    count = count + 1 # Increment the count to eventually break the loop
    
print("Loop finished.")


items = [1, 2, 3, 4, 5]
for item in items:
    print(item) # Print each item in the list
    
for num in range(6):
    print(num) # Print numbers from 0 to 5
    
for index, item in enumerate(items):
    print(index, item) # Print the index and item for each element in the list using enumerate()



#Break and Continue
for item in items:
    if item == 3:
        break # Exit the loop when item is 3
    print(item) # Print items until the loop is broken



#Classes
class Animal:
    def sleeping(self):
        print("The animal is sleeping.")

class Cat(Animal):
    def __init__(self, name, age, color):
        self.name = name
        self.age = age
        self.color = color

    def meow(self):
        print("Meow!")

XiaoHei = Cat("XiaoHei", 1, "black") # Create an instance of the Cat class

print(type(XiaoHei)) # Check the type of the instance
print(XiaoHei.color) # Access the name attribute of the instance

XiaoHei.meow() # Call the meow method of the instance
XiaoHei.sleeping() # Call the sleeping method inherited from the Animal class



#Modules
import cat
cat.meow() # Call the meow function from the cat module

from dog import bark
bark() # Call the bark function from the dog module

from lib import XiaoHei
XiaoHei.sleep() # Call the sleep function from the XiaoHei module in the lib package

from lib.XiaoBai import eating
eating() # Call the eating function from the XiaoBai module in the lib package

"""
math for math utilities
re for regular expressions
json to work with JSON
datetime to work with dates
sqlite3 to use SQLite databases
os for Operating system utilities
random for random number generation
statistics fpr statistical utilities
requests to perform HTTP network requests
http to create HTTP servers
urllib to manage URLs
"""

import math
from math import sqrt

print(math.sqrt(16)) # Use the sqrt function from the math module to calculate the square root of 16
print(sqrt(25)) # Use the sqrt function imported directly from the math module to calculate the square root of 25



#Arguments from Command Line
import sys

name = sys.argv[1] # Get the first command-line argument (after the script name)
print(sys.argv) # Print the list of command-line arguments passed to the script
print("Hello, " + name + "!") # Print a greeting using the command-line argument


import argparse

parser = argparse.ArgumentParser(
    description='This program prints the name of my cats.'
)

parser.add_argument('-c', '--color', metavar='color', required=True, choices={'black', 'white'}, 
                    help='the color to search for')

args = parser.parse_args(['-c', 'black']) # Parse the command-line arguments

print(args.color) # Print the value of the color argument



#Lambda Functions

lambda num: num * 2 # A lambda function that takes a number and returns it multiplied by 2

multiply = lambda a, b : a * b # A lambda function that takes two numbers and returns their product
print(multiply(3, 4)) # Call the lambda function to multiply 3 and 4



#Map, Filter, Reduce
number = [1, 2, 3, 4, 5]

def double(x):
    return x * 2

result = map(double, number) # Use map to apply the double function to each element in the list
print(list(result)) # Convert the map object to a list and print it

map_result = map(lambda x : x * 2, number) # Use map with a lambda function to double each element in the list
print(list(map_result)) # Convert the map object to a list and print it


numbers = [1, 2, 3, 4, 5]

def isEven(n):
    return n % 2 == 0

result = filter(isEven, numbers) # Use filter to get only the even numbers from the list
print(list(result)) # Convert the filter object to a list and print it

filter_result = filter(lambda x : x % 2 == 0, numbers) # Use filter with a lambda function to get only the even numbers from the list
print(list(filter_result)) # Convert the filter object to a list and print it


expenses = [
    ('Cat Food', 80),
    ('Cat Toys', 120)
]

sum = 0
for expense in expenses:
    sum += expense[1] # Add the cost of each expense to the sum
    
print(sum) # Print the total expenses

from functools import reduce

reduce_sum = reduce(lambda a, b : a[1] + b[1], expenses) # Use reduce with a lambda function to sum the costs of the expenses
print(reduce_sum) # Print the total expenses calculated using reduce



#Recursion
"""
3! = 3 * 2 * 1 = 6
4! = 4 * 3 * 2 * 1 = 24
5! = 5 * 4 * 3 * 2 * 1 = 120
"""

def factorial(n):
    if n == 1: return 1 # Base case: factorial of 1 is 1
    return n*factorial(n-1) # Recursive case: n! = n * (n-1)!

print(factorial(5)) # Call the factorial function to calculate 5!



#Decorators
def logtime(func):
    def wrapper():
        print("before") # Print a message before calling the original function
        val = func() # Call the original function and store its return value
        print("after") # Print a message after calling the original function
        return val
    return wrapper

@logtime
def hello():
    print("Hello!")

hello() # Call the decorated function to see the log messages before and after the function execution



#Docstrings
def increment(x):
    """Increment a number"""
    return x + 1


"""Cat module

This module does ... bla bla bla and provides the following classes:

- Cat
...
"""

class Cat:
    """A class representing a cat"""
    def __init__(self, name, age):
        """Initialize a new cat"""
        self.name = name
        self.age = age
        
    def meow(self):
        """Make the cat meow"""
        print("Meow!")
        
print(help(Cat)) # Print the docstring of the Cat class



#Annotations
def increment(num: int) -> int:
    return num + 1

count: int = 0 # Annotate the variable count as an integer



#Exceptions
"""
try:
    # some lines of code
except <ERROR1>:
    # handler <ERROR1>
except <ERROR2>:
    # handler <ERROR2>
else:
    # no exceptions were raised, the code ran successfully
finally:
    # do something in any case
"""


try:
    result = 10 / 0 # This will raise a ZeroDivisionError
except ZeroDivisionError:
    print("Cannot divide by zero!") # Handle the exception by printing a message
finally:
    result = 1
    
print(result) # Print the result, which will be 1 due to the finally block

try:
    raise Exception('An error occurred!') # Manually raise an exception with a custom message
except Exception as error:
    print(error) # Catch the exception and print the error message
    
    
class CatNotFoundException(Exception):
    print("This is a custom exception for when a cat is not found.")
    pass

try:
    raise CatNotFoundException()
except CatNotFoundException:
    print("Cat not found!") # Handle the custom exception by printing a message



#With
filename = 'lib/cats.txt'

try:
    file = open(filename, 'r') # Open the file for reading
    content = file.read() # Read the content of the file
    print(content) # Print the content of the file
finally:
    file.close() # Ensure the file is closed after reading
    

with open(filename, 'r') as file:
    content = file.read() # Read the content of the file
    print(content) # Print the content of the file
# The file is automatically closed after the with block, even if an error occurs



#Installing Packages with pip
"""
pypi.org is the Python Package Index
A repository of software for the Python programming language
It allows you to find and install packages developed by the Python community.

pip install <package_name> is the command to install a package from PyPI using pip
pip install -U <package_name> is the command to upgrade an already installed package to the latest version
pip uninstall <package_name> is the command to uninstall a package that was previously installed using pip
pip show <package_name> is the command to display information about an installed package, including its version and location
"""



#List Compression
numbers = [1, 2, 3, 4, 5]
numbers_power_2 = [n ** 2 for n in numbers] # Create a new list with the squares of the numbers using list comprehension

print(numbers_power_2) # Print the new list with the squares of the numbers

# For Loop version
numbers_power_2 = []
for n in numbers:
    numbers_power_2.append(n ** 2) 

# Map and Lambda version
numbers_power_2 = list(map(lambda n : n ** 2, numbers)) 



#Polymorphism
class Dog:
    def eat(self):
        print("The dog is eating.")
        
class Cat:
    def eat(self):
        print("The cat is eating.")
        
animal1 = Dog() # Create an instance of the Dog class
animal2 = Cat() # Create an instance of the Cat class

animal1.eat() # Call the eat method of the Dog instance
animal2.eat() # Call the eat method of the Cat instance



#Operator Overloading
class Cat:
    # the Cat class
    def __init__(self, name, age):
        self.name = name
        self.age = age
        
    def __gt__(self, other):
        return True if self.age > other.age else False
        
XiaoBai = Cat("XiaoBai", 1) # Create an instance of the Cat class
XiaoHei = Cat("XiaoHei", 1) # Create another instance of the Cat class

print(XiaoHei > XiaoBai) # Use the overloaded greater than operator to compare the ages of the two cats

"""
__add__() respond to the + operator
__sub__() respond to the - operator
__mul__() respond to the * operator
__truediv__() respond to the / operator
__floordiv__() respond to the // operator
__mod__() respond to the % operator
__pow__() respond to the ** operator
__rshift__() respond to the >> operator
__lshift__() respond to the << operator
__and__() respond to the & operator
__or__() respond to the | operator
__xor__() respond to the ^ operator
"""
```

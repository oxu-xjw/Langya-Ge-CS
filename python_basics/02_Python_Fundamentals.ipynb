## 1. Variables and Naming Conventions
* **Variables**: Dynamically typed labels used to store data in memory. You do not need to declare their type upfront.
* **Strict Rules**: Cannot start with a number, cannot contain special characters (except underscores `_`), and cannot use Python reserved keywords (like `if`, `for`, or `class`).
* **Pythonic Conventions**: Standard variables use `snake_case`. Naming a variable in `ALL_CAPS` is a convention to indicate a constant (a value that should not be changed). A leading underscore (e.g., `_name`) indicates the variable is intended for internal/private use.

```python
# Standard variable assignments
cat1 = "XiaoHei"
age = 20

# Numbers are allowed, just not at the very beginning
name1 = 1

# ALL_CAPS indicates a constant (a variable you do not intend to modify)
HEIGHT = "one"

# Snake_case is the standard for multi-word variables in Python
my_name = 2

# A leading underscore indicates an "internal" variable
_name = "two"

print(f"My cat is {cat1} and he is {age} years old.")
```

## 2. Expressions and Statements
* **Expression**: Any piece of code that evaluates to a value (e.g., `1 + 1` evaluates to `2`, and `"XiaoBai"` evaluates to a string). On their own, expressions do not *do* anything or save any data.
* **Statement**: An instruction that tells the Python interpreter to perform an action. Assigning a variable (`cat1 = "XiaoHei"`) or printing to the console (`print(cat1)`) are statements.
* **Semicolons**: Python normally relies on line breaks to separate statements. However, you can use a semicolon `;` to place multiple statements on a single line (though this is generally discouraged in Python for readability).

```python
# --- Expressions ---
# These evaluate to a value, but do not save or output anything on their own.
1 + 1
"XiaoBai"

# --- Statements ---
# These perform an action (assignment and printing).
cat1 = "XiaoHei" 
print(cat1)

# Using a semicolon to execute two statements on the same line
cat2 = "XiaoBai"; print(cat2)
```

## 4. Core Data Types and Type Checking
* **Core Types**: Python includes several built-in data types, such as `str` (String), `int` (Integer), `float` (Decimal), `bool` (Boolean), `list`, `tuple`, `dict` (Dictionary), `set`, and `NoneType`.
* **Type Checking (`isinstance` vs `type`)**: While `type(variable) == int` works, the Pythonic standard is to use `isinstance(variable, int)`. `isinstance()` is preferred in production code because it properly handles class inheritance.
* **Type Casting**: You can convert data from one type to another using constructor functions like `int()`, `float()`, or `str()`.

```python
# --- Core Data Types ---
my_string = "string"       # str
my_int = 123               # int
my_float = 3.14            # float
my_bool = True             # bool
my_list = [1, 2, 3]        # list 
my_tuple = (1, 2, 3)       # tuple (immutable)
my_dict = {"key": "value"} # dict (key-value pairs)
my_set = {1, 2, 3}         # set (unique elements)
my_none = None             # NoneType (represents the absence of a value)

# --- Type Checking ---
name = "XiaoHei"
age = 20

# Checking strings
print(type(name) == str)         # Output: True
print(isinstance(name, str))     # Output: True (Preferred)

# Checking integers
print(type(age) == int)          # Output: True
print(isinstance(age, int))      # Output: True 

# --- Type Casting (Conversion) ---
length = 2.9
print(isinstance(length, float)) # Output: True

# Casting an int to a float
width = float(3)
print(isinstance(width, float))  # Output: True

# Casting a string to an int
number_string = "20"
number_int = int(number_string) 
print(isinstance(number_int, int)) # Output: True
```

## 5. Arithmetic and Assignment Operators
* **Standard Math**: Addition (`+`), Subtraction (`-`), Multiplication (`*`), Division (`/`), and Exponentiation (`**`). Note that standard division (`/`) *always* returns a float, even if the numbers divide evenly.
* **Floor Division (`//`)**: Divides the numbers and chops off the decimal, returning the nearest whole integer rounded down.
* **Modulo (`%`)**: Returns the *remainder* of a division operation. This is incredibly useful for algorithms, such as checking if a number is even (`num % 2 == 0`).
* **String Concatenation**: The `+` operator can be used to merge strings together.
* **Augmented Assignment**: Shorthand operators like `+=` or `*=` update a variable in place without needing to write out the full equation (e.g., `age += 1` instead of `age = age + 1`).

```python
# --- Arithmetic Operators ---
print(1 + 1)   # Addition
print(5 - 2)   # Subtraction
print(3 * 4)   # Multiplication
print(10 / 2)  # Division (Returns a float: 5.0)
print(2 ** 3)  # Exponentiation (2 to the power of 3)

# --- Floor Division and Modulo ---
print(10 // 3) # Floor Division (Returns 3, drops the .333)
print(10 % 3)  # Modulo (Returns 1, which is the remainder of 10 divided by 3)

# --- String Concatenation ---
print("XiaoBai" + " is a good cat.") 

# --- Augmented Assignment Operators ---
age = 20

age += 1 # Same as: age = age + 1
print(f"Age after addition: {age}")

age *= 2 # Same as: age = age * 2
print(f"Age after multiplication: {age}")
```

## 6. Comparison Operators
* **Evaluation**: Comparison operators evaluate two values and *always* return a Boolean (`True` or `False`).
* **Assignment vs. Equality**: A single equals sign `=` assigns a value to a variable. A double equals sign `==` compares two values to see if they are identical. Confusing the two is a very common syntax error.
* **Operators**: Equal to (`==`), Not equal to (`!=`), Greater than (`>`), Less than (`<`), Greater than or equal to (`>=`), Less than or equal to (`<=`).

```python
# --- Comparison Operators ---
a = 5
b = 10

print(f"Is a equal to b? {a == b}")
print(f"Is a not equal to b? {a != b}")
print(f"Is a greater than b? {a > b}")
print(f"Is a less than b? {a < b}")
print(f"Is a greater than or equal to b? {a >= b}")
print(f"Is a less than or equal to b? {a <= b}")
```

## 7. Boolean Operators and Short-Circuiting
* **Logical Operators**: `not` (inverts the boolean), `and` (both must be true), and `or` (at least one must be true).
* **Truthy and Falsy**: In Python, values like `0`, `""` (empty string), `[]` (empty list), and `None` are considered "Falsy". Almost everything else is "Truthy".
* **Short-Circuit Evaluation**: `and` and `or` do not just return `True` or `False`; they return the actual evaluated objects.
  * **`or`**: Returns the *first Truthy* value it finds. If all are Falsy, it returns the *last Falsy* value.
  * **`and`**: Returns the *first Falsy* value it finds. If all are Truthy, it returns the *last Truthy* value.

```python
# --- Standard Logical Operators ---
condition1 = True
condition2 = False

print(f"NOT condition1: {not condition1}")             # Logical NOT
print(f"condition1 AND condition2: {condition1 and condition2}") # Logical AND
print(f"condition1 OR condition2: {condition1 or condition2}")   # Logical OR

print("\n--- Short-Circuit Evaluation: OR ---")
print(0 or 1)             # 0 is Falsy, 1 is Truthy -> returns 1
print(False or 'hey')     # Returns the first Truthy value -> 'hey'
print('hi' or 'hey')      # Both are Truthy, returns the first one -> 'hi'
print([] or False)        # Both are Falsy, returns the last value -> False
print(False or [])        # Both are Falsy, returns the last value -> []

print("\n--- Short-Circuit Evaluation: AND ---")
print(0 and 1)            # 0 is Falsy, returns the first Falsy value -> 0
print(1 and 0)            # 1 is Truthy, evaluates next, 0 is Falsy -> 0
print(False and 'hey')    # Returns the first Falsy value -> False
print('hi' and 'hey')     # Both are Truthy, returns the last one -> 'hey'
print([] and False)       # Both are Falsy, returns the first Falsy value -> []
print(False and [])       # Both are Falsy, returns the first Falsy value -> False
```

## 8. Bitwise Operators
* **Functionality**: Bitwise operators evaluate integers at their fundamental binary level (1s and 0s) rather than their base-10 integer value.
* **Algorithm Utility**: Extremely useful for optimizing data structures and performing math operations with very low time and space complexity.
* **Operators**: 
  * `&` (AND): Sets each bit to 1 if both bits are 1.
  * `|` (OR): Sets each bit to 1 if one of two bits is 1.
  * `^` (XOR): Sets each bit to 1 if only one of two bits is 1.
  * `~` (NOT): Inverts all the bits.
  * `<<` (Left Shift): Shifts bits left, pushing in zeros from the right (effectively multiplying by 2).
  * `>>` (Right Shift): Shifts bits right, pushing copies of the leftmost bit in from the left (effectively dividing by 2).

```python
# --- Bitwise Operators ---
a = 5  # Binary: 0b101
b = 3  # Binary: 0b011

# The bin() function converts an integer into its binary string representation
print(f"a = {a}  | Binary: {bin(a)}")
print(f"b = {b}  | Binary: {bin(b)}\n")

print(f"AND (a & b): {a & b}  | Binary: {bin(a & b)}")
print(f"OR  (a | b): {a | b}  | Binary: {bin(a | b)}")
print(f"XOR (a ^ b): {a ^ b}  | Binary: {bin(a ^ b)}")

# Note: Python handles negative binary via Two's Complement, which is why ~5 is -6
print(f"NOT (~a)   : {~a} | Binary: {bin(~a)}")

# Bit shifting
print(f"Left Shift  (a << 1): {a << 1} | Binary: {bin(a << 1)}")
print(f"Right Shift (a >> 1): {a >> 1}  | Binary: {bin(a >> 1)}")
```

## 9. Identity and Membership Operators (`is` & `in`)
* **Identity (`is`)**: Checks if two variables point to the exact same object in memory. This is different from `==`, which only checks if their values are equivalent.
* **Membership (`in`)**: Checks if a specific value exists within an iterable sequence (such as a string, list, or tuple).

```python
# --- Identity Operator (is) ---
list1 = [1, 2, 3]
list2 = [1, 2, 3]
list3 = list1

# '==' checks the value (the contents of the lists are the same)
print(f"list1 == list2: {list1 == list2}") 

# 'is' checks the memory address
# False: They have the same contents, but are separate lists in memory
print(f"list1 is list2: {list1 is list2}") 

# True: list3 was explicitly assigned to point to the exact same memory object as list1
print(f"list1 is list3: {list1 is list3}") 


# --- Membership Operator (in) ---
cats = ["XiaoHei", "XiaoBai"]
cat_name = "XiaoHei"

# Checking a list
print(f"Is XiaoHei in the list? {'XiaoHei' in cats}")

# Checking a string
print(f"Is the letter 'X' in the name? {'X' in cat_name}")
```

## 10. Ternary Operator
* **Definition**: The ternary operator provides a one-line shorthand for a standard `if-else` statement. 
* **Syntax**: `[Value if True] if [Condition] else [Value if False]`.
* **Use Case**: Best used for simple, binary evaluations. If the logic requires multiple `elif` conditions or complex calculations, a standard `if-else` block is preferred for readability.

```python
# --- Standard If-Else ---
def is_adult(age):
    if age >= 18:
        return True
    else:
        return False
    
# --- Ternary Operator ---
def is_adult_ternary(age):
    return True if age >= 18 else False

# --- Execution & Comparison ---
age_to_check = 20

print(f"Standard function output: {is_adult(age_to_check)}")
print(f"Ternary function output:  {is_adult_ternary(age_to_check)}")
```

## 12. String Methods
* **Methods vs. Functions**: Methods are specific functions attached to an object, called using dot notation (e.g., `string.upper()`).
* **Immutability**: Strings in Python are immutable. Methods like `.lower()` or `.replace()` return a *new* string; they do not change the original string variable unless you explicitly reassign it (e.g., `cat = cat.lower()`).
* **Common Methods**:
  * `upper()` / `lower()` / `title()`: Alters the casing of the string.
  * `isupper()` / `islower()`: Checks if the entire string matches a specific case.
  * `isalpha()`: Checks if the string contains *only* alphabetical characters.
  * `isalnum()`: Checks if the string contains characters or digits (no special characters).
  * `isdecimal()`: Checks if the string contains *only* digits.
  * `startswith(sub)` / `endswith(sub)`: Checks if the string begins or ends with a specific substring.
  * `replace(old, new)`: Swaps a specific part of a string with new text.
  * `split(separator)`: Breaks a string apart into a List, splitting it at the specified character.
  * `strip()`: Trims invisible whitespace from the very beginning and end of a string.
  * `join(list)`: Glues a list of strings together into a single string.
  * `find(substring)`: Returns the numerical index position of the first time a substring appears.

```python
# --- Case Formatting ---
print("XiaoHei".upper()) # XIAOHEI
print("XiaoHei".lower()) # xiaohei
print("xIaOhEi".title()) # Xiaohei

# --- Content Checking ---
print("XiaoHei".islower())    # False
print("XiaoHei123".isalnum()) # True (Alphanumeric)
print("12345".isdecimal())    # True

# --- String Manipulation ---
cat_info = "  XiaoBai is a good cat  "

# Strip removes the blank spaces at the start and end
print(f"Stripped: '{cat_info.strip()}'") 

# Replace swaps text
print(f"Replaced: '{cat_info.replace('good', 'great')}'")

# Split breaks a string into a list based on a separator (like a comma)
words = "XiaoBai,XiaoHei,Oscar".split(",")
print(f"Split List: {words}") 

# Join glues a list back together using a string as the connector
joined_string = " & ".join(words)
print(f"Joined String: {joined_string}") 

# --- Built-in Functions & Membership ---
cat = "XiaoBai"

# Calling a method does NOT change the original string
print(cat.lower()) 
print(cat) # Still prints "XiaoBai"

print(len(cat))     # Returns the number of characters (7)
print("Bai" in cat) # Membership check (Returns True)
```

## 13. Escaping Characters and String Slicing
* **Escaping (`\`)**: The backslash tells Python to ignore the special meaning of the next character. This is used to include quotes inside a string without breaking it, or to trigger special text formatting like `\n` (new line) and `\t` (tab).
* **Zero-Indexing**: Python strings are zero-indexed, meaning the first character is at position `0`.
* **Negative Indexing**: Python allows you to count backward from the end of the string using negative numbers (`-1` is always the last character).
* **Slicing Syntax**: `[start:stop:step]`. 
  * The `start` index is inclusive. 
  * The `stop` index is **exclusive** (it stops *before* this index).
  * If you leave `start` or `stop` blank, Python automatically assumes the very beginning or the very end of the string.

```python
# --- Escaping Characters ---
# The backslash prevents the internal quotes from ending the string early
print("She said, \"Hello!\"")

# \n creates a new line, \t creates an indented tab
print("XiaoBai\n\tis a good cat.\n")

# --- String Indexing ---
name = "XiaoBai"

print(f"First character [0]: {name[0]}")
print(f"Last character [-1]: {name[-1]}\n")

# --- String Slicing ---
# Slicing from index 1 to 3 (4 is exclusive)
print(f"Slice [1:4]: {name[1:4]}") 

# Slicing from the beginning to index 3
print(f"Slice [:4]:  {name[:4]}") 

# Slicing from index 4 to the very end
print(f"Slice [4:]:  {name[4:]}") 

# Using the 'step' parameter (-1) to step backward and reverse the string
print(f"Reversed [::-1]: {name[::-1]}")
```

## 14. Booleans, `any()`, and `all()`
* **Booleans**: Represent one of two values: `True` or `False`.
* **Pythonic Evaluation**: Because Booleans inherently represent truth, you do not need to write `if done == True:`. Simply writing `if done:` is the standard best practice.
* **`any(iterable)`**: Takes a list (or other iterable) of booleans and returns `True` if **at least one** element evaluates to True (similar to a chain of `or` operators).
* **`all(iterable)`**: Takes a list of booleans and returns `True` **only if every single** element evaluates to True (similar to a chain of `and` operators).

```python
# --- Basic Booleans ---
done = True

# Checking the type
print(f"Is 'done' a boolean? {isinstance(done, bool)}\n")

# Pythonic way to evaluate a boolean in a control statement
if done:
    print("Task is completed.\n")
else:
    print("Task is not completed.\n")

# --- The any() Function ---
book_1_read = True
book_2_read = False

# Evaluates to True because at least one condition in the list is True
read_any_book = any([book_1_read, book_2_read]) 
print(f"Has at least one book been read? {read_any_book}")

# --- The all() Function ---
ingredients_purchased = True
meal_cooked = False

# Evaluates to False because meal_cooked is False (ALL must be True)
ready_to_eat = all([ingredients_purchased, meal_cooked]) 
print(f"Is the meal ready to eat? {ready_to_eat}")
```

## 15. Number Data Types (Complex Numbers)
* **Native Support**: Python natively supports complex numbers without needing to import any external math libraries.
* **The Imaginary Unit**: While standard mathematics uses $i$ to represent the imaginary unit ($\sqrt{-1}$), Python was heavily influenced by engineering conventions and uses `j` instead.
* **Creation**: You can define a complex number using literal syntax (e.g., `2 + 3j`) or by using the built-in `complex(real, imaginary)` function.
* **Attributes**: You can extract the individual mathematical components using the `.real` and `.imag` properties. Python returns these components as floats.

```python
# --- Creating Complex Numbers ---
# Using the literal 'j' syntax
num1 = 2 + 3j 

# Using the built-in complex() function
num2 = complex(2, 3) 

print(f"Literal syntax: {num1}")
print(f"Function syntax: {num2}")
print(f"Are they equal? {num1 == num2}\n")

# --- Accessing Components ---
# The .real and .imag properties return floating-point numbers
print(f"Real part of num2: {num2.real}")
print(f"Imaginary part of num2: {num2.imag}")
```

## 16. Built-in Functions
* **Availability**: Python comes with several built-in functions that are always available without needing to `import` any external modules (e.g., `print()`, `len()`, `type()`).
* **`abs(number)`**: Returns the absolute value of a number. This simply means its mathematical distance from zero, stripping away any negative signs.
* **`round(number, ndigits)`**: Rounds a floating-point number to a specified number of decimal places. If you do not provide the second argument, it will round to the nearest whole integer.

```python
# --- Built-in Math Functions ---

# abs() returns the absolute (positive) value
print(f"Absolute value of -5.5: {abs(-5.5)}")

# round() rounds a float to a specific number of decimal places
print(f"Rounded to 2 decimals: {round(3.14159, 2)}")

# If you omit the second argument, round() returns the nearest integer
print(f"Rounded to nearest integer: {round(3.14159)}")
```

## 17. Enums (Enumerations)
* **Definition**: An Enum is a set of symbolic names bound to unique, constant values. It allows you to create readable, descriptive labels for variables that should never change.
* **Importing**: Enums are not available by default; you must import the `Enum` class from the built-in `enum` module.
* **Conventions**: The class name should use `PascalCase`, and the enum members inside should use `ALL_CAPS` to signify they are constants.
* **Use Case**: Excellent for representing specific states (like ON/OFF, directions like NORTH/SOUTH, or menu screens like MAIN/PAUSED).

```python
from enum import Enum

# --- Defining an Enum ---
class State(Enum):
    ON = 1
    OFF = 0

# --- Accessing Enums ---
# Accessing the Enum member directly
print(f"Direct access: {State.ON}")

# Accessing the underlying integer value
print(f"Value access: {State.ON.value}")

# Accessing dynamically via its string name (Key)
print(f"String key access: {State['ON']}")

# Accessing dynamically via its underlying value
print(f"Value lookup: {State(0)}")

# --- Listing Enums ---
# You can convert the entire Enum into a list to see all available options
print(f"\nFull List: {list(State)}")
```

## 18. User Input
* **The `input()` Function**: Pauses the execution of the program and displays a prompt, waiting for the user to type a response and press `Enter`.
* **The Data Type Trap**: `input()` *always* returns a string (`str`), regardless of what the user types. If the user enters `20`, Python reads it as the text `"20"`.
* **Type Casting Input**: If you need to perform mathematical operations on a user's numerical input, you must explicitly cast it to an integer or float using `int()` or `float()`.

```python
# --- Getting Input ---
# In a Jupyter Notebook, this will open a text box for you to type into
age_string = input("Please enter your age: ")

# String concatenation works perfectly because age_string is a string
print("Your age is " + age_string)


# --- The Data Type Trap ---
print(f"\nThe data type returned by input() is: {type(age_string)}")

# To perform math, we must cast the string into an integer
age_int = int(age_string)

print(f"In 5 years, you will be {age_int + 5} years old.")
```

## 19. Control Statements (`if`, `elif`, `else`)
* **Execution Flow**: Control statements evaluate conditions from top to bottom. The moment Python finds a condition that evaluates to `True`, it runs that specific block of code and skips the rest.
* **Indentation**: Python relies strictly on indentation (usually 4 spaces or a Tab) to determine which lines of code belong inside the `if` block. If the indentation is wrong, the program will crash.
* **The Catch-All (`else`)**: The `else` statement requires no condition. It serves as a final fallback that only runs if every single `if` and `elif` above it evaluated to `False`.
* **Pythonic Syntax**: You should evaluate booleans directly (`if condition:`) rather than comparing them to a boolean value (`if condition == True:`).

```python
# --- Basic If Statement ---
condition = True

# The indented lines only run if the condition is True
if condition:
    print("The condition")
    print("was true\n")

# --- If / Else ---
if condition:
    print("The condition was true\n")
else:
    print("The condition was false\n")
    
# --- If / Elif / Else ---
# You can chain as many 'elif' (else if) statements as you need
if condition:
    print("The condition was true")
elif not condition: 
    print("The condition was false")
else:
    # This acts as a catch-all if none of the above conditions were met
    print("The condition was neither true nor false")
```

## 20. Lists
* **Properties**: Lists are ordered, mutable (changeable), and allow duplicate elements. They can hold mixed data types (e.g., strings, integers, and booleans in the same list).
* **Adding Elements**: 
  * `append(item)`: Adds a single item to the end.
  * `extend(list)`: Merges another list onto the end.
  * `insert(index, item)`: Places an item at a specific position.
* **Removing Elements**:
  * `remove(value)`: Deletes the first occurrence of a specific value.
  * `pop()`: Removes and returns the very last item in the list.
* **Sorting**: 
  * `list.sort()`: Modifies the original list in-place. By default, it sorts alphabetically, but capital letters take priority over lowercase. You can pass `key=str.lower` to ignore casing.
  * `sorted(list)`: Leaves the original list alone and returns a brand-new sorted list.
* **Copying**: Using the slice operator `[:]` is a quick way to create a shallow copy of a list.

```python
# --- Creation & Access ---
cats = ["XiaoHei", "XiaoBai", 1, True]
dogs = []

print(f"Is XiaoHei in cats? {'XiaoHei' in cats}")
print(f"First element: {cats[0]}")
print(f"Last element: {cats[-1]}")
print(f"Slice [1:3]: {cats[1:3]}\n")

# --- Modification ---
# Lists are mutable, so we can change elements directly
cats[2] = 2  
print(f"Modified cats: {cats}")
print(f"Length of cats: {len(cats)}\n")

# --- Adding Elements ---
cats.append("new cat")           # Adds a single element to the very end
dogs.extend(["Fido", "Spot"])    # Merges another list
dogs += ["Buddy"]                # Concatenation (functions similarly to extend)
dogs.insert(0, "Rover")          # Inserts at a specific index

cats[3:3] = ["Cat 1", "Cat 2"] 
print(f"Cats after additions: {cats}")
print(f"Dogs after additions: {dogs}\n")

# --- Removing Elements ---
dogs.remove("Fido")     # Removes the first exact match of a value
popped_cat = cats.pop() # Removes and returns the last element

print(f"Dogs after removal: {dogs}")
print(f"Cats after popping '{popped_cat}': {cats}\n")

# --- Sorting & Copying ---
items = ["apple", "banana", "cherry", "Cherry", "banana"]

# In-place sort (capital letters take priority by default)
items.sort() 
print(f"Standard sort: {items}")

# In-place sort, ignoring case
items.sort(key=str.lower) 
print(f"Case-insensitive sort: {items}\n")

# Create a shallow copy via slicing
items_copy = items[:] 

# sorted() returns a completely new list, leaving the original intact
print(f"New sorted list: {sorted(items_copy, key=str.lower)}")
```

## 21. Tuples
* **Properties**: Tuples are ordered, allow duplicate elements, but are **immutable** (they cannot be changed, added to, or removed from after creation).
* **Syntax**: Created using parentheses `()` instead of square brackets `[]`.
* **Single-Element Tuples**: To create a tuple with only one item, you *must* include a trailing comma (e.g., `("New Cat",)`). Without the comma, Python treats it as a standard variable enclosed in mathematical parentheses.
* **Sorting**: Because tuples cannot be changed, you cannot use `.sort()`. However, you can pass a tuple into the built-in `sorted()` function, which will return a brand-new, sorted **list**.
* **Concatenation**: Since you cannot `.append()` to a tuple, the only way to add data is to use the `+` operator to merge two tuples into a completely new variable.

```python
# --- Creation & Access ---
names = ("XiaoHei", "XiaoBai", "Oscar")

print(f"First element: {names[0]}")
# Slice from index 0 to 1 (2 is exclusive)
print(f"Slice [0:2]: {names[0:2]}") 
print(f"Length of tuple: {len(names)}\n")

# --- Methods & Membership ---
# Get the numerical index position of a specific element
print(f"Index of 'XiaoBai': {names.index('XiaoBai')}")
print(f"Is XiaoHei in the tuple? {'XiaoHei' in names}\n")

# --- Sorting & Concatenation ---
# sorted() evaluates the tuple, but returns a brand-new LIST
print(f"Sorted list from tuple: {sorted(names)}\n") 

# Tuples are immutable. To "add" to one, you must concatenate it into a new tuple.
# Note the required trailing comma for a single-element tuple!
new_names = names + ("New Cat",) 
print(f"New concatenated tuple: {new_names}")
```

## 22. Dictionaries
* **Properties**: Dictionaries store data in `Key: Value` pairs. They are mutable, and keys must be strictly unique. (Note: As of Python 3.7+, dictionaries officially remember the order in which items were inserted).
* **Accessing Data**: 
  * Bracket notation `cat["name"]` will crash the program with a `KeyError` if the key does not exist.
  * The `.get("key")` method is much safer. It returns `None` (or a custom default value) if the key is missing, preventing crashes.
* **Membership (`in`)**: Using the `in` operator on a dictionary *only searches the keys*. To search the values, you must explicitly use `.values()`.
* **Dictionary Views**: Methods like `.keys()`, `.values()`, and `.items()` return dynamic view objects. If the dictionary changes, these views update automatically. It is common to cast them to lists (e.g., `list(cat.keys())`) to manipulate the data.

```python
# --- Creation & Updating ---
cat = {"name": "XiaoHei", "age": 1, "color": "white"}

# Updating an existing key
cat["name"] = "XiaoBai" 
print(f"Updated dictionary: {cat}\n")

# Adding a brand new key-value pair
cat["favorite_food"] = "fish" 


# --- Accessing Data ---
print(f"Direct access (Name): {cat['name']}")

# get() is safer because it won't crash if the key is missing
print(f"Safe access (Age): {cat.get('age')}")

# get() allows you to set a fallback value if the key doesn't exist
print(f"Safe access with default (Breed): {cat.get('breed', 'Unknown')}\n")


# --- Removing Data ---
# pop() removes a specific key and returns its value
removed_color = cat.pop("color") 
print(f"Removed color: {removed_color}")

# popitem() removes and returns the last inserted key-value pair as a tuple
last_item = cat.popitem() 
print(f"Popped last item: {last_item}")

# del permanently deletes a key without returning anything
del cat["age"]
print(f"Dictionary after deletions: {cat}\n")


# --- Views & Membership ---
cats = {"cat1": "XiaoHei", "cat2": "XiaoBai"}

# TRAP: 'in' only checks KEYS, not values!
print(f"Is 'XiaoHei' a KEY in cats? {'XiaoHei' in cats}") # Returns False

# CORRECT: How to check if a value exists
print(f"Is 'XiaoHei' a VALUE in cats? {'XiaoHei' in cats.values()}\n") # Returns True

# --- Extracting Views ---
print(f"Keys view: {cats.keys()}")
print(f"Keys as a list: {list(cats.keys())}\n")

print(f"Values view: {cats.values()}")
print(f"Values as a list: {list(cats.values())}\n")

print(f"Items (Key-Value tuples): {cats.items()}")
print(f"Total pairs in dictionary: {len(cats)}\n")

# --- Copying ---
# Create a shallow copy to modify without altering the original
cats_copy = cats.copy()
```

## 23. Sets
* **Properties**: Sets are unordered, unindexed, and contain **strictly unique** elements. If you try to add a duplicate, Python simply ignores it.
* **Deduplication**: Passing a list into the `set()` constructor is the fastest, most Pythonic way to remove all duplicate values from that list.
* **Mathematical Operations**: Sets support standard mathematical evaluations:
  * `&` (Intersection): Elements that exist in *both* sets.
  * `|` (Union): Combines all unique elements from *both* sets.
  * `-` (Difference): Elements in the first set that are *not* in the second set.
* **Comparisons**: You can use `<` and `>` to check if a set is a strict subset (fully contained within another) or a superset (fully contains another).

```python
# --- Creation & Deduplication ---
# Notice how the second "red" is automatically removed
set1 = {"red", "green", "blue", "red"} 
set2 = {"red"}
set3 = {"yellow", "green", "blue"}

print(f"set1 after deduplication: {set1}\n")

# --- Set Operations ---
intersection = set1 & set2 # Elements in BOTH sets
print(f"Intersection (set1 & set2): {intersection}")

union = set2 | set3        # ALL unique elements from both sets
print(f"Union (set2 | set3): {union}")

difference = set1 - set3   # Elements in set1 that are NOT in set3
print(f"Difference (set1 - set3): {difference}\n")

# --- Subsets & Supersets ---
superset = set1 > set2 # Does set1 fully contain set2?
print(f"Is set1 a superset of set2? {superset}")

subset = set2 < set1   # Is set2 fully contained within set1?
print(f"Is set2 a subset of set1? {subset}\n")

# --- Type Conversion & Membership ---
# Converting a set back into a list is a common practice after deduplicating
print(f"Set converted to list: {list(set1)}") 

# Membership check
print(f"Is 'red' in set1? {'red' in set1}")
```

## 24. Functions
* **Definition**: Functions are reusable blocks of code defined using the `def` keyword. They only run when explicitly called.
* **Parameters vs. Arguments**: 
  * *Parameters* are the variables listed inside the parentheses in the function definition.
  * *Arguments* are the actual data values passed into the function when you call it.
* **Default Parameters**: You can assign a default value to a parameter (e.g., `name="stranger"`). If the caller doesn't provide an argument, the function falls back to the default.
* **Mutability & Scope (Pass by Object Reference)**: 
  * **Immutable** objects (integers, strings, tuples) cannot be altered inside a function. Attempting to change them just creates a temporary local variable.
  * **Mutable** objects (lists, dictionaries, sets) *can* be modified inside a function. Changes made to them will affect the original variable globally.
* **Return Values**: Functions can send data back to the caller using `return`. If you return multiple comma-separated values, Python automatically packs them into a Tuple.

```python
# --- Basic Functions & Arguments ---
def hello():
    print("Hello!")
    
hello() # Call the function to execute its code

def hello_name(name):
    print(f"Hello, {name}!")
    
hello_name("XiaoHei") # Call with one argument

def hello_name_and_age(name, age):
    print(f"Hello, {name}! You are {age} years old.")
    
hello_name_and_age("XiaoHei", 20) # Call with multiple arguments


# --- Default Parameters ---
def hello_name_empty(name="stranger"):
    print(f"Hello, {name}!")

hello_name_empty()          # Falls back to default ("stranger")
hello_name_empty("XiaoBai") # Overrides default with argument


# --- Mutability & Variable Scope ---
def change_integer(value):
    value = 2 # This only creates a local variable; it does not alter the original

val = 1
change_integer(val)
print(f"\nInteger after function: {val}") # Remains 1 (Immutable)

def change_dictionary(value):
    value["name"] = "XiaoBai" # This modifies the actual object in memory

val_dict = {"name": "XiaoHei"}
change_dictionary(val_dict)
print(f"Dictionary after function: {val_dict}\n") # Changes to XiaoBai (Mutable)


# --- Return Statements ---
def hello_return_1(name):
    # If the string is empty, it evaluates to False
    if not name: 
        return "No name provided"
    return f"Hello, {name}!"

# We wrap these in print() so the returned string displays in the notebook
print(hello_return_1("")) 
print(hello_return_1("XiaoHei")) 

def hello_return_2(name):
    return name, "cat", 1

# Returning multiple values automatically generates a Tuple
print(f"\nMultiple returns: {hello_return_2('XiaoHei')}")
```

## 25. Variable Scope (Global vs. Local)
* **Scope**: The specific region of your code where a variable is recognized and accessible.
* **Global Variables**: Defined outside of any function. They can be read by any function anywhere in the script.
* **Local Variables**: Defined *inside* a function. They only exist while the function is actively running and are completely destroyed from memory the moment the function finishes.
* **Shadowing**: If you create a local variable with the exact same name as a global variable, the function will prioritize its own local version. It does *not* overwrite the global variable.
* **The `global` Keyword**: If you explicitly need to alter a global variable from inside a local function, you must declare it using the `global` keyword first.

```python
# --- Global Scope ---
age = 20 # Global variable, accessible everywhere

def read_global():
    # Reads the global 'age' because there is no local 'age' defined inside
    print(f"Inside read_global(): {age}") 

def create_local():
    # Creates a brand new local variable that "shadows" the global one
    age = 30 
    print(f"Inside create_local(): {age}")

def modify_global():
    # Explicitly tells Python we want to alter the global variable, not make a new one
    global age
    age = 99
    print(f"Inside modify_global(): {age}")

# --- Execution & Proof ---
print(f"Initial global age: {age}\n")

read_global()
create_local()

# Proof that the global variable was untouched by the local one
print(f"\nGlobal age after create_local(): {age} (Unchanged)")

modify_global()

# Proof that the 'global' keyword successfully altered the outer variable
print(f"\nGlobal age after modify_global(): {age} (Changed!)")
```

## 26. Nested Functions and `nonlocal`
* **Encapsulation**: You can define functions inside of other functions. These inner functions are "hidden" from the global scope and can only be called by the outer function. This is great for creating single-use helper functions.
* **Enclosing Scope**: An inner function can freely *read* variables created by its outer function.
* **The `nonlocal` Keyword**: If an inner function needs to *modify* a variable from the outer function, you must declare it as `nonlocal`. This is functionally very similar to the `global` keyword, but it specifically targets the immediate enclosing scope rather than the entire script.

```python
# --- Basic Nested Function ---
def talk(phrase):
    
    # This inner helper function cannot be called from outside talk()
    def say(word):
        print(word)
        
    words = phrase.split(' ')
    
    for word in words:
        say(word)
        
# Executing the outer function
talk("Hello, how are you?") 
print() # Adds a blank line for readability


# --- Modifying Outer Variables (nonlocal) ---
def count():
    # Renamed to 'current_count' so it doesn't clash with the function name 'count'
    current_count = 0 
    
    def increment():
        # Tells Python to modify the variable in the outer function, not create a new local one
        nonlocal current_count 
        current_count += 1
        return current_count
    
    # Call the inner function twice to prove the variable's state is changing
    increment()
    increment()
    
    # Return the final modified variable
    return current_count

# --- Execution ---
print(f"Final count after increments: {count()}")
```

## 27. Closures
* **Definition**: A closure is a nested function that captures and "remembers" the variables from its enclosing scope, even after the outer function has completely finished executing.
* **State Retention**: Closures act like functions equipped with their own private memory. This makes them an excellent, lightweight alternative to creating an entire Class just to track a few pieces of changing data.
* **The Three Criteria**: To create a closure, you must have:
  1. A nested (inner) function.
  2. The inner function must use a variable defined in the outer function.
  3. The outer function must **return** the inner function itself (without calling it).

```python
# --- Defining a Closure ---
def count_closure():
    current_count = 0 
    
    def increment():
        nonlocal current_count 
        current_count += 1
        return current_count
    
    # Notice there are no parentheses here! 
    # We are returning the function object itself, not its execution result.
    return increment 


# --- Executing a Closure ---
# counter_a now holds the increment() function, along with its own private 'current_count'
counter_a = count_closure() 

print(f"First call:  {counter_a()}") 
print(f"Second call: {counter_a()}") 
print(f"Third call:  {counter_a()}\n") 

# --- Independent State ---
# If we create a second closure, it gets its own completely separate memory state
counter_b = count_closure()

print(f"New closure call: {counter_b()}")
print(f"Original closure call: {counter_a()}")
```

## 28. Objects and the `id()` Function
* **The Golden Rule**: In Python, absolutely everything is an object. Even fundamental primitive data types like integers, strings, and booleans are objects under the hood.
* **Attributes vs. Methods**: 
  * **Attributes**: Properties or data stored *inside* the object. They are accessed without parentheses (e.g., `age.real`).
  * **Methods**: Functions bound to the object that perform actions. They are accessed with parentheses (e.g., `items.append()`).
* **Identity (`id()`)**: The `id()` function returns the unique, physical memory address of an object. This is what the `is` operator checks against when comparing two variables.

```python
# --- Primitive Objects (Integers) ---
age = 8

# Attributes (No parentheses)
print(f"Real part of integer 8: {age.real}") 
print(f"Imaginary part of integer 8: {age.imag}") 

# Method (Requires parentheses)
# 8 in binary is 1000, which takes 4 bits
print(f"Bits needed to represent 8: {age.bit_length()}\n") 


# --- Complex Objects & Memory (Lists) ---
items = [1, 2]

items.append(3)     # Method call (modifies the object)
items.pop()         # Method call (modifies the object and returns the last item)

print(f"Current list state: {items}")

# Every time you run this script, this ID will change because 
# the OS assigns it a new location in your computer's RAM!
print(f"Unique Memory Address (ID): {id(items)}")
```

## 29. Loops (`while` and `for`)
* **`while` Loops**: Execute a block of code repeatedly as long as a specific condition remains `True`. You must manually update the condition inside the loop, or it will run infinitely and crash the program.
* **`for` Loops**: Iterate directly over a sequence (like a list, tuple, dictionary, or string). The loop automatically stops when it reaches the end of the sequence.
* **`range(stop)`**: A built-in function that generates a sequence of numbers from `0` up to, but not including, the `stop` number. Highly useful for running a loop a specific number of times.
* **`enumerate(iterable)`**: A built-in function that takes an iterable and returns both the numerical `index` position and the `item` itself on every pass. This is the most Pythonic way to track index numbers in a `for` loop.

```python
# --- While Loops ---
condition = True

while condition:
    print("The condition is true.")
    condition = False # Update the condition to break the loop

count = 0
while count < 4: # Reduced to 4 to keep the notebook output concise
    print(f"Count is: {count}")
    count += 1 # Increment the count to eventually break the loop
    
print("While loop finished.\n")


# --- For Loops ---
items = ["apple", "banana", "cherry"]

# Iterating directly over the items
print("Standard For Loop:")
for item in items:
    print(item) 
    
# Using range() to generate numbers 0 through 4 (5 is exclusive)
print("\nRange Loop (range(5)):")
for num in range(5):
    print(num, end=" ") # end=" " prints them on the same line
print("\n")


# --- Enumerate ---
# enumerate() unpacks both the index and the value simultaneously
print("Enumerate Loop:")
for index, item in enumerate(items):
    print(f"Index {index} holds the value '{item}'")
```

## 30. Loop Control (`break`, `continue`, `pass`)
* **`break`**: Immediately and completely terminates the loop. Python stops iterating and moves on to the next line of code outside the loop.
* **`continue`**: Skips the rest of the code inside the current iteration and instantly jumps back to the top of the loop to evaluate the next item.
* **`pass`**: A null operation that does absolutely nothing. It is used as a placeholder to prevent Python from throwing an `IndentationError` when you have a loop or function that you intentionally want to leave empty for now.

```python
# --- The 'break' Statement ---
items = [1, 2, 3, 4, 5]

print("Break Example (Stops at 3):")
for item in items:
    if item == 3:
        break # The loop dies the moment it hits 3
    print(item) 
    
print("\n")


# --- The 'continue' Statement ---
print("Continue Example (Skips 3):")
for item in items:
    if item == 3:
        continue # Skips 3, but keeps the loop alive for 4 and 5
    print(item)

print("\n")


# --- The 'pass' Statement ---
print("Pass Example:")
for item in items:
    if item == 3:
        pass # Does nothing, acts as a temporary placeholder
    print(item)
```

## 31. Classes and Inheritance (OOP)
* **Classes vs. Instances**: A `class` is the blueprint (e.g., the concept of a Cat). An *instance* is the actual, physical object built from that blueprint (e.g., a specific cat named XiaoHei).
* **The `__init__` Method**: This is the "constructor." It is a special method that runs automatically the exact moment you create a new instance, used to set up the initial attributes.
* **The `self` Keyword**: Inside a class, `self` represents the specific instance being created or modified. It is required as the first parameter in almost all instance methods.
* **Inheritance**: By passing another class into the parentheses of a new class (e.g., `class Cat(Animal):`), the child class automatically inherits all the methods and attributes of the parent class.

```python
# --- Parent Class ---
class Animal:
    def sleeping(self):
        print("The animal is sleeping.")

# --- Child Class (Inherits from Animal) ---
class Cat(Animal):
    # The constructor method initializes the attributes of the new object
    def __init__(self, name, age, color):
        self.name = name
        self.age = age
        self.color = color

    # An instance method specific to the Cat class
    def meow(self):
        print(f"{self.name} says: Meow!")

# --- Instantiation ---
# PEP 8 standard: Variable names should be lowercase/snake_case
xiao_hei = Cat("XiaoHei", 1, "black") 

# --- Accessing Attributes & Methods ---
print(f"What type of object is xiao_hei? {type(xiao_hei)}\n")

# Accessing a specific attribute
print(f"Color attribute: {xiao_hei.color}") 

# Calling a method belonging to the Cat class
xiao_hei.meow() 

# Calling a method inherited from the parent Animal class
xiao_hei.sleeping()
```

## 34. Lambda Functions (Anonymous Functions)
* **Definition**: A lambda is a small, anonymous function that is defined without a name. 
* **Syntax**: `lambda arguments: expression`. They can take any number of arguments, but they can only ever contain **one single expression** (no loops, no multiple statements). They automatically return the result of that expression.
* **PEP 8 Convention**: You should *never* assign a lambda to a variable identifier (e.g., `multiply = lambda x, y: x * y`). If a function needs a name, use standard `def` syntax.
* **Primary Use Case**: Lambdas are best used as temporary, inline functions passed directly into higher-order functions (like `map()`, `filter()`, or `sort()`).

```python
# --- Basic Lambda Syntax ---
# A lambda that takes one number and returns it multiplied by 2.
# We wrap it in parentheses and pass (5) immediately to execute it on the spot.
result = (lambda num: num * 2)(5)
print(f"Inline execution result: {result}")


# --- The PEP 8 Anti-Pattern ---
# This works, but it violates Python's official style guide.
# If it has a name, it should be a 'def' function!
multiply = lambda a, b: a * b 
print(f"Multiply variable result: {multiply(3, 4)}\n")


# --- The Real-World Use Case (Sorting) ---
# Imagine we have a list of tuples representing (Name, Age)
people = [("XiaoHei", 1), ("Oscar", 5), ("XiaoBai", 2)]

# We want to sort them by AGE (the second element at index 1), not alphabetically by name.
# We pass a temporary lambda directly into the 'key' argument!
people.sort(key=lambda person: person[1])

print(f"Sorted by age using a lambda: {people}")
```

## 35. Map, Filter, and Reduce
* **Functional Programming**: These three functions allow you to process collections of data iteratively without writing standard `for` loops. They pair exceptionally well with `lambda` functions.
* **`map(function, iterable)`**: Applies a specific function to every single item in a list and returns a new map object.
* **`filter(function, iterable)`**: Runs every item through a function that returns `True` or `False`. It only keeps the items that evaluate to `True`.
* **`reduce(function, iterable, [initial])`**: imported from the `functools` module. It rolls up a list into a single value by applying a function cumulatively. 
  * *Warning:* If you are reducing complex data (like a list of tuples or dictionaries), always provide an `initial` starting value (like `0`) so the accumulator doesn't get confused by the changing data types.
* **Casting**: `map` and `filter` return memory objects, not standard lists. You must wrap them in `list()` to see the actual data.

```python
# --- Map ---
numbers = [1, 2, 3, 4, 5]

def double(x):
    return x * 2

# Using a standard function
result = map(double, numbers) 
print(f"Standard map: {list(result)}")

# Using an inline lambda function
map_result = map(lambda x: x * 2, numbers) 
print(f"Lambda map: {list(map_result)}\n")


# --- Filter ---
def is_even(n):
    return n % 2 == 0

# Using a standard function
even_result = filter(is_even, numbers) 
print(f"Standard filter: {list(even_result)}")

# Using an inline lambda function
filter_result = filter(lambda x: x % 2 == 0, numbers) 
print(f"Lambda filter: {list(filter_result)}\n")


# --- Reduce ---
from functools import reduce

expenses = [
    ('Cat Food', 80),
    ('Cat Toys', 120),
    ('Vet Bill', 50)
]

# Standard for-loop approach
total_sum = 0
for expense in expenses:
    total_sum += expense[1]
print(f"For-loop total: {total_sum}")

# Reduce approach
# We pass '0' as the initial value. 'a' starts as 0, and 'b' represents each tuple.
reduce_sum = reduce(lambda a, b: a + b[1], expenses, 0) 
print(f"Reduce total: {reduce_sum}")
```

## 36. Recursion
* **Definition**: Recursion occurs when a function calls *itself* from within its own code to solve a smaller piece of the same problem.
* **The Two Rules of Recursion**:
  1. **The Base Case**: The condition where the function stops calling itself and finally returns a concrete value. Without this, the function will run forever.
  2. **The Recursive Case**: The part where the function calls itself, modifying the arguments so that it gets closer and closer to the Base Case.
* **The Call Stack**: Every time a function calls itself, Python pauses the current function and puts it on a "stack" in your computer's memory. Once the base case is reached, the stack "unwinds," returning values back up the chain.
* **Recursion Limit**: Python has a built-in safety net (usually 1,000 calls). If your recursion goes deeper than that without hitting a base case, Python throws a `RecursionError` to prevent your computer from freezing.

```python
# --- Recursive Function ---
def factorial(n):
    # 1. The Base Case: Handles 1, but also protects against 0 and negative inputs
    if n <= 1:
        print(f"Base case reached! Returning 1")
        return 1
    
    # 2. The Recursive Case
    print(f"Pausing factorial({n}) to calculate {n} * factorial({n - 1})...")
    
    # The function calls itself, getting closer to n <= 1
    result = n * factorial(n - 1)
    
    print(f"Unwinding stack: factorial({n}) resolved to {result}")
    return result

# --- Execution ---
print("--- Starting Recursion ---")
final_answer = factorial(5)

print(f"\nFinal Result: 5! = {final_answer}")
```

## 37. Decorators
* **Definition**: A decorator is a function that takes another function as an argument and extends its behavior without permanently modifying the original function's code.
* **Syntactic Sugar (`@`)**: Placing `@decorator_name` directly above a function definition is just a shortcut. Under the hood, Python is doing this: `hello = log_time(hello)`.
* **The `*args` and `**kwargs` Requirement**: If you want your decorator to work on *any* function, the inner wrapper function must accept `*args` (positional arguments) and `**kwargs` (keyword arguments) and pass them down to the original function. Otherwise, decorating a function that takes parameters will cause a crash.

```python
# --- Basic Decorator ---
def log_time(func):
    def wrapper():
        print("--- Logging: Before execution ---")
        val = func() 
        print("--- Logging: After execution ---\n")
        return val
        
    return wrapper

@log_time
def hello():
    print("Hello, world!")

# Executes the wrapper, which in turn executes hello()
hello() 


# --- Production-Grade Decorator ---
def robust_log(func):
    # *args and **kwargs allow this wrapper to accept ANY arguments
    def wrapper(*args, **kwargs):
        print(f"Calling '{func.__name__}' with arguments: {args}")
        
        # Pass the arguments down into the original function
        result = func(*args, **kwargs) 
        
        print(f"Finished '{func.__name__}'\n")
        return result
        
    return wrapper

@robust_log
def add_numbers(a, b):
    return a + b

# The decorator seamlessly handles the arguments (5, 10)
sum_result = add_numbers(5, 10)
print(f"Final Sum: {sum_result}")
```

## 38. Docstrings (Documentation Strings)
* **Definition**: Docstrings are multi-line strings (`""" """`) used to explain the purpose of a module, class, or function. Unlike standard comments (`#`), docstrings are officially attached to the object and can be read by Python at runtime.
* **PEP 257 Conventions**:
  * **One-liners**: Should use the imperative mood ("Do this", not "Does this") and end with a period.
  * **Multi-liners**: Should consist of a one-line summary, a blank line, and then a more detailed description.
* **Placement**: They must be the very first statement directly beneath a `def` or `class` declaration, or at the absolute top of a `.py` file for a module.
* **Reading Docstrings**:
  * `help(object)`: Prints a highly readable, formatted manual to the console.
  * `object.__doc__`: Returns the raw string of the documentation, which is useful if you need to manipulate or print the text yourself.

```python
"""Cat module.

This module provides basic utilities for managing felines, 
including the primary Cat class and math helpers.
"""

def increment(x):
    """Increment a number by one."""
    return x + 1

class Cat:
    """A class representing a domestic cat."""
    
    def __init__(self, name, age):
        """Initialize a new cat with a name and age."""
        self.name = name
        self.age = age
        
    def meow(self):
        """Make the cat vocalize."""
        print(f"{self.name} says: Meow!")

# --- Accessing Documentation ---

# 1. The help() function generates a beautiful, formatted manual
print("--- Using help() ---")
help(Cat) 

# 2. The .__doc__ attribute returns just the raw string
print("--- Using .__doc__ ---")
print(f"Function docstring: {increment.__doc__}")
print(f"Method docstring: {Cat.meow.__doc__}")
```

* **Definition**: Annotations allow you to explicitly declare what data type a variable should be, or what a function should accept and return.
* **Syntax**: Use a colon (`:`) for variables and parameters, and a right arrow (`->`) for function return types.
* **The Runtime Trap**: Python does **not** enforce type annotations when the script is running. They are strictly "hints" meant for other developers, your IDE, and static analysis tools (like `mypy`) to catch potential bugs before execution.
* **Collections (Python 3.9+)**: You can hint exactly what goes inside data structures using standard brackets, such as `list[str]` for a list of strings, or `dict[str, int]` for a dictionary with string keys and integer values.

```python
# --- Basic Annotations ---
# 'num' expects an integer, and the function returns an integer
def increment(num: int) -> int:
    return num + 1

# Variable annotation
count: int = 0


# --- The Runtime Trap ---
# Because Python does not enforce hints at runtime, passing a float here 
# will NOT crash the program, but your code editor will flag it as a warning.
print(f"Passing a float to an int parameter: {increment(5.5)}\n")


# --- Complex Annotations ---
# Annotating a list that should only contain strings
cats: list[str] = ["XiaoHei", "XiaoBai"]

# Annotating a dictionary (Keys are strings, Values are integers)
cat_ages: dict[str, int] = {
    "XiaoHei": 1,
    "XiaoBai": 2
}

print(f"Annotated List: {cats}")
print(f"Annotated Dictionary: {cat_ages}")
```

## 40. Exceptions and Error Handling
* **Purpose**: Exceptions allow your program to handle errors gracefully instead of crashing entirely.
* **The Flow**:
  * `try`: The block of code to be tested for errors.
  * `except`: The block of code that executes *only* if an error is caught in the `try` block. You should specify exact error types (like `ZeroDivisionError`) rather than a blank `except:`.
  * `else`: Executes *only* if the `try` block succeeds without any errors.
  * `finally`: Executes absolutely no matter what happens. This is typically used for cleanup tasks, like closing a database connection or a file.
* **`raise`**: The keyword used to manually trigger an exception if a specific condition is met (or failed).
* **Custom Exceptions**: You can build your own error types by creating a class that inherits from Python's base `Exception` class.

```python
# --- The Complete Exception Flow ---
print("--- Standard Flow ---")
try:
    # Try changing the 0 to a 2 to see the 'else' block execute!
    result = 10 / 0 
except ZeroDivisionError:
    print("Error Caught: Cannot divide by zero!")
else:
    # This ONLY runs if no exceptions were raised
    print(f"Success! The result is {result}")
finally:
    # This runs regardless of success or failure
    print("Finally block executing (Simulating cleanup)...\n")


# --- Manually Raising Exceptions ---
print("--- Raising Exceptions ---")
try:
    # We can manually trigger an error with a custom message
    raise ValueError("Invalid user input detected.")
except ValueError as error:
    # We can capture that message using 'as'
    print(f"Caught a manual error: {error}\n")


# --- Custom Exceptions ---
# Inherit from the base Exception class
class CatNotFoundException(Exception):
    """Exception raised when a requested cat is not found."""
    pass

print("--- Custom Exceptions ---")
try:
    # We can pass a specific message into our custom exception
    raise CatNotFoundException("XiaoBai is nowhere to be found!")
except CatNotFoundException as error:
    print(f"Custom Error Caught: {error}")
```

## 41. Context Managers (The `with` Statement)
* **Resource Management**: When your code connects to an external resource (like a file or a database), you are borrowing memory from the operating system. If you do not explicitly return that memory by calling `.close()`, you create a memory leak or risk corrupting the file.
* **The Old Way**: Traditionally, developers used `try...finally` blocks to guarantee that `.close()` would execute even if the program crashed while reading the file.
* **The `with` Statement**: This is the modern, Pythonic way to handle resources. It acts as a **Context Manager**, meaning it automatically calls `.close()` the exact moment the indented block of code finishes, even if an error occurs inside the block.

```python
# --- Setup: Creating a file so our code has something to read ---
filename = 'cats.txt'

# Writing to the file (using 'w')
with open(filename, 'w') as setup_file:
    setup_file.write("1. XiaoHei\n2. XiaoBai\n3. Oscar")


# --- The Old Way (try...finally) ---
print("--- Reading via try/finally ---")
try:
    # Open the file for reading ('r')
    file = open(filename, 'r') 
    content = file.read() 
    print(content) 
finally:
    # Ensure the file is closed no matter what happens above
    file.close() 
    print("(File closed manually)\n")


# --- The Modern Way (Context Manager) ---
print("--- Reading via 'with' statement ---")

# The file is assigned to the variable 'file' just for this indented block
with open(filename, 'r') as file:
    content = file.read() 
    print(content) 

# The moment we un-indent, Python automatically closes the file for us!
print("(File closed automatically)")
```

## 43. List Comprehensions
* **Definition**: A highly concise, "Pythonic" way to create a new list by applying an expression to an existing sequence. 
* **Syntax**: `[expression for item in iterable if condition]`
* **Why use it?**: It is generally faster and much easier to read than writing out a full `for` loop with `.append()`, and it completely replaces the need for `map()` and `filter()` in most situations.

```python
# --- Basic List Comprehension (Mapping) ---
numbers = [1, 2, 3, 4, 5]

# 1. The Comprehension Way (Pythonic)
numbers_power_2 = [n ** 2 for n in numbers] 
print(f"Comprehension: {numbers_power_2}")

# 2. The For-Loop Way (Traditional)
loop_power_2 = []
for n in numbers:
    loop_power_2.append(n ** 2) 

# 3. The Map/Lambda Way (Functional)
map_power_2 = list(map(lambda n: n ** 2, numbers)) 


# --- Conditional List Comprehension (Filtering + Mapping) ---
# We can add an 'if' statement at the end to filter items simultaneously!
# Let's square the number, but ONLY if the number is even:

even_squares = [n ** 2 for n in numbers if n % 2 == 0]

print(f"Conditional Comprehension (Even squares only): {even_squares}")
```

## 44. Polymorphism and Duck Typing
* **Definition**: Polymorphism translates to "many forms." In OOP, it is the ability of different classes to respond to the exact same method call in their own unique way.
* **Duck Typing**: Python's specific flavor of polymorphism. Unlike strict languages (like Java or C++), Python does not require classes to share a parent class to use polymorphism. Python only cares about the *behavior* (methods) of an object, not its actual *type*.
* **Why use it?**: It allows you to write incredibly flexible, reusable code. You can iterate through a mixed list of different objects and call the same method on all of them without needing to write a bunch of complex `if type == Cat:` statements.

```python
# --- Defining Distinct Classes ---
class Dog:
    def eat(self):
        print("The dog greedily scarfs down its food.")
        
class Cat:
    def eat(self):
        print("The cat daintily picks at its food.")

# --- Instantiation ---
oscar = Dog() 
xiao_hei = Cat() 


# --- Standard Method Calls ---
print("--- Standard Calls ---")
oscar.eat() 
xiao_hei.eat() 
print("\n")


# --- The Power of Polymorphism (Duck Typing) ---
print("--- Polymorphic Loop ---")
# We put completely different object types into a single list
animals = [oscar, xiao_hei, Dog(), Cat()]

# Python doesn't care what type 'animal' is on each pass.
# It just trusts that the object has an .eat() method and runs it!
for animal in animals:
    animal.eat()
```

## 45. Operator Overloading (Dunder Methods)
* **Definition**: Operator overloading allows you to define how standard Python operators (like `+`, `-`, `>`, `==`) behave when used on your own custom objects.
* **Dunder Methods**: Short for "Double Underscore." These are special built-in methods (also called "Magic Methods") that Python calls automatically behind the scenes. `__init__` is the most common example.
* **Common Overloads**:
  * `__eq__()`: Responds to `==` (Equals)
  * `__gt__()` / `__lt__()`: Responds to `>` and `<` (Greater than / Less than)
  * `__add__()`: Responds to `+` (Addition)
  * `__sub__()`: Responds to `-` (Subtraction)
  * `__mul__()`: Responds to `*` (Multiplication)
  * `__str__()`: Defines what prints out if you call `print(object)`

```python
# --- Overloading Comparison Operators ---
class Cat:
    def __init__(self, name, age):
        self.name = name
        self.age = age
        
    # Overloading the > (Greater Than) operator
    def __gt__(self, other):
        # The comparison inherently returns True or False
        return self.age > other.age
        
    # Overloading the == (Equals) operator
    def __eq__(self, other):
        return self.age == other.age

# --- Instantiation ---
xiao_bai = Cat("XiaoBai", 1) 
xiao_hei = Cat("XiaoHei", 2) 


# --- Execution ---
# Python sees the '>' symbol and automatically looks for the __gt__ method!
print(f"Is XiaoHei older than XiaoBai? {xiao_hei > xiao_bai}")

# Python sees the '==' symbol and automatically looks for the __eq__ method!
print(f"Are they the exact same age? {xiao_hei == xiao_bai}")
```

### Problem 001: Contains Duplicate (LeetCode 217)

### Problem Definition and Constraints
Given an integer array `nums`, return `true` if any value appears more than once in the array, otherwise return `false`.

* Constraints: 
  * 0 <= nums.length <= 10^5
  * -10^9 <= nums[i] <= 10^9

### Examples
* **Example 1:**
  * Input: `nums = [1, 2, 3, 3]`
  * Output: `true`
* **Example 2:**
  * Input: `nums = [1, 2, 3, 4]`
  * Output: `false`

### Brute Force Approach
The baseline strategy utilizes a nested loop configuration. The outer loop selects an element, and the inner loop scans the remainder of the array to check for an identical value. 
* Time Complexity: O(n^2) — Each element is compared against every other element in a worst-case scenario.
* Space Complexity: O(1) — No additional tracking data structures are allocated.

### Optimized Approach
To achieve an optimal runtime, we bypass the nested scans by introducing a Hash Set to track our traversal history. We perform a single linear scan of the input array. For each element, we check our history in constant time. If the element exists in the set, a duplicate is immediately confirmed. If it does not exist, the element is recorded into the set, and the loop proceeds.
* Time Complexity: O(n) — The array is traversed exactly once, and Hash Set lookups/insertions operate in amortized O(1) time.
* Space Complexity: O(n) — In the worst-case scenario where all elements are unique, the data structure stores all n elements.

```python
from typing import List

class Solution:
    
    # --- BRUTE FORCE APPROACH ---
    def hasDuplicate_brute_force(self, nums: List[int]) -> bool:
        # We iterate through the array using an index 'i' to select our anchor number.
        # We stop at len(nums) - 1 because the final element has nothing after it to compare against.
        for i in range(len(nums) - 1):
            
            # For each anchor number, we scan the remainder of the array using index 'j'.
            # We start 'j' at 'i + 1' to ensure we only check numbers ahead of our anchor,
            # preventing us from comparing the anchor to itself or redundantly checking past pairs.
            for j in range(i + 1, len(nums)):
                
                # If the anchor number exactly matches the comparison number, a duplicate is confirmed.
                if nums[i] == nums[j]:
                    return True
                    
        # If both nested loops exhaust completely without returning True, every element is unique.
        return False


    # --- OPTIMIZED APPROACH ---
    def hasDuplicate(self, nums: List[int]) -> bool:
        # Initialize an empty hash set. This acts as our historical memory.ww
        # We are explicitly trading O(n) space to completely eliminate the inner loop above.
        seen = set()
        
        # We perform a single, linear traversal of the input array.
        for num in nums:
            
            # Before doing anything, we query our history. 
            # Because 'seen' is a hash set, this look-up happens in constant O(1) time.
            if num in seen:
                # The number exists in our history. Terminate execution and return True immediately.
                return True
            
            # The number is not in our history. We must record it into the set 
            # so that subsequent numbers in the array can check against it.
            seen.add(num)
            
        # If the single loop finishes without triggering the condition, no duplicates exist.
        return False
```

### Problem 002: Valid Anagram (LeetCode 242)

### Problem Definition and Constraints
Given two strings `s` and `t`, return `true` if the two strings are anagrams of each other, otherwise return `false`. An anagram is a string that contains the exact same characters as another string, but the order of the characters can be different.

* Constraints: 
  * 1 <= s.length, t.length <= 5 * 10^4
  * `s` and `t` consist of lowercase English letters.

### Examples
* **Example 1:**
  * Input: `s = "racecar", t = "carrace"`
  * Output: `true`
* **Example 2:**
  * Input: `s = "jar", t = "jam"`
  * Output: `false`

### Brute Force Approach
The most intuitive baseline is to sort both strings alphabetically and compare them. If they are anagrams, their sorted versions will be identical. 
* Time Complexity: O(n log n + m log m) — Standard sorting algorithms operate in log-linear time, where n and m are the lengths of the strings.
* Space Complexity: O(n + m) — In Python, strings are immutable. Sorting them requires generating new lists in memory.

### Optimized Approach
To achieve optimal linear runtime, we bypass sorting entirely by using Hash Maps (or dictionaries) to tally the frequency of every character. If the strings are of different lengths, they cannot be anagrams, allowing for an immediate O(1) exit. If the lengths match, we iterate through both strings simultaneously, counting occurrences. Finally, we compare the two frequency maps.
* Time Complexity: O(n + m) — We iterate through both strings exactly once to populate our counts.
* Space Complexity: O(1) — Because the problem constraints specify lowercase English letters, the Hash Maps will never exceed 26 key-value pairs, making the memory usage mathematically constant regardless of string size.

```python
class Solution:
    
    # --- BRUTE FORCE APPROACH ---
    def isAnagram_brute_force(self, s: str, t: str) -> bool:
        # A foundational check: If the strings differ in length, they cannot be anagrams.
        # This prevents us from doing unnecessary, expensive sorting.
        if len(s) != len(t):
            return False
            
        # In Python, sorted() takes a string, breaks it into a list of characters,
        # sorts them alphabetically, and returns the new list.
        # We simply compare the two resulting lists for equality.
        return sorted(s) == sorted(t)

    # --- OPTIMIZED APPROACH ---
    def isAnagram(self, s: str, t: str) -> bool:
        # Step 1: Length Validation
        # If the strings have a different number of characters, they cannot be anagrams.
        if len(s) != len(t):
            return False
            
        # Step 2: Initialize Frequency Counters
        # We use two separate hash maps (dictionaries) to count character occurrences.
        count_s, count_t = {}, {}
        
        # Step 3: Populate the Counters
        # Because we already proved len(s) == len(t), we can use a single loop to iterate
        # through both strings simultaneously using their indices.
        for i in range(len(s)):
            
            # Extract the current character from both strings
            char_s = s[i]
            char_t = t[i]
            
            # Add the character to the map. 
            # The .get(char, 0) method is crucial: it asks, "What is the current count of this char?"
            # If the char doesn't exist yet, it safely defaults to 0 instead of throwing a KeyError.
            # Then we add 1 to whatever that count was.
            count_s[char_s] = 1 + count_s.get(char_s, 0)
            count_t[char_t] = 1 + count_t.get(char_t, 0)
            
        # Step 4: Compare the Maps
        # In Python, evaluating two dictionaries with '==' checks if they contain the exact 
        # same keys and if the values for those keys match perfectly.
        return count_s == count_t
```

### Problem 003: Two Sum (LeetCode 1)

### Problem Definition and Constraints
Given an array of integers `nums` and an integer `target`, return the indices `i` and `j` such that `nums[i] + nums[j] == target` and `i != j`. You may assume that every input has exactly one pair of indices that satisfy the condition. Return the answer with the smaller index first.

* Constraints:
  * 2 <= nums.length <= 1000
  * -10,000,000 <= nums[i] <= 10,000,000
  * -10,000,000 <= target <= 10,000,000
  * Only one valid answer exists.

### Examples
* **Example 1:**
  * Input: `nums = [3, 4, 5, 6], target = 7`
  * Output: `[0, 1]`
  * Explanation: `nums[0] + nums[1] == 7` (3 + 4 = 7)
* **Example 2:**
  * Input: `nums = [4, 5, 6], target = 10`
  * Output: `[0, 2]`

### Brute Force Approach
The baseline approach involves checking every possible pair of numbers. We set an anchor number using an outer loop, and use an inner loop to scan the rest of the array. If the two numbers sum to the target, we return their indices.
* Time Complexity: O(n^2) — In the worst-case scenario, we iterate through the array quadratically.
* Space Complexity: O(1) — No auxiliary data structures are created; we only store the two index pointers.

### Optimized Approach
We trade space for time by utilizing a Hash Map. As we perform a single linear traversal of the array, we calculate the mathematical difference needed to reach the target (`target - current_number`). We query the Hash Map to see if we have already encountered that required difference. If we have, we instantly return the pair of indices. If we have not, we store the current number and its index in the Hash Map to serve as a potential puzzle piece for future numbers.
* Time Complexity: O(n) — We traverse the input array exactly once. Hash Map lookups operate in constant O(1) time.
* Space Complexity: O(n) — In the worst-case scenario (the pair is at the very end of the array), we will store almost every number and its index in the Hash Map.

```python
from typing import List

class Solution:
    
    # --- BRUTE FORCE APPROACH ---
    def twoSum_brute_force(self, nums: List[int], target: int) -> List[int]:
        # Outer loop sets the anchor index 'i'
        for i in range(len(nums) - 1):
            
            # Inner loop scans the remaining numbers to the right of 'i'
            for j in range(i + 1, len(nums)):
                
                # Check if the pair sums to our target
                if nums[i] + nums[j] == target:
                    return [i, j]
                    
        return []

    # --- OPTIMIZED APPROACH ---
    def twoSum(self, nums: List[int], target: int) -> List[int]:
        # Initialize an empty Hash Map (dictionary).
        # We will use this to store data in a {value: index} format.
        prev_map = {}
        
        # We use enumerate() to give us BOTH the index (i) and the number (num) simultaneously.
        for i, num in enumerate(nums):
            
            # Calculate the exact number needed to solve the equation
            difference = target - num
            
            # Query our historical Hash Map in O(1) time
            if difference in prev_map:
                # We found the missing puzzle piece! 
                # We return the index of the previously seen number, and our current index 'i'
                return [prev_map[difference], i]
            
            # If the missing piece isn't in our history, we record the current number 
            # and its index into the map. It might be the piece a future number is looking for.
            prev_map[num] = i
            
        return []
```

### Problem 004: Group Anagrams (LeetCode 49)

### Problem Definition and Constraints
Given an array of strings `strs`, group all anagrams together into sublists. You may return the output in any order. An anagram is a string that contains the exact same characters as another string, but the order of the characters can be different.

* Constraints:
  * 1 <= strs.length <= 1000
  * 0 <= strs[i].length <= 100
  * `strs[i]` is made up of lowercase English letters.

### Examples
* **Example 1:**
  * Input: `strs = ["act","pots","tops","cat","stop","hat"]`
  * Output: `[["hat"],["act", "cat"],["stop", "pots", "tops"]]`
* **Example 2:**
  * Input: `strs = ["x"]`
  * Output: `[["x"]]`

### Brute Force Approach
The baseline approach involves sorting each individual string. Because anagrams share the exact same characters, sorting them yields the exact same base string (e.g., "cat" and "act" both sort to "act"). We use a Hash Map where the key is the sorted string, and the value is a list of the original, unsorted strings. 
* Time Complexity: O(m * n log n) — Where `m` is the total number of strings, and `n` is the length of the longest string. Sorting each string takes O(n log n) time.
* Space Complexity: O(m * n) — We must store the resulting grouped strings inside our Hash Map.

### Optimized Approach
To bypass the expensive sorting operation, we rely on character frequencies. Because the constraints guarantee only lowercase English letters, we can represent every string as an array of 26 integers, where each index tallies the frequency of a specific letter ('a' through 'z'). We convert this frequency array into an immutable Tuple to serve as our Hash Map key. Strings with identical character counts will generate identical Tuples and group perfectly into the same Hash Map bucket.
* Time Complexity: O(m * n) — We process every character in every string exactly once. The Hash Map lookups operate in O(1) time.
* Space Complexity: O(m * n) — The Hash Map must store all `m` strings of length `n` as values. The Tuple keys take a constant O(26) space, which simplifies away.

```python
from typing import List
from collections import defaultdict

class Solution:
    
    # --- BRUTE FORCE APPROACH ---
    def groupAnagrams_brute_force(self, strs: List[str]) -> List[List[str]]:
        # Using defaultdict(list) automatically creates an empty list for any new key.
        # This saves us from writing the standard: "if key not in dict: dict[key] = []"
        anagram_map = defaultdict(list)
        
        for s in strs:
            # sorted(s) breaks the string into a list of sorted characters: ['a', 'c', 't'].
            # "".join() stitches them back into a single string: "act".
            sorted_s = "".join(sorted(s))
            
            # Group the original string under its sorted signature
            anagram_map[sorted_s].append(s)
            
        # We only care about the groups, so we extract just the values from the dictionary
        return list(anagram_map.values())

    # --- OPTIMIZED APPROACH ---
    def groupAnagrams(self, strs: List[str]) -> List[List[str]]:
        anagram_map = defaultdict(list)
        
        for s in strs:
            # Initialize an array of 26 zeros to act as our character counter
            count = [0] * 26
            
            for char in s:
                # ord() returns the underlying integer ASCII value of a character.
                # Subtracting ord('a') mathematically maps 'a' to index 0, 'b' to index 1, etc.
                index = ord(char) - ord('a')
                count[index] += 1
                
            # Python dictionaries strictly forbid mutable data structures (like Lists) as keys.
            # We cast the count array to a Tuple (which is immutable) so it can be hashed.
            anagram_map[tuple(count)].append(s)
            
        return list(anagram_map.values())
```

### Problem 005: Top K Frequent Elements (LeetCode 347)

### Problem Definition and Constraints
Given an integer array `nums` and an integer `k`, return the `k` most frequent elements within the array. The test cases are generated such that the answer is always unique. You may return the output in any order.

* Constraints:
  * 1 <= nums.length <= 10^4
  * -1000 <= nums[i] <= 1000
  * 1 <= k <= number of distinct elements in nums.

### Examples
* **Example 1:**
  * Input: `nums = [1,2,2,3,3,3], k = 2`
  * Output: `[2,3]`
* **Example 2:**
  * Input: `nums = [7,7], k = 1`
  * Output: `[7]`

### Brute Force Approach
The baseline approach calculates the frequency of every element using a Hash Map. Once the frequencies are tallied, we sort the unique elements based on their mapped frequency values in descending order. Finally, we slice the first `k` elements from the sorted list.
* Time Complexity: $O(n \log n)$ — While counting takes $O(n)$ time, sorting the unique elements dominates the time complexity. In the worst case (all elements are unique), sorting takes $O(n \log n)$.
* Space Complexity: $O(n)$ — The Hash Map and the resulting sorted array both require memory proportional to the number of unique elements.

### Optimized Approach
To achieve linear time, we utilize a modified Bucket Sort algorithm. We first tally the frequencies in a Hash Map. Then, we initialize an array of empty sublists ("buckets") where the array index represents the frequency count. We map each unique number into the bucket corresponding to its frequency. Finally, we iterate through the bucket array in reverse (from highest frequency down to zero), appending elements to our result array until we reach `k` elements.
* Time Complexity: $O(n)$ — We iterate over the input array to count frequencies, then iterate over the unique keys to populate the buckets, and finally do a bounded reverse sweep of the buckets. All operations scale linearly.
* Space Complexity: $O(n)$ — The Hash Map stores up to $n$ elements, and the bucket array requires $n + 1$ slots.

```python
from typing import List

class Solution:
    
    # --- BRUTE FORCE APPROACH ---
    def topKFrequent_brute_force(self, nums: List[int], k: int) -> List[int]:
        count = {}
        # Step 1: Tally the frequencies
        for num in nums:
            count[num] = 1 + count.get(num, 0)
            
        # Step 2: Sort the unique keys based on their dictionary values
        # The key parameter tells sorted() to look at count.get (the frequency) to determine the order.
        # reverse=True ensures we get descending order (most frequent first).
        sorted_nums = sorted(count, key=count.get, reverse=True)
        
        # Step 3: Return exactly k elements
        return sorted_nums[:k]


    # --- OPTIMIZED APPROACH ---
    def topKFrequent(self, nums: List[int], k: int) -> List[int]:
        count = {}
        # We need an array of lists. The maximum possible frequency is the length of nums.
        # We make the array size len(nums) + 1 so the indices range from 0 to len(nums).
        freq_buckets = [[] for _ in range(len(nums) + 1)]
        
        # Step 1: Tally the frequencies
        for num in nums:
            count[num] = 1 + count.get(num, 0)
            
        # Step 2: Populate the buckets
        # For every number we counted, we place it into the bucket that matches its frequency
        for num, frequency in count.items():
            freq_buckets[frequency].append(num)
            
        # Step 3: Gather the top k elements
        result = []
        
        # Iterate backwards from the end of the array down to 0
        for i in range(len(freq_buckets) - 1, 0, -1):
            # Some buckets might contain multiple numbers (if they had the exact same frequency)
            for num in freq_buckets[i]:
                result.append(num)
                # Once our result array is exactly length k, we are completely done
                if len(result) == k:
                    return result
```

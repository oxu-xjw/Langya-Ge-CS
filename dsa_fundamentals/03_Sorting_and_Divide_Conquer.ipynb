## 1. O(N²) Sorting Algorithms (Brute Force)
Before using advanced techniques, it is important to understand the baseline algorithms. These are too slow for large datasets (e.g., sorting 10,000 items takes ~12 seconds) but require very little memory.

### Bubble Sort
* **Strategy:** Iterate through the list. Compare adjacent elements and swap them if they are in the wrong order. Repeat this $N$ times. Largest elements "bubble" to the top.
* **Complexity:** Time: $O(N^2)$ | Space: $O(1)$.

### Insertion Sort
* **Strategy:** Keep the left side of the array sorted. Pick the next element and scan backwards through the sorted left side, inserting the element into its correct sequential position.
* **Complexity:** Time: $O(N^2)$ | Space: $O(1)$.

```python
def bubble_sort(nums):
    nums = list(nums) # Create a copy to avoid mutating the original input
    
    # Outer loop dictates how many passes we make
    for j in range(len(nums) - 1):
        # Inner loop performs the adjacent comparisons
        for i in range(len(nums) - 1):
            if nums[i] > nums[i+1]:
                # Pythonic tuple unpacking to swap values in place
                nums[i], nums[i+1] = nums[i+1], nums[i]
                
    return nums

def insertion_sort(nums):
    nums = list(nums)
    
    for i in range(len(nums)):
        cur = nums.pop(i) # Remove the current element
        j = i - 1
        
        # Scan backward through the sorted portion of the list
        while j >= 0 and nums[j] > cur:
            j -= 1
            
        # Insert the element in its correct position
        nums.insert(j + 1, cur)
        
    return nums
```

## 2. Merge Sort (Divide & Conquer)
To drastically reduce sorting time, we use a recursive strategy called **Divide and Conquer**.

### Core Mechanic: Merge Sort
1. **Divide:** Split the array perfectly in half.
2. **Conquer:** Recursively pass the left half and right half back into the `merge_sort` function until the sub-lists are only 1 element long (a 1-element list is inherently sorted).
3. **Combine (`merge`):** Take two sorted arrays and stitch them together by comparing their first elements, popping the smaller one into a new array.

### Complexity Analysis
* **Time Complexity:** $O(N \log N)$. The array is halved $\log N$ times. At each level of division, we do $N$ total comparisons during the merge process. $N \times \log N = O(N \log N)$.
* **Space Complexity:** $O(N)$. We must allocate a new array during the `merge` process to hold the combined sorted elements.

```python
def merge(nums1, nums2):
    """Helper function to combine two sorted arrays into one sorted array."""
    merged = []
    i, j = 0, 0
    
    # Compare the front elements of both arrays
    while i < len(nums1) and j < len(nums2):
        if nums1[i] <= nums2[j]:
            merged.append(nums1[i])
            i += 1 
        else:
            merged.append(nums2[j])
            j += 1
    
    # Append any remaining elements if one array was exhausted before the other
    return merged + nums1[i:] + nums2[j:]

def merge_sort(nums):
    """Recursively divides the array and sorts via merging."""
    # Base Case: Lists of length 0 or 1 are already sorted
    if len(nums) <= 1:
        return nums
    
    # Divide
    mid = len(nums) // 2
    left = nums[:mid]
    right = nums[mid:]
    
    # Conquer (Recursion)
    left_sorted = merge_sort(left)
    right_sorted = merge_sort(right)
    
    # Combine
    return merge(left_sorted, right_sorted)
```

## 3. Quicksort
Merge Sort is extremely fast but uses $O(N)$ extra memory. Quicksort solves the same problem but sorts the elements **in-place** to save memory.

### Core Mechanic: Partitioning
1. **Choose a Pivot:** Pick an element (usually the last element) to act as the pivot point.
2. **Partition:** Rearrange the array so that every number strictly *less* than the pivot is moved to its left, and every number *greater* is moved to its right.
3. **Recurse:** The pivot is now locked into its final, correct sorted position. Recursively call `quicksort` on the left partition and right partition.

### Complexity Analysis
* **Time Complexity:** * **Average:** $O(N \log N)$. 
  * **Worst Case:** $O(N^2)$ if the array is already sorted and we continually pick the worst pivot (the very last element).
* **Space Complexity:** $O(1)$ additional space for the algorithm, plus $O(\log N)$ for the recursive call stack.

```python
def partition(nums, start=0, end=None):
    """Rearranges elements around a pivot and returns the pivot's final index."""
    if end is None:
        end = len(nums) - 1
    
    # Initialize pointers. 'l' searches for larger elements, 'r' searches for smaller
    l, r = start, end - 1
    
    while r > l:
        if nums[l] <= nums[end]:
            l += 1 # Left element is correct, move forward
        elif nums[r] > nums[end]:
            r -= 1 # Right element is correct, move backward
        else:
            # Both elements are on the wrong side of the pivot. Swap them.
            nums[l], nums[r] = nums[r], nums[l]
            
    # Place the pivot in its final correct location between the two partitions
    if nums[l] > nums[end]:
        nums[l], nums[end] = nums[end], nums[l]
        return l
    else:
        return end

def quicksort(nums, start=0, end=None):
    """Recursively partitions the array to sort it in-place."""
    if end is None:
        nums = list(nums) # Base copy to prevent mutating test inputs
        end = len(nums) - 1
    
    if start < end:  # Base case: Only sort if there are at least 2 elements
        pivot_idx = partition(nums, start, end)
        quicksort(nums, start, pivot_idx - 1)  # Sort left of pivot
        quicksort(nums, pivot_idx + 1, end)    # Sort right of pivot

    return nums
```

## 4. Generic Sorting via Custom Comparators
Sorting algorithms aren't just for numbers. By passing a custom `compare` function into our algorithm, we can teach it how to sort complex Python objects (like ordering `Notebook` objects by `likes` or `title`).

This allows us to write the $O(N \log N)$ sorting logic once, and reuse it for any data type.

```python
class Notebook:
    """A custom data object representing a web application notebook."""
    def __init__(self, title, username, likes):
        self.title = title
        self.username = username
        self.likes = likes
        
    def __repr__(self):
        return f'Notebook <"{self.username}/{self.title}", {self.likes} likes>'

# Sample Data
nb1 = Notebook('linear-regression', 'siddhant', 532)
nb2 = Notebook('pytorch-basics', 'aakashns', 373)
nb3 = Notebook('feedforward-nn', 'sonaksh', 94)
notebooks = [nb1, nb2, nb3]

# --- Custom Comparators ---
def compare_likes(nb1, nb2):
    """Comparator to sort in DECREASING order of likes."""
    if nb1.likes > nb2.likes:
        return 'lesser' # Return 'lesser' so higher likes get sorted to the front
    elif nb1.likes == nb2.likes:
        return 'equal'
    else:
        return 'greater'

def merge_generic(left, right, compare_func):
    """Generic merge function that defers comparison logic to the passed function."""
    i, j, merged = 0, 0, []
    while i < len(left) and j < len(right):
        result = compare_func(left[i], right[j])
        if result == 'lesser' or result == 'equal':
            merged.append(left[i])
            i += 1
        else:
            merged.append(right[j])
            j += 1
    return merged + left[i:] + right[j:]

def merge_sort_generic(objs, compare_func):
    """Generic Merge Sort."""
    if len(objs) < 2:
        return objs
    mid = len(objs) // 2
    return merge_generic(merge_sort_generic(objs[:mid], compare_func), 
                         merge_sort_generic(objs[mid:], compare_func), 
                         compare_func)

# Execution:
# sorted_notebooks = merge_sort_generic(notebooks, compare_likes)
```

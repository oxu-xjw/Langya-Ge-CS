### Problem 001: Reverse Linked List (LeetCode 206)

### Problem Definition and Constraints
Given the beginning of a singly linked list `head`, reverse the list, and return the new beginning of the list.

* Constraints:
  * 0 <= The length of the list <= 1000
  * -1000 <= Node.val <= 1000

### Examples
* **Example 1:**
  * Input: `head = [0,1,2,3]`
  * Output: `[3,2,1,0]`
* **Example 2:**
  * Input: `head = []`
  * Output: `[]`

### Brute Force Approach
The naive approach is to traverse the linked list from start to finish, appending every value into a standard array. Once the array is populated, we iterate through it backwards, creating brand new `ListNode` objects and linking them together to form a new reversed list. 
* Time Complexity: $O(n)$ — We traverse the list once to build the array, and once more to build the new linked list.
* Space Complexity: $O(n)$ — We create an array of size $n$ to hold the values, and allocate memory for $n$ brand new nodes.

### Optimized Approach (In-Place Pointer Reversal)
To achieve $O(1)$ space, we must manipulate the existing pointers without creating any new objects. We use two main pointers: `prev` (initialized to `None`) and `curr` (initialized to `head`). As we traverse the list, we temporarily store the next node in `nxt = curr.next` so we don't sever the rest of the list. We then reverse the current node's pointer by pointing it backwards (`curr.next = prev`). Finally, we slide both pointers forward one step (`prev = curr`, then `curr = nxt`). When `curr` hits `None`, `prev` will be resting on the new head of the list.
* Time Complexity: $O(n)$ — We traverse the linked list exactly once.
* Space Complexity: $O(1)$ — We only use three temporary pointer variables (`prev`, `curr`, `nxt`), regardless of the size of the list.

```python
from typing import Optional

# Definition for singly-linked list.
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

class Solution:
    
    # --- BRUTE FORCE APPROACH ---
    def reverseList_brute_force(self, head: Optional[ListNode]) -> Optional[ListNode]:
        if not head:
            return None
            
        # Step 1: Extract all values into a standard Python list
        values = []
        curr = head
        while curr:
            values.append(curr.val)
            curr = curr.next
            
        # Step 2: Build a brand new linked list in reverse
        # We start by creating the new head using the last value in our array
        new_head = ListNode(values[-1])
        curr = new_head
        
        # Loop through the rest of the array backwards
        for i in range(len(values) - 2, -1, -1):
            curr.next = ListNode(values[i])
            curr = curr.next
            
        return new_head


    # --- OPTIMIZED APPROACH ---
    def reverseList(self, head: Optional[ListNode]) -> Optional[ListNode]:
        # prev starts as None because the very first node (the old head) 
        # needs to point to None (becoming the new tail).
        prev = None
        curr = head
        
        while curr:
            # 1. SAVE: Temporarily store the rest of the list so we don't lose it
            nxt = curr.next
            
            # 2. REVERSE: Break the forward link and point it backward
            curr.next = prev
            
            # 3. ADVANCE: Shift both pointers forward for the next loop iteration
            prev = curr
            curr = nxt
            
        # Once curr falls off the end (becomes None), prev is sitting on the last node,
        # which is the brand new head of our reversed list.
        return prev
```

### Problem 002: Merge Two Sorted Linked Lists (LeetCode 21)

### Problem Definition and Constraints
You are given the heads of two sorted linked lists `list1` and `list2`. Merge the two lists into one sorted linked list and return the head of the new sorted linked list. The new list should be made up of nodes spliced together from the first two lists.

* Constraints:
  * 0 <= The length of each list <= 100
  * -100 <= Node.val <= 100
  * Both `list1` and `list2` are sorted in non-decreasing order.

### Examples
* **Example 1:**
  * Input: `list1 = [1,2,4], list2 = [1,3,5]`
  * Output: `[1,1,2,3,4,5]`
* **Example 2:**
  * Input: `list1 = [], list2 = [1,2]`
  * Output: `[1,2]`

### Brute Force Approach
The naive approach abandons the advantage of the lists already being sorted. We traverse both linked lists, extract all their integer values, and store them in a standard array. We then run a sorting algorithm on the array. Finally, we iterate through the sorted array, creating brand new `ListNode` objects to build the resulting list.
* Time Complexity: $O((n + m) \log(n + m))$ — Extracting takes linear time, but sorting the combined array dictates the overall time complexity.
* Space Complexity: $O(n + m)$ — We allocate memory for the array holding all values, and allocate memory for $n + m$ brand new nodes.

### Optimized Approach (Two Pointers & Dummy Node)
To achieve $O(1)$ space and $O(n+m)$ time, we restructure the existing pointers in-place. We instantiate a `Dummy` node to act as the anchor for our new list, and a `tail` pointer to track the end of our growing list. We compare the current nodes of `list1` and `list2`. The `tail` pointer connects to the node with the smaller value, and we advance that specific list's pointer forward. Once one of the lists is fully exhausted, we attach the entire remaining segment of the other list directly to the `tail`. We return `Dummy.next` as the true head of the merged list.
* Time Complexity: $O(n + m)$ — We traverse each node in both lists at most once. 
* Space Complexity: $O(1)$ — We reuse the existing nodes, allocating no scaling auxiliary memory. Only the constant space of the `Dummy` node and tracking pointers is required.

```python
from typing import Optional

# Definition for singly-linked list.
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

class Solution:
    
    # --- BRUTE FORCE APPROACH ---
    def mergeTwoLists_brute_force(self, list1: Optional[ListNode], list2: Optional[ListNode]) -> Optional[ListNode]:
        values = []
        
        # Extract from first list
        while list1:
            values.append(list1.val)
            list1 = list1.next
            
        # Extract from second list
        while list2:
            values.append(list2.val)
            list2 = list2.next
            
        # Sort the combined array
        values.sort()
        
        # Build new list using a dummy node
        dummy = ListNode()
        curr = dummy
        for val in values:
            curr.next = ListNode(val)
            curr = curr.next
            
        return dummy.next


    # --- OPTIMIZED APPROACH ---
    def mergeTwoLists(self, list1: Optional[ListNode], list2: Optional[ListNode]) -> Optional[ListNode]:
        # The Dummy node acts as the unbreakable foundation for our new list
        dummy = ListNode()
        # The tail pointer is the "construction worker" building the track
        tail = dummy
        
        # We can only compare values if BOTH lists still have nodes left
        while list1 and list2:
            
            if list1.val < list2.val:
                # Point the tail to list1's node
                tail.next = list1
                # Advance list1 to its next node
                list1 = list1.next
            else:
                # Point the tail to list2's node
                tail.next = list2
                # Advance list2 to its next node
                list2 = list2.next
                
            # Move our construction worker forward onto the newly added track piece
            tail = tail.next
            
        # Once the loop breaks, at least one list is completely empty.
        # Because the lists were already sorted, the remaining list is perfectly 
        # ordered and can just be snapped onto the end of our tail.
        if list1:
            tail.next = list1
        elif list2:
            tail.next = list2
            
        # Return the real head, skipping the fake Dummy node
        return dummy.next
```

### Problem 003: Linked List Cycle (LeetCode 141)

### Problem Definition and Constraints
Given the `head` of a linked list, return `true` if there is a cycle in the linked list. Otherwise, return `false`. A cycle exists if at least one node in the list can be visited again by continuously following the `next` pointer.

* Constraints:
  * 0 <= Length of the list <= 10^4
  * -10^5 <= Node.val <= 10^5
  * `index` (internal representation of cycle start) is -1 or a valid index.

### Examples
* **Example 1:**
  * Input: `head = [1,2,3,4], index = 1`
  * Output: `true`
  * Explanation: There is a cycle in the linked list, where the tail connects to the 1st node (0-indexed).
* **Example 2:**
  * Input: `head = [1,2], index = -1`
  * Output: `false`

### Brute Force Approach
The baseline approach utilizes a Hash Set. As we traverse the linked list node by node, we check if the current node already exists in our Set. If it does, a cycle exists. If it does not, we add the node to the Set and move forward. If we reach the end of the list (`None`), no cycle exists.
* Time Complexity: $O(n)$ — We visit each node at most once, and Hash Set lookups are $O(1)$.
* Space Complexity: $O(n)$ — We allocate auxiliary memory to store a reference to every single node in the list.

### Optimized Approach (Fast & Slow Pointers)
To achieve $O(1)$ space, we use Floyd's Tortoise and Hare algorithm. We initialize two pointers, `slow` and `fast`, both starting at the head. In a loop, `slow` moves forward one node, while `fast` moves forward two nodes. If there is no cycle, `fast` will eventually reach the end of the list (`None`). If there is a cycle, `fast` will become trapped in the loop and will eventually "lap" `slow`. When `slow == fast`, we have definitively proven a cycle exists.
* Time Complexity: $O(n)$ — If there is no cycle, the fast pointer reaches the end in $n/2$ steps. If there is a cycle, the fast pointer will catch the slow pointer in at most $n$ steps after the slow pointer enters the cycle.
* Space Complexity: $O(1)$ — We only maintain two pointer variables, requiring no scaling auxiliary memory.

```python
from typing import Optional

# Definition for singly-linked list.
class ListNode:
    def __init__(self, x):
        self.val = x
        self.next = None

class Solution:
    
    # --- BRUTE FORCE APPROACH ---
    def hasCycle_brute_force(self, head: Optional[ListNode]) -> bool:
        # A Hash Set to track the physical memory addresses of nodes we've seen
        seen = set()
        
        curr = head
        while curr:
            # If we step on a node that is already in our set, we found a loop
            if curr in seen:
                return True
                
            # Otherwise, record that we visited this node and move forward
            seen.add(curr)
            curr = curr.next
            
        return False


    # --- OPTIMIZED APPROACH ---
    def hasCycle(self, head: Optional[ListNode]) -> bool:
        # Initialize both runners at the starting line
        slow = head
        fast = head
        
        # We must check 'fast' and 'fast.next' because fast takes TWO steps at a time.
        # If fast.next is None, trying to read fast.next.next will crash the program.
        while fast and fast.next:
            
            # The tortoise takes 1 step
            slow = slow.next
            
            # The hare takes 2 steps
            fast = fast.next.next
            
            # If they are ever standing on the exact same node, a cycle exists
            if slow == fast:
                return True
                
        # If the while loop breaks, the hare found a finish line (None).
        return False
```

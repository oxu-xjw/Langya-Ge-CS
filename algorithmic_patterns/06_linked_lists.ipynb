### Problem 001: Reverse Linked List (LeetCode 206)

### Problem Definition and Constraints
Given the beginning of a singly linked list `head`, reverse the list, and return the new beginning of the list.

* Constraints:
  * 0 <= The length of the list <= 1000
  * -1000 <= Node.val <= 1000

### Examples
* **Example 1:**
  * Input: `head = [0,1,2,3]`
  * Output: `[3,2,1,0]`
* **Example 2:**
  * Input: `head = []`
  * Output: `[]`

### Brute Force Approach
The naive approach is to traverse the linked list from start to finish, appending every value into a standard array. Once the array is populated, we iterate through it backwards, creating brand new `ListNode` objects and linking them together to form a new reversed list. 
* Time Complexity: $O(n)$ — We traverse the list once to build the array, and once more to build the new linked list.
* Space Complexity: $O(n)$ — We create an array of size $n$ to hold the values, and allocate memory for $n$ brand new nodes.

### Optimized Approach (In-Place Pointer Reversal)
To achieve $O(1)$ space, we must manipulate the existing pointers without creating any new objects. We use two main pointers: `prev` (initialized to `None`) and `curr` (initialized to `head`). As we traverse the list, we temporarily store the next node in `nxt = curr.next` so we don't sever the rest of the list. We then reverse the current node's pointer by pointing it backwards (`curr.next = prev`). Finally, we slide both pointers forward one step (`prev = curr`, then `curr = nxt`). When `curr` hits `None`, `prev` will be resting on the new head of the list.
* Time Complexity: $O(n)$ — We traverse the linked list exactly once.
* Space Complexity: $O(1)$ — We only use three temporary pointer variables (`prev`, `curr`, `nxt`), regardless of the size of the list.

```python
from typing import Optional

# Definition for singly-linked list.
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

class Solution:
    
    # --- BRUTE FORCE APPROACH ---
    def reverseList_brute_force(self, head: Optional[ListNode]) -> Optional[ListNode]:
        if not head:
            return None
            
        # Step 1: Extract all values into a standard Python list
        values = []
        curr = head
        while curr:
            values.append(curr.val)
            curr = curr.next
            
        # Step 2: Build a brand new linked list in reverse
        # We start by creating the new head using the last value in our array
        new_head = ListNode(values[-1])
        curr = new_head
        
        # Loop through the rest of the array backwards
        for i in range(len(values) - 2, -1, -1):
            curr.next = ListNode(values[i])
            curr = curr.next
            
        return new_head


    # --- OPTIMIZED APPROACH ---
    def reverseList(self, head: Optional[ListNode]) -> Optional[ListNode]:
        # prev starts as None because the very first node (the old head) 
        # needs to point to None (becoming the new tail).
        prev = None
        curr = head
        
        while curr:
            # 1. SAVE: Temporarily store the rest of the list so we don't lose it
            nxt = curr.next
            
            # 2. REVERSE: Break the forward link and point it backward
            curr.next = prev
            
            # 3. ADVANCE: Shift both pointers forward for the next loop iteration
            prev = curr
            curr = nxt
            
        # Once curr falls off the end (becomes None), prev is sitting on the last node,
        # which is the brand new head of our reversed list.
        return prev
```

### Problem 002: Merge Two Sorted Linked Lists (LeetCode 21)

### Problem Definition and Constraints
You are given the heads of two sorted linked lists `list1` and `list2`. Merge the two lists into one sorted linked list and return the head of the new sorted linked list. The new list should be made up of nodes spliced together from the first two lists.

* Constraints:
  * 0 <= The length of each list <= 100
  * -100 <= Node.val <= 100
  * Both `list1` and `list2` are sorted in non-decreasing order.

### Examples
* **Example 1:**
  * Input: `list1 = [1,2,4], list2 = [1,3,5]`
  * Output: `[1,1,2,3,4,5]`
* **Example 2:**
  * Input: `list1 = [], list2 = [1,2]`
  * Output: `[1,2]`

### Brute Force Approach
The naive approach abandons the advantage of the lists already being sorted. We traverse both linked lists, extract all their integer values, and store them in a standard array. We then run a sorting algorithm on the array. Finally, we iterate through the sorted array, creating brand new `ListNode` objects to build the resulting list.
* Time Complexity: $O((n + m) \log(n + m))$ — Extracting takes linear time, but sorting the combined array dictates the overall time complexity.
* Space Complexity: $O(n + m)$ — We allocate memory for the array holding all values, and allocate memory for $n + m$ brand new nodes.

### Optimized Approach (Two Pointers & Dummy Node)
To achieve $O(1)$ space and $O(n+m)$ time, we restructure the existing pointers in-place. We instantiate a `Dummy` node to act as the anchor for our new list, and a `tail` pointer to track the end of our growing list. We compare the current nodes of `list1` and `list2`. The `tail` pointer connects to the node with the smaller value, and we advance that specific list's pointer forward. Once one of the lists is fully exhausted, we attach the entire remaining segment of the other list directly to the `tail`. We return `Dummy.next` as the true head of the merged list.
* Time Complexity: $O(n + m)$ — We traverse each node in both lists at most once. 
* Space Complexity: $O(1)$ — We reuse the existing nodes, allocating no scaling auxiliary memory. Only the constant space of the `Dummy` node and tracking pointers is required.

```python
from typing import Optional

# Definition for singly-linked list.
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

class Solution:
    
    # --- BRUTE FORCE APPROACH ---
    def mergeTwoLists_brute_force(self, list1: Optional[ListNode], list2: Optional[ListNode]) -> Optional[ListNode]:
        values = []
        
        # Extract from first list
        while list1:
            values.append(list1.val)
            list1 = list1.next
            
        # Extract from second list
        while list2:
            values.append(list2.val)
            list2 = list2.next
            
        # Sort the combined array
        values.sort()
        
        # Build new list using a dummy node
        dummy = ListNode()
        curr = dummy
        for val in values:
            curr.next = ListNode(val)
            curr = curr.next
            
        return dummy.next


    # --- OPTIMIZED APPROACH ---
    def mergeTwoLists(self, list1: Optional[ListNode], list2: Optional[ListNode]) -> Optional[ListNode]:
        # The Dummy node acts as the unbreakable foundation for our new list
        dummy = ListNode()
        # The tail pointer is the "construction worker" building the track
        tail = dummy
        
        # We can only compare values if BOTH lists still have nodes left
        while list1 and list2:
            
            if list1.val < list2.val:
                # Point the tail to list1's node
                tail.next = list1
                # Advance list1 to its next node
                list1 = list1.next
            else:
                # Point the tail to list2's node
                tail.next = list2
                # Advance list2 to its next node
                list2 = list2.next
                
            # Move our construction worker forward onto the newly added track piece
            tail = tail.next
            
        # Once the loop breaks, at least one list is completely empty.
        # Because the lists were already sorted, the remaining list is perfectly 
        # ordered and can just be snapped onto the end of our tail.
        if list1:
            tail.next = list1
        elif list2:
            tail.next = list2
            
        # Return the real head, skipping the fake Dummy node
        return dummy.next
```

### Problem 003: Linked List Cycle (LeetCode 141)

### Problem Definition and Constraints
Given the `head` of a linked list, return `true` if there is a cycle in the linked list. Otherwise, return `false`. A cycle exists if at least one node in the list can be visited again by continuously following the `next` pointer.

* Constraints:
  * 0 <= Length of the list <= 10^4
  * -10^5 <= Node.val <= 10^5
  * `index` (internal representation of cycle start) is -1 or a valid index.

### Examples
* **Example 1:**
  * Input: `head = [1,2,3,4], index = 1`
  * Output: `true`
  * Explanation: There is a cycle in the linked list, where the tail connects to the 1st node (0-indexed).
* **Example 2:**
  * Input: `head = [1,2], index = -1`
  * Output: `false`

### Brute Force Approach
The baseline approach utilizes a Hash Set. As we traverse the linked list node by node, we check if the current node already exists in our Set. If it does, a cycle exists. If it does not, we add the node to the Set and move forward. If we reach the end of the list (`None`), no cycle exists.
* Time Complexity: $O(n)$ — We visit each node at most once, and Hash Set lookups are $O(1)$.
* Space Complexity: $O(n)$ — We allocate auxiliary memory to store a reference to every single node in the list.

### Optimized Approach (Fast & Slow Pointers)
To achieve $O(1)$ space, we use Floyd's Tortoise and Hare algorithm. We initialize two pointers, `slow` and `fast`, both starting at the head. In a loop, `slow` moves forward one node, while `fast` moves forward two nodes. If there is no cycle, `fast` will eventually reach the end of the list (`None`). If there is a cycle, `fast` will become trapped in the loop and will eventually "lap" `slow`. When `slow == fast`, we have definitively proven a cycle exists.
* Time Complexity: $O(n)$ — If there is no cycle, the fast pointer reaches the end in $n/2$ steps. If there is a cycle, the fast pointer will catch the slow pointer in at most $n$ steps after the slow pointer enters the cycle.
* Space Complexity: $O(1)$ — We only maintain two pointer variables, requiring no scaling auxiliary memory.

```python
from typing import Optional

# Definition for singly-linked list.
class ListNode:
    def __init__(self, x):
        self.val = x
        self.next = None

class Solution:
    
    # --- BRUTE FORCE APPROACH ---
    def hasCycle_brute_force(self, head: Optional[ListNode]) -> bool:
        # A Hash Set to track the physical memory addresses of nodes we've seen
        seen = set()
        
        curr = head
        while curr:
            # If we step on a node that is already in our set, we found a loop
            if curr in seen:
                return True
                
            # Otherwise, record that we visited this node and move forward
            seen.add(curr)
            curr = curr.next
            
        return False


    # --- OPTIMIZED APPROACH ---
    def hasCycle(self, head: Optional[ListNode]) -> bool:
        # Initialize both runners at the starting line
        slow = head
        fast = head
        
        # We must check 'fast' and 'fast.next' because fast takes TWO steps at a time.
        # If fast.next is None, trying to read fast.next.next will crash the program.
        while fast and fast.next:
            
            # The tortoise takes 1 step
            slow = slow.next
            
            # The hare takes 2 steps
            fast = fast.next.next
            
            # If they are ever standing on the exact same node, a cycle exists
            if slow == fast:
                return True
                
        # If the while loop breaks, the hare found a finish line (None).
        return False
```

### Problem 004: Reorder Linked List (LeetCode 143)

### Problem Definition and Constraints
You are given the `head` of a singly linked list. 
Currently, the nodes are ordered sequentially: $[L_0, L_1, L_2, \dots, L_{n-2}, L_{n-1}]$
You need to physically reorder the nodes so that they fold in on each other, alternating between the front and the back of the list: $[L_0, L_{n-1}, L_1, L_{n-2}, L_2, L_{n-3}, \dots]$
* You may **not** modify the values inside the nodes. You must physically detach and reattach the nodes themselves.
* Constraints:
  * 1 <= Length of the list <= 1000
  * **Requirement:** Must run in $O(n)$ time and $O(1)$ extra space.

### Examples
* **Example 1:**
  * Input: `head = [2, 4, 6, 8]`
  * Output: `[2, 8, 4, 6]` (1st, 4th, 2nd, 3rd)
* **Example 2:**
  * Input: `head = [2, 4, 6, 8, 10]`
  * Output: `[2, 10, 4, 8, 6]` (1st, 5th, 2nd, 4th, 3rd)

### Brute Force Approach
Since Linked Lists don't have indices, it's really hard to jump to the back of the list. A brute-force way to solve this is to iterate through the entire linked list and save every single node into a standard Python array. Once they are in an array, you can use a `left` pointer at the start and a `right` pointer at the end, and simply wire them together alternatingly.
* Time Complexity: $O(n)$
* Space Complexity: $O(n)$ — Storing all nodes in an array violates the $O(1)$ space requirement.

### Optimized Approach (The "Three-Part Harmony")
To do this entirely in-place without extra memory, we can combine three massive Linked List techniques you have already learned into one mega-algorithm:
1. **Find the Middle:** Use the Tortoise and Hare (slow/fast pointers) to find the exact center of the list. We then physically cut the list into two separate halves.
2. **Reverse the Second Half:** Take the second half of the list and reverse its arrows (using the `prev`, `curr`, `nxt` pointer dance). Now, the nodes from the back of the list are easily accessible from front to back.
3. **Merge the Two Halves:** Take the front half and the reversed back half, and zip them together by alternating nodes, similar to the Dummy Node merge trick (but without the Dummy).

* Time Complexity: $O(n)$ — Finding the middle, reversing, and merging each take linear time.
* Space Complexity: $O(1)$ — We just use a handful of pointers.

```python
from typing import Optional

# Definition for singly-linked list.
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

class Solution:
    
    # --- BRUTE FORCE APPROACH ---
    def reorderList_brute_force(self, head: Optional[ListNode]) -> None:
        if not head: return
        
        # Step 1: Dump all nodes into an array
        nodes = []
        curr = head
        while curr:
            nodes.append(curr)
            curr = curr.next
            
        # Step 2: Use two pointers to re-wire them
        left = 0
        right = len(nodes) - 1
        
        while left < right:
            # Connect the left node to the right node
            nodes[left].next = nodes[right]
            left += 1
            
            # If we cross the middle, break early to avoid a cycle
            if left == right:
                break
                
            # Connect the right node to the NEXT left node
            nodes[right].next = nodes[left]
            right -= 1
            
        # Ensure the very last node points to None to close the list
        nodes[left].next = None


    # --- OPTIMIZED APPROACH (In-Place) ---
    def reorderList(self, head: Optional[ListNode]) -> None:
        if not head or not head.next: return
        
        # STEP 1: Find the middle using Tortoise and Hare
        slow = head
        fast = head.next
        while fast and fast.next:
            slow = slow.next
            fast = fast.next.next
            
        # slow is now standing on the last node of the first half.
        # second points to the first node of the second half.
        second = slow.next
         
        # CRITICAL: We must snip the connection between the two halves!
        slow.next = None
        
        # STEP 2: Reverse the second half
        prev = None
        while second:
            nxt = second.next
            second.next = prev
            prev = second
            second = nxt
            
        # 'prev' is now the head of our reversed second half
        first = head
        second = prev
        
        # STEP 3: Merge the two halves by alternating
        while second:
            # Save the next nodes before we break the arrows
            tmp1 = first.next
            tmp2 = second.next
            
            # Wire 'first' to 'second'
            first.next = second
            # Wire 'second' to the next 'first'
            second.next = tmp1
            
            # Shift our main pointers forward
            first = tmp1
            second = tmp2
```

### Problem 005: Remove Nth Node From End of List (LeetCode 19)

### Problem Definition and Constraints
You are given the `head` of a linked list and an integer `n`. You need to remove the `n`-th node from the **end** of the list and return the new head.
* Constraints:
  * The number of nodes is between 1 and 30.
  * 1 <= n <= number of nodes.
  * **Requirement:** Try to do this in a single pass (without walking the list twice).

### Examples
* **Example 1:**
  * Input: `head = [1, 2, 3, 4, 5]`, `n = 2`
  * Output: `[1, 2, 3, 5]` (The 2nd node from the end was '4', so we snipped it out).
* **Example 2:**
  * Input: `head = [5]`, `n = 1`
  * Output: `[]` (We removed the 1st node from the end, leaving an empty list).

### Brute Force Approach (The "Two-Pass" Method)
If you need to find the 2nd node from the *end*, but you only know how to walk from the *front*, the easiest way is to just measure the entire list first!
1. **Pass 1:** Walk a pointer all the way to the end just to count how many nodes there are (let's say the total length is `L`).
2. **Pass 2:** Now you know exactly which node from the front you need to delete. You do some quick math (`L - n`), walk from the front to the node right *before* your target, and change its `.next` pointer to skip the target.
* Time Complexity: $O(N)$ — You walk the entire list twice, which is technically $O(2N)$, but simplifies to $O(N)$.
* Space Complexity: $O(1)$ — No extra memory is used.

### Optimized Approach (The "Measuring Tape" Trick)
Can we do this by only walking the list once? Yes! We use two pointers (`left` and `right`) and create a fixed "gap" between them.

Imagine a wooden stick that is exactly `n` nodes long. 
1. Place the `right` pointer at the front, and walk it forward `n` times. Now the distance between the `left` pointer (still at the start) and the `right` pointer is exactly your stick's length.
2. Now, push the stick forward! Move both `left` and `right` at the exact same speed.
3. The moment `right` falls off the end of the list, your stick has hit the back wall. Because the stick is exactly `n` nodes long, your `left` pointer is standing perfectly right in front of the `n`-th node from the end!

**The Return of the Dummy Node:**
What if `n` is equal to the total length of the list? That means we have to delete the very first node (the `head`). If we delete the head, where does our `left` pointer stand to do the snipping? It would have to stand *before* the list even starts! 
This is exactly why we use a **Dummy Node** again. By placing `left` on a fake node before the `head`, it always has a safe place to stand, even if we are deleting the very first real node.

```python
from typing import Optional

# Definition for singly-linked list.
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

class Solution:
    
    # --- BRUTE FORCE APPROACH (Two-Pass) ---
    def removeNthFromEnd_brute_force(self, head: Optional[ListNode], n: int) -> Optional[ListNode]:
        # Pass 1: Find the total length of the list
        length = 0
        curr = head
        while curr:
            length += 1
            curr = curr.next
            
        # Edge Case: If we are removing the very first node
        if length == n:
            return head.next
            
        # Pass 2: Walk to the node right BEFORE the one we want to delete
        curr = head
        for _ in range(length - n - 1):
            curr = curr.next
            
        # Snip the target node out of the list
        curr.next = curr.next.next
        
        return head


    # --- OPTIMIZED APPROACH (One-Pass with Dummy Node) ---
    def removeNthFromEnd(self, head: Optional[ListNode], n: int) -> Optional[ListNode]:
        # Set up the Dummy Node attached to the front of our list
        dummy = ListNode(0, head)
        
        # Both pointers start at the Dummy Node
        left = dummy
        right = dummy
        
        # Step 1: Create the "gap". Move 'right' forward exactly 'n' + 1 times.
        # (We do n + 1 so that 'left' lands right BEFORE the node we want to delete)
        for _ in range(n + 1):
            right = right.next
            
        # Step 2: Slide the whole "measuring tape" forward at the same speed
        while right is not None:
            left = left.next
            right = right.next
            
        # Step 3: 'right' has hit the wall. 'left' is sitting right before the target.
        # Snip it out!
        left.next = left.next.next
        
        # Return the true head of the list (skipping the fake engine)
        return dummy.next
```

### Problem 006: Copy List with Random Pointer (LeetCode 138)

### Problem Definition and Constraints
You are given a Linked List where every node has **two** pointers:
1. `next`: Points to the next node in the sequence (just like normal).
2. `random`: Points to *any* random node in the entire list, or it can be `None`.

Your mission is to create a **Deep Copy** of this list. 
* A "deep copy" means you cannot just reuse the existing nodes. You must physically create brand new nodes in memory that perfectly mirror the original list's structure and arrows.
* Constraints:
  * 0 <= Length of the list <= 100
  * **Requirement:** Aim for $O(n)$ time and $O(n)$ space.

### The Core Problem: The "Chicken and Egg"
Imagine you are building a clone of a city. You are currently building Clone House #1. 
House #1's `random` pointer points to House #4. You want to wire Clone House #1's random pointer to Clone House #4... but **Clone House #4 hasn't been built yet!** You can't point to something that doesn't exist.

### Standard Approach (The Hash Map Registry)
To solve the chicken-and-egg problem, we do this in **Two Passes** using a Hash Map (a standard Python dictionary).
* **Pass 1 (Build the Clones):** Walk through the original list. For every original node you see, build a brand new isolated clone node. Don't worry about the arrows yet. Write them down in a registry (Hash Map): `{Original_Node : Clone_Node}`.
* **Pass 2 (Wire the Arrows):** Walk through the original list again. Look at an original node. Find its clone in the registry. Look at where the original's `next` and `random` point to, find *those* clones in the registry, and wire them up!
* Time Complexity: $O(n)$ — We walk the list twice.
* Space Complexity: $O(n)$ — We use a dictionary to store a reference to every single node.

```python
from typing import Optional

# Definition for a Node.
class Node:
    def __init__(self, x: int, next: 'Node' = None, random: 'Node' = None):
        self.val = int(x)
        self.next = next
        self.random = random

class Solution:
    
    # --- STANDARD APPROACH (Hash Map Registry) ---
    def copyRandomList(self, head: 'Optional[Node]') -> 'Optional[Node]':
        if not head: 
            return None
        
        # Registry mapping {Original Node : Clone Node}
        # We pre-fill it with {None: None} to handle edge cases where pointers point to None
        old_to_new = {None: None}
        
        # PASS 1: Build all the clones and put them in the registry
        curr = head
        while curr:
            clone = Node(curr.val)
            old_to_new[curr] = clone
            curr = curr.next
            
        # PASS 2: Wire up the 'next' and 'random' arrows
        curr = head
        while curr:
            clone = old_to_new[curr]
            
            # Look up the clone of the 'next' target in our dictionary
            clone.next = old_to_new[curr.next]
            
            # Look up the clone of the 'random' target in our dictionary
            clone.random = old_to_new[curr.random]
            
            curr = curr.next
            
        # Return the head of the brand new clone list
        return old_to_new[head] 
```

### Problem 007: Add Two Numbers (LeetCode 2)

### Problem Definition and Constraints
You are given two non-empty linked lists, `l1` and `l2`, representing two non-negative integers. 
The digits are stored in **reverse order**, meaning the 1s column is at the head, the 10s column is next, and so on.
Add the two numbers together and return the sum as a new linked list (also in reverse order).

**Constraints:**
* 1 <= Length of lists <= 100
* 0 <= Node.val <= 9
* No leading zeros except the number 0 itself.

**Examples:**
* **Example 1:**
  * **Input:** `l1 = [1,2,3]`, `l2 = [4,5,6]` (Represents 321 + 654)
  * **Output:** `[5,7,9]` (Represents 975)
* **Example 2:**
  * **Input:** `l1 = [9]`, `l2 = [9]` (Represents 9 + 9)
  * **Output:** `[8,1]` (Represents 18)

### Core Logic: Grade-School Addition
Because the lists are stored in reverse order, the heads of the lists are already perfectly aligned at the ones column. We can just walk forward and do standard grade-school addition column by column.

1. **The Setup:** We use the **Dummy Node** pattern to anchor our new result list, and a `curr` pointer to act as our construction worker building the track. We also need a `carry` variable initialized to 0.
2. **The Loop:** We loop as long as `l1` has nodes, `l2` has nodes, **OR** there is a `carry` left over. (The carry check is crucial for cases like 9 + 9 = 18, where a brand new column needs to be created at the very end).
3. **The Math:**
   * Extract the values (if a list is exhausted, treat its value as `0`).
   * `total = val1 + val2 + carry`
   * `carry = total // 10` (Extracts the tens place to carry over to the next loop)
   * `digit = total % 10` (Extracts the ones place for the current node)
4. **The Build:** Create a new node with `digit`, attach it to `curr.next`, and move all pointers forward.

* **Time Complexity:** $O(\max(m, n))$ — We iterate through the longest list exactly once.
* **Space Complexity:** $O(1)$ auxiliary space — We only use a few tracking variables. (The memory required to build the answer list itself is not counted against us in space complexity).

```python
from typing import Optional

# Definition for singly-linked list.
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

class Solution:
    def addTwoNumbers(self, l1: Optional[ListNode], l2: Optional[ListNode]) -> Optional[ListNode]:
        # The Dummy node anchors our new list so we don't lose the head
        dummy = ListNode()
        curr = dummy
        carry = 0
        
        # Keep going if l1 has nodes, l2 has nodes, OR we have a leftover carry.
        while l1 or l2 or carry:
            
            # Extract values, defaulting to 0 if one list is shorter than the other
            v1 = l1.val if l1 else 0
            v2 = l2.val if l2 else 0
            
            # Calculate the exact column sum
            total = v1 + v2 + carry
            
            # Update the carry for the NEXT loop (e.g., 15 // 10 = 1)
            carry = total // 10
            
            # Get the single digit for the CURRENT node (e.g., 15 % 10 = 5)
            digit = total % 10
            
            # Build the new track piece and snap it onto the list
            curr.next = ListNode(digit)
            
            # Move our construction worker forward
            curr = curr.next
            
            # Move l1 and l2 forward (if they still exist)
            if l1:
                l1 = l1.next
            if l2:
                l2 = l2.next
                
        # Return the true head, skipping the fake Dummy node
        return dummy.next

        """
        =========================================================
        SIMULATION NOTES: Adding 9 + 9
        Initial State: l1 = [9], l2 = [9], carry = 0, curr = Dummy
        =========================================================
        
        LOOP 1 (l1 and l2 exist):
        v1 = 9, v2 = 9
        total = 9 + 9 + 0 = 18
        carry = 18 // 10 = 1
        digit = 18 % 10  = 8
        curr.next = ListNode(8)
        Move pointers: l1 = None, l2 = None, curr = Node(8)
        
        LOOP 2 (l1 is None, l2 is None, but carry == 1!):
        v1 = 0, v2 = 0
        total = 0 + 0 + 1 = 1
        carry = 1 // 10 = 0
        digit = 1 % 10  = 1
        curr.next = ListNode(1)
        Move pointers: l1 = None, l2 = None, curr = Node(1)
        
        END OF LOOP (l1, l2, and carry are all empty/0)
        Return: Dummy.next -> [8 -> 1] (which represents 18)
        """
```

### Problem 008: Find the Duplicate Number (LeetCode 287)

### Problem Definition and Constraints
You are given an array `nums` of size `n + 1`. Every integer in the array is between `1` and `n` inclusive.
There is exactly one duplicate number, but it could be repeated multiple times.
You must find and return the duplicate number.

**Strict Constraints:**
* You cannot modify the array (so no `.sort()`).
* You must use $O(1)$ extra space (so no Hash Sets).
* You must run in $O(n)$ time.

**Examples:**
* **Example 1:**
  * **Input:** `nums = [1,2,3,2,2]`
  * **Output:** `2`
* **Example 2:**
  * **Input:** `nums = [1,2,3,4,4]`
  * **Output:** `4`

### Core Logic: The "Array as a Linked List" Trick
Because the array has `n + 1` slots, but only contains numbers from `1` to `n`, **every single value in the array is also a valid index.** 
This means we can treat the array exactly like a Linked List! The current index is the "Node", and the value inside that slot is the "Pointer" telling us which index to jump to next (`next_node = nums[curr_index]`).

Because multiple slots contain the duplicate number, multiple "pointers" will point to the exact same index. This forces the path to merge, creating a guaranteed **Linked List Cycle**. The start of that cycle is our duplicate number!

We solve this using **Floyd's Tortoise and Hare**:
1. **Phase 1 (Find the Intersection):** We put a `slow` pointer and a `fast` pointer at index `0`. `slow` takes 1 jump (`nums[slow]`), while `fast` takes 2 jumps (`nums[nums[fast]]`). Because there is a cycle, they will eventually crash into each other.
2. **Phase 2 (Find the Cycle Entrance):** The spot where they crash isn't necessarily the cycle entrance (the duplicate). Floyd's math proves that the distance from the *start of the array* to the entrance is perfectly equal to the distance from the *crash point* to the entrance. 
   * We leave `slow` at the crash point, put a new pointer `slow2` at the start of the array, and move them both forward 1 step at a time. The exact index where they meet is the duplicate number!

* **Time Complexity:** $O(n)$ — The fast and slow pointers will find the cycle and the entrance in linear time.
* **Space Complexity:** $O(1)$ — We are only using three pointer variables (`slow`, `fast`, and `slow2`).

```python
from typing import List

class Solution:
    def findDuplicate(self, nums: List[int]) -> int:
        # Pointers start at index 0 (the starting line)
        slow = 0
        fast = 0
        
        # -------------------------------------------------------------
        # PHASE 1: Tortoise and Hare (Find the crash point)
        # -------------------------------------------------------------
        while True:
            # slow takes 1 step (looks at the value in its current slot)
            slow = nums[slow]
            
            # fast takes 2 steps (uses the first value to find the next value)
            fast = nums[nums[fast]]
            
            # When they step on the exact same index, we found the loop
            if slow == fast:
                break
                
        # -------------------------------------------------------------
        # PHASE 2: Find the Cycle Entrance (The Duplicate Number)
        # -------------------------------------------------------------
        # Put a new tortoise at the starting line
        slow2 = 0
        
        # Move both tortoises 1 step at a time until they collide
        while True:
            # If they are on the same index, we found the entrance!
            if slow == slow2:
                return slow
                
            slow = nums[slow]
            slow2 = nums[slow2]

        
        """
        =========================================================
        SIMULATION NOTES: nums = [1, 2, 3, 2, 2]
        =========================================================
        
        PHASE 1 (Find Crash Point):
        Start: slow = 0, fast = 0
        Loop 1: 
          slow = nums[0] -> 1
          fast = nums[nums[0]] -> nums[1] -> 2
        Loop 2:
          slow = nums[1] -> 2
          fast = nums[nums[2]] -> nums[3] -> 2
        CRASH! Both are at index 2. Break.
        
        PHASE 2 (Find Entrance):
        Start: slow2 = 0, slow = 2
        Check: slow2 (0) != slow (2)
        Loop 1:
          slow2 = nums[0] -> 1
          slow  = nums[2] -> 3
        Check: slow2 (1) != slow (3)
        Loop 2:
          slow2 = nums[1] -> 2
          slow  = nums[3] -> 2
        Check: slow2 (2) == slow (2)! 
        
        RETURN: 2 (The duplicate number).
        """
```

### Problem 009: LRU Cache (LeetCode 146)

### Problem Definition and Constraints
Implement the Least Recently Used (LRU) cache class.
* `LRUCache(int capacity)`: Initializes the cache with a maximum capacity.
* `get(int key)`: Returns the value of the `key` if it exists, otherwise returns `-1`. Using this makes the key the "Most Recently Used".
* `put(int key, int value)`: Updates or inserts the key-value pair. Using this makes the key the "Most Recently Used". If inserting pushes the cache over `capacity`, evict the "Least Recently Used" key.

**Constraints:**
* 1 <= capacity <= 3000
* Must run in $O(1)$ average time complexity for both `get` and `put`.

### Core Logic: The "Hash Map + Doubly Linked List" Combo
To achieve $O(1)$ time for both operations, we are mathematically forced to combine two separate data structures:
1.  **Hash Map:** Gives us $O(1)$ lookups to instantly check if a key exists and grab its value.
2.  **Doubly Linked List (DLL):** Gives us $O(1)$ insertions and deletions. If we used a standard array to track the usage history, shifting all the elements over when a key was deleted would take $O(n)$ time. In a DLL, if we know a node's exact memory address (which we store in our Hash Map!), we can sever its left and right arrows and physically pull it out of the chain in exactly $O(1)$ time.

**The Architecture:**
We use a DLL ordered by usage. The **Head** of the list represents the Least Recently Used (LRU), and the **Tail** represents the Most Recently Used (MRU).
To prevent our program from crashing when the list is empty or when we delete the very last node, we use **Two Dummy Nodes**: a `left` dummy (locked at the LRU end) and a `right` dummy (locked at the MRU end). 

Every time a node is touched (via `get` or `put`), we completely `remove` it from its current spot in the chain, and `insert` it right before the `right` dummy node, crowning it the new MRU. 

### Approach
1. **Node Class:** Create a custom node containing `key`, `val`, `prev`, and `next`. (We *must* store the `key` inside the node so that when we delete the LRU node, we know which key to delete from our Hash Map).
2. **Initialization:** Set up the empty Hash Map. Create the `left` and `right` dummy nodes and wire them to each other.
3. **Helpers:** Write a `remove(node)` function that severs a node from the chain, and an `insert(node)` function that always wires a node just before the `right` dummy.
4. **Get:** If the key doesn't exist, return -1. If it does, fetch the node from the Hash Map, `remove` it, `insert` it (updating it to MRU), and return its value.
5. **Put:** If the key already exists, `remove` the old node. Create the new node, add it to the Hash Map, and `insert` it. If `len(cache) > capacity`, grab the LRU node (`left.next`), `remove` it from the list, and delete its key from the Hash Map.

* **Time Complexity:** $O(1)$ for both `get` and `put`.
* **Space Complexity:** $O(c)$ where $c$ is the `capacity`. Our Hash Map and DLL will never exceed the allowed capacity size.

```python
class Node:
    def __init__(self, key, val):
        self.key = key      # Required so we can delete from the Hash Map later
        self.val = val
        self.prev = None
        self.next = None

class LRUCache:
    def __init__(self, capacity: int):
        self.cap = capacity
        self.cache = {}  # Map: key -> Node

        # Left dummy node = LRU boundary
        self.left = Node(0, 0)
        
        # Right dummy node = MRU boundary
        self.right = Node(0, 0)
        
        # Initially, the dummies just point at each other
        self.left.next = self.right
        self.right.prev = self.left

    # Helper function to remove a node from the linked list
    def remove(self, node: Node):
        prev_node = node.prev
        next_node = node.next
        
        # Sever the node by rewiring its neighbors to each other
        prev_node.next = next_node
        next_node.prev = prev_node

    # Helper function to insert a node at the MRU position (right before the right dummy)
    def insert(self, node: Node):
        prev_node = self.right.prev
        next_node = self.right
        
        # Wire the neighbors to point at the new node
        prev_node.next = node
        next_node.prev = node
        
        # Wire the new node to point at the neighbors
        node.next = next_node
        node.prev = prev_node

    def get(self, key: int) -> int:
        if key in self.cache:
            # We accessed it, so it becomes the Most Recently Used
            node = self.cache[key]
            self.remove(node)
            self.insert(node)
            return node.val
        return -1

    def put(self, key: int, value: int) -> None:
        # If it already exists, remove the old version from the chain
        if key in self.cache:
            self.remove(self.cache[key])
            
        # Create the new updated node and insert it as the MRU
        new_node = Node(key, value)
        self.cache[key] = new_node
        self.insert(new_node)
        
        # Check if we exceeded capacity
        if len(self.cache) > self.cap:
            # The LRU is always the node sitting immediately after the left dummy
            lru_node = self.left.next
            
            # Remove it from the linked list
            self.remove(lru_node)
            
            # Remove it from the Hash Map (this is why the Node class MUST store 'key')
            del self.cache[lru_node.key]


    """
    =========================================================
    SIMULATION NOTES: capacity = 2
    =========================================================
    put(1, 10): 
    - Insert Node(1). Map = {1: Node(1)}.
    - Chain: Left <-> (1, 10) <-> Right
    
    put(2, 20):
    - Insert Node(2). Map = {1: Node(1), 2: Node(2)}.
    - Chain: Left <-> (1, 10) <-> (2, 20) <-> Right
    
    get(1):
    - Node(1) exists. Remove it. Insert it.
    - Chain: Left <-> (2, 20) <-> (1, 10) <-> Right
    
    put(3, 30):
    - Insert Node(3). Map = {1: Node(1), 2: Node(2), 3: Node(3)}.
    - Chain: Left <-> (2, 20) <-> (1, 10) <-> (3, 30) <-> Right
    - Capacity exceeded (3 > 2)! 
    - LRU is left.next -> Node(2).
    - Remove Node(2) from chain. Delete key '2' from map.
    - Map = {1: Node(1), 3: Node(3)}.
    - Final Chain: Left <-> (1, 10) <-> (3, 30) <-> Right
    """
```

### Problem 010: Merge K Sorted Linked Lists (LeetCode 23) [HARD]

### Problem Definition and Constraints
You are given an array of `k` linked lists `lists`, where each list is sorted in ascending order.
Return the sorted linked list that is the result of merging all of the individual linked lists into one single list.

**Examples:**
* **Example 1:**
  * **Input:** `lists = [[1,2,4], [1,3,5], [3,6]]`
  * **Output:** `[1,1,2,3,3,4,5,6]`
* **Example 2:**
  * **Input:** `lists = []`
  * **Output:** `[]`

* Constraints:
  * 0 <= k <= 10,000
  * 0 <= lists[i].length <= 500
  * Total nodes across all lists $\le 10,000$

### Core Logic: Divide and Conquer (Merge Sort style)
The hints suggest a sequential merge: merge List 1 into List 2, then merge that result into List 3, then into List 4, etc. While this works, it is incredibly inefficient ($O(n \cdot k)$) because the growing merged list gets dragged through every single operation over and over again.

Instead, we use **Divide and Conquer**, which is the exact engine that powers Merge Sort. 
Rather than merging one by one, we merge them in **pairs**. 
* Round 1: Merge List 1+2, List 3+4, List 5+6, etc. (We just cut `k` lists down to `k/2` lists).
* Round 2: Merge the results together in pairs. (`k/4` lists).
* We repeat this until exactly 1 list remains. 

Because we halve the number of lists every round, we only need to perform $\log k$ rounds of merging.

### Approach: Pairwise Merging
1. **The Base Cases:** If the `lists` array is empty, return `None`.
2. **The Helper Function:** We reuse the exact `mergeTwoLists` logic from Problem 002. It takes two sorted lists and weaves them together using a Dummy Node in $O(n)$ time.
3. **The Tournament Bracket:** We use a `while len(lists) > 1` loop.
   * Inside, we create an empty `merged_lists` array for the current round.
   * We iterate through the current `lists` array in steps of 2 (`i` and `i + 1`).
   * We merge the pair using our helper function and append the result to `merged_lists`.
   * (If there is an odd number of lists, the last list just gets a "bye" round and moves forward un-merged).
   * We overwrite our main `lists` variable with `merged_lists` and start the next round.
4. When the `while` loop finishes, `lists[0]` holds the final, massive merged list.

* **Time Complexity:** $O(N \log k)$ — Where $N$ is the total number of nodes across all lists, and $k$ is the number of lists. We process all $N$ nodes during each round of merging, and there are exactly $\log k$ rounds. This completely destroys the sub-optimal $O(N \cdot k)$ approach.
* **Space Complexity:** $O(1)$ auxiliary space for the nodes themselves. We reuse the existing nodes. (Storing the `merged_lists` array takes $O(k)$ memory, which scales logarithmically down to $1$).

```python
from typing import List, Optional

# Definition for singly-linked list.
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

class Solution:
    def mergeKLists(self, lists: List[Optional[ListNode]]) -> Optional[ListNode]:
        if not lists or len(lists) == 0:
            return None
            
        # Keep running "tournament rounds" until only 1 winner remains
        while len(lists) > 1:
            merged_lists = []
            
            # Step by 2 to grab pairs of lists
            for i in range(0, len(lists), 2):
                l1 = lists[i]
                # If 'i' is the last index (odd number of lists), l2 is None
                l2 = lists[i + 1] if (i + 1) < len(lists) else None
                
                # Merge the pair and add it to the winners bracket
                merged_lists.append(self.mergeTwoLists(l1, l2))
                
            # The winners bracket becomes the new starting lineup for the next round
            lists = merged_lists
            
        return lists[0]

    # Helper function: Exactly the same as "Merge Two Sorted Lists"
    def mergeTwoLists(self, l1: Optional[ListNode], l2: Optional[ListNode]) -> Optional[ListNode]:
        dummy = ListNode()
        tail = dummy
        
        while l1 and l2:
            if l1.val < l2.val:
                tail.next = l1
                l1 = l1.next
            else:
                tail.next = l2
                l2 = l2.next
            tail = tail.next
            
        if l1:
            tail.next = l1
        if l2:
            tail.next = l2
            
        return dummy.next


        """
        =========================================================
        SIMULATION NOTES: lists = [[1,4], [2,5], [3,6], [0,7]]
        =========================================================
        INITIAL: k = 4
        
        ROUND 1:
        - i = 0: Merge lists[0] and lists[1] -> [1,4] + [2,5] = [1,2,4,5]
        - i = 2: Merge lists[2] and lists[3] -> [3,6] + [0,7] = [0,3,6,7]
        - merged_lists = [[1,2,4,5], [0,3,6,7]]
        - lists = merged_lists (k is now 2)
        
        ROUND 2:
        - i = 0: Merge lists[0] and lists[1] -> [1,2,4,5] + [0,3,6,7] = [0,1,2,3,4,5,6,7]
        - merged_lists = [[0,1,2,3,4,5,6,7]]
        - lists = merged_lists (k is now 1)
        
        LOOP BREAKS (len(lists) == 1).
        Return lists[0].
        
        Total time saved: We didn't have to drag the elements of list[0] through 
        three separate sequential merges!
        """
```

### Problem 011: Reverse Nodes in k-Group (LeetCode 25) [HARD]

### Problem Definition and Constraints
You are given the `head` of a singly linked list and a positive integer `k`.
You must reverse the first `k` nodes in the linked list, and then reverse the next `k` nodes, and so on. If there are fewer than `k` nodes left at the very end, leave those nodes exactly as they are.
Return the modified list.
*You are only allowed to modify the nodes' next pointers, not the values inside the nodes.*

**Examples:**
* **Example 1:**
  * **Input:** `head = [1,2,3,4,5,6], k = 3`
  * **Output:** `[3,2,1,6,5,4]`
  * **Explanation:** Group 1 (`[1,2,3]`) is reversed to `[3,2,1]`. Group 2 (`[4,5,6]`) is reversed to `[6,5,4]`.
* **Example 2:**
  * **Input:** `head = [1,2,3,4,5], k = 3`
  * **Output:** `[3,2,1,4,5]`
  * **Explanation:** Group 1 (`[1,2,3]`) is reversed to `[3,2,1]`. The remaining nodes (`[4,5]`) are fewer than `k=3`, so they remain untouched.

* Constraints:
  * 1 <= k <= length of list <= 5000
  * 0 <= Node.val <= 100

### Core Logic: The "Chunked Pointer Dance"
This problem perfectly combines the **Dummy Node** pattern with the **Pointer Dance** (In-Place Reversal). 

Instead of reversing the whole list until `None`, we reverse it in perfectly measured "chunks". 
1. **Measure the Chunk:** For every group, we send a scout pointer ahead exactly `k` steps. If the scout hits `None` before taking `k` steps, we don't have enough nodes to form a complete group, so we immediately stop and return our list.
2. **Isolate the Chunk:** If the scout successfully finds the `k`-th node, we save the node *immediately after* it (`groupNext`). We need this to ensure the rest of the list isn't lost.
3. **Reverse the Chunk:** We run our standard `prev`, `curr`, `nxt` pointer dance. The brilliant twist? Instead of initializing `prev = None` like a normal reverse, we initialize `prev = groupNext`. This way, as the first node in the chunk gets reversed (becoming the new tail), it automatically points forward to the rest of the unreversed list!
4. **Reconnect the Chunk:** We use a `groupPrev` pointer (starting at our Dummy Node) to connect the previous part of the list to the brand new head of the reversed chunk. We then slide `groupPrev` forward to prepare for the next chunk.

### Approach: Dummy Node + Modular Reversal
1. Create a `dummy` node pointing to `head`.
2. Initialize a pointer `groupPrev = dummy`. This node sits immediately *before* the `k`-group we are about to reverse.
3. Enter a `while True` loop:
   * **Scout:** Find the `k`-th node from `groupPrev`. If it returns `None`, break the loop.
   * **Save:** Store the `k`-th node's next node as `groupNext`.
   * **Reverse:** Set `prev = groupNext` and `curr = groupPrev.next`. Run a while loop until `curr == groupNext`, reversing the arrows.
   * **Rewire:** The node at `kth` is the new head of this chunk. Set `groupPrev.next` to `kth`.
   * **Advance:** The old head of the chunk (which is now the tail) is the new `groupPrev` for the next round.

* **Time Complexity:** $O(n)$ — We walk forward to count `k` nodes, and walk forward again to reverse them. Every node is processed at most twice.
* **Space Complexity:** $O(1)$ — We only manipulate existing pointers, using no extra memory.

```python
from typing import Optional

# Definition for singly-linked list.
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

class Solution:
    def reverseKGroup(self, head: Optional[ListNode], k: int) -> Optional[ListNode]:
        dummy = ListNode(0, head)
        # groupPrev points to the node right BEFORE the k-group we are working on
        groupPrev = dummy
        
        while True:
            # 1. SCOUT: Find the k-th node of the current group
            kth = self.getKth(groupPrev, k)
            
            # If we don't have k nodes left, the problem says to leave them as is
            if not kth:
                break
                
            # 2. SAVE: We must remember where the NEXT group starts
            groupNext = kth.next
            
            # 3. REVERSE: Standard pointer dance, but bounded to this chunk
            # 'prev' starts as groupNext so the tail automatically hooks into the rest of the list
            prev = groupNext
            curr = groupPrev.next
            
            while curr != groupNext:
                nxt = curr.next
                curr.next = prev
                prev = curr
                curr = nxt
                
            # 4. REWIRE: 
            # groupPrev is still pointing to the OLD head (which is now the tail).
            # We save that old head so we can jump our groupPrev to it later.
            tmp = groupPrev.next
            
            # Connect the previous chunk's tail to the NEW head of this chunk (which is 'kth')
            groupPrev.next = kth
            
            # 5. ADVANCE: Move our anchor to the tail of the chunk we just reversed
            groupPrev = tmp
            
        return dummy.next

    # Helper function to find the exact k-th node from a starting point
    def getKth(self, curr: Optional[ListNode], k: int) -> Optional[ListNode]:
        while curr and k > 0:
            curr = curr.next
            k -= 1
        return curr


        """
        =========================================================
        SIMULATION NOTES: head = [1, 2, 3, 4, 5], k = 3
        =========================================================
        dummy -> 1 -> 2 -> 3 -> 4 -> 5
        groupPrev = dummy
        
        ROUND 1:
        - getKth(dummy, 3) lands on Node(3). kth = Node(3).
        - groupNext = Node(4).
        - Reverse chunk [1, 2, 3]:
          - prev starts as Node(4).
          - Node(1).next becomes Node(4). (1 -> 4)
          - Node(2).next becomes Node(1). (2 -> 1 -> 4)
          - Node(3).next becomes Node(2). (3 -> 2 -> 1 -> 4)
        - Rewire:
          - dummy.next = kth (Node(3)). (dummy -> 3 -> 2 -> 1 -> 4 -> 5)
        - Advance:
          - groupPrev jumps to the old head, which is now the tail: Node(1).
        
        ROUND 2:
        - getKth(Node(1), 3) searches for 3 nodes. 
        - It steps to Node(4), then Node(5), then hits None!
        - Returns None. Loop breaks.
        
        Final list: dummy.next -> [3, 2, 1, 4, 5]
        """
```

### Problem 001: Single Number (LeetCode 136)

### Problem Definition and Constraints
You are given a non-empty array of integers `nums`. Every integer appears exactly twice except for one.
Return the integer that appears only once.
*Note: You must implement a solution with a linear runtime complexity and use only constant extra space.*

* Constraints:
  * 1 <= nums.length <= 10^4
  * -10^4 <= nums[i] <= 10^4

### Examples
* **Example 1:**
  * Input: `nums = [3,2,3]`
  * Output: `2`
* **Example 2:**
  * Input: `nums = [7,6,6,7,8]`
  * Output: `8`

### Brute Force Approach (Hash Set)
We can iterate through the array and use a Hash Set to track numbers. If a number is not in the set, we add it. If it is already in the set, we remove it. At the end of the loop, the only number remaining in the set is our single number.
* Time Complexity: $O(n)$ — We iterate through the array once.
* Space Complexity: $O(n)$ — The Hash Set will grow to store up to half the elements in the array. This violates the $O(1)$ space constraint of the problem.

### Optimized Approach (Bitwise XOR)
To achieve $O(1)$ extra space, we use the Bitwise XOR operator (`^`). 
XOR has two vital properties:
1. `a ^ a = 0` (XORing a number by itself cancels it out to zero).
2. `a ^ 0 = a` (XORing a number by zero leaves it unchanged).

Because XOR operations can be done in any order, if we XOR every single number in the array together, all the identical pairs will naturally find each other and cancel out to `0`. The only number left standing will be the unique number XORed with `0`, which is just the unique number itself.
* Time Complexity: $O(n)$ — We iterate through the array exactly once.
* Space Complexity: $O(1)$ — We only use a single integer variable `res` to store our running XOR total.


```python
from typing import List

class Solution:
    def singleNumber(self, nums: List[int]) -> int:
        # Initialize our running total to 0. 
        # (Remember: 0 ^ any_number = any_number)
        res = 0
        
        # Iterate through every number in the array
        for n in nums:
            # Apply the XOR operator to our running total.
            # - If 'n' is a new number, its bits get added to 'res'.
            # - If 'n' is a duplicate we've seen before, its bits perfectly 
            #   cancel out the bits it left in 'res' the first time!
            res = res ^ n
            
            # Note: In Python, you can write this shorthand as: res ^= n
            
        # The pairs have all annihilated each other. 
        # The lone survivor is left in 'res'.
        return res
```

### Problem 002: Number of 1 Bits (LeetCode 191)

### Problem Definition and Constraints
You are given an unsigned integer `n`. Return the number of `1` bits in its binary representation (also known as the Hamming weight).
*Note: You may assume `n` is a non-negative integer which fits within 32-bits.*

* Constraints:
  * The input must be a binary string of length `32`.

### Examples
* **Example 1:**
  * Input: `n = 00000000000000000000000000010111` (Decimal 23)
  * Output: `4`
* **Example 2:**
  * Input: `n = 01111111111111111111111111111101` (Decimal 2147483645)
  * Output: `30`

### Approach 1: Bit Shifting & AND Masking
We can loop 32 times. At each step, we use the Bitwise AND operator (`&`) with the number `1` to check if the rightmost bit is a `1`. `n & 1` will return `1` if the rightmost bit is `1`, and `0` if it is `0`. We add that result to our count, and then use the right-shift operator (`>>`) to push all the bits in `n` to the right by one position, discarding the bit we just checked.
* Time Complexity: $O(1)$ — The loop always runs exactly 32 times regardless of how big the number is.
* Space Complexity: $O(1)$

### Approach 2: Brian Kernighan's Algorithm (Optimized Trick)
Instead of checking every single zero, what if we could just "turn off" the 1s one by one and count how many times we do it? 
The operation `n & (n - 1)` has a magical mathematical property: it always flips the right-most `1` bit in `n` to a `0`. We can just run this operation in a `while n > 0:` loop. The number of times the loop runs before `n` becomes `0` is exactly our number of `1` bits!
* Time Complexity: $O(1)$ — Runs exactly as many times as there are `1` bits in the number. In the worst case (all 1s), it runs 32 times.
* Space Complexity: $O(1)$

```python
class Solution:
    def hammingWeight(self, n: int) -> int:
        count = 0
        
        # Loop until all bits are shifted out (n becomes 0)
        while n > 0:
            # 'n & 1' checks the rightmost bit. 
            # If it's a 1, it adds 1. If it's a 0, it adds 0.
            count += n & 1
            
            # Shift all bits to the right by 1, effectively deleting the rightmost bit
            # Example: 1011 >> 1 becomes 101
            n = n >> 1
            
        return count


    # ---------------------------------------------------------
    # BONUS: Brian Kernighan's Algorithm (The Clever Trick)
    # ---------------------------------------------------------
    def hammingWeightOptimized(self, n: int) -> int:
        count = 0
        
        while n > 0:
            # This specific math operation ALWAYS deletes the right-most '1' bit
            # Try it with n = 5 (101). 
            # n - 1 = 4 (100). 
            # 101 & 100 = 100 (The rightmost 1 was deleted!)
            n = n & (n - 1)
            count += 1
            
        return count
```

### Problem 003: Counting Bits (LeetCode 338)

### Problem Definition and Constraints
Given an integer `n`, count the number of `1`s in the binary representation of every number in the range `[0, n]`.
Return an array `output` where `output[i]` is the number of `1`s in the binary representation of `i`.

* Constraints:
  * 0 <= n <= 10^5

### Examples
* **Example 1:**
  * Input: `n = 2`
  * Output: `[0,1,1]`
  * Explanation: 0 (0 bits), 1 (1 bit), 2 (10 -> 1 bit).
* **Example 2:**
  * Input: `n = 4`
  * Output: `[0,1,1,2,1]`
  * Explanation: 0 (0), 1 (1), 2 (10), 3 (11 -> 2 bits), 4 (100 -> 1 bit).

### Brute Force Approach
We can iterate from `0` to `n`, and run our $O(1)$ `hammingWeight` function from the previous problem on every single number, adding the result to an array.
* Time Complexity: $O(n \log n)$ — For each of the $n$ numbers, checking its bits takes time proportional to its length in binary (which is $\log n$).
* Space Complexity: $O(n)$ — To store the output array.

### Optimized Approach (Dynamic Programming & Pattern Recognition)
If we look closely at binary numbers, they repeat themselves perfectly:
* `0` to `3` is: `0`, `1`, `10`, `11`
* `4` to `7` is: `100`, `101`, `110`, `111`
Notice that `4` through `7` are literally the exact same patterns as `0` through `3`, just with a brand new `1` attached to the front!

**The Offset DP Pattern:**
The number of `1` bits in any number `i` is exactly `1` plus the number of bits in `i - offset` (where the `offset` is the largest power of 2 we have seen so far). 
For example, to find the bits in `7`: The highest power of 2 below 7 is `4`. `7 - 4 = 3`. So, `dp[7] = 1 + dp[3]`. We just look up `dp[3]` in our DP array!
* Time Complexity: $O(n)$ — We fill out our DP array in a single linear pass.
* Space Complexity: $O(n)$ — We return an array of size $n + 1$.

```python
from typing import List

class Solution:
    def countBits(self, n: int) -> List[int]:
        # Initialize our DP array (our notebook) with 0s.
        # It's size n + 1 because we need to store an answer for 0 all the way up to n.
        dp = [0] * (n + 1)
        
        # 'offset' tracks the highest power of 2 we have reached (1, 2, 4, 8, 16...)
        offset = 1
        
        for i in range(1, n + 1):
            # If our current number hits exactly double our offset, it's a new power of 2!
            if offset * 2 == i:
                offset = i
                
            # The number of 1 bits is: 1 (for the power of 2 offset) + whatever was left over.
            dp[i] = 1 + dp[i - offset]
            
        return dp

    # ---------------------------------------------------------
    # BONUS: The "Right-Shift" DP Trick (Even Shorter)
    # ---------------------------------------------------------
    def countBitsShift(self, n: int) -> List[int]:
        dp = [0] * (n + 1)
        
        for i in range(1, n + 1):
            # What does i >> 1 do? It divides the number by 2 (chopping off the right-most bit).
            # What does i & 1 do? It returns 1 if the number is odd, and 0 if it's even.
            # So: bits in current number = bits in (number // 2) + 1 if the number is odd.
            dp[i] = dp[i >> 1] + (i & 1)
            
        return dp
```

### Problem 004: Reverse Bits (LeetCode 190)

### Problem Definition and Constraints
Given a 32-bit unsigned integer `n`, reverse the bits of the binary representation of `n` and return the result as an integer.

* Constraints:
  * The input must be a binary string of length `32`.

### Examples
* **Example 1:**
  * Input: `n = 00000000000000000000000000010101` (Decimal 21)
  * Output: `2818572288`
  * Explanation: Reversing the bits gives us `10101000000000000000000000000000`.

### Optimized Approach (Bit-by-Bit Shift and Build)
We can solve this by constructing a completely new number `res` initialized to `0`. 
Since `n` is guaranteed to be exactly 32 bits, we will loop exactly 32 times. In each iteration:
1. **Extract**: Grab the right-most bit from `n` using `n & 1`.
2. **Make Room**: Left-shift `res` (`res << 1`) to shift all previously added bits to the left, opening up a `0` on the right-most side.
3. **Insert**: Use Bitwise OR (`|`) to place the extracted bit into that newly opened spot on `res`.
4. **Discard**: Right-shift `n` (`n >> 1`) to discard the bit we just extracted, prepping the next bit for the next iteration.
* Time Complexity: $O(1)$ — The loop always runs exactly 32 times.
* Space Complexity: $O(1)$ — We only use one integer variable to store the result.

```python
class Solution:
    def reverseBits(self, n: int) -> int:
        res = 0
        
        # We know it's exactly a 32-bit integer, so we loop 32 times
        for i in range(32):
            
            # 1. Pop the right-most bead off the old string
            bit = n & 1
            
            # 2. Shift all beads on the new string to the left to make room
            res = res << 1
            
            # 3. Put the popped bead onto the new string
            # We use '|' (Bitwise OR), but a standard '+' works here too!
            res = res | bit
            
            # 4. Shift the old string to the right, deleting the bead we just popped
            n = n >> 1
            
        return res
```

### Problem 005: Missing Number (LeetCode 268)

### Problem Definition and Constraints
Given an array `nums` containing `n` integers in the range `[0, n]` without any duplicates, return the single number in the range that is missing from the array.

* Constraints:
  * 1 <= nums.length <= 1000

### Examples
* **Example 1:**
  * Input: `nums = [3,0,1]`
  * Output: `2`
  * Explanation: `n = 3`, so the range should be `[0,1,2,3]`. 2 is missing.
* **Example 2:**
  * Input: `nums = [0,1]`
  * Output: `2`
  * Explanation: `n = 2`, so the range should be `[0,1,2]`. 2 is missing.

### Approach 1: Bitwise XOR (The Single Number Trick)
We can solve this problem using the exact same logic from Problem 001 ("Single Number"). If we take a list of all the numbers from `0` to `n`, and mash it together with the `nums` array, we create a giant list where every number appears perfectly twice in pairs... except for the missing number, which appears only once!
Because `XOR` cancels out identical pairs to `0`, we can just `XOR` every number from `0` to `n`, and then `XOR` that result with every number inside the `nums` array. The pairs will annihilate each other, and the only number left standing will be the missing one.
* Time Complexity: $O(n)$ — We iterate through the array once.
* Space Complexity: $O(1)$ — We only use a single `res` variable.

### Approach 2: Math (Gauss's Formula)
There is an alternative math trick. The sum of all numbers from `0` up to `n` can be instantly calculated using Gauss's formula: `n * (n + 1) / 2`. If we calculate the expected sum, and subtract the actual sum of the numbers in our array, the difference is exactly the missing number!
* Time Complexity: $O(n)$ — We only have to loop once to find the actual sum of the array.
* Space Complexity: $O(1)$

```python
from typing import List

class Solution:
    def missingNumber(self, nums: List[int]) -> int:
        # Initialize our running XOR total.
        # We start it at the length of the array (n) because our loop below 
        # only iterates up to n-1, so we need to make sure the max number 'n' is included!
        res = len(nums)
        
        # Iterate through the indices of the array
        for i in range(len(nums)):
            
            # XOR the expected number (the index 'i')
            res ^= i
            
            # XOR the actual number found in the array
            res ^= nums[i]
            
        # The identical pairs canceled out. The missing number remains!
        return res


    # ---------------------------------------------------------
    # BONUS: The Math Approach (Gauss's Sum Formula)
    # ---------------------------------------------------------
    def missingNumberMath(self, nums: List[int]) -> int:
        # The number of expected elements
        n = len(nums)
        
        # The sum if NO numbers were missing: n * (n + 1) / 2
        expected_sum = (n * (n + 1)) // 2
        
        # The sum of the numbers we were actually given
        actual_sum = sum(nums)
        
        # The difference is exactly the number that is missing
        return expected_sum - actual_sum
```

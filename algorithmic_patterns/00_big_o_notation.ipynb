### 1) Constant Time - O(1)

**Definition:**
An algorithm runs in **Constant Time**, or O(1), if the execution time is completely independent of the size of the input data (`n`). Whether the input contains 10 items or 10 million items, the algorithm takes the exact same amount of time to complete.

**Key Characteristics:**
* **No Loops Dependent on `n`:** O(1) algorithms do not iterate over the input size. 
* **Direct Access:** They usually involve direct memory access, such as grabbing an item from an array by its index or retrieving a value from a hash map (dictionary) by its key.
* **Not Always "One Step":** An O(1) algorithm might take 1 operation, or it might take 50 operations. The defining factor is that the number of operations is a **constant bound** that does not grow as the input grows. O(50) simplifies to O(1).

**Common O(1) Operations:**
* Accessing an array element by its index (`arr[5]`).
* Adding, removing, or looking up a key in a Dictionary/Set.
* Basic arithmetic equations (e.g., checking if a number is even using modulo `%`).

```python
# --- EXAMPLE 1: Array Indexing ---
def get_first_element(arr: list) -> int:
    # Accessing an index in a list is an O(1) operation.
    # The computer calculates the exact memory address instantly.
    if not arr:
        return None
    return arr[0]


# --- EXAMPLE 2: Dictionary (Hash Map) Lookup ---
def check_user_exists(user_dict: dict, username: str) -> bool:
    # Looking up a key in a dictionary is O(1) on average.
    # It does NOT need to scan the entire dictionary to find the key.
    return username in user_dict


# --- EXAMPLE 3: Mathematical Formula ---
def sum_first_n_numbers(n: int) -> int:
    # Instead of looping from 1 to n (which would take O(n) time), 
    # we use Gauss's formula to get the answer instantly in O(1) time.
    return n * (n + 1) // 2


# --- EXAMPLE 4: Fixed Number of Operations ---
def print_first_three(arr: list):
    # Even though there is a loop here, it only ever runs a maximum of 3 times.
    # Because the loop's execution limit does not scale with the size of 'arr', 
    # the time complexity is O(3), which simplifies down to O(1).
    loop_limit = min(3, len(arr))
    for i in range(loop_limit):
        print(arr[i])
```

### 2) Logarithmic Time - O(log n)

**Definition:**
An algorithm runs in **Logarithmic Time**, or O(log n), if its execution time grows logarithmically as the input size (`n`) increases. In practical terms, this usually means the algorithm cuts the problem size in half (or by some other fraction) during each step of its execution. 

**Key Characteristics:**
* **Divide and Conquer:** The most common indicator of O(log n) is an algorithm that repeatedly divides the remaining search space. 
* **Extreme Efficiency:** Logarithmic algorithms are incredibly fast, especially for massive datasets. For example, if you have an input size of 1,000,000 items, an O(log n) algorithm (base 2) will take at most about 20 steps to find the answer. If you double the input to 2,000,000 items, it only takes **one** extra step (21 steps).
* **Requires Sorted/Structured Data:** To eliminate half the data at each step without looking at it, the data usually needs to be pre-sorted (like a sorted array) or structured in a specific way (like a Binary Search Tree).

**Common O(log n) Operations:**
* Binary Search on a sorted array.
* Searching, inserting, or deleting a node in a balanced Binary Search Tree (BST).
* Pushing or popping an element from a Heap (Priority Queue).

```python
# --- EXAMPLE 1: Binary Search ---
def binary_search(arr: list, target: int) -> int:
    # This is the classic O(log n) algorithm.
    # Instead of checking every single element, we look at the middle.
    # If the target is smaller than the middle, we completely discard the right half.
    # We repeat this halving process until we find the target.
    
    left = 0
    right = len(arr) - 1
    
    while left <= right:
        mid = (left + right) // 2
        
        if arr[mid] == target:
            return mid          # Target found!
        elif arr[mid] < target:
            left = mid + 1      # Discard the left half
        else:
            right = mid - 1     # Discard the right half
            
    return -1


# --- EXAMPLE 2: Iterative Halving (Mathematical) ---
def count_halves(n: int) -> int:
    # How many times can we divide a number by 2 before it reaches 1?
    # Because 'n' is reduced by half on every iteration, 
    # the loop runs O(log n) times.
    
    steps = 0
    while n > 1:
        n = n // 2
        steps += 1
        
    return steps
```

### 3) Linear Time - O(n)

**Definition:**
An algorithm runs in **Linear Time**, or O(n), if its execution time grows directly in proportion to the size of the input (`n`). If the input size doubles, the time it takes to execute the algorithm roughly doubles. 

**Key Characteristics:**
* **Full Traversal:** You usually see O(n) when an algorithm must examine every single item in a dataset at least once to compute the result.
* **Non-Nested Loops:** A single loop over an array of size `n` is O(n).
* **Dropping Constants:** If an algorithm iterates through an array of size `n` twice (in two separate, non-nested loops), the time complexity is technically O(2n). However, in Big O notation, we drop the constants because we only care about the *growth rate*. Therefore, O(2n), O(3n), and O(100n) all simplify to just O(n).

**Common O(n) Operations:**
* Looping through an array to find a specific element (Linear Search).
* Finding the minimum or maximum value in an unsorted array.
* Summing all elements in an array.
* Reversing an array or string.

```python
# --- EXAMPLE 1: Simple Traversal ---
def find_maximum(arr: list) -> int:
    # We must check every single element in the array to guarantee 
    # we have found the absolute maximum value. 
    # If the array has 'n' elements, this loop runs 'n' times.
    if not arr:
        return None
        
    max_val = arr[0]
    for num in arr:
        if num > max_val:
            max_val = num
            
    return max_val


# --- EXAMPLE 2: Early Exit (Worst-Case Scenario) ---
def linear_search(arr: list, target: int) -> bool:
    # Even though we might find the target on the very first try (O(1) best case),
    # Big O notation always describes the WORST-case scenario.
    # In the worst case, the target is at the very end (or not in the list at all),
    # meaning we must check all 'n' elements. Thus, it is O(n).
    for item in arr:
        if item == target:
            return True
    return False


# --- EXAMPLE 3: Dropping Constants ---
def print_twice(arr: list):
    # This function loops through the array twice.
    # Time Complexity: O(n) + O(n) = O(2n)
    # However, Big O ignores constants, so this is simply O(n).
    
    print("First pass:")
    for item in arr:
        print(item)
        
    print("Second pass:")
    for item in arr:
        print(item)
```

### 4) Square Root Time - O(√n)

**Definition:**
An algorithm runs in **Square Root Time**, or O(√n), if its execution time grows proportionally to the square root of the input size (`n`). It sits comfortably between O(log n) and O(n) in terms of efficiency. 

**Key Characteristics:**
* **Math and Number Theory:** You will almost exclusively see O(√n) in algorithms dealing with properties of numbers, such as factorization, finding divisors, or primality testing.
* **The "Pair" Optimization:** The logic usually relies on the mathematical principle that factors come in pairs. For any number `n`, if it has a factor greater than its square root, the corresponding pair factor must be smaller than the square root. Therefore, you only ever need to check up to √n.
* **Loop Condition:** In code, this often looks like a `while` or `for` loop that continues as long as `i * i <= n`.

**Common O(√n) Operations:**
* Checking if a number is prime.
* Finding all divisors of a given integer.

```python
import math

# --- EXAMPLE 1: Primality Test ---
def is_prime(n: int) -> bool:
    # To check if a number is prime, we only need to test divisibility 
    # up to the square root of n. 
    # For example, to check if 100 is prime, we only check up to 10.
    # Time Complexity: O(√n)
    
    if n < 2:
        return False
        
    # We iterate from 2 up to the integer value of √n
    limit = int(math.sqrt(n))
    for i in range(2, limit + 1):
        if n % i == 0:
            return False  # We found a divisor, so it's not prime
            
    return True


# --- EXAMPLE 2: The "i * i" Loop approach (No Math Module) ---
def is_prime_no_math(n: int) -> bool:
    # We can achieve O(√n) without importing the math module 
    # by squaring the iterator in the while loop condition.
    
    if n < 2:
        return False
        
    i = 2
    # This loop stops once i squared is strictly greater than n
    while i * i <= n:
        if n % i == 0:
            return False
        i += 1
        
    return True


# --- EXAMPLE 3: Finding all Divisors ---
def get_all_divisors(n: int) -> list:
    # Factors come in pairs! 
    # If n = 36, when we find 4, we also automatically find 9 (since 36 // 4 = 9).
    # Because we only loop up to √n, this algorithm runs in O(√n) time.
    
    divisors = set()
    limit = int(math.sqrt(n))
    
    for i in range(1, limit + 1):
        if n % i == 0:
            divisors.add(i)          # Add the smaller factor
            divisors.add(n // i)     # Add the corresponding larger factor
            
    return list(divisors)
```

### 5) Loglinear Time - O(n log n)

**Definition:**
An algorithm runs in **Loglinear Time**, or O(n log n), if its execution time is a combination of linear and logarithmic time. You can think of it as doing a linear amount of work, O(n), inside a logarithmic number of steps, O(log n).

**Key Characteristics:**
* **The Sorting Standard:** This is the baseline time complexity for the most efficient general-purpose comparison sorting algorithms. If you are sorting data, you are almost always dealing with O(n log n) time.
* **Divide, then Merge:** It frequently appears in "divide and conquer" algorithms. The dataset is repeatedly divided in half (the `log n` part), and then a linear pass is made to process or merge the divided parts together (the `n` part).
* **Performance:** It is slightly slower than linear time O(n) but significantly faster than quadratic time O(n²), making it highly scalable for large datasets.

**Common O(n log n) Operations:**
* Efficient sorting algorithms like Merge Sort, Heap Sort, and Quick Sort (average case).
* Python's built-in `.sort()` and `sorted()` functions (which use an O(n log n) algorithm called Timsort).

```python
# --- EXAMPLE 1: Built-in Sorting ---
def sort_array(arr: list) -> list:
    # Python's built-in sorting method uses Timsort.
    # The time complexity of this operation is inherently O(n log n).
    # In algorithmic interviews, if you sort an array first to solve a problem, 
    # your algorithm's overall time complexity becomes at least O(n log n).
    return sorted(arr)


# --- EXAMPLE 2: Merge Sort (The Classic O(n log n) Algorithm) ---
def merge_sort(arr: list) -> list:
    # BASE CASE: Arrays of size 0 or 1 are already sorted.
    if len(arr) <= 1:
        return arr
        
    # DIVIDE (The 'log n' part): 
    # We keep splitting the array in half until we get single elements.
    mid = len(arr) // 2
    left_half = merge_sort(arr[:mid])
    right_half = merge_sort(arr[mid:])
    
    # CONQUER / MERGE (The 'n' part): 
    # We iterate through the sub-arrays to combine them in sorted order.
    return merge(left_half, right_half)


def merge(left: list, right: list) -> list:
    # This merge helper function runs in O(n) time.
    # It loops through both halves, doing a linear amount of work.
    sorted_result = []
    i = 0
    j = 0
    
    while i < len(left) and j < len(right):
        if left[i] < right[j]:
            sorted_result.append(left[i])
            i += 1
        else:
            sorted_result.append(right[j])
            j += 1
            
    # Append any remaining elements
    sorted_result.extend(left[i:])
    sorted_result.extend(right[j:])
    
    return sorted_result
```

### 6) Bilinear Time - O(n * m)

**Definition:**
An algorithm runs in **Bilinear Time**, or O(n * m), when its execution time is proportional to the product of two *different* input sizes. This typically occurs when you have a nested loop, but the outer loop iterates over one dataset (size `n`) and the inner loop iterates over a completely different dataset (size `m`).

**Key Characteristics:**
* **Two Independent Inputs:** It is crucial to differentiate this from Quadratic Time O(n²). If you iterate through an array of size `n` inside another loop of the *same* array, that is O(n²). If you iterate through array `A` (size `n`) and inside that loop you iterate through array `B` (size `m`), it is O(n * m).
* **Matrix / Grid Traversal:** This time complexity is extremely common when dealing with 2D arrays or grids, where `n` represents the number of rows and `m` represents the number of columns.
* **Combinations:** Any algorithm that requires checking every element of one set against every element of another set will result in O(n * m) time.

**Common O(n * m) Operations:**
* Traversing a 2D matrix (rows * columns).
* Finding all possible pairs between two different lists.
* Certain string matching algorithms (comparing a string of length `n` against a pattern of length `m`).

```python
# --- EXAMPLE 1: Pairs from Two Different Lists ---
def get_all_combinations(colors: list, clothing_items: list) -> list:
    # We have two distinct inputs. Let n = len(colors) and m = len(clothing_items).
    # For every single color, we loop through all clothing items.
    # The inner loop runs exactly 'm' times for each of the 'n' iterations of the outer loop.
    # Total operations: n * m.
    
    combinations = []
    for color in colors:               # Outer loop runs 'n' times
        for item in clothing_items:    # Inner loop runs 'm' times
            combinations.append(f"{color} {item}")
            
    return combinations


# --- EXAMPLE 2: 2D Matrix / Grid Traversal ---
def print_matrix_elements(matrix: list) -> None:
    # A standard 2D grid traversal.
    # Let n = number of rows, and m = number of columns.
    # To visit every single cell in the grid, it takes O(n * m) time.
    
    if not matrix:
        return
        
    num_rows = len(matrix)
    num_cols = len(matrix[0])
    
    for row in range(num_rows):            # Runs 'n' times
        for col in range(num_cols):        # Runs 'm' times
            # Print the value at the current row and column
            print(matrix[row][col])


# --- EXAMPLE 3: String Comparison (Brute Force Pattern Matching) ---
def count_character_matches(text: str, pattern: str) -> int:
    # Let n = len(text) and m = len(pattern).
    # We compare every character in the text against every character in the pattern.
    
    matches = 0
    for t_char in text:            # Runs 'n' times
        for p_char in pattern:     # Runs 'm' times
            if t_char == p_char:
                matches += 1
                
    return matches
```

### 7) Quadratic Time - O(n²)

**Definition:**
An algorithm runs in **Quadratic Time**, or O(n²), if its execution time is proportional to the square of the input size (`n`). If the input size doubles, the time it takes to execute doesn't just double—it quadruples ($2^2 = 4$). If the input size increases by 10x, the execution time increases by 100x. 

**Key Characteristics:**
* **Nested Loops on the Same Data:** This usually happens when you have a loop *inside* another loop, and both loops iterate over the exact same array (or datasets of the exact same size `n`). For each of the `n` elements, we do `n` operations, resulting in `n * n = n²` operations.
* **Brute Force Solutions:** O(n²) is often the time complexity of the "brute force" approach to a problem (like comparing every element to every other element to find pairs or duplicates). 
* **Crucial Distinction:** Two *separate*, non-nested passes over an array of size `n` are O(n) + O(n) = O(2n), which simplifies to **O(n)**. Nested loops are O(n) * O(n), which is **O(n²)**. 

**Common O(n²) Operations:**
* Inefficient sorting algorithms like Bubble Sort, Insertion Sort, and Selection Sort.
* Checking for duplicates in an array without using a Hash Set.
* Creating all possible subarrays from a given array.

```python
# --- EXAMPLE 1: Checking all pairs (Brute Force) ---
def find_duplicates(arr: list) -> bool:
    # We want to see if there are any duplicate values.
    # The outer loop runs 'n' times. For EACH of those times, 
    # the inner loop also runs 'n' times. 
    # Time Complexity: O(n * n) = O(n²)
    
    n = len(arr)
    for i in range(n):
        for j in range(n):
            # We must make sure we aren't comparing the exact same index to itself
            if i != j and arr[i] == arr[j]:
                return True
    return False


# --- EXAMPLE 2: Bubble Sort ---
def bubble_sort(arr: list) -> list:
    # A classic O(n²) sorting algorithm.
    # It repeatedly steps through the list, compares adjacent elements, 
    # and swaps them if they are in the wrong order.
    
    n = len(arr)
    for i in range(n):
        # The inner loop bounds decrease slightly as the array gets sorted,
        # but in Big O notation (worst-case), this is still mathematically bounded by O(n²)
        for j in range(0, n - i - 1):
            if arr[j] > arr[j + 1]:
                # Swap the elements
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                
    return arr


# --- EXAMPLE 3: Sequential vs. Nested Loops ---
def sequential_vs_nested(arr: list):
    n = len(arr)
    
    # THIS IS O(n) + O(n) = O(n)
    # The loops are completely separate.
    for i in range(n):
        print(arr[i])
    for j in range(n):
        print(arr[j])
        
    # THIS IS O(n) * O(n) = O(n²)
    # The loop is nested. This will be drastically slower for large arrays.
    for i in range(n):
        for j in range(n):
            print(arr[i], arr[j])
```

### 8) Cubic Time - O(n³)

**Definition:**
An algorithm runs in **Cubic Time**, or O(n³), if its execution time is proportional to the cube of the input size (`n`). If the input size doubles, the execution time increases by a factor of 8 ($2^3 = 8$). If the input size increases by 10x, the execution time takes 1,000 times longer.

**Key Characteristics:**
* **Three Nested Loops:** This is the most common hallmark of O(n³) algorithms. You have a loop inside a loop inside a loop, all iterating over the same dataset (or datasets of size `n`).
* **Severe Inefficiency:** Algorithms with O(n³) complexity scale very poorly. They are usually only acceptable for very small inputs. If you end up with an O(n³) solution in an interview, there is almost always a way to optimize it down to O(n²) or better.
* **Combinatorics and Matrices:** Often found in algorithms calculating combinations of three items (triplets) or mathematically standard operations like naive matrix multiplication.

**Common O(n³) Operations:**
* The brute-force solution for the "3Sum" problem (finding three numbers that add up to a target).
* Naive multiplication of two `n x n` matrices.
* Parsing algorithms like the CYK algorithm for context-free grammars.

```python
# --- EXAMPLE 1: Brute Force 3Sum ---
def find_three_sum_zero(arr: list) -> list:
    # We want to find all unique triplets in the array that sum to 0.
    # The brute force approach uses three nested loops to check every possible combination.
    # Time Complexity: O(n * n * n) = O(n³)
    
    n = len(arr)
    triplets = []
    
    for i in range(n):
        for j in range(i + 1, n):
            for k in range(j + 1, n):
                if arr[i] + arr[j] + arr[k] == 0:
                    triplets.append((arr[i], arr[j], arr[k]))
                    
    return triplets


# --- EXAMPLE 2: 3D Grid Traversal ---
def build_3d_coordinates(n: int) -> list:
    # Imagine a 3D cube where each side has a length of 'n'.
    # To find every single coordinate inside this cube, we need 3 nested loops.
    
    coords = []
    for x in range(n):
        for y in range(n):
            for z in range(n):
                coords.append((x, y, z))
                
    return coords


# --- EXAMPLE 3: Naive Matrix Multiplication ---
def multiply_square_matrices(matrix_a: list, matrix_b: list) -> list:
    # Standard multiplication of two n x n matrices.
    # We iterate through rows of A, columns of B, and then calculate the dot product.
    
    n = len(matrix_a)
    # Initialize an n x n result matrix with zeros
    result = [[0 for _ in range(n)] for _ in range(n)]
    
    for i in range(n):              # 1st loop: rows of A
        for j in range(n):          # 2nd loop: cols of B
            for k in range(n):      # 3rd loop: calculate dot product
                result[i][j] += matrix_a[i][k] * matrix_b[k][j]
                
    return result
```

### 9) Exponential Time - O(2ⁿ)

**Definition:**
An algorithm runs in **Exponential Time**, or $O(2^n)$, if its execution time doubles with each addition to the input dataset. If $n = 10$, it takes ~1,000 operations. If $n = 20$, it takes ~1,000,000 operations. If $n = 30$, it takes over 1 billion operations. 

**Key Characteristics:**
* **Include / Exclude Branching:** This complexity frequently appears in algorithms that use a decision tree where at each step, you make a binary choice (e.g., "Do I include this element in the subset, or do I exclude it?").
* **Multiple Recursive Calls:** If a recursive function calls itself twice (or more) for every execution without reducing the problem size logarithmically, it will likely result in exponential time. 
* **The "Intractable" Threshold:** $O(2^n)$ algorithms scale so poorly that they are practically useless for any input size larger than 20 or 30. In coding interviews, if you write an $O(2^n)$ solution, you will almost certainly be asked to optimize it using Dynamic Programming or Memoization.

**Common O(2ⁿ) Operations:**
* Generating the Power Set (all possible subsets of a set).
* Naive recursive calculation of the Fibonacci sequence.
* Solving the "Traveling Salesperson" problem via brute force (though this can sometimes be worse, creeping into factorial time).

```python
# --- EXAMPLE 1: Naive Recursive Fibonacci ---
def fibonacci(n: int) -> int:
    # A classic example of an O(2ⁿ) algorithm.
    # For every call, the function branches out into two more calls.
    # This creates a massive, exponentially growing tree of recursive calls,
    # recalculating the same subproblems over and over.
    
    if n <= 1:
        return n
        
    # The branching factor is 2. 
    return fibonacci(n - 1) + fibonacci(n - 2)


# --- EXAMPLE 2: Generating All Subsets (Power Set) ---
def generate_subsets(arr: list) -> list:
    # We want to find every possible combination of elements in the array.
    # For a list of size n, there are exactly 2ⁿ possible subsets.
    # Why? Because for every element, we make a binary choice: 
    # "Include it" or "Exclude it".
    
    subsets = []
    
    def backtrack(index: int, current_subset: list):
        # Base case: We've made a decision for every element
        if index == len(arr):
            subsets.append(current_subset[:])
            return
            
        # Decision 1: EXCLUDE the current element
        backtrack(index + 1, current_subset)
        
        # Decision 2: INCLUDE the current element
        current_subset.append(arr[index])
        backtrack(index + 1, current_subset)
        
        # Backtrack to clean up before the next recursion bubbles up
        current_subset.pop()
        
    backtrack(0, [])
    return subsets
```

### 10) Factorial Time - $O(n!)$

**Definition:**
An algorithm runs in **Factorial Time**, or $O(n!)$, if its execution time grows proportionally to the factorial of the input size. The factorial of $n$ (written as $n!$) is the product of all positive integers less than or equal to $n$ ($n \times (n-1) \times (n-2) \dots \times 1$).

**Key Characteristics:**
* **The Absolute Worst:** This is the most inefficient standard time complexity. It scales horrifyingly fast. For $n = 5$, it takes 120 operations. For $n = 10$, it jumps to 3,628,800 operations. By $n = 15$, it takes over 1.3 trillion operations.
* **"Every Possible Arrangement":** You will almost exclusively see $O(n!)$ when an algorithm needs to calculate all possible permutations of a dataset where the **order matters**.
* **Brute Force Permutations:** When you place the first item in a sequence, you have $n$ choices. For the second position, you have $n-1$ choices remaining. For the third, $n-2$ choices, and so on.

**Common $O(n!)$ Operations:**
* Generating all permutations (arrangements) of a string or array.
* The naive, brute-force solution to the Traveling Salesperson Problem (checking every single possible route between a set of cities).

```python
# --- EXAMPLE 1: Generating All Permutations ---
def get_permutations(arr: list) -> list:
    # We want to find every possible arrangement of the elements.
    # Unlike subsets (where [1, 2] is the same as [2, 1]), in permutations order matters.
    # The number of possible arrangements for an array of size n is exactly n!
    # Time Complexity: O(n!)
    
    # Base case: An empty array or a single element has only 1 permutation
    if len(arr) <= 1:
        return [arr]
        
    permutations = []
    
    # For every element in the array, we place it at the front, 
    # and then recursively find all permutations of the remaining elements.
    for i in range(len(arr)):
        current_element = arr[i]
        
        # Extract all OTHER elements except the current one
        remaining_elements = arr[:i] + arr[i+1:]
        
        # Recursively find permutations of the remaining elements
        for p in get_permutations(remaining_elements):
            permutations.append([current_element] + p)
            
    return permutations


# --- EXAMPLE 2: Heap's Algorithm (In-Place Permutations) ---
def generate_permutations_heaps(n: int, arr: list):
    # Heap's algorithm is a highly optimized, standard way to generate permutations.
    # However, because it still must generate and process n! distinct results, 
    # its time complexity remains fundamentally bounded by O(n!).
    
    if n == 1:
        # We have formed a complete permutation, print or save it
        print(arr)
        return
        
    for i in range(n):
        generate_permutations_heaps(n - 1, arr)
        
        # Swap choice depends on whether 'n' is even or odd
        if n % 2 == 0:
            arr[i], arr[n - 1] = arr[n - 1], arr[i]
        else:
            arr[0], arr[n - 1] = arr[n - 1], arr[0]
```

### 1) Quick Sort

**Time Complexity:**
* **Best:** $O(n \log n)$ - Occurs when the pivot consistently divides the array into two nearly equal halves.
* **Average:** $O(n \log n)$ - On average, the pivot creates reasonably balanced partitions.
* **Worst:** $O(n^2)$ - Occurs when the array is already sorted (or reverse sorted) and the pivot chosen is always the first or last element, causing one partition to be empty and the other to have $n-1$ elements.

**Space Complexity:** 
* **Worst:** $O(n)$ - In the worst-case scenario (unbalanced partitions), the recursive call stack goes $n$ levels deep. 
* **Average/Best:** $O(\log n)$ - With balanced partitions, the recursion tree is only $\log n$ levels deep.

**Key Characteristics:**
* **Divide and Conquer:** Quick Sort works by selecting a "pivot" element and partitioning the array into two sub-arrays: elements less than the pivot, and elements greater than the pivot. It then recursively sorts the sub-arrays.
* **In-Place (Mostly):** While the recursive call stack takes up memory, standard implementations sort the array by swapping elements within the original array, requiring no massive external data structures.
* **Highly Practical:** Despite the $O(n^2)$ worst-case, Quick Sort is often faster in practice than other $O(n \log n)$ algorithms like Merge Sort because its inner loop can be efficiently implemented on most architectures, and it has excellent cache locality.

```python
# --- EXAMPLE 1: Pythonic Quick Sort (For Conceptual Understanding) ---
def quick_sort_conceptual(arr: list) -> list:
    # This version is incredibly easy to read but uses extra memory 
    # to create the 'left', 'middle', and 'right' arrays.
    
    if len(arr) <= 1:
        return arr
        
    pivot = arr[len(arr) // 2]
    left = [x for x in arr if x < pivot]
    middle = [x for x in arr if x == pivot]
    right = [x for x in arr if x > pivot]
    
    return quick_sort_conceptual(left) + middle + quick_sort_conceptual(right)


# --- EXAMPLE 2: In-Place Quick Sort (Interview Standard) ---
def quick_sort_inplace(arr: list, low: int = 0, high: int = None):
    # This is the standard interview approach. It sorts the array entirely in-place
    # by swapping elements, meaning we only use memory for the recursive call stack.
    
    if high is None:
        high = len(arr) - 1

    if low < high:
        # Partition the array and get the final resting index of the pivot
        pivot_index = partition(arr, low, high)
        
        # Recursively sort the elements before and after the partition
        quick_sort_inplace(arr, low, pivot_index - 1)
        quick_sort_inplace(arr, pivot_index + 1, high)

def partition(arr: list, low: int, high: int) -> int:
    # Choose the rightmost element as the pivot
    pivot = arr[high]
    
    # Pointer for the greater element
    i = low - 1
    
    # Traverse through all elements and compare them with the pivot
    for j in range(low, high):
        if arr[j] <= pivot:
            # If we find an element smaller than the pivot, swap it with the greater element
            i += 1
            arr[i], arr[j] = arr[j], arr[i]
            
    # Swap the pivot element with the greater element specified by i
    arr[i + 1], arr[high] = arr[high], arr[i + 1]
    
    # Return the position from where partition is done
    return i + 1
```

### 2) Merge Sort

**Time Complexity:**
* **Best:** $O(n \log n)$
* **Average:** $O(n \log n)$
* **Worst:** $O(n \log n)$
* **Why?** Merge Sort is incredibly consistent. It *always* divides the array in half until it reaches single elements ($\log n$ steps), and it *always* merges every element back together ($n$ operations per level). It does not care if the array is already sorted or completely reversed; it does the exact same amount of work.

**Space Complexity:** 
* **Worst:** $O(n)$ - This is Merge Sort's biggest drawback. Because it cannot easily merge elements in-place like Quick Sort, it requires allocating new, temporary arrays to hold the divided halves during the merging process.

**Key Characteristics:**
* **Divide and Conquer:** It recursively cuts the array in half until it reaches sub-arrays of size 1 (which are inherently sorted), then stitches them back together in the correct order.
* **Stable Sort:** If two elements have the same value, Merge Sort guarantees they will stay in the same relative order they were in originally. (Quick Sort is generally unstable).
* **Guaranteed Performance:** Because its worst-case time is still $O(n \log n)$, it is often used in systems where you absolutely cannot risk a sudden drop in performance (like Python's Timsort, which is heavily based on Merge Sort).

```python
# --- EXAMPLE 1: Standard Merge Sort ---
def merge_sort(arr: list) -> list:
    # BASE CASE: An array of zero or one elements is already sorted.
    if len(arr) <= 1:
        return arr
        
    # DIVIDE: Find the middle index and split the array into two halves.
    mid = len(arr) // 2
    left_half = arr[:mid]
    right_half = arr[mid:]
    
    # RECURSIVELY SORT BOTH HALVES
    left_sorted = merge_sort(left_half)
    right_sorted = merge_sort(right_half)
    
    # CONQUER: Merge the two sorted halves back together.
    return merge(left_sorted, right_sorted)


def merge(left: list, right: list) -> list:
    # Helper function to combine two sorted arrays into one sorted array.
    merged_result = []
    i = 0  # Pointer for the left array
    j = 0  # Pointer for the right array
    
    # Compare elements from both arrays and append the smaller one
    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            merged_result.append(left[i])
            i += 1
        else:
            merged_result.append(right[j])
            j += 1
            
    # Once one of the arrays is exhausted, we append the remaining 
    # elements of the OTHER array. (Only one of these slices will have items left).
    merged_result.extend(left[i:])
    merged_result.extend(right[j:])
    
    return merged_result
```

### 3) Heap Sort

**Time Complexity:**
* **Best:** $O(n \log n)$
* **Average:** $O(n \log n)$
* **Worst:** $O(n \log n)$
* **Why?** Heap Sort operates in two phases. First, it builds a Max-Heap out of the array, which takes $O(n)$ time. Second, it repeatedly extracts the maximum element and restores the heap property (called "heapifying"). Extracting an element and heapifying takes $O(\log n)$ time, and it does this $n$ times, resulting in a strict $O(n \log n)$ overall time complexity regardless of the initial array order.

**Space Complexity:** 
* **Worst:** $O(1)$ - This is Heap Sort's biggest advantage over Merge Sort. It sorts the array entirely in-place by treating the array itself as a complete binary tree. No recursive call stacks or temporary arrays are needed.

**Key Characteristics:**
* **Array as a Tree:** It represents a binary tree using flat array indices. For any node at index `i`, its left child is at `2i + 1`, and its right child is at `2i + 2`.
* **Not Stable:** Like Quick Sort, Heap Sort is not a stable sort. Two identical elements might have their relative order swapped during the heapify process.
* **Cache Locality:** While it has the optimal $O(n \log n)$ worst-case time and $O(1)$ space, Heap Sort is often slightly slower in real-world practice than Quick Sort. This is because it constantly jumps around the array to compare parents and children, which leads to poor CPU cache locality compared to Quick Sort's linear scans.

```python
# --- EXAMPLE 1: In-Place Heap Sort (Interview Standard) ---
def heapify(arr: list, n: int, i: int):
    # Helper function to maintain the max-heap property.
    # 'n' is the size of the heap, 'i' is the root index of the current subtree.
    
    largest = i          # Initialize largest as root
    left = 2 * i + 1     # Left child index
    right = 2 * i + 2    # Right child index
    
    # See if left child of root exists and is greater than root
    if left < n and arr[left] > arr[largest]:
        largest = left
        
    # See if right child of root exists and is greater than the largest so far
    if right < n and arr[right] > arr[largest]:
        largest = right
        
    # If the largest is not the root, swap them and continue heapifying
    if largest != i:
        arr[i], arr[largest] = arr[largest], arr[i]  # Swap
        # Recursively heapify the affected sub-tree
        heapify(arr, n, largest)

def heap_sort(arr: list) -> list:
    n = len(arr)
    
    # PHASE 1: Build a Max-Heap.
    # We start from the last non-leaf node and move backwards to the root.
    # The last non-leaf node is always at index (n // 2) - 1.
    for i in range(n // 2 - 1, -1, -1):
        heapify(arr, n, i)
        
    # PHASE 2: Extract elements one by one.
    for i in range(n - 1, 0, -1):
        # Swap the current root (the maximum value) with the end of the array
        arr[i], arr[0] = arr[0], arr[i]
        
        # Call heapify on the reduced heap to find the new maximum
        # Notice we pass 'i' as the size, shrinking the heap boundary each time
        heapify(arr, i, 0)
        
    return arr


# --- EXAMPLE 2: Using Python's built-in heapq (Not O(1) space) ---
import heapq

def pythonic_heap_sort(arr: list) -> list:
    # Python has a built-in heap queue module, but it implements a MIN-heap.
    # While highly readable, this specific implementation uses O(n) space 
    # to store the newly sorted elements.
    
    # heapify transforms the list into a min-heap in-place in O(n) time
    heapq.heapify(arr)
    
    # heappop extracts the smallest element in O(log n) time
    sorted_result = []
    while arr:
        sorted_result.append(heapq.heappop(arr))
        
    return sorted_result
```

### 4) Insertion Sort

**Time Complexity:**
* **Best:** $O(n)$ - This is the "holy grail" scenario. It occurs when the array is already perfectly sorted. The algorithm only needs to make one comparison per element to confirm it's in the right place, resulting in a single linear pass.
* **Average:** $O(n^2)$ - Elements are in random order, requiring a nested loop to shift elements around.
* **Worst:** $O(n^2)$ - Occurs when the array is sorted in reverse order. Every single element must be checked and shifted all the way to the beginning of the array.

**Space Complexity:** 
* **Worst:** $O(1)$ - It is an in-place sorting algorithm. It only requires a single variable to temporarily hold the current value being inserted, regardless of the array's size.

**Key Characteristics:**
* **The "Playing Cards" Analogy:** It builds the final sorted array one item at a time, exactly how you might sort a hand of playing cards. You pick up a card, look at the cards you are already holding in your left hand, and insert the new card into its correct ordered slot.
* **Highly Efficient for Small/Nearly Sorted Data:** Even though it is $O(n^2)$, it has incredibly low overhead. Advanced algorithms like Python's built-in Timsort actually switch to Insertion Sort when dealing with small "chunks" of an array because it's faster for tiny datasets than the complex logic of Merge/Quick sort.
* **Stable:** If two elements have the same value, their relative order is preserved.

```python
# --- EXAMPLE 1: Standard Insertion Sort ---
def insertion_sort(arr: list) -> list:
    # We start at index 1 because we consider the first element (index 0) 
    # to already be a "sorted" sub-list of size 1.
    
    for i in range(1, len(arr)):
        # The 'key' is the current card we just picked up
        key = arr[i]
        
        # 'j' is the index of the last card in our sorted hand (to the left of 'i')
        j = i - 1
        
        # We look at the cards in our sorted hand from right to left.
        # If the card in our hand (arr[j]) is greater than our key card,
        # we shift the hand card one space to the right to make room.
        while j >= 0 and arr[j] > key:
            arr[j + 1] = arr[j]
            j -= 1
            
        # Once we find a card smaller than our key (or we reach the very beginning),
        # we drop the key card into the empty space we created.
        arr[j + 1] = key
        
    return arr


# --- EXAMPLE 2: Tracking the Shifts (Conceptual Demo) ---
def insertion_sort_verbose(arr: list) -> list:
    # Same logic, but prints out the shifting process so you can visually 
    # see the "sorted" portion of the array growing on the left side.
    
    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1
        print(f"Inserting {key} into sorted portion {arr[:i]}")
        
        while j >= 0 and arr[j] > key:
            arr[j + 1] = arr[j]
            j -= 1
            
        arr[j + 1] = key
        print(f"Current array state: {arr}")
        print("-" * 20)
        
    return arr
```

### 5) Timsort

**Time Complexity:**
* **Best:** $O(n)$ - Occurs when the data is already sorted (or reverse-sorted). Timsort detects these naturally occurring ordered sequences and effortlessly zips through them in a single linear pass.
* **Average:** $O(n \log n)$ - The standard performance for randomized data.
* **Worst:** $O(n \log n)$ - Like Merge Sort, it mathematically guarantees a worst-case limit of $O(n \log n)$.

**Space Complexity:** 
* **Worst:** $O(n)$ - Because it relies on Merge Sort to combine its sub-arrays, it requires additional memory for the temporary arrays during the merge phase.

**Key Characteristics:**
* **The Real-World Champion:** Timsort is a hybrid sorting algorithm derived from Merge Sort and Insertion Sort. It was designed specifically for Python in 2002 by Tim Peters and is now the default sorting algorithm for Python (`.sort()` and `sorted()`), Java, Android, and V8 (JavaScript).
* **Exploits Natural Order ("Runs"):** Real-world data is rarely 100% random; it usually contains sub-segments that are already sorted. Timsort searches for these segments (called "runs"). 
* **The Hybrid Approach:** If a run is too small, Timsort uses **Insertion Sort** (which is blazing fast for small datasets) to boost its size. Once the array is divided into these sorted runs, it uses **Merge Sort** to combine them together.
    * **Stable Sort:** It preserves the relative order of equal elements.

```python
# --- EXAMPLE 1: Python's Built-In Sort (Actual Timsort) ---
def python_timsort(arr: list) -> list:
    # Under the hood, Python's built-in sorted() function and .sort() method 
    # are implemented in C using Timsort.
    # In an interview setting, this is how you "use" Timsort.
    return sorted(arr)


# --- EXAMPLE 2: Conceptual "Mini-Timsort" Implementation ---
# Note: A true production Timsort is incredibly complex. This is a simplified 
# version to demonstrate the hybrid Insertion/Merge logic.

MIN_MERGE = 32

def calc_min_run(n: int) -> int:
    # Calculates a minimum run length (usually between 32 and 64)
    r = 0
    while n >= MIN_MERGE:
        r |= n & 1
        n >>= 1
    return n + r

def insertion_sort_for_tim(arr: list, left: int, right: int):
    # Standard insertion sort, but heavily localized to just sort a specific "chunk"
    for i in range(left + 1, right + 1):
        j = i
        while j > left and arr[j] < arr[j - 1]:
            arr[j], arr[j - 1] = arr[j - 1], arr[j]
            j -= 1

def merge_for_tim(arr: list, l: int, m: int, r: int):
    # Standard merge function to combine the sorted chunks
    len1, len2 = m - l + 1, r - m
    left, right = [], []
    for i in range(0, len1): left.append(arr[l + i])
    for i in range(0, len2): right.append(arr[m + 1 + i])

    i, j, k = 0, 0, l
    while i < len1 and j < len2:
        if left[i] <= right[j]:
            arr[k] = left[i]
            i += 1
        else:
            arr[k] = right[j]
            j += 1
        k += 1

    while i < len1:
        arr[k] = left[i]
        k += 1
        i += 1
    while j < len2:
        arr[k] = right[j]
        k += 1
        j += 1

def simplified_timsort(arr: list):
    n = len(arr)
    min_run = calc_min_run(n)
    
    # 1. Chunk the array into pieces of size 'min_run' and sort them using Insertion Sort
    for start in range(0, n, min_run):
        end = min(start + min_run - 1, n - 1)
        insertion_sort_for_tim(arr, start, end)
        
    # 2. Merge those chunks together using Merge Sort
    size = min_run
    while size < n:
        for left in range(0, n, 2 * size):
            mid = min(n - 1, left + size - 1)
            right = min(left + 2 * size - 1, n - 1)
            if mid < right:
                merge_for_tim(arr, left, mid, right)
        size = 2 * size
        
    return arr
```

### 6) Bubble Sort

**Time Complexity:**
* **Best:** $O(n)$ - This happens when the array is already sorted. By adding a simple "swapped" flag, the algorithm can detect that no elements were moved during its first pass and terminate immediately.
* **Average:** $O(n^2)$ - Elements are jumbled, requiring the nested loops to repeatedly traverse the array and swap items.
* **Worst:** $O(n^2)$ - The array is in reverse order, meaning every single element must be swapped all the way across the array.

**Space Complexity:** 
* **Worst:** $O(1)$ - It is an in-place sorting algorithm. Elements are swapped directly within the existing array structure, requiring only a single temporary variable for the swap operation.

**Key Characteristics:**
* **The "Bubbling" Effect:** It gets its name because the largest elements "bubble" up to the end of the array during each pass, much like bubbles rising in a glass of soda. It does this by repeatedly comparing adjacent pairs and swapping them if they are in the wrong order.
* **Educational Use:** Bubble Sort is incredibly inefficient for real-world large datasets. In professional software engineering, it is almost never used in production. It is primarily taught to introduce the concept of sorting algorithms and nested loops.
* **Stable Sort:** It preserves the relative order of identical elements because it strictly swaps strictly when one is *greater* than the other (not equal).

```python
# --- EXAMPLE 1: Optimized Bubble Sort (Interview Standard) ---
def bubble_sort(arr: list) -> list:
    n = len(arr)
    
    # Traverse through all array elements
    for i in range(n):
        # We use a flag to track if any swaps happen in this pass.
        # This is what allows the Best Case Time Complexity to be O(n).
        swapped = False
        
        # Last 'i' elements are already in place, so we don't need to check them.
        for j in range(0, n - i - 1):
            
            # Compare adjacent elements
            if arr[j] > arr[j + 1]:
                # Swap them if they are in the wrong order
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                swapped = True
                
        # If no two elements were swapped by the inner loop, the array is perfectly sorted.
        # We can break early to save time.
        if not swapped:
            break
            
    return arr


# --- EXAMPLE 2: Unoptimized Bubble Sort (For Comparison) ---
def bubble_sort_unoptimized(arr: list) -> list:
    # This version lacks the 'swapped' flag and does not shrink the inner loop.
    # Therefore, even if the array is perfectly sorted, it will ALWAYS run in O(n²) time.
    # Never write this version in an interview!
    
    n = len(arr)
    for i in range(n):
        for j in range(n - 1):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                
    return arr
```

### 7) Shellsort

**Time Complexity:**
* **Best:** $O(n \log n)$ - When the array is already sorted, the inner loops do very little work, making the passes highly efficient.
* **Average:** *Depends on gap sequence* - Unlike other algorithms, Shellsort's average time complexity varies wildly based on the mathematical sequence used to define its "gaps". It typically ranges from $O(n^{1.25})$ to $O(n^{1.5})$.
* **Worst:** $O(n^2)$ - This is the worst-case scenario when using Donald Shell's original gap sequence ($n/2, n/4 \dots 1$). More advanced gap sequences (like Sedgewick's or Hibbard's) can lower the worst-case bound to $O(n^{4/3})$ or $O(n \log^2 n)$.

**Space Complexity:** 
* **Worst:** $O(1)$ - It is strictly an in-place sorting algorithm. It only requires a single temporary variable for swapping, regardless of the dataset's size.

**Key Characteristics:**
* **Upgraded Insertion Sort:** Shellsort is essentially a highly optimized version of Insertion Sort. Standard Insertion Sort is slow because it only shifts elements one position at a time. Shellsort fixes this by allowing elements to take massive leaps across the array.
* **The "Gap" Concept:** It works by comparing elements separated by a specific "gap" (or interval). It sorts these distanced elements first. Then, the gap is shrunk, and the process repeats. 
* **The Final Pass:** The gap continuously shrinks until it reaches exactly `1`. When the gap is `1`, the algorithm performs a standard Insertion Sort. However, because the larger gaps have already moved the elements roughly into the correct vicinity, the array is "nearly sorted". Insertion Sort runs in highly efficient $O(n)$ time on nearly sorted arrays, making this final pass lightning fast.
* **Not Stable:** Because elements leapfrog over each other across large gaps, identical elements can lose their original relative order.

```python
# --- EXAMPLE 1: Shellsort (Using Shell's Original Gap Sequence) ---
def shell_sort(arr: list) -> list:
    n = len(arr)
    
    # Start with a large gap, then reduce the gap by half each iteration.
    # (This is Donald Shell's original sequence: n/2, n/4, ..., 1)
    gap = n // 2
    
    # We continue until the gap reaches 0
    while gap > 0:
        
        # Perform a "gapped" insertion sort for this specific gap size.
        # We start at index 'gap' because the elements before it are 
        # considered single-element sorted sublists for their respective gaps.
        for i in range(gap, n):
            
            # Store the current element to be inserted in its correct gapped position
            temp = arr[i]
            
            # Shift earlier gap-sorted elements up until the correct location 
            # for 'temp' is found.
            j = i
            while j >= gap and arr[j - gap] > temp:
                arr[j] = arr[j - gap]
                j -= gap
                
            # Place the 'temp' element into its correct slot
            arr[j] = temp
            
        # Shrink the gap for the next pass
        gap //= 2
        
    return arr
```

### 8) Bucket Sort

**Time Complexity:**
* **Best:** $O(n + k)$ - Occurs when the data is perfectly uniformly distributed across the $k$ buckets. Each bucket ends up with very few elements (or just one), making the sorting step inside each bucket nearly instantaneous.
* **Average:** $O(n + k)$ (often simplified to $O(n)$) - When elements are evenly distributed, each bucket contains a small, constant number of items. Distributing the items takes $O(n)$, and sorting all the tiny buckets takes roughly $O(n)$ time combined.
* **Worst:** $O(n^2)$ - Occurs when the data distribution is heavily skewed, causing all elements to fall into a *single* bucket. If that overloaded bucket is then sorted using an $O(n^2)$ algorithm (like Insertion Sort), the entire algorithm collapses to quadratic time.

**Space Complexity:** 
* **Worst:** $O(n + k)$ (often written just as $O(n)$) - We must allocate memory for $k$ empty buckets (arrays/lists) and then store all $n$ original elements inside those buckets.

**Key Characteristics:**
* **Scatter-Gather:** The algorithm works by distributing ("scattering") the input elements into a series of "buckets" based on their value. Each bucket is sorted individually, and then the buckets are concatenated ("gathered") in order.
* **Distribution Dependent:** Bucket Sort is rarely used as a general-purpose sorting algorithm. It is incredibly fast *only* if you know the data is uniformly distributed across a specific range (like floating-point values between 0.0 and 1.0, or ages between 1 and 100).
* **Breaks the Barrier:** Because it groups items mathematically rather than purely comparing them (`A > B`), it can achieve linear $O(n)$ time, breaking the traditional $O(n \log n)$ barrier of comparison-based sorts.

```python
# --- EXAMPLE 1: Classic Bucket Sort (For floats between 0.0 and 1.0) ---
def bucket_sort(arr: list) -> list:
    # This is the classic textbook implementation for Bucket Sort, 
    # assuming all numbers in 'arr' are uniformly distributed floats 
    # in the range [0.0, 1.0).
    
    if len(arr) == 0:
        return arr
        
    n = len(arr)
    k = n  # Usually, we create as many buckets as there are elements
    
    # 1. Create 'k' empty buckets
    buckets = [[] for _ in range(k)]
    
    # 2. SCATTER: Put array elements into different buckets
    for num in arr:
        # Multiply the decimal by k to find its bucket index.
        # e.g., if k=10, the float 0.23 goes into bucket int(10 * 0.23) = index 2.
        index = int(k * num)
        
        # Handle edge case where num might be exactly 1.0
        if index == k:
            index -= 1
            
        buckets[index].append(num)
        
    # 3. SORT: Sort individual buckets
    # For small buckets, Python's built-in Timsort behaves like Insertion Sort, 
    # which is blazing fast for small datasets.
    for bucket in buckets:
        bucket.sort()
        
    # 4. GATHER: Concatenate all sorted buckets into the final array
    sorted_array = []
    for bucket in buckets:
        sorted_array.extend(bucket)
        
    return sorted_array


# --- EXAMPLE 2: Conceptual Bucket Sort for Integers ---
def bucket_sort_integers(arr: list, bucket_size: int = 5) -> list:
    # If dealing with integers, we calculate the bucket index using 
    # the maximum and minimum values in the array.
    if not arr:
        return arr
        
    min_val, max_val = min(arr), max(arr)
    
    # Determine the number of buckets (k) based on the chosen bucket_size
    k = (max_val - min_val) // bucket_size + 1
    buckets = [[] for _ in range(k)]
    
    # Scatter
    for num in arr:
        index = (num - min_val) // bucket_size
        buckets[index].append(num)
        
    # Sort and Gather
    sorted_array = []
    for bucket in buckets:
        bucket.sort()
        sorted_array.extend(bucket)
        
    return sorted_array
```

### 9) Radix Sort

**Time Complexity:**
* **Best:** $O(nk)$
* **Average:** $O(nk)$
* **Worst:** $O(nk)$
* **Why?** Radix Sort is a non-comparison algorithm. It does not compare `A > B`. Instead, it groups numbers digit by digit. If you have $n$ numbers, and the longest number has $k$ digits, the algorithm will perform exactly $k$ passes over the $n$ numbers. Therefore, it always takes $O(nk)$ time.

**Space Complexity:** 
* **Worst:** $O(n + b)$ (Often written as $O(n + k)$ depending on notation) - It requires an output array of size $n$ to hold the sorted elements during each pass, plus an array of size $b$ (where $b$ is the base, e.g., 10 for the decimal system) to count the digit occurrences.

**Key Characteristics:**
* **Non-Comparison Sort:** It completely bypasses the $O(n \log n)$ mathematical limit of comparison-based sorts (like Merge/Quick sort) by exploiting the specific structure of integers or strings.
* **Digit by Digit:** It usually works from the Least Significant Digit (LSD) to the Most Significant Digit (MSD). For example, sorting 103, 802, and 401: it first sorts by the 1s place, then the 10s place, then the 100s place.
* **Requires a Stable Subroutine:** Radix sort *must* use a stable sorting algorithm internally to sort each individual digit pass. If it isn't stable, the work done sorting the previous digits will be ruined. It almost exclusively uses **Counting Sort** for this subroutine.

```python
# --- EXAMPLE 1: Radix Sort (Using Counting Sort as a subroutine) ---

def counting_sort_for_radix(arr: list, exp: int):
    # This is a specialized version of Counting Sort designed strictly 
    # to sort an array based on a specific digit place (exp).
    
    n = len(arr)
    output = [0] * n    # The temporary array to hold sorted values
    count = [0] * 10    # Base 10 (digits 0-9)
    
    # 1. Store the count of occurrences for the current digit
    for i in range(n):
        # To isolate the digit: divide by the exponent (1, 10, 100...) and modulo 10
        digit = (arr[i] // exp) % 10
        count[digit] += 1
        
    # 2. Modify the count array to store the actual position of these digits in output
    for i in range(1, 10):
        count[i] += count[i - 1]
        
    # 3. Build the output array
    # We MUST loop backwards to ensure the sort remains STABLE. 
    # Stability is required for Radix Sort to function correctly across multiple digits.
    i = n - 1
    while i >= 0:
        digit = (arr[i] // exp) % 10
        output[count[digit] - 1] = arr[i]
        count[digit] -= 1
        i -= 1
        
    # 4. Copy the sorted output back into the original array
    for i in range(n):
        arr[i] = output[i]


def radix_sort(arr: list) -> list:
    # Radix sort only works natively on positive integers.
    if not arr:
        return arr
        
    # Find the maximum number to determine how many digits we need to process
    max_val = max(arr)
    
    # 'exp' is 10^i where i is the current digit place (1, 10, 100, 1000...)
    exp = 1
    
    # Continue sorting digit by digit until we exceed the largest number
    while max_val // exp > 0:
        counting_sort_for_radix(arr, exp)
        exp *= 10
        
    return arr
```

### 10) Counting Sort

**Time Complexity:**
* **Best:** $O(n + k)$
* **Average:** $O(n + k)$
* **Worst:** $O(n + k)$
* **Why?** The algorithm takes $O(n)$ time to iterate through the input array to count the occurrences of each element. It then takes $O(k)$ time to iterate through the count array to reconstruct the sorted output. Since there are no nested loops depending on $n$ entirely, the time is strictly bounded by the sum of $n$ and $k$.

**Space Complexity:** 
* **Worst:** $O(k)$ - You must allocate a new "count" array of size $k$ (where $k$ is the difference between the maximum and minimum values in the dataset). *Note: If a stable sort is required, it also requires an output array of size $n$, making the space complexity $O(n + k)$.*

**Key Characteristics:**
* **Non-Comparison Sort:** Like Radix Sort, it completely ignores the $O(n \log n)$ barrier because it never compares `A > B`. Instead, it uses the actual values of the elements as direct indices in a new array.
* **Integer Restrictive:** It generally only works for discrete values like integers or categorical data. You cannot easily counting-sort floating-point numbers or complex strings.
* **The "k" Trap:** Counting Sort is incredibly fast (linear time) *only* if $k$ (the range of values) is relatively small compared to $n$. If you have an array of just two elements, `[1, 1000000]`, Counting Sort will waste massive amounts of time and memory creating and iterating through an array of one million empty zeroes. It trades space for time.

```python
# --- EXAMPLE 1: Stable Counting Sort (Interview/Textbook Standard) ---
def counting_sort_stable(arr: list) -> list:
    if not arr:
        return arr
        
    # 1. Find the maximum value to determine the size of the count array ('k')
    # If the array has negative numbers, we would also find the min and offset the indices.
    # For simplicity, we assume positive integers here.
    max_val = max(arr)
    
    # Initialize the count array with zeros
    # Size is max_val + 1 to accommodate 0-based indexing
    count = [0] * (max_val + 1)
    output = [0] * len(arr)
    
    # 2. Store the count of each unique element
    for num in arr:
        count[num] += 1
        
    # 3. Modify the count array to store the cumulative sum.
    # This tells us the exact ending position of each element in the sorted array.
    for i in range(1, len(count)):
        count[i] += count[i - 1]
        
    # 4. Build the output array.
    # We iterate backwards through the original array to maintain STABILITY 
    # (identical elements retain their relative original order).
    for i in range(len(arr) - 1, -1, -1):
        num = arr[i]
        # count[num] - 1 gives us the correct 0-based index for this number
        output[count[num] - 1] = num
        # Decrement the count so the next identical number goes to the left of this one
        count[num] -= 1
        
    return output


# --- EXAMPLE 2: Simplified Unstable Counting Sort (In-Place Overwrite) ---
def counting_sort_simple(arr: list) -> list:
    # If stability does not matter (e.g., we are just sorting primitive integers 
    # and not complex objects), we can skip the cumulative sum and output array.
    
    if not arr:
        return arr
        
    max_val = max(arr)
    count = [0] * (max_val + 1)
    
    # Count occurrences
    for num in arr:
        count[num] += 1
        
    # Rebuild the original array by simply overwriting it based on the counts
    index = 0
    for num in range(len(count)):
        # While we still have occurrences of 'num' left to place...
        while count[num] > 0:
            arr[index] = num
            index += 1
            count[num] -= 1
            
    return arr
```

### 11) Selection Sort

**Time Complexity:**
* **Best:** $O(n^2)$
* **Average:** $O(n^2)$
* **Worst:** $O(n^2)$
* **Why?** Unlike Insertion or Bubble Sort, Selection Sort has no early exit mechanism. It strictly divides the array into a "sorted" and "unsorted" section. To find the next smallest element, it *must* scan every single remaining item in the unsorted section. Even if the array is perfectly sorted from the start, it will still blindly execute its nested loops to double-check everything, resulting in a strict $O(n^2)$ time in all cases.

**Space Complexity:** 
* **Worst:** $O(1)$ - It sorts in-place, requiring only a couple of integer variables to keep track of indices during the swaps.

**Key Characteristics:**
* **Search and Swap:** It works by repeatedly finding the absolute minimum element from the unsorted part of the array and putting it at the beginning. 
* **Minimizes Swaps:** While it does $O(n^2)$ comparisons, it only performs a maximum of $O(n)$ *swaps* (exactly one swap per outer loop iteration). In rare systems where writing data to memory is significantly more computationally expensive than reading it, Selection Sort can actually outperform other $O(n^2)$ algorithms.
* **Unstable Sort:** Standard Selection Sort is generally not stable. Because it swaps elements over long distances, an identical element might be inadvertently pushed backwards past another identical element.

```python
# --- EXAMPLE 1: Standard Selection Sort ---
def selection_sort(arr: list) -> list:
    n = len(arr)
    
    # We step through the entire array.
    # 'i' represents the boundary between the sorted (left) and unsorted (right) portions.
    for i in range(n):
        
        # Assume the first element of the unsorted section is the smallest
        min_idx = i
        
        # Iterate through the REST of the unsorted section to find the true minimum
        for j in range(i + 1, n):
            if arr[j] < arr[min_idx]:
                min_idx = j
                
        # Swap the found minimum element with the first element of the unsorted section
        # (This moves the boundary of the sorted section one step to the right)
        arr[i], arr[min_idx] = arr[min_idx], arr[i]
        
    return arr


# --- EXAMPLE 2: Verbose Selection Sort (To visualize the boundary) ---
def selection_sort_verbose(arr: list) -> list:
    n = len(arr)
    
    for i in range(n):
        min_idx = i
        for j in range(i + 1, n):
            if arr[j] < arr[min_idx]:
                min_idx = j
                
        # Perform swap
        arr[i], arr[min_idx] = arr[min_idx], arr[i]
        
        # Visualize the sorted vs. unsorted portions after each pass
        sorted_part = arr[:i+1]
        unsorted_part = arr[i+1:]
        print(f"Pass {i+1}: Sorted {sorted_part} | Unsorted {unsorted_part}")
        
    return arr
```

## Common Data Structures: Time Complexity Cheat Sheet

### Static Array
* **Access (Index):** $O(1)$
* **Search (Value):** $O(n)$
* **Insertion:** $O(n)$ — requires shifting elements to make room.
* **Deletion:** $O(n)$ — requires shifting elements to close the gap.

### Dynamic Array (Python `list`)
* **Access (Index):** $O(1)$
* **Search (Value):** $O(n)$
* **Insertion (Append at End):** $O(1)$ average — $O(n)$ worst-case when memory reallocation/resizing is triggered.
* **Insertion / Deletion (Middle):** $O(n)$ — requires shifting elements.

### Linked List (Singly / Doubly)
* **Access (Index):** $O(n)$ — must traverse sequentially from the head.
* **Search (Value):** $O(n)$
* **Insertion / Deletion (at Head):** $O(1)$
* **Insertion / Deletion (Middle):** $O(n)$ to find the node, though the actual pointer swap is $O(1)$. 
* *(Note: Doubly Linked Lists allow $O(1)$ deletion at the tail if a tail pointer is maintained).*

### Hash Table (Python `dict` / `set`)
* **Access / Search:** $O(1)$ average — $O(n)$ worst-case (if massive hash collisions occur).
* **Insertion:** $O(1)$ average — $O(n)$ worst-case.
* **Deletion:** $O(1)$ average — $O(n)$ worst-case.

### Stack (LIFO - Last In, First Out)
* **Access (Peek Top):** $O(1)$
* **Search:** $O(n)$
* **Insertion (Push):** $O(1)$
* **Deletion (Pop):** $O(1)$

### Queue (FIFO - First In, First Out)
* **Access (Peek Front):** $O(1)$
* **Search:** $O(n)$
* **Insertion (Enqueue / Push):** $O(1)$
* **Deletion (Dequeue / Pop):** $O(1)$

### Binary Search Tree (BST)
* **Access / Search:** $O(\log n)$ average — $O(n)$ worst-case (if the tree becomes unbalanced and forms a straight line).
* **Insertion:** $O(\log n)$ average — $O(n)$ worst-case.
* **Deletion:** $O(\log n)$ average — $O(n)$ worst-case.

### Min-Heap / Max-Heap (Priority Queue)
* **Access (Find Min/Max):** $O(1)$ — the root node is always the min/max.
* **Search (Specific Value):** $O(n)$ — heaps are not rigidly sorted, requiring a full scan.
* **Insertion:** $O(\log n)$ — must "bubble up" to restore the heap property.
* **Deletion (Extract Min/Max):** $O(\log n)$ — must "bubble down" to restore the heap property.

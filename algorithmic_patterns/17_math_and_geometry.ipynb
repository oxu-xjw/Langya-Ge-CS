### Problem 001: Non-Cyclical Number / Happy Number (LeetCode 202)

### Problem Definition and Constraints
A non-cyclical (or "Happy") number is determined by the following process:
1. Replace the number with the sum of the squares of its digits.
2. Repeat the process until the number equals 1 (where it will stay 1), or it loops endlessly in a cycle that does not include 1.
3. Return `true` if it reaches 1, otherwise return `false`.

* Constraints:
  * 1 <= n <= 1000 (though standard LeetCode constraints go up to 2^31 - 1)

### Examples
* **Example 1:**
  * Input: `n = 100`
  * Output: `true`
  * Explanation: $1^2 + 0^2 + 0^2 = 1$. It reaches 1 immediately.
* **Example 2:**
  * Input: `n = 101`
  * Output: `false`
  * Explanation: 101 -> 2 -> 4 -> 16 -> 37 -> 58 -> 89 -> 145 -> 42 -> 20 -> 4. The number 4 has already been seen, so it is stuck in an infinite cycle.

### Optimized Approach (Hash Set Cycle Detection)
We need two distinct pieces of logic:
1. **A Helper Function:** To extract the digits of a number, square them, and sum them. We can do this mathematically using modulo `% 10` (to grab the last digit) and integer division `// 10` (to chop off the last digit).
2. **Cycle Detection:** We use a `HashSet` to keep track of every number we have generated. At each step, we check if the new number is already in our set. If it is, we've detected an infinite loop and return `False`. If the number becomes 1, we successfully break out and return `True`.

* Time Complexity: $O(\log n)$ — The number of digits in a number $n$ is given by $\log_{10}(n)$. Calculating the sum of squares takes time proportional to the number of digits. The cycle length is strictly bounded (numbers quickly drop below 243).
* Space Complexity: $O(\log n)$ — To store the numbers in our Hash Set.

```python
class Solution:
    def isNonCyclical(self, n: int) -> bool:
        # Our "notebook" of dropped breadcrumbs
        seen = set()
        
        # Keep wandering as long as we haven't hit the target (1)
        while n != 1:
            # If we see our own breadcrumb, we are walking in a circle!
            if n in seen:
                return False
                
            # Drop a breadcrumb at the current clearing
            seen.add(n)
            
            # Teleport to the next clearing using our helper function
            n = self.sumOfSquares(n)
            
        # If the while loop naturally exits, it means n == 1!
        return True

    # --- HELPER FUNCTION: The "Mathy" Way ---
    def sumOfSquares(self, n: int) -> int:
        total_sum = 0
        
        while n > 0:
            # 1. Grab the right-most digit (e.g., 145 % 10 = 5)
            digit = n % 10
            
            # 2. Square it and add to our running total
            total_sum += digit ** 2
            
            # 3. Chop off the right-most digit (e.g., 145 // 10 = 14)
            n = n // 10
            
        return total_sum

    # --- HELPER FUNCTION: The "Pythonic" Shortcut ---
    # Alternatively, Python lets us treat the number as a string to easily grab digits.
    # This is slightly less efficient but very fast to write in an interview!
    def sumOfSquaresStringMethod(self, n: int) -> int:
        total_sum = 0
        for char in str(n):
            total_sum += int(char) ** 2
        return total_sum
```

### Problem 002: Plus One (LeetCode 66)

### Problem Definition and Constraints
You are given a large integer represented as an integer array `digits`, where each `digits[i]` is the $i$-th digit of the integer. The digits are ordered from most significant to least significant in left-to-right order.
Return the array of digits after incrementing the large integer by one.

* Constraints:
  * 1 <= digits.length <= 100
  * 0 <= digits[i] <= 9
  * The array does not contain any leading zeros (except for the number 0 itself).

### Examples
* **Example 1:**
  * Input: `digits = [1,2,3]`
  * Output: `[1,2,4]`
* **Example 2:**
  * Input: `digits = [1,9,9]`
  * Output: `[2,0,0]`
* **Example 3:**
  * Input: `digits = [9,9,9]`
  * Output: `[1,0,0,0]`

### Brute Force Approach (Type Conversion)
Convert the array of digits into a string, convert the string into a single large integer, add 1 mathematically, convert it back to a string, and then split it back out into an array of integers. 
* Time Complexity: $O(n)$ — String operations and conversions iterate over all digits.
* Space Complexity: $O(n)$ — Requires memory to store the intermediate strings.
* *Note: While easy in Python, this approach defeats the algorithmic purpose of the question and can overflow in strongly-typed languages like Java or C++.*

### Optimized Approach (Reverse Iteration Simulation)
We simulate how a physical odometer works by iterating through the array backwards (from right to left). 
If the current digit is less than 9, we simply add 1 to it and instantly return the array (no further carries are needed). 
If the digit is exactly 9, we change it to 0, and let the loop continue to the next digit to the left to "carry the one".
If the loop finishes checking the entire array and hasn't returned yet, it means *every* digit was a 9 (like `[9,9,9]`), which all became `[0,0,0]`. We handle this edge case by prepending a `1` to the start of the array.
* Time Complexity: $O(n)$ — In the worst-case scenario (all 9s), we iterate through every digit once.
* Space Complexity: $O(1)$ — We modify the array in-place. (Prepending an element is $O(n)$ space in Python, but conceptually the simulation is $O(1)$ extra space).

```python
from typing import List

class Solution:
    def plusOne(self, digits: List[int]) -> List[int]:
        
        # Iterate backwards through the array (start at last index, go to -1, step -1)
        for i in range(len(digits) - 1, -1, -1):
            
            # If the digit is NOT a 9, we just add 1 and we are completely done!
            if digits[i] < 9:
                digits[i] += 1
                return digits
                
            # If the digit IS a 9, it rolls over to 0.
            # The loop will naturally continue to the next digit to the left.
            digits[i] = 0
            
        # If we successfully made it all the way out of the loop without returning,
        # it means every single digit was a 9 and rolled over to 0 (e.g., [0,0,0]).
        # We need to add a brand new 1 to the front of the array.
        return [1] + digits
```

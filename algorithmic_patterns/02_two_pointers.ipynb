### Problem 001: Valid Palindrome (LeetCode 125)

### Problem Definition and Constraints
Given a string `s`, return `true` if it is a palindrome, otherwise return `false`. A palindrome is a string that reads the same forward and backward. It is also case-insensitive and ignores all non-alphanumeric characters (A-Z, a-z, 0-9).

* Constraints:
  * 1 <= s.length <= 1000
  * `s` is made up of only printable ASCII characters.

### Examples
* **Example 1:**
  * Input: `s = "Was it a car or a cat I saw?"`
  * Output: `true`
  * Explanation: After removing punctuation and spaces, and lowering the case, we get "wasitacaroracatisaw", which is a palindrome.
* **Example 2:**
  * Input: `s = "tab a cat"`
  * Output: `false`

### Brute Force Approach
The intuitive approach is to create a brand new, filtered string. We iterate through the original string, checking if each character is alphanumeric. If it is, we convert it to lowercase and append it to our new string. Finally, we compare the new string against a reversed copy of itself.
* Time Complexity: O(n) — We iterate through the string to clean it, and Python's string reversal also takes O(n) time.
* Space Complexity: O(n) — We are actively allocating new memory to build the cleaned string and the reversed string, scaling directly with the input size.

### Optimized Approach (Two Pointers)
To achieve O(1) space complexity, we must perform the check "in-place" without creating any new strings. We initialize two pointers: one at the start of the string (left) and one at the end (right). We move the pointers inward. If a pointer lands on a non-alphanumeric character, we advance it until it finds a valid character. Once both pointers are on valid characters, we compare their lowercase values. If they mismatch, it is not a palindrome. If the pointers meet in the middle without any mismatches, the string is a valid palindrome.
* Time Complexity: O(n) — The pointers traverse the string from opposite ends, meeting in the middle. Every character is evaluated exactly once.
* Space Complexity: O(1) — We only store two integer variables for the index pointers, requiring no scaling auxiliary memory.

```python
 class Solution:
    
    # --- BRUTE FORCE APPROACH ---
    def isPalindrome_brute_force(self, s: str) -> bool:
        cleaned_string = ""
        
        for char in s:
            # .isalnum() is a built-in Python method that returns True 
            # if the character is a letter or a number.
            if char.isalnum():
                cleaned_string += char.lower()
                
        # In Python, [::-1] is the idiomatic way to slice a string backwards (reverse it)
        return cleaned_string == cleaned_string[::-1]


    # --- OPTIMIZED APPROACH ---
    def isPalindrome(self, s: str) -> bool:
        # Initialize our two pointers at opposite ends of the string
        left = 0
        right = len(s) - 1
        
        # The loop runs until the two pointers cross in the middle
        while left < right:
            
            # If the left pointer is on a space/punctuation, step forward.
            # We must include 'left < right' inside this inner loop to ensure 
            # the pointer doesn't accidentally run off the end of the string.
            while left < right and not s[left].isalnum():
                left += 1
                
            # If the right pointer is on a space/punctuation, step backward.
            while left < right and not s[right].isalnum():
                right -= 1
                
            # Both pointers are now guaranteed to be on valid letters or numbers.
            # Compare them in lowercase. If they don't match, we fail immediately.
            if s[left].lower() != s[right].lower():
                return False
                
            # If they matched, successfully move both fingers inward for the next check.
            left += 1
            right -= 1
            
        # If the pointers touch/cross without ever hitting the 'return False' condition,
        # the entire string mirrors itself perfectly.
        return True
```

### Problem 002: Two Integer Sum II (LeetCode 167)

### Problem Definition and Constraints
Given a **1-indexed** array of integers `numbers` that is sorted in non-decreasing order, find two numbers that add up to a specific `target` number.
* Return the indices of the two numbers as an array `[index1, index2]`.
* **Constraints:** * You must use exactly $O(1)$ extra space.
  * There is exactly one valid solution.
  * You may not use the same element twice (i.e., `index1` cannot equal `index2`).

### Examples
* **Example 1:**
  * Input: `numbers = [1, 2, 3, 4]`, `target = 3`
  * Output: `[1, 2]` (The numbers at index 1 and index 2 are 1 and 2, which sum to 3).

### Brute Force & Sub-optimal Approaches
1. **Brute Force:** Nested loops to check every possible pair. Takes $O(n^2)$ time. Too slow.
2. **Standard Two Sum (Hash Map):** We could use the exact same Hash Map strategy from standard Two Sum to solve this in $O(n)$ time. However, a Hash Map requires $O(n)$ extra memory to store the values, which directly violates the $O(1)$ space constraint of this specific problem.

### Optimized Approach (Two Pointers)
Because the array is **already sorted**, we can use two pointers starting at opposite ends of the array: a `left` pointer at the beginning (smallest numbers) and a `right` pointer at the end (largest numbers).

We calculate the sum of the numbers at our two pointers and compare it to the target:
1. **If `sum == target`**: We found our answer! We just add 1 to our pointers (since the problem demands 1-based indexing) and return them.
2. **If `sum > target`**: The total is too big. The only way to make the sum smaller is to move our `right` pointer one step to the left (pointing to a smaller number).
3. **If `sum < target`**: The total is too small. The only way to make the sum larger is to move our `left` pointer one step to the right (pointing to a bigger number).

* **Time Complexity:** $O(n)$ — In the worst-case scenario, the pointers meet in the exact middle, meaning we scan the array exactly once.
* **Space Complexity:** $O(1)$ — We are only using two integer variables (`left` and `right`) to track our positions, completely eliminating the need for a Hash Map.

```python
from typing import List

class Solution:
    def twoSum(self, numbers: List[int], target: int) -> List[int]:
        # Initialize our two pointers at the absolute ends of the array
        left = 0
        right = len(numbers) - 1
        
        # The pointers will move towards each other until they meet
        while left < right:
            current_sum = numbers[left] + numbers[right]
            
            if current_sum > target:
                # Sum is too big, shrink it by moving the right pointer to a smaller number
                right -= 1
                
            elif current_sum < target:
                # Sum is too small, grow it by moving the left pointer to a larger number
                left += 1
                
            else:
                # We found the exact target! 
                # The problem requires 1-indexed output, so we add 1 to both indices
                return [left + 1, right + 1]
```

### Problem 003: 3Sum (LeetCode 15)

### Problem Definition and Constraints
Given an integer array `nums`, return all the triplets `[nums[i], nums[j], nums[k]]` such that they add up exactly to `0`. 
* The indices must be distinct ($i \neq j \neq k$).
* The output **must not contain any duplicate triplets**. 

### Examples
* **Example 1:**
  * Input: `nums = [-1, 0, 1, 2, -1, -4]`
  * Output: `[[-1, -1, 2], [-1, 0, 1]]`
* **Example 2:**
  * Input: `nums = [0, 1, 1]`
  * Output: `[]` (No triplets add to 0)

### Brute Force Approach
The naive approach is to use three nested loops to check every single combination of three numbers. 
* **Time Complexity:** $O(n^3)$. This will instantly trigger a "Time Limit Exceeded" error on LeetCode. 

### Optimized Approach (Sort + Two Pointers)
We can reduce this to $O(n^2)$ by leaning on the exact same logic we used in **Two Integer Sum II**. 
1. **Sort the array:** This takes $O(n \log n)$ time, but it is entirely worth it. Sorting groups duplicate numbers together (making them easy to skip) and enables our two-pointer math.
2. **Lock down one number:** We iterate through the array with a standard `for` loop. We treat the current number (`nums[i]`) as a locked anchor.
3. **Run Two Pointers:** If we need `nums[i] + left_num + right_num == 0`, that is the exact same math as finding two numbers that add up to a target, where our `target = -nums[i]`. We place a `left` pointer just after our anchor, and a `right` pointer at the end of the array, and squeeze them together just like the previous problem.

* **Handling Duplicates:** Because the array is sorted, duplicates are sitting right next to each other. If our `left` pointer lands on a number we just processed, we just shift it right again to avoid calculating the same exact triplet twice.
* **Time Complexity:** $O(n^2)$ — We iterate through the array once ($O(n)$) and for every element, we do a two-pointer sweep of the remaining array ($O(n)$). Sorting takes $O(n \log n)$, but $O(n^2)$ dominates it, so the final time is $O(n^2)$.
* **Space Complexity:** $O(1)$ or $O(n)$ — Depends heavily on the language's sorting algorithm. If the sorting is done purely in-place, it uses $O(1)$ auxiliary space. Python's built-in `sort()` requires $O(n)$ memory.

```python
from typing import List

class Solution:
    def threeSum(self, nums: List[int]) -> List[List[int]]:
        res = []
        
        # Step 1: Sort the array so we can use Two Pointers and skip duplicates
        nums.sort()
        
        for i, a in enumerate(nums):
            # Optimization: If the smallest number we are looking at is greater than 0,
            # it's mathematically impossible to sum to 0. We can stop searching entirely.
            if a > 0:
                break
                
            # Step 2: Skip duplicates for our "anchor" number
            # If this number is the same as the one before it, we've already found all 
            # possible triplets for it. Skip to avoid duplicate answers.
            if i > 0 and a == nums[i - 1]:
                continue
                
            # Step 3: Run standard Two Pointers on the remainder of the array
            left = i + 1
            right = len(nums) - 1
            
            while left < right:
                three_sum = a + nums[left] + nums[right]
                
                if three_sum > 0:
                    # Sum is too big, shrink it from the right
                    right -= 1
                elif three_sum < 0:
                    # Sum is too small, grow it from the left
                    left += 1
                else:
                    # We found a valid triplet!
                    res.append([a, nums[left], nums[right]])
                    
                    # Move our left pointer to start searching for another potential pair
                    left += 1
                    
                    # Step 4: Skip duplicates for the left pointer
                    # We must ensure we don't use the exact same left number again
                    while nums[left] == nums[left - 1] and left < right:
                        left += 1
                        
                    # (Note: We don't need to manually update the right pointer for duplicates. 
                    # Once left changes, the 'three_sum' will naturally be too big or too small 
                    # in the next loop, forcing the pointers to adjust correctly).
                    
        return res
```

### Problem 004: Container With Most Water (LeetCode 11)

### Problem Definition and Constraints
You are given an integer array `heights` where `heights[i]` represents the height of a vertical line drawn at index `i`. 
Find two lines that, together with the x-axis, form a container that holds the most water (i.e., has the maximum area).
* **Constraints:** * $2 \le \text{height.length} \le 1000$
  * $0 \le \text{height[i]} \le 1000$
  * You cannot slant the container.

### Examples
* **Example 1:**
  * Input: `height = [1, 7, 2, 5, 4, 7, 3, 6]`
  * Output: `36` (The container is formed by the line at index 1 of height 7, and the line at index 7 of height 6. Width = 6, Height = 6. Area = 36).

### Brute Force Approach
Check every single possible pair of lines. 
* **Time Complexity:** $O(n^2)$. You would need nested loops to calculate the area for every combination. This is too slow.

### Optimized Approach (Two Pointers)
We want to maximize the area. The formula for the area of water is:
**$\text{Area} = \text{width} \times \text{height}$**
* **Width:** The distance between the two lines (`right - left`).
* **Height:** The water can only fill up to the top of the *shorter* line, otherwise it spills over. So, `min(height[left], height[right])`.

We can use the Two Pointer approach to solve this in a single pass:
1. Start with the **maximum possible width** by placing a `left` pointer at index 0 and a `right` pointer at the very end of the array.
2. Calculate the current area and update our `max_area` variable.
3. **The Core Logic:** Which pointer do we move inward? 
   * As we move pointers inward, the **width** will *always* decrease. 
   * The only way to find a *larger* area is if we find a **taller** line to make up for the lost width.
   * If we move the taller line, the water height is still bottlenecked by the shorter line, so the area will strictly decrease.
   * Therefore, we **always discard the shorter line** and move its pointer inward, hoping to find a taller line.

* **Time Complexity:** $O(n)$ — We scan the array exactly once from the outside in.
* **Space Complexity:** $O(1)$ — We only use a few integer variables (`left`, `right`, `max_area`).

```python
from typing import List

class Solution:
    def maxArea(self, heights: List[int]) -> int:
        left = 0
        right = len(heights) - 1
        max_area = 0
        
        while left < right:
            # The width is the distance between the two pointers
            width = right - left
            # The water level is limited by the shorter line
            current_height = min(heights[left], heights[right])
            
            # Calculate current area and update max_area if it's the biggest we've seen
            current_area = width * current_height
            max_area = max(max_area, current_area)
            
            # Move the pointer of the shorter line inward
            if heights[left] < heights[right]:
                left += 1
            else:
                # If right is shorter, or if they are perfectly equal, it doesn't 
                # matter which one we move, so we just move right.
                right -= 1
                
        return max_area
```

### Problem 005: Trapping Rain Water (LeetCode 42) [HARD]

### Problem Definition and Constraints
You are given an array of non-negative integers `height` which represents an elevation map. Each value `height[i]` represents the height of a solid bar, which has a width of 1.
Return the total amount of water that can be trapped between the bars after a rainstorm.

**Examples:**
* **Example 1:**
  * **Input:** `height = [0,2,0,3,1,0,1,3,2,1]`
  * **Output:** `9`
  * **Explanation:** Water gets trapped in the valleys between the taller bars. For instance, the `0` at index 2 is trapped between a `2` on the left and a `3` on the right, trapping exactly 2 units of water above it.

* Constraints:
  * 1 <= height.length <= 20,000
  * 0 <= height[i] <= 100,000

### Core Logic: The "Invisible Walls" (Dynamic Bottleneck)
To figure out how much water sits directly on top of *one specific bar*, you only need to know three things:
1.  The highest bar anywhere to its left (`maxLeft`).
2.  The highest bar anywhere to its right (`maxRight`).
3.  The height of the bar itself (`height[i]`).

The water level above any bar is completely determined by the **shorter** of the two highest walls surrounding it. If the highest wall on the left is 2, and the highest wall on the right is 10, the water will spill over the left side at height 2. 
The mathematical formula for water at any specific index is: 
`water = min(maxLeft, maxRight) - height[i]`
*(If this calculation is negative, it just means the bar itself is a peak, so it holds 0 water).*

### Approach: Two Pointers (O(1) Space Optimization)
The hints suggest building two arrays to store the `maxLeft` and `maxRight` for every index. This takes $O(n)$ memory. 
We can completely eliminate that memory usage by using the **Greedy Bottleneck** Two Pointer logic from *Container With Most Water*!

We place a `left` pointer at the start and a `right` pointer at the end. We maintain two running variables: `maxLeft` and `maxRight`.
Here is the genius realization: We don't actually need to know *both* maximums to calculate the water. We only care about the **bottleneck** (the smaller one).
* If `maxLeft < maxRight`, we know for an absolute fact that the water level at the `left` pointer is bottlenecked by `maxLeft`. (Even if there is a massive skyscraper hidden somewhere in the middle, it won't matter, because the water will just spill out over `maxLeft` anyway). We calculate the water for `left`, add it to our total, and move `left` inward.
* If `maxRight <= maxLeft`, the exact opposite is true. The water at the `right` pointer is bottlenecked by `maxRight`. We process `right` and move inward.

* **Time Complexity:** $O(n)$ — The two pointers start at the edges and step inward until they meet. Every bar is evaluated exactly once.
* **Space Complexity:** $O(1)$ — We completely bypass the prefix/suffix arrays and only use four integer variables (`left`, `right`, `maxLeft`, `maxRight`).

```python
from typing import List

class Solution:
    def trap(self, height: List[int]) -> int:
        if not height:
            return 0
            
        # 1. Initialize Pointers at the absolute edges
        left = 0
        right = len(height) - 1
        
        # 2. Track the highest walls we have seen so far from both sides
        maxLeft = height[left]
        maxRight = height[right]
        
        total_water = 0
        
        # 3. Step inward until the pointers meet in the middle
        while left < right:
            
            # THE GREEDY BOTTLENECK:
            # We process whichever side has the smaller "highest wall".
            if maxLeft < maxRight:
                # The left side is the bottleneck. We can safely process the left pointer.
                left += 1
                
                # Update the highest wall seen on the left
                maxLeft = max(maxLeft, height[left])
                
                # Add trapped water 
                # (We don't need to check for negative numbers because we JUST updated maxLeft. 
                # If height[left] was taller, maxLeft equals height[left], resulting in 0 water).
                total_water += maxLeft - height[left]
                
            else:
                # The right side is the bottleneck (or they are equal). Process the right pointer.
                right -= 1
                
                # Update the highest wall seen on the right
                maxRight = max(maxRight, height[right])
                
                # Add trapped water
                total_water += maxRight - height[right]
                
        return total_water


        """
        =========================================================
        SIMULATION NOTES: height = [0, 2, 0, 3]
        =========================================================
        left = 0 (height 0), right = 3 (height 3)
        maxLeft = 0, maxRight = 3
        
        Step 1:
        maxLeft (0) < maxRight (3). Left is the bottleneck.
        Move left -> index 1 (height 2).
        maxLeft updates to max(0, 2) = 2.
        Water added: maxLeft (2) - height (2) = 0.
        
        Step 2:
        maxLeft (2) < maxRight (3). Left is still the bottleneck.
        Move left -> index 2 (height 0).
        maxLeft updates to max(2, 0) = 2.
        Water added: maxLeft (2) - height (0) = 2!  (Total = 2)
        
        Step 3:
        maxLeft (2) < maxRight (3). Left is still the bottleneck.
        Move left -> index 3.
        left == right. Loop terminates.
        
        Final Answer: 2 units of water trapped perfectly in the valley at index 2.
        """
```

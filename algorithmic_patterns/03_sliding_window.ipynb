### Problem 001: Best Time to Buy and Sell Stock (LeetCode 121)

### Problem Definition and Constraints
You are given an integer array `prices` where `prices[i]` is the price of a given stock on the `ith` day. You want to maximize your profit by choosing a single day to buy one stock and choosing a different day in the future to sell that stock. Return the maximum profit you can achieve. If no profit is possible, return 0.

* Constraints:
  * 1 <= prices.length <= 10^5
  * 0 <= prices[i] <= 10^4

### Examples
* **Example 1:**
  * Input: `prices = [7, 1, 5, 3, 6, 4]`
  * Output: `5`
  * Explanation: Buy on day 2 (price = 1) and sell on day 5 (price = 6), profit = 6 - 1 = 5.
* **Example 2:**
  * Input: `prices = [7, 6, 4, 3, 1]`
  * Output: `0`
  * Explanation: In this case, no transactions are done and the max profit = 0.

### Brute Force Approach
The baseline strategy considers every single valid buy/sell combination. We use an outer loop to select a buy day, and an inner loop to evaluate every subsequent day as a potential sell day, calculating the profit for each pair.
* Time Complexity: $O(n^2)$ — We check every possible pair in the array. For large constraints (e.g., $10^5$), this will result in a Time Limit Exceeded (TLE) error.
* Space Complexity: $O(1)$ — No auxiliary data structures are used.

### Optimized Approach (Sliding Window)
We use a dynamic sliding window approach utilizing two pointers: a Left pointer (Buy) and a Right pointer (Sell). Both pointers start at the beginning of the array. The Right pointer scans forward day-by-day. If the price at the Right pointer is ever lower than the price at the Left pointer, we have found a new historical low. We immediately snap the Left pointer to the Right pointer's position. If the price is higher, we calculate the profit and update our maximum profit tracker. 
* Time Complexity: $O(n)$ — The Right pointer traverses the array exactly once.
* Space Complexity: $O(1)$ — We only maintain pointer variables and a maximum profit integer.

```python
from typing import List

class Solution:
    
    # --- BRUTE FORCE APPROACH ---
    def maxProfit_brute_force(self, prices: List[int]) -> int:
        max_profit = 0
        
        # Outer loop sets the buy day
        for i in range(len(prices)):
            # Inner loop sets the sell day (must be in the future)
            for j in range(i + 1, len(prices)):
                profit = prices[j] - prices[i]
                
                # If this profit is better than our current best, update it
                if profit > max_profit:
                    max_profit = profit
                    
        return max_profit


    # --- OPTIMIZED APPROACH ---
    def maxProfit(self, prices: List[int]) -> int:
        # Left pointer = Buy, Right pointer = Sell
        left = 0
        right = 1
        max_profit = 0
        
        # The window expands until the Right pointer hits the end of the timeline
        while right < len(prices):
            
            # Is this a profitable transaction?
            if prices[left] < prices[right]:
                profit = prices[right] - prices[left]
                
                # Update max_profit using the built-in max() function for cleaner code
                max_profit = max(max_profit, profit)
                
            else:
                # We found a price LOWER than our current buy price.
                # Shift the Left pointer all the way to the Right pointer's current day.
                left = right
                
            # Regardless of what happened, advance the Sell pointer to the next day
            right += 1
            
        return max_profit
```

### Problem 002: Longest Substring Without Repeating Characters (LeetCode 3)

### Problem Definition and Constraints
Given a string `s`, find the length of the longest substring without duplicate characters. A substring is a contiguous sequence of characters within a string.

* Constraints:
  * 0 <= s.length <= 1000
  * `s` may consist of printable ASCII characters.

### Examples
* **Example 1:**
  * Input: `s = "zxyzxyz"`
  * Output: `3`
  * Explanation: The string "xyz" is the longest without duplicate characters.
* **Example 2:**
  * Input: `s = "xxxx"`
  * Output: `1`
  * Explanation: The longest substring is "x".

### Brute Force Approach
The brute force approach evaluates every possible substring. We use an outer loop to set a starting index, and an inner loop to expand the substring character by character. We use a Hash Set to track seen characters. The moment we encounter a character already in our Hash Set, we stop expanding that substring and move to the next starting index.
* Time Complexity: $O(n^2)$ — In the worst case, we check substrings starting from every single character and iterate until we find a duplicate or hit the end.
* Space Complexity: $O(m)$ — Where $m$ is the number of unique characters in the string (the maximum size of our Hash Set).

### Optimized Approach (Sliding Window)
We use a dynamic sliding window utilizing two pointers (`left` and `right`) and a Hash Set. As the `right` pointer iterates through the string, we add each character to our set. If the `right` pointer lands on a character that is *already* in the set, our window is no longer valid. To fix this, we remove the character at the `left` pointer from the set and slide the `left` pointer forward, repeating this until the duplicate character is completely removed from our window. We update the maximum length at each valid step.
* Time Complexity: $O(n)$ — Both the `left` and `right` pointers only ever move forward. In the worst-case scenario, each character is visited exactly twice (once by the `right` pointer, once by the `left` pointer).
* Space Complexity: $O(m)$ — Where $m$ is the number of unique characters in the string, which dictates the size of the Hash Set.

```python
class Solution:
    
    # --- BRUTE FORCE APPROACH ---
    def lengthOfLongestSubstring_brute_force(self, s: str) -> int:
        max_length = 0
        
        # Outer loop sets the starting character of the substring
        for i in range(len(s)):
            seen_chars = set()
            current_length = 0
            
            # Inner loop expands the substring one character at a time
            for j in range(i, len(s)):
                # If we find a duplicate, this substring is done
                if s[j] in seen_chars:
                    break
                    
                # Otherwise, add it to our set and track the length
                seen_chars.add(s[j])
                current_length += 1
                max_length = max(max_length, current_length)
                
        return max_length


    # --- OPTIMIZED APPROACH ---
    def lengthOfLongestSubstring(self, s: str) -> int:
        char_set = set()
        left = 0
        max_length = 0
        
        # The right pointer expands the window character by character
        for right in range(len(s)):
            
            # If we encounter a duplicate, our window is currently invalid.
            # We shrink the window from the left until the duplicate is gone.
            while s[right] in char_set:
                char_set.remove(s[left])
                left += 1
                
            # Add the new, unique character to the set
            char_set.add(s[right])
            
            # Calculate the current window size and update max_length
            current_window_size = right - left + 1
            max_length = max(max_length, current_window_size)
            
        return max_length
```

### Problem 003: Longest Repeating Character Replacement (LeetCode 424)

### Problem Definition and Constraints
You are given a string `s` consisting only of uppercase English characters, and an integer `k`. You can replace up to `k` characters in the string with any other uppercase English character. Return the length of the longest substring containing only one distinct character after performing at most `k` replacements.
* Constraints:
  * 1 <= s.length <= 1000
  * 0 <= k <= s.length

### Examples
* **Example 1:**
  * Input: `s = "XYYX"`, `k = 2`
  * Output: `4`
  * Explanation: Either replace the two 'X's with 'Y's, or the two 'Y's with 'X's to make the whole string identical.
* **Example 2:**
  * Input: `s = "AAABABB"`, `k = 1`
  * Output: `5`
  * Explanation: Replace the one 'B' in the middle to get "AAAAA".

### Brute Force Approach
The brute force approach evaluates every possible substring. For every starting index, we expand the substring character by character. We use a Hash Map to count the frequencies of each character in our current substring. A substring is valid if its total length minus the count of its most frequent character is less than or equal to `k`. If it exceeds `k`, we stop and move to the next starting index.
* Time Complexity: $O(n^2)$ — We use nested loops to evaluate substrings starting from every single character.
* Space Complexity: $O(m)$ — Where $m$ is the number of unique characters in the string (up to 26 for English letters), making this effectively $O(1)$.

### Optimized Approach (Sliding Window)
We use a sliding window with two pointers (`left` and `right`) and a Hash Map to keep track of character frequencies. As the `right` pointer expands the window, we update our frequency map.

The core logic relies on a simple formula: `(Total Window Size) - (Count of Most Frequent Character)`. 
This formula tells us exactly how many "other" characters are in our window. If this number is `<=` `k`, our window is valid! If it is `>` `k`, we have too many characters to replace. When that happens, we shrink the window by decrementing the count of the character at the `left` pointer and moving `left` forward.
* Time Complexity: $O(n)$ — The `right` pointer traverses the array exactly once. The `left` pointer also only moves forward. Scanning the map to find the max frequency takes at most 26 operations, which is $O(1)$ constant time.
* Space Complexity: $O(1)$ — The Hash Map will never store more than 26 key-value pairs (the English alphabet).

### The "Max Frequency" Trick
You might notice in the optimized code that when we shrink the window from the left, we decrease the character count in our dictionary (`char_counts[s[left]] -= 1`), but we **never decrease `max_frequent_char_count`**, even if the most frequent character just fell out of the window! 

Why? Because we only care about finding a **new historical maximum length**. 
To beat our current record, we mathematically *must* find a window that has an even higher `max_frequent_char_count`. If the frequency drops, it can never beat our high score anyway, so it doesn't matter if the variable is temporarily "inaccurate". The code will only expand `max_length` again when we encounter a frequency that beats the historical high score. This trick saves us from having to rescan the dictionary every time we shrink the window.

```python
from typing import List

class Solution:
    
    # --- BRUTE FORCE APPROACH ---
    def characterReplacement_brute_force(self, s: str, k: int) -> int:
        max_length = 0
        
        # Outer loop sets the start of our substring
        for i in range(len(s)):
            char_counts = {}
            max_frequent_char_count = 0
            
            # Inner loop expands the substring
            for j in range(i, len(s)):
                # Update the count for the current character
                char_counts[s[j]] = char_counts.get(s[j], 0) + 1
                
                # Keep track of the most frequent character in this specific substring
                max_frequent_char_count = max(max_frequent_char_count, char_counts[s[j]])
                
                current_length = j - i + 1
                
                # Formula: (Length of Window) - (Most Frequent Char) = Number of Replacements Needed
                if current_length - max_frequent_char_count <= k:
                    max_length = max(max_length, current_length)
                else:
                    # If we need more than k replacements, this substring is invalid.
                    # Stop expanding and move to the next starting character.
                    break
                    
        return max_length


    # --- OPTIMIZED APPROACH ---
    def characterReplacement(self, s: str, k: int) -> int:
        char_counts = {}
        left = 0
        max_length = 0
        max_frequent_char_count = 0
        
        # Right pointer expands the window
        for right in range(len(s)):
            char_counts[s[right]] = char_counts.get(s[right], 0) + 1
            
            # Update the max frequency of any character seen in the window so far
            max_frequent_char_count = max(max_frequent_char_count, char_counts[s[right]])
            
            # Is our current window invalid? 
            # (Total Window Size) - (Most Frequent Character Count) > k
            while (right - left + 1) - max_frequent_char_count > k:
                # The window is invalid, so we shrink it from the left
                char_counts[s[left]] -= 1
                left += 1
                
            # If we make it here, the window is valid. Update our max length.
            max_length = max(max_length, right - left + 1)
            
        return max_length
```

### Problem 004: Permutation in String (LeetCode 567)

### Problem Definition and Constraints
You are given two strings `s1` and `s2`. Return `True` if `s2` contains a permutation of `s1`, or `False` otherwise. In other words, return `True` if one of `s1`'s permutations is the substring of `s2`.
* Both strings only contain lowercase letters.
* Constraints:
  * 1 <= s1.length, s2.length <= 1000

### Examples
* **Example 1:**
  * Input: `s1 = "abc"`, `s2 = "lecabee"`
  * Output: `True`
  * Explanation: The substring "cab" is a permutation of "abc" and is present in "lecabee".
* **Example 2:**
  * Input: `s1 = "abc"`, `s2 = "lecaabee"`
  * Output: `False`
  * Explanation: There is no contiguous substring in `s2` that contains exactly one 'a', one 'b', and one 'c'.

### Brute Force Approach
The baseline strategy is to check every possible substring in `s2` that has the exact same length as `s1`. For each of these substrings, we count the frequency of its characters and compare it to the frequency of characters in `s1`. 
* Time Complexity: $O(n \cdot m)$ — Where $n$ is the length of `s2` and $m$ is the length of `s1`. We check $n - m + 1$ substrings, and building the frequency map for each takes $m$ steps.
* Space Complexity: $O(1)$ — Because the problem specifies lowercase English letters only, our frequency maps will never hold more than 26 key-value pairs.

### Optimized Approach (Fixed Sliding Window)
Since a permutation must have the exact same length and character frequencies as `s1`, we can use a **fixed-size sliding window** of length `len(s1)` to scan across `s2`. 

Instead of rebuilding a frequency map from scratch for every substring, we maintain a running count using an array of size 26 (representing 'a' through 'z'). 
1. We first count the frequencies of `s1` to establish our target.
2. We initialize our first window in `s2` by counting the first `len(s1)` characters.
3. We slide the window one character to the right: we add the new character on the right to our running count, and remove the character that fell off the left side.
4. If our current window's count ever exactly matches the target count, we have found a permutation.

* Time Complexity: $O(n)$ — We process each character of `s2` exactly once. Comparing two arrays of size 26 takes a constant 26 operations, which simplifies to $O(1)$ per step, making the total run time $O(n)$.
* Space Complexity: $O(1)$ — We use two arrays of size 26 to store character counts, which requires constant extra space.

```python
class Solution:
    
    # --- BRUTE FORCE APPROACH ---
    def checkInclusion_brute_force(self, s1: str, s2: str) -> bool:
        if len(s1) > len(s2):
            return False
            
        # Helper function to get a frequency dictionary for a string
        def get_counts(s: str) -> dict:
            counts = {}
            for char in s:
                counts[char] = counts.get(char, 0) + 1
            return counts
            
        target_counts = get_counts(s1)
        window_size = len(s1)
        
        # Check every substring in s2 of length len(s1)
        for i in range(len(s2) - window_size + 1):
            current_substring = s2[i:i + window_size]
            
            # If the frequencies match, we found a permutation
            if get_counts(current_substring) == target_counts:
                return True
                
        return False

    # --- OPTIMIZED APPROACH ---
    def checkInclusion(self, s1: str, s2: str) -> bool:
        if len(s1) > len(s2):
            return False
            
        # Use arrays of size 26 to represent 'a' through 'z' counts
        s1_counts = [0] * 26
        window_counts = [0] * 26
        
        window_size = len(s1)
        
        # Initialize the target counts and the first window
        for i in range(window_size):
            # ord(char) - ord('a') maps 'a' to index 0, 'b' to 1, etc.
            s1_counts[ord(s1[i]) - ord('a')] += 1
            window_counts[ord(s2[i]) - ord('a')] += 1
            
        # Check if the very first window is a match
        if s1_counts == window_counts:
            return True
            
        # Slide the window across the rest of s2
        for right in range(window_size, len(s2)):
            # 1. Add the new character on the right to our window
            new_char_index = ord(s2[right]) - ord('a')
            window_counts[new_char_index] += 1
            
            # 2. Remove the character that fell out of the left side of the window
            left = right - window_size
            old_char_index = ord(s2[left]) - ord('a')
            window_counts[old_char_index] -= 1
            
            # 3. Check if our updated window matches the target
            if s1_counts == window_counts:
                return True
                
        return False
```

### Problem 005: Minimum Window Substring (LeetCode 76) [HARD]

### Problem Definition and Constraints
Given two strings `s` and `t`, return the shortest substring of `s` such that every character in `t` (including duplicates) is present in the substring. If no such substring exists, return an empty string `""`.
You may assume that the correct output is always unique.

**Examples:**
* **Example 1:**
  * **Input:** `s = "OUZODYXAZV", t = "XYZ"`
  * **Output:** `"YXAZ"`
  * **Explanation:** `"YXAZ"` is the shortest substring that contains "X", "Y", and "Z" from string `t`.
* **Example 2:**
  * **Input:** `s = "xyz", t = "xyz"`
  * **Output:** `"xyz"`
* **Example 3:**
  * **Input:** `s = "x", t = "xy"`
  * **Output:** `""`

* Constraints:
  * 1 <= s.length <= 100,000
  * 1 <= t.length <= 100,000
  * `s` and `t` consist of uppercase and lowercase English letters.

### Core Logic: The "Caterpillar" (Shrink-to-Optimize)
This is the ultimate evolution of the dynamic "Caterpillar" window. In previous problems, we shrank the window when it became *invalid*. Here, the logic flips: we stretch the right pointer until the window becomes *valid*, and then we aggressively shrink the left pointer to make it as small as possible while **keeping** it valid.

**The $O(1)$ Verification Trick (`have` vs `need`)**
A brute-force sliding window would pause at every step and compare two Hash Maps (`window == target`). Comparing two maps can take up to 52 operations (upper and lower case letters) per loop.
To make it purely $O(1)$, we use two simple integers: `have` and `need`.
* `need`: The total number of *unique* characters we need from `t`.
* `have`: The total number of *unique* characters we have fully satisfied in our current window.
When `have == need`, the window is valid.

### Approach: HashMap + Two Pointers
1. Build a Hash Map `countT` containing the frequencies of all characters in `t`.
2. Initialize `have = 0`, `need = len(countT)`, and an empty `window` Hash Map.
3. **Expand:** The `right` pointer moves forward, adding characters to `window`. If the count of a character in `window` perfectly matches its requirement in `countT`, increment `have`.
4. **Shrink:** The moment `have == need`, record the current window size. Then, start moving the `left` pointer inward to shrink the window. 
5. As `left` moves, decrement the character from `window`. If its count drops below the requirement in `countT`, decrement `have`. This breaks the `have == need` loop and forces `right` to start expanding again to find the missing character.

* **Time Complexity:** $O(n + m)$ — Where $n$ is the length of `s` and $m$ is the length of `t`. Building the target map takes $O(m)$. The `left` and `right` pointers sweep across `s` at most once, taking $O(n)$ time.
* **Space Complexity:** $O(k)$ — Where $k$ is the number of unique characters in the strings (bounded by 52 English letters), used by the Hash Maps.

```python
class Solution:
    def minWindow(self, s: str, t: str) -> str:
        if t == "":
            return ""

        # 1. Build the target inventory
        countT, window = {}, {}
        for c in t:
            countT[c] = 1 + countT.get(c, 0)

        # 2. Setup the 'have' vs 'need' trackers
        have, need = 0, len(countT)
        
        # res stores the starting and ending indices of the best window
        res, resLen = [-1, -1], float("infinity")
        left = 0

        # 3. The Greedy Expansion
        for right in range(len(s)):
            c = s[right]
            window[c] = 1 + window.get(c, 0)

            # If this character is required AND we just hit the exact required count,
            # we have fully satisfied one of our unique character conditions!
            if c in countT and window[c] == countT[c]:
                have += 1

            # 4. The Shrink-to-Optimize Phase
            # While our window is totally valid, try to squeeze it smaller from the left
            while have == need:
                
                # Update our historical shortest record
                if (right - left + 1) < resLen:
                    res = [left, right]
                    resLen = right - left + 1

                # Attempt to pop the left-most character out of the window
                left_char = s[left]
                window[left_char] -= 1
                
                # If we just popped a required character and it dropped below the needed amount,
                # our window is now officially broken. 'have' goes down, breaking the while loop.
                if left_char in countT and window[left_char] < countT[left_char]:
                    have -= 1
                    
                left += 1

        l, r = res
        return s[l : r + 1] if resLen != float("infinity") else ""


        """
        =========================================================
        SIMULATION NOTES: s = "ADOBECODEBANC", t = "ABC"
        =========================================================
        countT = {'A':1, 'B':1, 'C':1}, need = 3
        
        right pointer expands...
        right = 0 ('A'): window={'A':1}. have=1.
        right = 3 ('B'): window={'A':1, 'D':1, 'O':1, 'B':1}. have=2.
        right = 5 ('C'): window={'A':1, 'D':1, 'O':1, 'B':1, 'E':1, 'C':1}. have=3.
        
        have == need! (3 == 3)
        Valid window found: "ADOBEC" (Length 6). Record it.
        
        Shrink phase:
        Pop 'A' (left=0). 
        'A' count drops to 0. 0 < 1, so have becomes 2. 
        Loop breaks.
        
        right pointer resumes expanding to find another 'A'...
        Finds 'A' at index 10. Valid again! "DOBECODEBA" (Length 10 - not shorter).
        Shrinks until it pops 'B' or 'C'.
        
        Eventually finds "BANC" (Length 4). Record it.
        Final Answer: "BANC"
        """
```

### Problem 006: Sliding Window Maximum (LeetCode 239) [HARD]

### Problem Definition and Constraints
You are given an array of integers `nums` and an integer `k`. There is a sliding window of size `k` that starts at the left edge of the array. The window slides one position to the right until it reaches the right edge of the array.
Return a list that contains the maximum element in the window at each step.

**Examples:**
* **Example 1:**
  * **Input:** `nums = [1,2,1,0,4,2,6], k = 3`
  * **Output:** `[2,2,4,4,6]`
  * **Explanation:** 
    * `[1 2 1] 0 4 2 6` -> Max is 2
    * `1 [2 1 0] 4 2 6` -> Max is 2
    * `1 2 [1 0 4] 2 6` -> Max is 4
    * `1 2 1 [0 4 2] 6` -> Max is 4
    * `1 2 1 0 [4 2 6]` -> Max is 6

* Constraints:
  * 1 <= nums.length <= 100,000
  * -10,000 <= nums[i] <= 10,000
  * 1 <= k <= nums.length

### Core Logic: The "Monotonic Queue" (The Line of Sight)
While the hints suggest a Max-Heap ($O(n \log n)$), there is a legendary $O(n)$ solution using a **Monotonic Deque** (Double-Ended Queue).

Imagine people standing in a line, looking to the right. If a very tall person steps into the back of the line, any shorter people standing in front of them are completely blocked from view. In the context of our sliding window: **If a new number enters the window, any older numbers that are smaller than it are completely useless.** They will *never* be the maximum for this window, nor any future window, because the new taller number will outlive them.

**The Greedy Choice:**
We maintain a Deque that stores the *indices* of the array. We enforce one strict rule: the values corresponding to these indices must be strictly decreasing. 
Whenever a new number enters, we ruthlessly kick out all smaller numbers from the back of the Deque. This guarantees that the absolute largest number in the current window is *always* sitting perfectly at the front of the Deque.

### Approach: Monotonic Deque
1. Initialize a `deque` to store indices, and an `output` array.
2. Iterate through the array with a `right` pointer.
3. **The Purge:** Before adding `nums[right]` to the Deque, check the back of the Deque. While the number at the back is smaller than `nums[right]`, pop it out. They are obsolete.
4. Add the `right` index to the back of the Deque.
5. **The Boundary Check:** Check the front of the Deque. If the index at the front is outside our current window bounds (`index < right - k + 1`), pop it off the front. It has expired.
6. **The Result:** Once our window has expanded to at least size `k`, the maximum for the current window is guaranteed to be the value at the front of the Deque. Append it to the output.

* **Time Complexity:** $O(n)$ — Even though there is a `while` loop inside the `for` loop, every single index is pushed into the Deque exactly once and popped out exactly once. This mathematically guarantees a strictly linear runtime.
* **Space Complexity:** $O(k)$ — The Deque will only ever hold a maximum of `k` indices at any given time.

```python
import collections
from typing import List

class Solution:
    def maxSlidingWindow(self, nums: List[int], k: int) -> List[int]:
        output = []
        # The deque stores indices, NOT the actual values.
        # This allows us to easily check if a value has fallen out of the window.
        q = collections.deque()
        
        # We start sliding our window
        for right in range(len(nums)):
            
            # 1. THE PURGE (Maintain strictly decreasing order)
            # If the new guy entering the window is taller than the guys at the back of the line,
            # kick the shorter guys out. They are useless now.
            while q and nums[q[-1]] < nums[right]:
                q.pop()
                
            # Add the new guy's index to the line
            q.append(right)
            
            # 2. THE BOUNDARY CHECK (Remove expired numbers)
            # The left boundary of our current window is (right - k + 1).
            # If the index at the front of our queue is smaller than the left boundary,
            # it has fallen out of the window and must be removed.
            if q[0] < right - k + 1:
                q.popleft()
                
            # 3. RECORD THE RESULT
            # We only start recording results once our window has reached size 'k'.
            # The maximum value is ALWAYS sitting at the front of the deque.
            if right + 1 >= k:
                output.append(nums[q[0]])
                
        return output


        """
        =========================================================
        SIMULATION NOTES: nums = [1, 3, -1, -3, 5], k = 3
        =========================================================
        
        right = 0 (val: 1). 
        - Purge: q is empty. 
        - Append index 0. q = [0]
        - Window size < 3.
        
        right = 1 (val: 3). 
        - Purge: nums[q[-1]] (1) < 3. POP! q is empty.
        - Append index 1. q = [1]
        - Window size < 3.
        
        right = 2 (val: -1).
        - Purge: nums[q[-1]] (3) is NOT < -1. Do nothing.
        - Append index 2. q = [1, 2]. (Values: [3, -1])
        - Window size is 3! Append nums[q[0]] (3). output = [3]
        
        right = 3 (val: -3).
        - Purge: nums[q[-1]] (-1) is NOT < -3. Do nothing.
        - Append index 3. q = [1, 2, 3]. (Values: [3, -1, -3])
        - Expired check: q[0] (1) < right - k + 1 (3 - 3 + 1 = 1)? No.
        - Append nums[q[0]] (3). output = [3, 3]
        
        right = 4 (val: 5).
        - Purge: -3 < 5 (POP!), -1 < 5 (POP!), 3 < 5 (POP!). q is empty.
        - Append index 4. q = [4].
        - Append nums[q[0]] (5). output = [3, 3, 5]
        """
```

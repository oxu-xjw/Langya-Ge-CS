### Problem 001: Best Time to Buy and Sell Stock (LeetCode 121)

### Problem Definition and Constraints
You are given an integer array `prices` where `prices[i]` is the price of a given stock on the `ith` day. You want to maximize your profit by choosing a single day to buy one stock and choosing a different day in the future to sell that stock. Return the maximum profit you can achieve. If no profit is possible, return 0.

* Constraints:
  * 1 <= prices.length <= 10^5
  * 0 <= prices[i] <= 10^4

### Examples
* **Example 1:**
  * Input: `prices = [7, 1, 5, 3, 6, 4]`
  * Output: `5`
  * Explanation: Buy on day 2 (price = 1) and sell on day 5 (price = 6), profit = 6 - 1 = 5.
* **Example 2:**
  * Input: `prices = [7, 6, 4, 3, 1]`
  * Output: `0`
  * Explanation: In this case, no transactions are done and the max profit = 0.

### Brute Force Approach
The baseline strategy considers every single valid buy/sell combination. We use an outer loop to select a buy day, and an inner loop to evaluate every subsequent day as a potential sell day, calculating the profit for each pair.
* Time Complexity: $O(n^2)$ — We check every possible pair in the array. For large constraints (e.g., $10^5$), this will result in a Time Limit Exceeded (TLE) error.
* Space Complexity: $O(1)$ — No auxiliary data structures are used.

### Optimized Approach (Sliding Window)
We use a dynamic sliding window approach utilizing two pointers: a Left pointer (Buy) and a Right pointer (Sell). Both pointers start at the beginning of the array. The Right pointer scans forward day-by-day. If the price at the Right pointer is ever lower than the price at the Left pointer, we have found a new historical low. We immediately snap the Left pointer to the Right pointer's position. If the price is higher, we calculate the profit and update our maximum profit tracker. 
* Time Complexity: $O(n)$ — The Right pointer traverses the array exactly once.
* Space Complexity: $O(1)$ — We only maintain pointer variables and a maximum profit integer.

```python
from typing import List

class Solution:
    
    # --- BRUTE FORCE APPROACH ---
    def maxProfit_brute_force(self, prices: List[int]) -> int:
        max_profit = 0
        
        # Outer loop sets the buy day
        for i in range(len(prices)):
            # Inner loop sets the sell day (must be in the future)
            for j in range(i + 1, len(prices)):
                profit = prices[j] - prices[i]
                
                # If this profit is better than our current best, update it
                if profit > max_profit:
                    max_profit = profit
                    
        return max_profit


    # --- OPTIMIZED APPROACH ---
    def maxProfit(self, prices: List[int]) -> int:
        # Left pointer = Buy, Right pointer = Sell
        left = 0
        right = 1
        max_profit = 0
        
        # The window expands until the Right pointer hits the end of the timeline
        while right < len(prices):
            
            # Is this a profitable transaction?
            if prices[left] < prices[right]:
                profit = prices[right] - prices[left]
                
                # Update max_profit using the built-in max() function for cleaner code
                max_profit = max(max_profit, profit)
                
            else:
                # We found a price LOWER than our current buy price.
                # Shift the Left pointer all the way to the Right pointer's current day.
                left = right
                
            # Regardless of what happened, advance the Sell pointer to the next day
            right += 1
            
        return max_profit
```

### Problem 002: Longest Substring Without Repeating Characters (LeetCode 3)

### Problem Definition and Constraints
Given a string `s`, find the length of the longest substring without duplicate characters. A substring is a contiguous sequence of characters within a string.

* Constraints:
  * 0 <= s.length <= 1000
  * `s` may consist of printable ASCII characters.

### Examples
* **Example 1:**
  * Input: `s = "zxyzxyz"`
  * Output: `3`
  * Explanation: The string "xyz" is the longest without duplicate characters.
* **Example 2:**
  * Input: `s = "xxxx"`
  * Output: `1`
  * Explanation: The longest substring is "x".

### Brute Force Approach
The brute force approach evaluates every possible substring. We use an outer loop to set a starting index, and an inner loop to expand the substring character by character. We use a Hash Set to track seen characters. The moment we encounter a character already in our Hash Set, we stop expanding that substring and move to the next starting index.
* Time Complexity: $O(n^2)$ — In the worst case, we check substrings starting from every single character and iterate until we find a duplicate or hit the end.
* Space Complexity: $O(m)$ — Where $m$ is the number of unique characters in the string (the maximum size of our Hash Set).

### Optimized Approach (Sliding Window)
We use a dynamic sliding window utilizing two pointers (`left` and `right`) and a Hash Set. As the `right` pointer iterates through the string, we add each character to our set. If the `right` pointer lands on a character that is *already* in the set, our window is no longer valid. To fix this, we remove the character at the `left` pointer from the set and slide the `left` pointer forward, repeating this until the duplicate character is completely removed from our window. We update the maximum length at each valid step.
* Time Complexity: $O(n)$ — Both the `left` and `right` pointers only ever move forward. In the worst-case scenario, each character is visited exactly twice (once by the `right` pointer, once by the `left` pointer).
* Space Complexity: $O(m)$ — Where $m$ is the number of unique characters in the string, which dictates the size of the Hash Set.

```python
class Solution:
    
    # --- BRUTE FORCE APPROACH ---
    def lengthOfLongestSubstring_brute_force(self, s: str) -> int:
        max_length = 0
        
        # Outer loop sets the starting character of the substring
        for i in range(len(s)):
            seen_chars = set()
            current_length = 0
            
            # Inner loop expands the substring one character at a time
            for j in range(i, len(s)):
                # If we find a duplicate, this substring is done
                if s[j] in seen_chars:
                    break
                    
                # Otherwise, add it to our set and track the length
                seen_chars.add(s[j])
                current_length += 1
                max_length = max(max_length, current_length)
                
        return max_length


    # --- OPTIMIZED APPROACH ---
    def lengthOfLongestSubstring(self, s: str) -> int:
        char_set = set()
        left = 0
        max_length = 0
        
        # The right pointer expands the window character by character
        for right in range(len(s)):
            
            # If we encounter a duplicate, our window is currently invalid.
            # We shrink the window from the left until the duplicate is gone.
            while s[right] in char_set:
                char_set.remove(s[left])
                left += 1
                
            # Add the new, unique character to the set
            char_set.add(s[right])
            
            # Calculate the current window size and update max_length
            current_window_size = right - left + 1
            max_length = max(max_length, current_window_size)
            
        return max_length
```

### Problem 001: Climbing Stairs (LeetCode 70)

### Problem Definition and Constraints
You are given an integer `n` representing the number of steps to reach the top of a staircase. You can climb with either 1 or 2 steps at a time.
Return the number of distinct ways to climb to the top of the staircase.

* Constraints:
  * 1 <= n <= 45

### Examples
* **Example 1:**
  * Input: `n = 2`
  * Output: `2`
  * Explanation: (1 step + 1 step), or (2 steps).
* **Example 2:**
  * Input: `n = 3`
  * Output: `3`
  * Explanation: (1+1+1), (1+2), or (2+1).

### Brute Force Approach (Recursion)
At every step, you make a recursive call for `take 1 step` and `take 2 steps`. This creates a massive, branching decision tree.
* Time Complexity: $O(2^n)$ — The number of branches doubles at every step. For `n=45`, this will result in a Time Limit Exceeded (TLE) error.

### Optimized Approach (1-D Dynamic Programming / Tabulation)
Instead of top-down recursion, we build the solution bottom-up. We create an array `dp` where `dp[i]` represents the number of ways to reach step `i`. 
We know the base cases:
* `dp[1] = 1` way (just take 1 step)
* `dp[2] = 2` ways (1+1, or 2)

For any step `i` beyond 2, the only way to arrive at `i` is by jumping from `i-1` or jumping from `i-2`. Therefore, the total number of ways to reach `i` is simply `dp[i] = dp[i-1] + dp[i-2]`. We fill the array up to `n` in a single pass.
* Time Complexity: $O(n)$ — We iterate through a loop from 3 to `n` exactly once.
* Space Complexity: $O(n)$ — We use an array of size `n+1` to store our cached answers.


```python
class Solution:
    def climbStairs(self, n: int) -> int:
        # BASE CASES: If the staircase is very short, the answer is just n.
        if n <= 2:
            return n
            
        # Create our "DP Array" (our notebook) to cache our answers.
        # We make it size n + 1 so the indices align perfectly with the step numbers.
        dp = [0] * (n + 1)
        
        # Write down the base cases we already know for a fact
        dp[1] = 1
        dp[2] = 2
        
        # Start calculating from Step 3 all the way up to Step n
        for i in range(3, n + 1):
            # To reach step i, you either stepped from i-1 or i-2.
            # So, add the cached results of those two previous steps!
            dp[i] = dp[i - 1] + dp[i - 2]
            
        # By the end of the loop, the answer for the top step is calculated and cached.
        return dp[n]
```

### Problem 002: Min Cost Climbing Stairs (LeetCode 746)

### Problem Definition and Constraints
You are given an array of integers `cost` where `cost[i]` is the cost of taking a step from the $i$-th floor. After paying the cost, you can step to either the $(i + 1)$-th floor or the $(i + 2)$-th floor.
You may choose to start at the index 0 or the index 1 floor.
Return the minimum cost to reach the top of the staircase (just past the last index in `cost`).

* Constraints:
  * 2 <= cost.length <= 1000
  * 0 <= cost[i] <= 999

### Examples
* **Example 1:**
  * Input: `cost = [10,15,20]`
  * Output: `15`
  * Explanation: Start at index 1 (cost 15). Pay 15 and take two steps to reach the top.
* **Example 2:**
  * Input: `cost = [1,100,1,1,1,100,1,1,100,1]`
  * Output: `6`

### Optimized Approach (1-D Dynamic Programming / Tabulation)
This builds directly on the logic of standard "Climbing Stairs". The only way to reach step `i` is by jumping from `i-1` or `i-2`. 
We create a `dp` array where `dp[i]` represents the minimum cost to reach step `i`. 
* **Base Cases:** Because we are allowed to start at index 0 or index 1 for free (we only pay when we *leave* the step), `dp[0] = 0` and `dp[1] = 0`.
* **The Transition:** For any step `i` from 2 up to `n`, we calculate two potential costs:
  1. The cost to reach the previous step (`dp[i-1]`) PLUS the cost to jump off it (`cost[i-1]`).
  2. The cost to reach two steps back (`dp[i-2]`) PLUS the cost to jump off it (`cost[i-2]`).
We simply take the `min()` of those two choices and record it in `dp[i]`.
* Time Complexity: $O(n)$ — We loop through the steps exactly once.
* Space Complexity: $O(n)$ — We use a DP array of size $n+1$. (This can be optimized to $O(1)$ by just keeping track of the last two steps).

```python
from typing import List

class Solution:
    def minCostClimbingStairs(self, cost: List[int]) -> int:
        n = len(cost)
        
        # dp[i] represents the minimum cost to REACH step i.
        # We need to reach step n (the "top" of the staircase, just past the last index)
        dp = [0] * (n + 1)
        
        # Base Cases: We can start at index 0 or index 1 without paying anything yet.
        dp[0] = 0
        dp[1] = 0
        
        # Build the DP array bottom-up
        for i in range(2, n + 1):
            
            # Choice 1: Walk from the step right below us
            take_one_step = dp[i - 1] + cost[i - 1]
            
            # Choice 2: Jump from two steps below us
            take_two_steps = dp[i - 2] + cost[i - 2]
            
            # Record the absolute cheapest option
            dp[i] = min(take_one_step, take_two_steps)
            
        # The final answer is the cheapest cost calculated to reach the top
        return dp[n]


    # ---------------------------------------------------------
    # BONUS: O(1) Space Optimization
    # Just like Climbing Stairs, we only need the last two values!
    # ---------------------------------------------------------
    def minCostClimbingStairsOpt(self, cost: List[int]) -> int:
        n = len(cost)
        two_steps_back = 0
        one_step_back = 0
        
        for i in range(2, n + 1):
            current_cost = min(one_step_back + cost[i - 1], two_steps_back + cost[i - 2])
            
            # Slide the window forward
            two_steps_back = one_step_back
            one_step_back = current_cost
            
        return one_step_back
```

### Problem 003: House Robber (LeetCode 198)

### Problem Definition and Constraints
You are given an integer array `nums` where `nums[i]` represents the amount of money the $i$-th house has. The houses are arranged in a straight line, i.e., the $i$-th house is the neighbor of the $(i-1)$-th and $(i+1)$-th house.

You are planning to rob money from the houses, but you cannot rob two adjacent houses because the security system will automatically alert the police if two adjacent houses were both broken into.

Return the maximum amount of money you can rob without alerting the police.

**Examples:**
* **Example 1:**
  * **Input:** `nums = [1,1,3,3]`
  * **Output:** `4`
  * *Explanation:* `nums[0] + nums[2] = 1 + 3 = 4`.
* **Example 2:**
  * **Input:** `nums = [2,9,8,3,6]`
  * **Output:** `16`
  * *Explanation:* `nums[0] + nums[2] + nums[4] = 2 + 8 + 6 = 16`.

* Constraints:
  * 1 <= nums.length <= 100
  * 0 <= nums[i] <= 100

### Approach 1: 1D Dynamic Programming ($O(n)$ Space)
This problem builds perfectly on the "conjunction" logic we just mastered with the stairs. 

Imagine you are standing in front of House `i`. You are looking at two different realities that merge right here:
1. **Reality A (You rob this house):** Because you cannot rob adjacent houses, you must have skipped the house right before this one (`i-1`). That means your total money is the money inside this current house (`nums[i]`) PLUS the maximum haul you had securely stashed away two houses ago (`dp[i-2]`).
2. **Reality B (You skip this house):** You decide not to touch this house. That means your total money is just whatever maximum haul you had from the house right before this one (`dp[i-1]`).

Because you want to be as rich as possible, you simply look at Reality A and Reality B, and take the `max()` of the two. We store these answers in a `dp` array where `dp[i]` represents the max money up to house `i`.

* **The Recurrence Relation:** `dp[i] = max(nums[i] + dp[i-2], dp[i-1])`
* **Time Complexity:** $O(n)$ — We walk down the street of houses exactly once.
* **Space Complexity:** $O(n)$ — We use an array of size $n$ to store the max money for every single house.

### Approach 2: Space Optimization ($O(1)$ Space)
If you look closely at the recurrence relation, to calculate the maximum haul for House `i`, we only ever need to look at the previous house (`dp[i-1]`) and the house before that (`dp[i-2]`). 

We do not care about `dp[i-3]` or `dp[i-4]`. We can completely drop the $O(n)$ array and just use two sliding variables (`rob1` and `rob2`) that track our last two states.

* **Time Complexity:** $O(n)$
* **Space Complexity:** $O(1)$ — Only two variables used, no matter how long the street is!

```python
from typing import List

class Solution:
    # ---------------------------------------------------------
    # Approach 1: Standard DP Array (O(n) Space)
    # ---------------------------------------------------------
    def rob_array(self, nums: List[int]) -> int:
        if not nums:
            return 0
        
        # Edge case: If there is only one house, just rob it.
        if len(nums) == 1:
            return nums[0]
            
        # Create our DP notebook (array)
        dp = [0] * len(nums)
        
        # Base Cases:
        # The max money at the first house is just the money in the first house.
        dp[0] = nums[0]
        # The max money at the second house is the better of the first two houses.
        dp[1] = max(nums[0], nums[1])
        
        # Start walking down the street from the 3rd house (index 2)
        for i in range(2, len(nums)):
            # Compare Reality A (rob it) vs Reality B (skip it)
            rob_current = nums[i] + dp[i - 2]
            skip_current = dp[i - 1]
            
            dp[i] = max(rob_current, skip_current)
            
        # The last slot in our array holds the maximum possible haul.
        return dp[-1]

    # ---------------------------------------------------------
    # Approach 2: Sliding Variables (O(1) Space)
    # ---------------------------------------------------------
    def rob(self, nums: List[int]) -> int:
        # rob1 represents dp[i-2] (two houses back)
        # rob2 represents dp[i-1] (one house back)
        rob1 = 0
        rob2 = 0
        
        # We walk down the street visiting each house 'n'
        for n in nums:
            # Choice A (Rob it): 'n' money + haul from two houses ago (rob1)
            # Choice B (Skip it): 0 new money, keep the haul from one house ago (rob2)
            current_max = max(n + rob1, rob2)
            
            # SLIDING THE WINDOW FORWARD
            # What used to be "one house back" (rob2) becomes "two houses back" (rob1)
            rob1 = rob2
            
            # Our newly calculated max becomes the new "one house back" for the next loop
            rob2 = current_max
            
        # By the end of the street, rob2 holds the absolute maximum possible haul.
        return rob2
```

### Problem 004: House Robber II (LeetCode 213)

### Problem Definition and Constraints
You are given an integer array `nums` where `nums[i]` represents the amount of money the $i$-th house has. 

This time, the houses are arranged in a **circle**. That means the first house and the last house are neighbors. 
You still cannot rob two adjacent houses. Return the maximum amount of money you can rob tonight without alerting the police.

**Examples:**
* **Example 1:**
  * **Input:** `nums = [3,4,3]`
  * **Output:** `4`
  * *Explanation:* You cannot rob `nums[0]` (3) + `nums[2]` (3) = 6 because they are adjacent in a circle. The maximum is to just rob `nums[1]` (4).
* **Example 2:**
  * **Input:** `nums = [2,9,8,3,6]`
  * **Output:** `15`
  * *Explanation:* You cannot rob `nums[0]` (2) + `nums[4]` (6) because they are connected. The optimal path is `nums[1]` (9) + `nums[4]` (6) = 15. *(Note: We skipped House 0 entirely to make this legal!)*

* Constraints:
  * 1 <= nums.length <= 100
  * 0 <= nums[i] <= 200

### 1D Dynamic Programming Approach (The "Two Linear Streets" Trick)
When you put the houses in a circle, it feels like the math gets insanely complicated. But there is a massive logical shortcut. 

Because House 0 and the Last House are connected, **it is physically impossible to rob both of them in the same night.** 

Therefore, the absolute best strategy in the universe MUST fall into one of two realities:
1. **Reality A:** We consider robbing House 0. That means the Last House is dead to us. We pretend it doesn't exist. We just run our normal "House Robber I" logic on the street from `nums[0]` to `nums[-2]`.
2. **Reality B:** We consider robbing the Last House. That means House 0 is dead to us. We pretend it doesn't exist. We just run our normal "House Robber I" logic on the street from `nums[1]` to `nums[-1]`.

Our final answer is simply `max(Reality A, Reality B)`. 
We don't need to write any new complicated circular logic; we literally just use our optimized `O(1)` space code from the previous problem and run it twice!

* **Time Complexity:** $O(n)$ — We walk down the street twice, which is $O(2n)$, which simplifies to $O(n)$.
* **Space Complexity:** $O(1)$ — We are still just using our two sliding variables `rob1` and `rob2`.

```python
from typing import List

class Solution:
    def rob(self, nums: List[int]) -> int:
        # EDGE CASE: If there is only one house, the "circle" doesn't matter.
        # We have to catch this here, because if we slice an array of length 1 
        # (like nums[:-1]), it becomes empty and messes up our helper function.
        if len(nums) == 1:
            return nums[0]
            
        # Reality A: Skip the Last House (nums[:-1] slices off the end)
        reality_a = self.rob_linear(nums[:-1])
        
        # Reality B: Skip the First House (nums[1:] slices off the beginning)
        reality_b = self.rob_linear(nums[1:])
        
        # Return the richest reality!
        return max(reality_a, reality_b)


    # ---------------------------------------------------------
    # HELPER FUNCTION: This is literally our exact House Robber I code!
    # ---------------------------------------------------------
    def rob_linear(self, nums: List[int]) -> int:
        rob1 = 0
        rob2 = 0
        
        for n in nums:
            current_max = max(n + rob1, rob2)
            
            # Slide the window
            rob1 = rob2
            rob2 = current_max
            
        return rob2
```

### Problem 005: Longest Palindromic Substring (LeetCode 5)

### Problem Definition and Constraints
Given a string `s`, return the longest substring of `s` that is a palindrome.
A palindrome is a string that reads the same forward and backward. If there are multiple palindromic substrings that have the same length, return any one of them.

**Examples:**
* **Example 1:**
  * **Input:** `s = "ababd"`
  * **Output:** `"bab"` (or `"aba"`)
  * *Explanation:* Both "aba" and "bab" are valid answers.
* **Example 2:**
  * **Input:** `s = "abbc"`
  * **Output:** `"bb"`

* Constraints:
  * 1 <= s.length <= 1000
  * `s` contains only digits and English letters.

### Expand Around Center Approach ($O(1)$ Space)
While this is categorized under Dynamic Programming, the standard DP array solution uses $O(n^2)$ space, which is incredibly wasteful. The absolute best way to solve this is by exploiting the physical nature of a palindrome: **Palindromes perfectly mirror outward from their center.**

Instead of checking every possible combination of letters, we just walk down the string, treat every single letter as the "center" of a palindrome, and try to expand outward. 

**The One Big Trap (Odd vs. Even)**
There are two types of palindromes we must check for at every stop:
1. **Odd Length Palindromes (e.g., "aba"):** The center is exactly one letter. We set our left pointer and right pointer to both start on the same letter (`i`), and expand outward.
2. **Even Length Palindromes (e.g., "abba"):** The center is technically the invisible space *between* two letters. To check for this, we set our left pointer to our current letter (`i`), and our right pointer to the letter right next to it (`i + 1`), and expand outward.

We just run both of these checks for every letter, keep track of the longest string we build, and return it at the end!

* **Time Complexity:** $O(n^2)$ — We check $n$ centers, and expanding outward can take at most $n$ steps.
* **Space Complexity:** $O(1)$ — We are only storing pointers (integers) and the final result string, no massive arrays!

```python
class Solution:
    def longestPalindrome(self, s: str) -> str:
        # These will store the best palindrome we've found so far
        res = ""
        resLen = 0
        
        for i in range(len(s)):
            # ---------------------------------------------------
            # CHECK 1: Odd Length Palindromes (Center is one char)
            # ---------------------------------------------------
            l, r = i, i
            # Expand outward as long as we don't go out of bounds 
            # AND the left and right characters match perfectly.
            while l >= 0 and r < len(s) and s[l] == s[r]:
                # If this new palindrome is longer than our previous best, save it!
                if (r - l + 1) > resLen:
                    res = s[l:r+1]      # Python slicing to grab the string
                    resLen = r - l + 1
                
                # Move pointers outward to check the next layer
                l -= 1
                r += 1
                
            # ---------------------------------------------------
            # CHECK 2: Even Length Palindromes (Center is between chars)
            # ---------------------------------------------------
            l, r = i, i + 1
            # Exact same expansion logic, we just started the pointers differently!
            while l >= 0 and r < len(s) and s[l] == s[r]:
                if (r - l + 1) > resLen:
                    res = s[l:r+1]
                    resLen = r - l + 1
                
                l -= 1
                r += 1
                
        # By the end of the loop, 'res' holds the undisputed champion
        return res
```

### Problem 006: Palindromic Substrings (LeetCode 647)

### Problem Definition and Constraints
Given a string `s`, return the number of substrings within `s` that are palindromes.
A palindrome is a string that reads the same forward and backward. Note that different substrings are counted as different palindromes even if the string contents are the same (e.g., in "aaa", the first "a" and the second "a" are counted separately).

**Examples:**
* **Example 1:**
  * **Input:** `s = "abc"`
  * **Output:** `3`
  * *Explanation:* "a", "b", "c".
* **Example 2:**
  * **Input:** `s = "aaa"`
  * **Output:** `6`
  * *Explanation:* "a", "a", "a", "aa", "aa", "aaa".

* Constraints:
  * 1 <= s.length <= 1000
  * `s` consists of lowercase English letters.

### Expand Around Center Approach ($O(1)$ Space)
This uses the exact same "Expand Around Center" logic as *Longest Palindromic Substring*. The only difference is the goal: instead of tracking the *longest* string, we simply count *how many* valid palindromes we find.

Every single time our `while` loop successfully runs, it means our left and right pointers are sitting on a valid palindrome. So, instead of doing length math (`r - l + 1`) and string slicing (`s[l:r+1]`), we just add `1` to a counter. 

* **Time Complexity:** $O(n^2)$ — We check $n$ centers, and expanding outward can take at most $n$ steps.
* **Space Complexity:** $O(1)$ — We just store a single integer counter.

```python
class Solution:
    def countSubstrings(self, s: str) -> int:
        # This will keep a running tally of every palindrome we find
        total_palindromes = 0
        
        for i in range(len(s)):
            # ---------------------------------------------------
            # CHECK 1: Odd Length Palindromes (Center is one char)
            # ---------------------------------------------------
            l, r = i, i
            # Expand outward as long as it's a valid palindrome
            while l >= 0 and r < len(s) and s[l] == s[r]:
                # We found one! Increment the counter.
                total_palindromes += 1
                
                # Move pointers outward to check the next layer
                l -= 1
                r += 1
                
            # ---------------------------------------------------
            # CHECK 2: Even Length Palindromes (Center is between chars)
            # ---------------------------------------------------
            l, r = i, i + 1
            # Exact same expansion logic
            while l >= 0 and r < len(s) and s[l] == s[r]:
                # We found one! Increment the counter.
                total_palindromes += 1
                
                l -= 1
                r += 1
                
        return total_palindromes
```

### Problem 007: Decode Ways (LeetCode 91)

### Problem Definition and Constraints
A string consisting of uppercase English characters can be encoded to a number using the following mapping:
*   'A' -> "1"
*   'B' -> "2"
*   ...
*   'Z' -> "26"

To decode a message, digits must be grouped and then mapped back into letters using the reverse of the mapping above. There may be multiple ways to decode a message. For example, "1012" can be mapped into:
*   "JAB" with the grouping (10 1 2)
*   "JL" with the grouping (10 12)

The grouping (1 01 2) is invalid because "01" cannot be mapped into a letter since it contains a leading zero.
Given a string `s` containing only digits, return the number of ways to decode it.

**Examples:**
* **Example 1:**
  * **Input:** `s = "12"`
  * **Output:** `2`
  * *Explanation:* "12" could be decoded as "AB" (1 2) or "L" (12).
* **Example 2:**
  * **Input:** `s = "01"`
  * **Output:** `0`
  * *Explanation:* "01" cannot be decoded because "01" cannot be mapped into a letter.

* Constraints:
  * 1 <= s.length <= 100
  * `s` consists of digits.

### Core Logic: Climbing Stairs with Traps
At any given digit, you have two doors (choices):
1.  **The 1-step door (Single digit):** Legal if the digit is not `'0'`.
2.  **The 2-step door (Double digit):** Legal if the two digits form a number between `10` and `26`.

**The "Tagging Along" Principle (Why we ADD past paths):**
If both doors are valid, we take all the successful paths from 1 step ago, and tag our new single letter onto the end of them. Then, we take all the paths from 2 steps ago, and tag our new double-letter onto the end of them. Because the paths from Door 1 and Door 2 end in completely different letters, there are zero duplicate paths. We simply add the two unique piles together.

**Edge Cases ("Guilty Until Proven Innocent"):**
We do not need special `if` statements for `"00"` or `"30"`. We assume every new step has `0` valid paths. We only pull paths forward if a door successfully opens. If a zero locks both doors at the same time, the path count stays `0`. This `0` naturally propagates forward through the sliding window, permanently killing the invalid string.

### Approach 1: 1D Dynamic Programming ($O(n)$ Space)
We build a 1-indexed `dp` array where `dp[i]` tracks the total valid ways to decode a string of *length* `i`. Because we need a `dp[0]` to represent the "empty space" before the string starts, our array indices are shifted forward by 1 compared to the raw Python string indices.
* **Time Complexity:** $O(n)$ — We iterate through the string exactly once.
* **Space Complexity:** $O(n)$ — We create a DP array of size $n + 1$.

### Approach 2: Space Optimization ($O(1)$ Space)
Calculating `dp[i]` strictly relies on `dp[i-1]` and `dp[i-2]`. We do not need to keep the entire historical array in memory. We can replace the $O(n)$ array with two variables (`one_back` and `two_back`) that slide forward. Because we drop the array, we no longer need length-based math offsets; our loop directly traverses the literal 0-indexed positions of the string.
* **Time Complexity:** $O(n)$
* **Space Complexity:** $O(1)$ — Only tracking two variables.

```python
class Solution:
    # ---------------------------------------------------------
    # Approach 1: Standard DP Array (O(n) Space)
    # ---------------------------------------------------------
    def numDecodings_array(self, s: str) -> int:
        # Instant failure: empty string or starts with a '0'
        if not s or s[0] == '0':
            return 0
            
        # dp[i] represents the number of ways to decode a string of LENGTH i.
        # It requires an extra slot at the front for the "empty space" (length 0).
        dp = [0] * (len(s) + 1)
        
        # Base Cases
        dp[0] = 1 # 1 way to decode the imaginary empty space
        dp[1] = 1 # 1 way to decode the first letter (we verified it isn't '0')
        
        # Loop i represents the LENGTH of the string we are currently checking
        for i in range(2, len(s) + 1):
            
            # The 1-Step Door
            # Because i is a length, we use i - 1 to look at the correct 0-indexed Python character
            if s[i - 1] != '0':
                dp[i] += dp[i - 1]
                
            # The 2-Step Door
            # Slicing from i - 2 up to i grabs the previous and current digit
            two_digit = int(s[i - 2 : i])
            if 10 <= two_digit <= 26:
                dp[i] += dp[i - 2]
                
        return dp[len(s)]

    # ---------------------------------------------------------
    # Approach 2: Sliding Variables (O(1) Space)
    # ---------------------------------------------------------
    def numDecodings(self, s: str) -> int:
        if not s or s[0] == '0':
            return 0
            
        # The Memory (Base Cases)
        two_back = 1 # Represents dp[0]
        one_back = 1 # Represents dp[1]
        
        # Loop i represents the literal 0-indexed position in the string
        for i in range(1, len(s)):
            # Guilty until proven innocent: reset paths to 0 for this step
            current_ways = 0
            
            # The 1-Step Door
            # We look directly at the current index i
            if s[i] != '0':
                current_ways += one_back
                
            # The 2-Step Door
            # We slice from i - 1 up to i + 1 to grab the 2-digit combo
            two_digit = int(s[i - 1 : i + 1])
            if 10 <= two_digit <= 26:
                current_ways += two_back
                
            # The Slide
            # If both doors failed (e.g., "00"), current_ways stays 0.
            # That 0 gets passed to one_back and propagates forward, killing the path.
            two_back = one_back
            one_back = current_ways
            
        return one_back
```

### Problem 008: Coin Change (LeetCode 322)

### Problem Definition and Constraints
You are given an integer array `coins` representing coins of different denominations and an integer `amount` representing a target amount of money. Return the fewest number of coins that you need to make up the exact target amount. 
If it is impossible to make up the amount, return `-1`. 
*Note: You may assume that you have an unlimited number of each coin.*

**Examples:**
* **Example 1:**
  * **Input:** `coins = [1,5,10], amount = 12`
  * **Output:** `3`
  * *Explanation:* 12 = 10 + 1 + 1. 
* **Example 2:**
  * **Input:** `coins = [2], amount = 3`
  * **Output:** `-1`
  * *Explanation:* 3 cannot be made up with coins of size 2.
* **Example 3:**
  * **Input:** `coins = [1], amount = 0`
  * **Output:** `0`

**Constraints:**
* 1 <= coins.length <= 10
* 1 <= coins[i] <= 2^31 - 1
* 0 <= amount <= 10000

### Core Logic: The "Time Travel" Look-Back
Unlike finding *all possible ways* (which requires addition, and defaulting to `0`), we are looking for the *minimum* (which requires the `min()` function, and defaulting to `infinity`). 

Imagine you are standing in a room labeled `$12`, holding a `$5` coin. How do you know the minimum coins needed to make `$12`? 
You "time travel" back to the room labeled `$7` (since $12 - $5 = $7). You don't need to recalculate how to make `$7`; the answer is already permanently recorded there. You simply take the best answer for `$7`, and **add 1** (representing the `$5` coin in your hand). 

You do this look-back for *every* coin denomination you have. Whichever coin's look-back gives you the smallest total, you write that number down in the `$12` room and move on.

**The "Infinity" Trap:**
Because we use `min()` to compare paths, if we default our DP array to `0`, `0` will always win (e.g., `min(0, 1 + dp[7])` is always 0). Therefore, we must initialize our array with a value that represents "impossible." `float('inf')` works, but a cleaner trick is `amount + 1`. The absolute worst-case scenario is making the amount using entirely `1` coins, which takes exactly `amount` coins. Therefore, `amount + 1` is mathematically guaranteed to act as an unreachable "infinity."

### Approach: 1D Dynamic Programming (Bottom-Up)
We build a 1D `dp` array where `dp[a]` stores the minimum coins needed to make amount `a`. We solve for every amount from `1` up to `amount`. By the time we evaluate a higher amount, all the smaller sub-problems it relies on are already fully solved.

* **Time Complexity:** $O(n \times t)$ — where $n$ is the number of coins and $t$ is the target amount. For every amount from 1 to $t$, we loop through all $n$ coins.
* **Space Complexity:** $O(t)$ — We create a DP array of size $t + 1$. Because we need to look back by varying, unpredictable coin amounts (we might need to look back 1 step, or 500 steps), we cannot use the $O(1)$ space optimization (sliding window) here. We must keep the whole array in memory.

```python
from typing import List

class Solution:
    # ---------------------------------------------------------
    # Approach: 1D Dynamic Programming Bottom-Up
    # ---------------------------------------------------------
    def coinChange(self, coins: List[int], amount: int) -> int:
        # The maximum possible coins we could ever need is exactly `amount` 
        # (if we only used 1-cent coins). 
        # Therefore, `amount + 1` acts as a safe "infinity" default.
        dp = [(amount + 1)] * (amount + 1)
        
        # Base Case: It takes 0 coins to make $0.
        dp[0] = 0
        
        # Loop through every amount from 1 up to our target
        for a in range(1, amount + 1):
            
            # For the current amount 'a', try using every coin we have
            for coin in coins:
                
                # We can only use the coin if it isn't larger than the amount we are trying to make
                if a - coin >= 0:
                    
                    # State Transition: 
                    # Do we keep our current best record for amount 'a'?
                    # OR is it better to take the best record from 'a - coin' and add 1 (for the current coin)?
                    dp[a] = min(dp[a], 1 + dp[a - coin])
                    
        # If the target amount's value is still our "infinity" default, it means 
        # it was impossible to reach. Return -1.
        return dp[amount] if dp[amount] != (amount + 1) else -1
```

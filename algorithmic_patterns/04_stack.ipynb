### Problem 001: Valid Parentheses (LeetCode 20)

### Problem Definition and Constraints
Given a string `s` consisting of the characters `'(', ')', '{', '}', '['` and `']'`, determine if the input string is valid.
A string is valid if:
1. Every open bracket is closed by the same type of close bracket.
2. Open brackets are closed in the correct order.
3. Every close bracket has a corresponding open bracket of the same type.

* Constraints:
  * 1 <= s.length <= 10^4
  * `s` consists of parentheses only.

### Examples
* **Example 1:**
  * Input: `s = "[]"`
  * Output: `true`
* **Example 2:**
  * Input: `s = "([{}])"`
  * Output: `true`
* **Example 3:**
  * Input: `s = "[(])"`
  * Output: `false`
  * Explanation: The brackets are not closed in the correct order. The `[` is closed by a `)`, which is invalid.

### Brute Force Approach
The baseline strategy relies on string manipulation. We continuously scan the string for valid, adjacent matching pairs (`"()"`, `"[]"`, `"{}"`) and replace them with an empty string. We repeat this process until the string stops shrinking. If the resulting string is completely empty, all brackets were validly matched.
* Time Complexity: $O(n^2)$ — In the worst-case scenario, the `replace` function scans the entire string repeatedly, resulting in quadratic time.
* Space Complexity: $O(n)$ — Strings are immutable in Python, meaning every replacement allocates memory for a brand new string.

### Optimized Approach (Stack & Hash Map)
We iterate through the string exactly once. We utilize a Stack (implemented as a Python list) to keep track of opening brackets. We also use a Hash Map to map every closing bracket to its corresponding opening bracket. 
When we encounter an opening bracket, we push it onto the stack. When we encounter a closing bracket, we check the top element of the stack. If the stack is empty or the top element does not match the Hash Map's required opening bracket, we immediately return `false`. If it matches, we pop the element off the stack. Finally, we return `true` only if the stack is completely empty at the end.
* Time Complexity: $O(n)$ — We traverse the string exactly once. Pushing and popping from the end of a list in Python are $O(1)$ operations.
* Space Complexity: $O(n)$ — In the worst-case scenario (e.g., `s = "(((((((("`), the stack will grow to store all $n$ characters.

```python
class Solution:
    
    # --- BRUTE FORCE APPROACH ---
    def isValid_brute_force(self, s: str) -> bool:
        # We loop continuously as long as a valid pair exists somewhere in the string
        while "()" in s or "[]" in s or "{}" in s:
            # Erase the valid pairs by replacing them with empty strings
            s = s.replace("()", "")
            s = s.replace("[]", "")
            s = s.replace("{}", "")
            
        # If the string is completely empty, it was a valid parenthesis string
        return len(s) == 0

    # --- OPTIMIZED APPROACH ---
    def isValid(self, s: str) -> bool:
        # In Python, a standard list functions perfectly as a Stack
        stack = []
        
        # A Hash Map to instantly check which opening bracket belongs to which closing bracket
        bracket_map = {")": "(", "]": "[", "}": "{"}
        
        for char in s:
            # Check if the character is a CLOSING bracket (exists as a key in our map)
            if char in bracket_map:
                
                # If the stack has items, 'pop' removes and returns the top item.
                # If the stack is empty, we assign a dummy value '#' so the comparison doesn't crash.
                top_element = stack.pop() if stack else '#'
                
                # If the plate we pulled off the top doesn't match the required opening bracket, fail.
                if bracket_map[char] != top_element:
                    return False
            else:
                # The character is an OPENING bracket. Push it onto the top of the stack.
                stack.append(char)
                
        # If the stack is empty at the end, every bracket found a valid partner.
        # If there are items left over (e.g., s = "["), it returns False.
        return len(stack) == 0
```

### Problem 001: Valid Parentheses (LeetCode 20)

### Problem Definition and Constraints
Given a string `s` consisting of the characters `'(', ')', '{', '}', '['` and `']'`, determine if the input string is valid.
A string is valid if:
1. Every open bracket is closed by the same type of close bracket.
2. Open brackets are closed in the correct order.
3. Every close bracket has a corresponding open bracket of the same type.

* Constraints:
  * 1 <= s.length <= 10^4
  * `s` consists of parentheses only.

### Examples
* **Example 1:**
  * Input: `s = "[]"`
  * Output: `true`
* **Example 2:**
  * Input: `s = "([{}])"`
  * Output: `true`
* **Example 3:**
  * Input: `s = "[(])"`
  * Output: `false`
  * Explanation: The brackets are not closed in the correct order. The `[` is closed by a `)`, which is invalid.

### Brute Force Approach
The baseline strategy relies on string manipulation. We continuously scan the string for valid, adjacent matching pairs (`"()"`, `"[]"`, `"{}"`) and replace them with an empty string. We repeat this process until the string stops shrinking. If the resulting string is completely empty, all brackets were validly matched.
* Time Complexity: $O(n^2)$ — In the worst-case scenario, the `replace` function scans the entire string repeatedly, resulting in quadratic time.
* Space Complexity: $O(n)$ — Strings are immutable in Python, meaning every replacement allocates memory for a brand new string.

### Optimized Approach (Stack & Hash Map)
We iterate through the string exactly once. We utilize a Stack (implemented as a Python list) to keep track of opening brackets. We also use a Hash Map to map every closing bracket to its corresponding opening bracket. 
When we encounter an opening bracket, we push it onto the stack. When we encounter a closing bracket, we check the top element of the stack. If the stack is empty or the top element does not match the Hash Map's required opening bracket, we immediately return `false`. If it matches, we pop the element off the stack. Finally, we return `true` only if the stack is completely empty at the end.
* Time Complexity: $O(n)$ — We traverse the string exactly once. Pushing and popping from the end of a list in Python are $O(1)$ operations.
* Space Complexity: $O(n)$ — In the worst-case scenario (e.g., `s = "(((((((("`), the stack will grow to store all $n$ characters.

```python
class Solution:
    
    # --- BRUTE FORCE APPROACH ---
    def isValid_brute_force(self, s: str) -> bool:
        # We loop continuously as long as a valid pair exists somewhere in the string
        while "()" in s or "[]" in s or "{}" in s:
            # Erase the valid pairs by replacing them with empty strings
            s = s.replace("()", "")
            s = s.replace("[]", "")
            s = s.replace("{}", "")
            
        # If the string is completely empty, it was a valid parenthesis string
        return len(s) == 0

    # --- OPTIMIZED APPROACH ---
    def isValid(self, s: str) -> bool:
        # In Python, a standard list functions perfectly as a Stack
        stack = []
        
        # A Hash Map to instantly check which opening bracket belongs to which closing bracket
        bracket_map = {")": "(", "]": "[", "}": "{"}
        
        for char in s:
            # Check if the character is a CLOSING bracket (exists as a key in our map)
            if char in bracket_map:
                
                # If the stack has items, 'pop' removes and returns the top item.
                # If the stack is empty, we assign a dummy value '#' so the comparison doesn't crash.
                top_element = stack.pop() if stack else '#'
                
                # If the plate we pulled off the top doesn't match the required opening bracket, fail.
                if bracket_map[char] != top_element:
                    return False
            else:
                # The character is an OPENING bracket. Push it onto the top of the stack.
                stack.append(char)
                
        # If the stack is empty at the end, every bracket found a valid partner.
        # If there are items left over (e.g., s = "["), it returns False.
        return len(stack) == 0
```

### Problem 002: Min Stack (LeetCode 155)

### Problem Definition and Constraints
Design a stack class that supports the `push`, `pop`, `top`, and `getMin` operations.
* `MinStack()` initializes the stack object.
* `void push(int val)` pushes the element `val` onto the stack.
* `void pop()` removes the element on the top of the stack.
* `int top()` gets the top element of the stack.
* `int getMin()` retrieves the minimum element in the stack.
* **Constraint:** Each function must run in exactly $O(1)$ time.

### Examples
* **Example 1:**
  * Input: `["MinStack", "push", 1, "push", 2, "push", 0, "getMin", "pop", "top", "getMin"]`
  * Output: `[null, null, null, null, 0, null, 2, 1]`

### Brute Force Approach
The simplest way to build this is to use a single standard array as our stack. `push`, `pop`, and `top` all run in $O(1)$ time natively on Python lists. However, for `getMin()`, we would have to scan the entire array to find the smallest number using the built-in `min()` function.
* Time Complexity: $O(1)$ for push/pop/top, but $O(n)$ for `getMin()`. This violates the strict $O(1)$ constraint of the problem.
* Space Complexity: $O(n)$ to store the elements.

### Optimized Approach (Two Stacks)
To achieve $O(1)$ time for `getMin()`, we can trade some space for speed by maintaining **two** separate stacks:
1. A standard `stack` to keep track of the actual elements.
2. A `min_stack` to keep track of the minimum value at each corresponding level.

When we `push` a new value, we add it to the `stack`. We also check the top of our `min_stack`: we take the smaller of the new value and the current minimum, and push that onto the `min_stack`.
When we `pop` a value, we pop from *both* stacks, ensuring they always stay in sync. Because the `min_stack` always has the historical minimum sitting right at the top, `getMin()` is just a direct $O(1)$ lookup.
* Time Complexity: $O(1)$ for all operations (`push`, `pop`, `top`, `getMin`).
* Space Complexity: $O(n)$ — We maintain an additional stack of the same size as the main stack.

```python
class MinStack_BruteForce:
    
    # --- BRUTE FORCE APPROACH ---
    def __init__(self):
        self.stack = []

    def push(self, val: int) -> None:
        self.stack.append(val)

    def pop(self) -> None:
        self.stack.pop()

    def top(self) -> int:
        return self.stack[-1]

    def getMin(self) -> int:
        # Scanning the entire list takes O(n) time, failing the problem constraint
        return min(self.stack)


class MinStack:
    
    # --- OPTIMIZED APPROACH ---
    def __init__(self):
        # We maintain two separate lists
        self.stack = []
        self.min_stack = []

    def push(self, val: int) -> None:
        self.stack.append(val)
        
        # If min_stack is empty, the new value is inherently the minimum.
        # Otherwise, the new minimum is the smaller of 'val' and the current minimum.
        if not self.min_stack:
            self.min_stack.append(val)
        else:
            current_min = self.min_stack[-1]
            self.min_stack.append(min(val, current_min))

    def pop(self) -> None:
        # We must pop from both stacks to keep them perfectly synced
        self.stack.pop()
        self.min_stack.pop()

    def top(self) -> int:
        return self.stack[-1]

    def getMin(self) -> int:
        # The top of the min_stack always holds the correct minimum for the current stack state
        return self.min_stack[-1]
```

### Problem 003: Evaluate Reverse Polish Notation (LeetCode 150)

### Problem Definition and Constraints
You are given an array of strings `tokens` that represents a valid arithmetic expression in Reverse Polish Notation (RPN). Return the integer that represents the evaluation of the expression.
* The operands may be integers or the results of other operations.
* The operators include `+`, `-`, `*`, and `/`.
* Assume that division between integers always **truncates toward zero**.
* Constraints:
  * 1 <= tokens.length <= 1000
  * `tokens[i]` is an operator or an integer in the range [-200, 200].

### Examples
* **Example 1:**
  * Input: `tokens = ["1","2","+","3","*","4","-"]`
  * Output: `5`
  * Explanation: `((1 + 2) * 3) - 4 = 5`

### Brute Force Approach
A brute-force method involves scanning the array from left to right to find the first operator. Once found, we take the two numbers immediately to its left, perform the math operation, and replace all three elements (the two numbers and the operator) with the new result. We then start scanning the array all over again from the beginning until only one number remains.
* Time Complexity: $O(n^2)$ — In the worst case, we repeatedly scan the array and shift elements around after every single operation.
* Space Complexity: $O(n)$ — Modifying the array or creating new array slices takes extra memory.

### Optimized Approach (Stack)
Reverse Polish Notation is naturally designed to be solved using a stack. We iterate through the `tokens` array one by one:
1. If the token is a number, we push it onto the stack.
2. If the token is an operator, we pop the top two numbers off the stack. 
3. We apply the operator to those two numbers (being careful with the order: the first popped is the right operand, the second popped is the left operand).
4. We push the result back onto the stack.

When the loop finishes, the final answer will be the only item left in the stack.
* Time Complexity: $O(n)$ — We iterate through the `tokens` array exactly once. Pushing and popping from a stack takes $O(1)$ constant time.
* Space Complexity: $O(n)$ — In the worst-case scenario (e.g., all numbers and no operators until the very end), the stack will hold up to $n$ elements.

```python
from typing import List

class Solution:
    
    # --- BRUTE FORCE APPROACH ---
    def evalRPN_brute_force(self, tokens: List[str]) -> int:
        # Create a copy so we don't destroy the original input
        arr = tokens[:]
        operators = {"+", "-", "*", "/"}
        
        # Keep reducing the array until only the final answer remains
        while len(arr) > 1:
            for i in range(len(arr)):
                if arr[i] in operators:
                    # Grab the two numbers to the left of the operator
                    left_operand = int(arr[i - 2])
                    right_operand = int(arr[i - 1])
                    operator = arr[i]
                    
                    # Perform the calculation
                    if operator == "+":
                        result = left_operand + right_operand
                    elif operator == "-":
                        result = left_operand - right_operand
                    elif operator == "*":
                        result = left_operand * right_operand
                    elif operator == "/":
                        # int() truncates toward zero in Python
                        result = int(left_operand / right_operand) 
                        
                    # Replace the 3 items (num, num, operator) with the result
                    arr[i - 2 : i + 1] = [str(result)]
                    
                    # Break the inner loop to start scanning from the beginning again
                    break
                    
        return int(arr[0])


    # --- OPTIMIZED APPROACH ---
    def evalRPN(self, tokens: List[str]) -> int:
        stack = []
        operators = {"+", "-", "*", "/"}
        
        for token in tokens:
            if token not in operators:
                # It's a number. Push it to the stack.
                stack.append(int(token))
            else:
                # It's an operator. Pop the top two numbers.
                # The first number popped is the RIGHT operand.
                # The second number popped is the LEFT operand.
                right_operand = stack.pop()
                left_operand = stack.pop()
                
                if token == "+":
                    stack.append(left_operand + right_operand)
                elif token == "-":
                    stack.append(left_operand - right_operand)
                elif token == "*":
                    stack.append(left_operand * right_operand)
                elif token == "/":
                    # int(a / b) ensures we truncate toward zero
                    stack.append(int(left_operand / right_operand))
                    
        # The final result is the only item left in the stack
        return stack[0]
```

### Problem 004: Daily Temperatures (LeetCode 739)

### Problem Definition and Constraints
You are given an array of integers `temperatures` where `temperatures[i]` represents the daily temperature on the `ith` day.
Return an array `result` where `result[i]` is the number of days you have to wait after the `ith` day to get a warmer temperature. If there is no future day with a warmer temperature, set `result[i]` to `0` instead.
* Constraints:
  * 1 <= temperatures.length <= 1000
  * 1 <= temperatures[i] <= 100

### Examples
* **Example 1:**
  * Input: `temperatures = [30, 38, 30, 36, 35, 40, 28]`
  * Output: `[1, 4, 1, 2, 1, 0, 0]`
  * Explanation: 
    * Day 0 (30): Next warmer is 38 (1 day later).
    * Day 1 (38): Next warmer is 40 (4 days later).
* **Example 2:**
  * Input: `temperatures = [22, 21, 20]`
  * Output: `[0, 0, 0]`
  * Explanation: Temperatures only go down, so there is never a warmer day.

### Brute Force Approach
For every single day in the array, we use an inner loop to scan all the future days to its right. The moment we find a temperature strictly greater than our current day's temperature, we calculate the distance between the two days, save it to our result array, and stop scanning.
* Time Complexity: $O(n^2)$ — In the worst-case scenario (e.g., temperatures are strictly decreasing), we have to scan the rest of the array for every single day, doing massive amounts of redundant work.
* Space Complexity: $O(1)$ — Excluding the space needed for the output array, we don't use any extra data structures.

### Optimized Approach (Monotonic Stack)
We use a **Monotonic Decreasing Stack**. Instead of looking forward to find a warmer day, we store days in a stack while we wait for a warmer day to arrive. 

We iterate through the array day by day. If the current day's temperature is colder than or equal to the previous days, we just push its **index** onto the stack (they are still waiting for a warm day). 
However, if the current day is *warmer* than the day at the top of the stack, we found a match! We pop that older day off the stack, calculate how many days it waited (`current_index - popped_index`), and record the answer. We keep popping until the stack is empty or the top of the stack is warmer than the current day.
* Time Complexity: $O(n)$ — Even though there is a `while` loop inside a `for` loop, every single index is pushed onto the stack exactly once and popped exactly once. The total number of operations scales linearly with the size of the array.
* Space Complexity: $O(n)$ — In the worst case (temperatures constantly dropping), the stack will hold the indices of every single day.

```python
from typing import List

class Solution:
    
    # --- BRUTE FORCE APPROACH ---
    def dailyTemperatures_brute_force(self, temperatures: List[int]) -> List[int]:
        n = len(temperatures)
        # Initialize the result array with 0s. 
        # If we never find a warmer day, it correctly defaults to 0.
        result = [0] * n
        
        for i in range(n):
            # Scan future days 
            for j in range(i + 1, n):
                if temperatures[j] > temperatures[i]:
                    # Calculate how many days we waited
                    result[i] = j - i
                    # We found the NEXT warmer day, so stop scanning
                    break
                    
        return result


    # --- OPTIMIZED APPROACH ---
    def dailyTemperatures(self, temperatures: List[int]) -> List[int]:
        n = len(temperatures)
        result = [0] * n
        
        # The stack will store INDICES of temperatures, not the temperatures themselves.
        # This is because we need the indices to calculate the distance (number of days).
        stack = []
        
        for current_day_index in range(n):
            current_temp = temperatures[current_day_index]
            
            # While the stack has items AND the current day is warmer than the 
            # temperature of the day sitting at the top of the stack:
            while stack and current_temp > temperatures[stack[-1]]:
                # We found a warmer day for the index at the top of the stack!
                popped_index = stack.pop()
                
                # Calculate the wait time and save it
                wait_time = current_day_index - popped_index
                result[popped_index] = wait_time
                
            # After resolving any older days that were waiting, 
            # push the current day onto the stack to wait for its own warmer day.
            stack.append(current_day_index)
            
        return result
```

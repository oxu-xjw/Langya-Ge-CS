### Problem 001: Maximum Subarray (LeetCode 53)

### Problem Definition and Constraints
Given an integer array `nums`, find the contiguous subarray (containing at least one number) which has the largest sum and return its sum.

**Examples:**
* **Example 1:**
  * **Input:** `nums = [2,-3,4,-2,2,1,-1,4]`
  * **Output:** `8`
  * *Explanation:* The subarray `[4,-2,2,1,-1,4]` has the largest sum `8`.
* **Example 2:**
  * **Input:** `nums = [-1]`
  * **Output:** `-1`

**Constraints:**
* 1 <= nums.length <= 100,000
* -10,000 <= nums[i] <= 10,000

### Core Logic: Kadane's Algorithm (The "Toxic Baggage" Analogy)
This is one of the most famous algorithms in computer science, known as **Kadane's Algorithm**. While it is often categorized as Dynamic Programming, its logic is highly Greedy. 

Imagine you are walking through the array, picking up numbers and keeping a "running sum" in your backpack. 
*   If the numbers you've collected so far add up to a **positive** number, they are an asset. Even if the next number is negative, your backpack can help absorb the blow. You should keep carrying it.
*   If the numbers you've collected so far add up to a **negative** number, your backpack has become "toxic baggage." If you add a negative backpack to *any* future number, it will just drag that new number down. 

**The Greedy Choice:** The moment your running sum drops below zero, you immediately throw the backpack away. You reset your running sum to `0` and start a brand new subarray with the very next element. At every single step, you check if your current running sum is the biggest you've ever seen, and record it.

### Approach: Kadane's Algorithm
Instead of a double loop checking every combination, we maintain a single running sum and make a greedy choice at each step to either keep extending the subarray or start fresh.
1. Initialize `max_sum` to the first number (to handle arrays with all negative numbers).
2. Initialize `cur_sum` to `0`.
3. Loop through every number in the array.
4. If `cur_sum` is negative, reset it to `0` (drop the toxic baggage).
5. Add the current number to `cur_sum`.
6. Update `max_sum` to be the maximum of itself and `cur_sum`.

* **Time Complexity:** O(n) — We iterate through the `nums` array exactly once, completely avoiding the nested loop of the brute force approach.
* **Space Complexity:** O(1) — We only track two variables (`cur_sum` and `max_sum`), using constant extra space regardless of how large the input array is.

```python
from typing import List

class Solution:
    def maxSubArray(self, nums: List[int]) -> int:
        # max_sum tracks the absolute best subarray we've ever seen.
        # Initialize to the first element to handle arrays that are entirely negative.
        max_sum = nums[0] 
        
        # cur_sum tracks our current backpack of numbers.
        cur_sum = 0 
        
        for n in nums:
            # THE GREEDY CHOICE: 
            # If our previous running sum is negative, it is a liability.
            # Throw it away before picking up the current number 'n'.
            if cur_sum < 0:
                cur_sum = 0
                
            # Add the current number to our backpack
            cur_sum += n
            
            # Record the highest score we've ever seen
            max_sum = max(max_sum, cur_sum)
            
        return max_sum
```

### Problem 002: Jump Game (LeetCode 55)

### Problem Definition and Constraints
You are given an integer array `nums` where each element `nums[i]` indicates your maximum jump length at that position.
Return `true` if you can reach the last index starting from index 0, or `false` otherwise.

**Examples:**
* **Example 1:**
  * **Input:** `nums = [1,2,0,1,0]`
  * **Output:** `true`
  * *Explanation:* First jump from index 0 to 1, then from index 1 to 3, and lastly from index 3 to 4.
* **Example 2:**
  * **Input:** `nums = [1,2,1,0,1]`
  * **Output:** `false`

**Constraints:**
* 1 <= nums.length <= 1000
* 0 <= nums[i] <= 1000

### Core Logic: The "Moving Goalpost" (Reverse Greedy)
If you try to solve this from left-to-right, it becomes a nightmare of branching paths. You jump to index 1, which gives you 3 new jump options. You have to explore all of them, leading to an exponential $O(2^n)$ brute force solution. 

The brilliant Greedy realization here is to **think in reverse**. 
Imagine the last index is the final "Goalpost." We don't care about how to get there from the start yet. We just ask: *"Who is the closest person standing in front of the Goalpost who can successfully jump to it?"*

If the person at index `3` can reach the Goalpost at index `4`, we don't need to aim for index `4` anymore. We can literally pick up the Goalpost and move it to index `3`. Why? Because we logically know that if *anybody* can just reach index `3`, they are guaranteed to reach the end. 

**The Greedy Choice:** 
We start at the end of the array and work backward. At every step, we check if the current index can reach the Goalpost. If it can, we shift the Goalpost to the current index. If we successfully shift the Goalpost all the way back to index `0`, it means the game is winnable!

### Approach: Greedy (Iterating Backwards)
1. Initialize `goal` to the very last index of the array (`len(nums) - 1`).
2. Loop through the array in reverse, starting from the second-to-last index down to `0`.
3. At each step `i`, calculate your maximum reach: `i + nums[i]`.
4. If your reach is greater than or equal to the `goal`, update `goal = i`.
5. Once the loop finishes, check if `goal == 0`. If it is, return `True`.

* **Time Complexity:** $O(n)$ — We iterate through the `nums` array exactly once in reverse.
* **Space Complexity:** $O(1)$ — We only track a single integer variable (`goal`), regardless of the input size.

```python
from typing import List

class Solution:
    def canJump(self, nums: List[int]) -> bool:
        # Place the initial goalpost at the very last index
        goal = len(nums) - 1
        
        # Iterate backwards starting from the second-to-last index down to 0
        for i in range(len(nums) - 2, -1, -1):
            # Your current index (i) + your jump power (nums[i]) is your max reach.
            # If your max reach can hit or pass the current goal, 
            # shift the goalpost to where you are standing right now.
            if i + nums[i] >= goal:
                goal = i
                
        # If the goalpost successfully made it all the way to the starting line, 
        # the end is reachable from the beginning.
        return True if goal == 0 else False
```

### Problem 003: Jump Game II (LeetCode 45)

### Problem Definition and Constraints
You are given an array of integers `nums`, where `nums[i]` represents the maximum length of a jump towards the right from index `i`. 
You are initially positioned at `nums[0]`.
Return the **minimum** number of jumps to reach the last position in the array. You may assume there is always a valid answer.

**Examples:**
* **Example 1:**
  * **Input:** `nums = [2,4,1,1,1,1]`
  * **Output:** `2`
  * *Explanation:* Jump from index 0 to index 1 (since you can jump 2 spaces), then jump from index 1 to the last index (since you can jump 4 spaces).
* **Example 2:**
  * **Input:** `nums = [2,1,2,1,0]`
  * **Output:** `2`

**Constraints:**
* 1 <= nums.length <= 1000
* 0 <= nums[i] <= 100

### Core Logic: The "Exploration Windows" (1D Breadth-First Search)
In Jump Game I, we just needed to know *if* we could reach the end, so we worked backward. 
Here, we need the *minimum number of jumps*, which completely changes the strategy. We must work forward and think in "levels" or "windows" of reachability. 

Imagine you are standing at the start. 
*   **Jump 0:** You are at index 0. Your window is just `[0]`. From here, you check how far you can possibly jump. Let's say you can reach up to index 2.
*   **Jump 1:** Your new window is `[1, 2]`. You don't know exactly which of these two spots is the best to jump from yet, so you evaluate *both* of them. You find the absolute farthest index you can reach from *anywhere* inside this window. Let's say index 1 can reach index 4, and index 2 can reach index 3. The absolute farthest is 4.
*   **Jump 2:** Your new window becomes `[3, 4]`. 

**The Greedy Choice:** 
You group the array into contiguous "windows" that represent how many jumps it took to get there. You use a Left (`l`) and Right (`r`) pointer to define the current window. You iterate through the current window to find the absolute maximum reach, and then shift your window forward to start the next jump. 

### Approach: Greedy Window / Two Pointers
1. Initialize `jumps = 0`.
2. Initialize `l = 0` and `r = 0`. This is our first window.
3. While our right pointer `r` hasn't reached the end of the array:
   * Scan every index inside the current window (`l` to `r`).
   * Find the `farthest` index we can reach from this entire window.
   * After checking the whole window, it's time to make a jump! Increment `jumps += 1`.
   * Update the window: the new Left boundary starts immediately after the old Right boundary (`l = r + 1`). The new Right boundary becomes the `farthest` point we just calculated.
4. Return `jumps`.

* **Time Complexity:** $O(n)$ — Even though there is a `while` loop and a `for` loop, the `l` and `r` pointers simply slide across the array from left to right. We visit each index exactly once.
* **Space Complexity:** $O(1)$ — We only maintain a few integer variables (`l`, `r`, `farthest`, `jumps`).

```python
from typing import List

class Solution:
    def jump(self, nums: List[int]) -> int:
        jumps = 0
        l = 0  # Left boundary of our current jump window
        r = 0  # Right boundary of our current jump window
        
        # We keep going until our window's right boundary reaches the end of the array
        while r < len(nums) - 1:
            farthest = 0
            
            # Scan every index inside our current jump window
            for i in range(l, r + 1):
                # Calculate the absolute farthest we can reach from this specific index
                farthest = max(farthest, i + nums[i])
                
            # Now that we've scanned the current window, we transition to the next window.
            # This represents officially taking 1 jump.
            l = r + 1
            r = farthest
            jumps += 1
            
        return jumps
```

### Problem 004: Gas Station (LeetCode 134)

### Problem Definition and Constraints
There are `n` gas stations along a circular route. You are given two integer arrays `gas` and `cost`:
*   `gas[i]` is the amount of gas you get at the `i`th station.
*   `cost[i]` is the gas needed to travel from the `i`th station to the `(i + 1)`th station.

You have a car with an unlimited gas tank, starting empty. 
Return the starting gas station's index such that you can travel around the circuit once in the clockwise direction. If it's impossible, return `-1`. (It is guaranteed that if a solution exists, it is unique).

**Examples:**
* **Example 1:**
  * **Input:** `gas = [1,2,3,4], cost = [2,2,4,1]`
  * **Output:** `3`
  * *Explanation:* Start at station 3, fill up with 4 gas, and you will have enough to make it all the way around the circuit.
* **Example 2:**
  * **Input:** `gas = [1,2,3], cost = [2,3,2]`
  * **Output:** `-1`
  * *Explanation:* Total gas available is 6, but total cost is 7. You can never complete the circuit.

**Constraints:**
* 1 <= gas.length == cost.length <= 100,000
* 0 <= gas[i], cost[i] <= 1,000

### Core Logic: The "Global Guarantee" and the "Doomed Segment"
This problem requires two massive logical leaps that completely eliminate the need for a brute-force $O(n^2)$ simulation.

**1. The Global Guarantee:** 
If the total amount of gas on the entire map is greater than or equal to the total cost to travel the map, **a valid starting point is mathematically guaranteed to exist.** Therefore, we can do a quick global check at the very beginning. If `sum(gas) < sum(cost)`, we instantly return `-1`. If it passes, we know an answer exists.

**2. The Doomed Segment (The Greedy Choice):**
Let's say you start at Station A. You make it through Station B and Station C, but when you try to drive to Station D, your tank drops below zero. 
The brute force method would say: *"A failed. Let's try starting at B instead."*

The Greedy logic realizes: **If A couldn't reach D, then B and C can't reach D either!** 
Why? Because when you started at A, you arrived at B with a *surplus* of gas in your tank. Even with that extra surplus, you still couldn't reach D. If you started fresh at B with an empty tank, your chances of reaching D are even worse!

Therefore, the exact moment your tank hits empty, the entire segment from your starting point up to your current point is doomed. You instantly reset your starting point to the very next station (`i + 1`).

### Approach: Greedy (Single Pass)
1. Check the global guarantee: if `sum(gas) < sum(cost)`, return `-1`.
2. Initialize `tank = 0` (your current gas) and `start_index = 0`.
3. Loop through the stations. At each station, add the net gas (`gas[i] - cost[i]`) to your `tank`.
4. If your `tank` drops below zero, it means the current `start_index` (and everything after it up to `i`) is doomed. 
5. Reset your `tank` to `0` and set `start_index = i + 1`.
6. Return `start_index`.

* **Time Complexity:** $O(n)$ — We do one pass to get the sums, and one pass to find the starting index.
* **Space Complexity:** $O(1)$ — We only use integer variables (`tank`, `start_index`).

```python
from typing import List

class Solution:
    def canCompleteCircuit(self, gas: List[int], cost: List[int]) -> int:
        # 1. The Global Guarantee Check
        # If the total cost is strictly greater than the total gas available,
        # it is mathematically impossible to complete the loop.
        if sum(gas) < sum(cost):
            return -1
            
        tank = 0
        start_index = 0
        
        # 2. Finding the exact starting point
        for i in range(len(gas)):
            # Add the net change in gas for this specific station
            tank += gas[i] - cost[i]
            
            # The Greedy Choice: The Doomed Segment
            # If our tank drops below zero, we cannot reach the next station.
            if tank < 0:
                # Reset our tank to empty for our next attempt
                tank = 0
                # Since everything from our old start_index up to 'i' failed,
                # the earliest possible valid starting point must be the next station.
                start_index = i + 1
                
        # Because our Global Guarantee passed at the top of the function,
        # we are absolutely certain that this start_index is the correct and only answer.
        return start_index
```

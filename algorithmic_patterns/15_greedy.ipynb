### Problem 001: Maximum Subarray (LeetCode 53)

### Problem Definition and Constraints
Given an integer array `nums`, find the contiguous subarray (containing at least one number) which has the largest sum and return its sum.

**Examples:**
* **Example 1:**
  * **Input:** `nums = [2,-3,4,-2,2,1,-1,4]`
  * **Output:** `8`
  * *Explanation:* The subarray `[4,-2,2,1,-1,4]` has the largest sum `8`.
* **Example 2:**
  * **Input:** `nums = [-1]`
  * **Output:** `-1`

**Constraints:**
* 1 <= nums.length <= 100,000
* -10,000 <= nums[i] <= 10,000

### Core Logic: Kadane's Algorithm (The "Toxic Baggage" Analogy)
This is one of the most famous algorithms in computer science, known as **Kadane's Algorithm**. While it is often categorized as Dynamic Programming, its logic is highly Greedy. 

Imagine you are walking through the array, picking up numbers and keeping a "running sum" in your backpack. 
*   If the numbers you've collected so far add up to a **positive** number, they are an asset. Even if the next number is negative, your backpack can help absorb the blow. You should keep carrying it.
*   If the numbers you've collected so far add up to a **negative** number, your backpack has become "toxic baggage." If you add a negative backpack to *any* future number, it will just drag that new number down. 

**The Greedy Choice:** The moment your running sum drops below zero, you immediately throw the backpack away. You reset your running sum to `0` and start a brand new subarray with the very next element. At every single step, you check if your current running sum is the biggest you've ever seen, and record it.

### Approach: Kadane's Algorithm
Instead of a double loop checking every combination, we maintain a single running sum and make a greedy choice at each step to either keep extending the subarray or start fresh.
1. Initialize `max_sum` to the first number (to handle arrays with all negative numbers).
2. Initialize `cur_sum` to `0`.
3. Loop through every number in the array.
4. If `cur_sum` is negative, reset it to `0` (drop the toxic baggage).
5. Add the current number to `cur_sum`.
6. Update `max_sum` to be the maximum of itself and `cur_sum`.

* **Time Complexity:** O(n) — We iterate through the `nums` array exactly once, completely avoiding the nested loop of the brute force approach.
* **Space Complexity:** O(1) — We only track two variables (`cur_sum` and `max_sum`), using constant extra space regardless of how large the input array is.

```python
from typing import List

class Solution:
    def maxSubArray(self, nums: List[int]) -> int:
        # max_sum tracks the absolute best subarray we've ever seen.
        # Initialize to the first element to handle arrays that are entirely negative.
        max_sum = nums[0] 
        
        # cur_sum tracks our current backpack of numbers.
        cur_sum = 0 
        
        for n in nums:
            # THE GREEDY CHOICE: 
            # If our previous running sum is negative, it is a liability.
            # Throw it away before picking up the current number 'n'.
            if cur_sum < 0:
                cur_sum = 0
                
            # Add the current number to our backpack
            cur_sum += n
            
            # Record the highest score we've ever seen
            max_sum = max(max_sum, cur_sum)
            
        return max_sum
```

### Problem 002: Jump Game (LeetCode 55)

### Problem Definition and Constraints
You are given an integer array `nums` where each element `nums[i]` indicates your maximum jump length at that position.
Return `true` if you can reach the last index starting from index 0, or `false` otherwise.

**Examples:**
* **Example 1:**
  * **Input:** `nums = [1,2,0,1,0]`
  * **Output:** `true`
  * *Explanation:* First jump from index 0 to 1, then from index 1 to 3, and lastly from index 3 to 4.
* **Example 2:**
  * **Input:** `nums = [1,2,1,0,1]`
  * **Output:** `false`

**Constraints:**
* 1 <= nums.length <= 1000
* 0 <= nums[i] <= 1000

### Core Logic: The "Moving Goalpost" (Reverse Greedy)
If you try to solve this from left-to-right, it becomes a nightmare of branching paths. You jump to index 1, which gives you 3 new jump options. You have to explore all of them, leading to an exponential $O(2^n)$ brute force solution. 

The brilliant Greedy realization here is to **think in reverse**. 
Imagine the last index is the final "Goalpost." We don't care about how to get there from the start yet. We just ask: *"Who is the closest person standing in front of the Goalpost who can successfully jump to it?"*

If the person at index `3` can reach the Goalpost at index `4`, we don't need to aim for index `4` anymore. We can literally pick up the Goalpost and move it to index `3`. Why? Because we logically know that if *anybody* can just reach index `3`, they are guaranteed to reach the end. 

**The Greedy Choice:** 
We start at the end of the array and work backward. At every step, we check if the current index can reach the Goalpost. If it can, we shift the Goalpost to the current index. If we successfully shift the Goalpost all the way back to index `0`, it means the game is winnable!

### Approach: Greedy (Iterating Backwards)
1. Initialize `goal` to the very last index of the array (`len(nums) - 1`).
2. Loop through the array in reverse, starting from the second-to-last index down to `0`.
3. At each step `i`, calculate your maximum reach: `i + nums[i]`.
4. If your reach is greater than or equal to the `goal`, update `goal = i`.
5. Once the loop finishes, check if `goal == 0`. If it is, return `True`.

* **Time Complexity:** $O(n)$ — We iterate through the `nums` array exactly once in reverse.
* **Space Complexity:** $O(1)$ — We only track a single integer variable (`goal`), regardless of the input size.

```python
from typing import List

class Solution:
    def canJump(self, nums: List[int]) -> bool:
        # Place the initial goalpost at the very last index
        goal = len(nums) - 1
        
        # Iterate backwards starting from the second-to-last index down to 0
        for i in range(len(nums) - 2, -1, -1):
            # Your current index (i) + your jump power (nums[i]) is your max reach.
            # If your max reach can hit or pass the current goal, 
            # shift the goalpost to where you are standing right now.
            if i + nums[i] >= goal:
                goal = i
                
        # If the goalpost successfully made it all the way to the starting line, 
        # the end is reachable from the beginning.
        return True if goal == 0 else False
```

### Problem 003: Jump Game II (LeetCode 45)

### Problem Definition and Constraints
You are given an array of integers `nums`, where `nums[i]` represents the maximum length of a jump towards the right from index `i`. 
You are initially positioned at `nums[0]`.
Return the **minimum** number of jumps to reach the last position in the array. You may assume there is always a valid answer.

**Examples:**
* **Example 1:**
  * **Input:** `nums = [2,4,1,1,1,1]`
  * **Output:** `2`
  * *Explanation:* Jump from index 0 to index 1 (since you can jump 2 spaces), then jump from index 1 to the last index (since you can jump 4 spaces).
* **Example 2:**
  * **Input:** `nums = [2,1,2,1,0]`
  * **Output:** `2`

**Constraints:**
* 1 <= nums.length <= 1000
* 0 <= nums[i] <= 100

### Core Logic: The "Exploration Windows" (1D Breadth-First Search)
In Jump Game I, we just needed to know *if* we could reach the end, so we worked backward. 
Here, we need the *minimum number of jumps*, which completely changes the strategy. We must work forward and think in "levels" or "windows" of reachability. 

Imagine you are standing at the start. 
*   **Jump 0:** You are at index 0. Your window is just `[0]`. From here, you check how far you can possibly jump. Let's say you can reach up to index 2.
*   **Jump 1:** Your new window is `[1, 2]`. You don't know exactly which of these two spots is the best to jump from yet, so you evaluate *both* of them. You find the absolute farthest index you can reach from *anywhere* inside this window. Let's say index 1 can reach index 4, and index 2 can reach index 3. The absolute farthest is 4.
*   **Jump 2:** Your new window becomes `[3, 4]`. 

**The Greedy Choice:** 
You group the array into contiguous "windows" that represent how many jumps it took to get there. You use a Left (`l`) and Right (`r`) pointer to define the current window. You iterate through the current window to find the absolute maximum reach, and then shift your window forward to start the next jump. 

### Approach: Greedy Window / Two Pointers
1. Initialize `jumps = 0`.
2. Initialize `l = 0` and `r = 0`. This is our first window.
3. While our right pointer `r` hasn't reached the end of the array:
   * Scan every index inside the current window (`l` to `r`).
   * Find the `farthest` index we can reach from this entire window.
   * After checking the whole window, it's time to make a jump! Increment `jumps += 1`.
   * Update the window: the new Left boundary starts immediately after the old Right boundary (`l = r + 1`). The new Right boundary becomes the `farthest` point we just calculated.
4. Return `jumps`.

* **Time Complexity:** $O(n)$ — Even though there is a `while` loop and a `for` loop, the `l` and `r` pointers simply slide across the array from left to right. We visit each index exactly once.
* **Space Complexity:** $O(1)$ — We only maintain a few integer variables (`l`, `r`, `farthest`, `jumps`).

```python
from typing import List

class Solution:
    def jump(self, nums: List[int]) -> int:
        jumps = 0
        l = 0  # Left boundary of our current jump window
        r = 0  # Right boundary of our current jump window
        
        # We keep going until our window's right boundary reaches the end of the array
        while r < len(nums) - 1:
            farthest = 0
            
            # Scan every index inside our current jump window
            for i in range(l, r + 1):
                # Calculate the absolute farthest we can reach from this specific index
                farthest = max(farthest, i + nums[i])
                
            # Now that we've scanned the current window, we transition to the next window.
            # This represents officially taking 1 jump.
            l = r + 1
            r = farthest
            jumps += 1
            
        return jumps
```

### Problem 004: Gas Station (LeetCode 134)

### Problem Definition and Constraints
There are `n` gas stations along a circular route. You are given two integer arrays `gas` and `cost`:
*   `gas[i]` is the amount of gas you get at the `i`th station.
*   `cost[i]` is the gas needed to travel from the `i`th station to the `(i + 1)`th station.

You have a car with an unlimited gas tank, starting empty. 
Return the starting gas station's index such that you can travel around the circuit once in the clockwise direction. If it's impossible, return `-1`. (It is guaranteed that if a solution exists, it is unique).

**Examples:**
* **Example 1:**
  * **Input:** `gas = [1,2,3,4], cost = [2,2,4,1]`
  * **Output:** `3`
  * *Explanation:* Start at station 3, fill up with 4 gas, and you will have enough to make it all the way around the circuit.
* **Example 2:**
  * **Input:** `gas = [1,2,3], cost = [2,3,2]`
  * **Output:** `-1`
  * *Explanation:* Total gas available is 6, but total cost is 7. You can never complete the circuit.

**Constraints:**
* 1 <= gas.length == cost.length <= 100,000
* 0 <= gas[i], cost[i] <= 1,000

### Core Logic: The "Global Guarantee" and the "Doomed Segment"
This problem requires two massive logical leaps that completely eliminate the need for a brute-force $O(n^2)$ simulation.

**1. The Global Guarantee:** 
If the total amount of gas on the entire map is greater than or equal to the total cost to travel the map, **a valid starting point is mathematically guaranteed to exist.** Therefore, we can do a quick global check at the very beginning. If `sum(gas) < sum(cost)`, we instantly return `-1`. If it passes, we know an answer exists.

**2. The Doomed Segment (The Greedy Choice):**
Let's say you start at Station A. You make it through Station B and Station C, but when you try to drive to Station D, your tank drops below zero. 
The brute force method would say: *"A failed. Let's try starting at B instead."*

The Greedy logic realizes: **If A couldn't reach D, then B and C can't reach D either!** 
Why? Because when you started at A, you arrived at B with a *surplus* of gas in your tank. Even with that extra surplus, you still couldn't reach D. If you started fresh at B with an empty tank, your chances of reaching D are even worse!

Therefore, the exact moment your tank hits empty, the entire segment from your starting point up to your current point is doomed. You instantly reset your starting point to the very next station (`i + 1`).

### Approach: Greedy (Single Pass)
1. Check the global guarantee: if `sum(gas) < sum(cost)`, return `-1`.
2. Initialize `tank = 0` (your current gas) and `start_index = 0`.
3. Loop through the stations. At each station, add the net gas (`gas[i] - cost[i]`) to your `tank`.
4. If your `tank` drops below zero, it means the current `start_index` (and everything after it up to `i`) is doomed. 
5. Reset your `tank` to `0` and set `start_index = i + 1`.
6. Return `start_index`.

* **Time Complexity:** $O(n)$ — We do one pass to get the sums, and one pass to find the starting index.
* **Space Complexity:** $O(1)$ — We only use integer variables (`tank`, `start_index`).

```python
from typing import List

class Solution:
    def canCompleteCircuit(self, gas: List[int], cost: List[int]) -> int:
        # 1. The Global Guarantee Check
        # If the total cost is strictly greater than the total gas available,
        # it is mathematically impossible to complete the loop.
        if sum(gas) < sum(cost):
            return -1
            
        tank = 0
        start_index = 0
        
        # 2. Finding the exact starting point
        for i in range(len(gas)):
            # Add the net change in gas for this specific station
            tank += gas[i] - cost[i]
            
            # The Greedy Choice: The Doomed Segment
            # If our tank drops below zero, we cannot reach the next station.
            if tank < 0:
                # Reset our tank to empty for our next attempt
                tank = 0
                # Since everything from our old start_index up to 'i' failed,
                # the earliest possible valid starting point must be the next station.
                start_index = i + 1
                
        # Because our Global Guarantee passed at the top of the function,
        # we are absolutely certain that this start_index is the correct and only answer.
        return start_index
```

### Problem 005: Hand of Straights (LeetCode 846)

### Problem Definition and Constraints
You are given an integer array `hand` where `hand[i]` is the value written on the `i`th card and an integer `groupSize`.
You want to rearrange the cards into groups so that each group is of size `groupSize`, and card values are consecutively increasing by 1.
Return `true` if it's possible to rearrange the cards in this way, otherwise, return `false`.

**Examples:**
* **Example 1:**
  * **Input:** `hand = [1,2,4,2,3,5,3,4], groupSize = 4`
  * **Output:** `true`
  * **Explanation:** The cards can be rearranged as `[1,2,3,4]` and `[2,3,4,5]`.
* **Example 2:**
  * **Input:** `hand = [1,2,3,3,4,5,6,7], groupSize = 4`
  * **Output:** `false`
  * **Explanation:** The closest we can get is `[1,2,3,4]` and `[3,5,6,7]`, but the cards in the second group are not consecutive.

* Constraints:
  * 1 <= hand.length <= 10000
  * 0 <= hand[i] <= 1000
  * 1 <= groupSize <= hand.length

### Core Logic: The "Smallest Anchor"
This problem requires a very specific Greedy realization: **You must always build your next group starting with the absolute smallest available card.** 

Why? Imagine your smallest available card is `2`. That `2` cannot possibly be the second, third, or fourth card in a sequence, because that would require a `1`, `0`, or `-1` to exist in your hand (and if they did exist, `2` wouldn't be the smallest card!). Therefore, `2` *must* be the anchor (the starting card) of its own group. 

Once you lock in `2` as the anchor, the Greedy choice forces your hand: you are mathematically required to find a `3`, a `4`, and so on, until the `groupSize` is met. If you are missing any of those required cards, the whole system collapses, and you instantly return `false`.

### Approach: Hash Map + Min-Heap
1. **The Math Check:** If the total number of cards isn't perfectly divisible by `groupSize`, it's mathematically impossible to group them. Return `False` instantly.
2. **The Inventory:** Count the frequency of every card using a Hash Map (`Counter`). 
3. **The Anchor Finder:** We need a fast way to repeatedly find the smallest available card. We take all the unique cards from our map and throw them into a Min-Heap.
4. **The Greedy Assembly:** Look at the top of the Min-Heap to get the smallest card. Run a loop `groupSize` times to check if the next consecutive cards exist in our Hash Map. 
   * If a required card is missing, return `False`.
   * If it exists, decrement its count. 
   * If a card's count hits `0`, it is completely exhausted. We pop it from the Min-Heap so we don't try to use it as an anchor later.

* **Time Complexity:** $O(n \log n)$ — Counting the cards takes $O(n)$. Building the heap from unique cards takes $O(n)$. Popping from the heap takes $O(\log n)$, and we might pop up to $n$ times in the worst case (where every card is unique).
* **Space Complexity:** $O(n)$ — Our Hash Map and Min-Heap store the unique card values, which in the worst case will scale linearly with the size of the input array.

```python
import heapq
from collections import Counter
from typing import List

class Solution:
    def isNStraightHand(self, hand: List[int], groupSize: int) -> bool:
        # 1. The Math Check
        if len(hand) % groupSize != 0:
            return False
            
        # 2. The Inventory
        # Example: hand = [1,2,2,3], count = {1: 1, 2: 2, 3: 1}
        count = Counter(hand)
        
        # 3. The Anchor Finder
        # Extract unique keys and turn them into a Min-Heap
        min_heap = list(count.keys())
        heapq.heapify(min_heap)
        
        # 4. The Greedy Assembly
        while min_heap:
            # Look at (but don't pop) the smallest available card
            first = min_heap[0]
            
            # Attempt to build a consecutive group starting from 'first'
            for i in range(first, first + groupSize):
                # If a required consecutive card is completely missing, we fail!
                if count[i] == 0:
                    return False
                    
                # We used one of these cards, so reduce its inventory count
                count[i] -= 1
                
                # If we completely run out of a specific card...
                if count[i] == 0:
                    # THE SAFETY CHECK: 
                    # The card that just ran out MUST be the one currently at the top of the heap.
                    # If it's not, it means a card in the middle of our sequence ran out before 
                    # the anchor did, creating an unfillable hole for future groups!
                    if i != min_heap[0]:
                        return False
                        
                    # Safely remove the exhausted card from the heap
                    heapq.heappop(min_heap)
                    
        return True


        """
        =========================================================
        SIMULATION NOTES: hand = [1,2,4,2,3,5,3,4], groupSize = 4
        =========================================================
        Math check: 8 % 4 == 0. (Valid)
        count = {1:1, 2:2, 3:2, 4:2, 5:1}
        min_heap = [1, 2, 3, 4, 5]
        
        -- GROUP 1 --
        Smallest is 1. We need [1, 2, 3, 4].
        - check 1: count[1] becomes 0. Pop 1 from heap. 
        - check 2: count[2] becomes 1.
        - check 3: count[3] becomes 1.
        - check 4: count[4] becomes 1.
        min_heap is now [2, 3, 4, 5]
        
        -- GROUP 2 --
        Smallest is 2. We need [2, 3, 4, 5].
        - check 2: count[2] becomes 0. Pop 2 from heap.
        - check 3: count[3] becomes 0. Pop 3 from heap.
        - check 4: count[4] becomes 0. Pop 4 from heap.
        - check 5: count[5] becomes 0. Pop 5 from heap.
        min_heap is now []
        
        Heap is empty. All cards successfully grouped. Return True!
        """
```

### Problem 006: Merge Triplets to Form Target Triplet (LeetCode 1899)

### Problem Definition and Constraints
You are given a 2D array of integers `triplets`, where `triplets[i] = [a, b, c]`. You are also given an array `target = [x, y, z]` which is the exact triplet we want to obtain.
To obtain `target`, you may choose any two triplets and merge them. The new merged triplet takes the **maximum** value at each position:
* `[max(a1, a2), max(b1, b2), max(c1, c2)]`

Return `true` if it is possible to obtain `target` as an element, or `false` otherwise.

**Examples:**
* **Example 1:**
  * **Input:** `triplets = [[1,2,3],[7,1,1]], target = [7,2,3]`
  * **Output:** `true`
  * **Explanation:** Merge them: `[max(1,7), max(2,1), max(3,1)]` = `[7, 2, 3]`.
* **Example 2:**
  * **Input:** `triplets = [[2,5,6],[1,4,4],[5,7,5]], target = [5,4,6]`
  * **Output:** `false`
  * **Explanation:** No combination will ever result in exactly `[5, 4, 6]`.

* Constraints:
  * 1 <= triplets.length <= 1000
  * 1 <= ai, bi, ci, x, y, z <= 100

### Core Logic: The "Toxic Triplet" Filter
This problem is a pure Greedy filtering puzzle. The only operation available is `max()`. 
If you ever merge a triplet that has a value *strictly greater* than the target in any of the three slots, that slot will permanently exceed the target. Because you only have `max()`, there is no mathematical way to bring a number back down. 

Therefore, any triplet containing a value larger than the target is a **Toxic Triplet**. We must ruthlessly ignore them.

Once you filter out all the toxic triplets, every remaining triplet is **Safe** (all its values are $\le$ the target). Because they are safe, you could theoretically mash *every single safe triplet* together into one giant super-triplet without ever overshooting the target! 

The Greedy Choice: You just need to check if the exact target values (`x`, `y`, and `z`) exist *somewhere* among the safe triplets. If they do, merging them will eventually construct the exact target.

### Approach: Greedy Filtering
1. Initialize a Hash Set (or three boolean flags) called `good` to track which parts of the target we have successfully found.
2. Loop through every triplet.
3. **The Filter:** If `t[0] > target[0]` or `t[1] > target[1]` or `t[2] > target[2]`, `continue` to skip it. It is toxic.
4. **The Collector:** If the triplet is safe, check if `t[0] == target[0]`. If it does, add index `0` to our `good` set. Do the same for indices `1` and `2`.
5. If the length of our `good` set reaches 3, it means we found a safe path to build all three target components. Return `True`.

* **Time Complexity:** $O(n)$ — We iterate through the list of triplets exactly once.
* **Space Complexity:** $O(1)$ — Our `good` set will only ever hold a maximum of 3 integers (0, 1, and 2), regardless of how massive the input array is.

```python
from typing import List

class Solution:
    def mergeTriplets(self, triplets: List[List[int]], target: List[int]) -> bool:
        # A set to keep track of which target positions (0, 1, or 2) we have found
        good = set()
        
        for t in triplets:
            # 1. THE TOXIC FILTER
            # If ANY value in this triplet is strictly greater than the target,
            # using this triplet would permanently ruin our merged result.
            if t[0] > target[0] or t[1] > target[1] or t[2] > target[2]:
                continue
                
            # 2. THE COLLECTOR
            # We know this triplet is completely safe to merge.
            # Does it contain the exact target value for the first position?
            if t[0] == target[0]:
                good.add(0)
            
            # Does it contain the exact target value for the second position?
            if t[1] == target[1]:
                good.add(1)
                
            # Does it contain the exact target value for the third position?
            if t[2] == target[2]:
                good.add(2)
                
            # Early Exit: If we found all 3 pieces, we can stop searching.
            if len(good) == 3:
                return True
                
        # If we checked every safe triplet and couldn't piece together all 3 parts,
        # it is impossible to form the target.
        return len(good) == 3


        """
        =========================================================
        SIMULATION NOTES: target = [5, 4, 6]
        triplets = [[2, 5, 6], [1, 4, 4], [5, 7, 5]]
        =========================================================
        
        Triplet 1: [2, 5, 6]
        - Check: 5 > 4 (target[1]). 
        - TOXIC! Skip it. (Even though it has the '6' we need, it ruins the '4').
        
        Triplet 2: [1, 4, 4]
        - Check: 1 <= 5, 4 <= 4, 4 <= 6. 
        - SAFE! 
        - Does 1 == 5? No.
        - Does 4 == 4? Yes! Add 1 to `good`. (good = {1})
        - Does 4 == 6? No.
        
        Triplet 3: [5, 7, 5]
        - Check: 7 > 4 (target[1]). 
        - TOXIC! Skip it. 
        
        End of loop. good = {1}. Length is 1, not 3.
        Return False! 
        """
```

### Problem 007: Partition Labels (LeetCode 763)

### Problem Definition and Constraints
You are given a string `s` consisting of lowercase English letters.
We want to split the string into as many substrings as possible, while ensuring that each letter appears in at most one substring.
Return a list of integers representing the size of these substrings in the order they appear in the string.

**Examples:**
* **Example 1:**
  * **Input:** `s = "xyxxyzbzbbisl"`
  * **Output:** `[5, 5, 1, 1, 1]`
  * **Explanation:** The string splits into `["xyxxy", "zbzbb", "i", "s", "l"]`. Notice that all 'x's are in the first piece, all 'z's are in the second piece, etc.
* **Example 2:**
  * **Input:** `s = "abcabc"`
  * **Output:** `[6]`
  * **Explanation:** You cannot split it at all, because all characters are scattered across the entire string.

* Constraints:
  * 1 <= s.length <= 100
  * `s` consists of only lowercase English letters.

### Core Logic: The "Rubber Band" (Greedy Expansion)
To ensure that all copies of a letter stay in the exact same partition, the partition **must** stretch to include the absolute last occurrence of that letter. 

Imagine putting a rubber band around the letters. 
If the first letter is `'x'`, and the last `'x'` in the string is at index 4, your rubber band *must* stretch to at least index 4. 
But as you walk toward index 4, you might encounter a `'y'`. If the absolute last `'y'` is at index 7, your rubber band is forced to stretch even further to index 7 to capture it! 

The Greedy Choice: You keep stretching your `end` boundary to the maximum last occurrence of any character you step on. The exact moment your current steps (`i`) catch up to the stretched boundary (`end`), it means you have successfully encompassed all copies of every character inside the rubber band. You snap it shut, save the size, and start a brand new rubber band for the next letter.

### Approach: Hash Map + Two Pointers
1. **The Inventory (Last Seen Map):** Do a quick first pass through the string to record the absolute last index of every character. We store this in a Hash Map (e.g., `lastIndex['x'] = 4`).
2. **The Expansion:** Iterate through the string again. Track the `size` of your current partition, and an `end` pointer.
3. For every character you step on, look up its last index in the map, and update `end = max(end, lastIndex[c])`.
4. **The Snap:** If your current index `i == end`, the partition is perfect. Append the `size` to your result array, reset `size = 0`, and keep walking.

* **Time Complexity:** $O(n)$ — We iterate through the string exactly twice. Once to build the hash map, and once to draw the partitions.
* **Space Complexity:** $O(1)$ — Our Hash Map will only ever store up to 26 key-value pairs (for the 26 lowercase English letters), which uses constant space regardless of how massive the string gets.

```python
from typing import List

class Solution:
    def partitionLabels(self, s: str) -> List[int]:
        # 1. THE INVENTORY
        # Find the absolute last occurrence of every character.
        # Enumerate gives us both the index (i) and the character (c).
        last_idx = {}
        for i, c in enumerate(s):
            # This naturally overwrites earlier occurrences, 
            # leaving us with only the absolute last index.
            last_idx[c] = i
            
        res = []
        size = 0
        end = 0
        
        # 2. THE GREEDY EXPANSION
        for i, c in enumerate(s):
            size += 1
            
            # Stretch the rubber band if this character's last occurrence is further out
            end = max(end, last_idx[c])
            
            # 3. THE SNAP
            # If our current walking position perfectly matches the furthest boundary,
            # we have safely captured all necessary characters!
            if i == end:
                res.append(size)  # Save the length of this partition
                size = 0          # Reset the length for the next partition
                
        return res


        """
        =========================================================
        SIMULATION NOTES: s = "ababcbacadefegdehijhklij"
        (Let's trace just the first partition)
        =========================================================
        last_idx = {'a': 8, 'b': 5, 'c': 7, ...}
        
        i = 0, c = 'a':
        - size = 1
        - end = max(0, last_idx['a']) -> max(0, 8) = 8
        - Boundary stretched to index 8!
        
        i = 1, c = 'b':
        - size = 2
        - end = max(8, last_idx['b']) -> max(8, 5) = 8
        - Boundary stays at 8.
        
        ... walks forward ...
        
        i = 8, c = 'a':
        - size = 9
        - end = max(8, last_idx['a']) = 8
        - i (8) == end (8)! SNAP!
        - Append size 9. Reset size to 0.
        
        The first partition is exactly 9 characters long!
        """
```

### Problem 008: Valid Parenthesis String (LeetCode 678)

### Problem Definition and Constraints
You are given a string `s` containing only three types of characters: `'('`, `')'` and `'*'`.
Return `true` if `s` is valid, otherwise return `false`.
A string is valid if:
1. Every `'('` must have a corresponding `')'`.
2. Every `')'` must have a corresponding `'('`.
3. Left parenthesis `'('` must go before the corresponding right parenthesis `')'`.
4. A `'*'` could be treated as a single `')'`, a single `'('`, or an empty string `""`.

**Examples:**
* **Example 1:**
  * **Input:** `s = "((**)"`
  * **Output:** `true`
  * **Explanation:** Treat the first `'*'` as `""` and the second `'*'` as `')'`.
* **Example 2:**
  * **Input:** `s = "(((*)"`
  * **Output:** `false`
  * **Explanation:** There are too many open parentheses, even if the star is used as a closing one.

* Constraints:
  * 1 <= s.length <= 100

### Core Logic: The "Branching Reality" (Min/Max Range)
The hints suggest using two stacks ($O(n)$ space), but since we are in the Greedy section, there is a vastly superior $O(1)$ space mathematical approach.

In a standard parenthesis problem (without stars), you just keep a running counter. `(` adds 1, `)` subtracts 1. If it ever drops below 0, you fail. If it ends at exactly 0, you win.
Because the `'*'` can be three different things, our exact counter shatters into a **range of possibilities**. 

Instead of one counter, we track a `leftMin` and a `leftMax`:
*   `leftMax` tracks the absolute maximum number of open `(` we *could* have (assuming we turn every single `*` into a `(`). 
*   `leftMin` tracks the absolute minimum number of open `(` we *must* resolve (assuming we turn every `*` into a `)`).

**The Greedy Rules:**
1.  **The Hard Fail:** If `leftMax < 0` at any point, it means even if we used every single magic star as a `(`, we still don't have enough to match the `)` we just encountered. The string is hopelessly broken. Return `False`.
2.  **The Floor Reset:** What if `leftMin < 0`? This just means we were too greedy trying to turn stars into `)`. A negative `leftMin` is physically impossible (you can't have negative open parentheses). We simply reset `leftMin = 0`, logically forcing some of those stars to act as empty strings `""` instead.
3.  **The Final Check:** At the end of the string, if `leftMin == 0`, it means there is a valid reality where all parentheses perfectly cancel out.

### Approach: Greedy Min/Max
1. Initialize `leftMin = 0` and `leftMax = 0`.
2. Loop through the string.
   * If `(`, increment both `leftMin` and `leftMax`.
   * If `)`, decrement both `leftMin` and `leftMax`.
   * If `*`, decrement `leftMin` (treating it as `)`) and increment `leftMax` (treating it as `(`).
3. If `leftMax < 0`, return `False`.
4. If `leftMin < 0`, reset `leftMin = 0`.
5. Return `leftMin == 0`.

* **Time Complexity:** $O(n)$ — We scan the string exactly once.
* **Space Complexity:** $O(1)$ — We only use two integer variables, completely bypassing the $O(n)$ Stack memory suggested by the hints.

```python
class Solution:
    def checkValidString(self, s: str) -> bool:
        # leftMin: Minimum possible open left parentheses
        # leftMax: Maximum possible open left parentheses
        leftMin, leftMax = 0, 0
        
        for c in s:
            if c == "(":
                # Must increase both boundaries
                leftMin += 1
                leftMax += 1
            elif c == ")":
                # Must decrease both boundaries
                leftMin -= 1
                leftMax -= 1
            else: # c == "*"
                # leftMin decreases (pretend '*' is a right paren)
                # leftMax increases (pretend '*' is a left paren)
                leftMin -= 1
                leftMax += 1
                
            # THE HARD FAIL: 
            # Even if every '*' was a '(', we STILL don't have enough to match the ')'
            if leftMax < 0:
                return False
                
            # THE FLOOR RESET:
            # We can't have negative open parentheses. 
            # This just means we shouldn't have turned a '*' into a ')'.
            # We conceptually change it to a "" instead, resetting the floor to 0.
            if leftMin < 0:
                leftMin = 0
                
        # If the minimum possible open parentheses is exactly 0, 
        # it means there is a valid path that perfectly resolves the string.
        return leftMin == 0


        """
        =========================================================
        SIMULATION NOTES: s = "(*))"
        =========================================================
        Start: leftMin = 0, leftMax = 0
        
        1. c = '('
           leftMin = 1, leftMax = 1
           
        2. c = '*'
           leftMin = 0 (1 - 1), leftMax = 2 (1 + 1)
           (Reality branch: We either have 0 open, 1 open (if ""), or 2 open)
           
        3. c = ')'
           leftMin = -1, leftMax = 1
           leftMin dropped below 0! Reset leftMin = 0.
           (Reality branch: We can't have -1 open, so we force the '*' to be "")
           
        4. c = ')'
           leftMin = -1, leftMax = 0
           leftMin dropped below 0! Reset leftMin = 0.
           
        End of loop.
        leftMin == 0. Return True!
        """
```

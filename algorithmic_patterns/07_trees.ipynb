### Problem 001: Invert Binary Tree (LeetCode 226)

### Problem Definition and Constraints
Given the `root` of a binary tree, invert the binary tree and return its root. Inverting a binary tree means swapping the left and right children of all nodes in the tree.

* Constraints:
  * 0 <= The number of nodes in the tree <= 100
  * -100 <= Node.val <= 100

### Examples
* **Example 1:**
  * Input: `root = [4,2,7,1,3,6,9]`
  * Output: `[4,7,2,9,6,3,1]`
* **Example 2:**
  * Input: `root = [2,1,3]`
  * Output: `[2,3,1]`

### Optimized Approach (Recursive DFS)
Because a binary tree is a recursive data structure, the most elegant and optimal solution is to use Depth First Search (DFS) recursively. 
Our base case checks if the current node is `None`; if so, we return `None`. 
For a valid node, we temporarily store its `left` child. We then overwrite its `left` pointer with its `right` child, and its `right` pointer with the temporarily stored `left` child. Once the current node's children are swapped, we recursively call the `invertTree` function on both the new left child and the new right child to propagate the inversion down to the leaves. Finally, we return the `root`.
* Time Complexity: $O(n)$ — We must visit every single node in the tree exactly once to swap its children.
* Space Complexity: $O(n)$ — In the worst-case scenario (a completely unbalanced tree that looks like a linked list), the recursive call stack will grow to size $n$. In a perfectly balanced tree, the space complexity is bounded by the height of the tree, $O(\log n)$.

```python
from typing import Optional

# Definition for a binary tree node.
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

class Solution:
    
    # --- OPTIMIZED APPROACH (RECURSIVE) ---
    def invertTree(self, root: Optional[TreeNode]) -> Optional[TreeNode]:
        # BASE CASE: If the node doesn't exist, just stop and return None
        if not root:
            return None
            
        # 1. SWAP: Temporarily store the left child so we don't lose it
        temp = root.left
        
        # Overwrite the left child with the right child
        root.left = root.right
        
        # Overwrite the right child with the stored left child
        root.right = temp
        
        # 2. RECURSE: Tell the newly swapped children to go invert their own sub-trees
        self.invertTree(root.left)
        self.invertTree(root.right)
        
        # 3. RETURN: Pass the fully inverted tree back up the chain
        return root
```

### Problem 002: Maximum Depth of Binary Tree (LeetCode 104)

### Problem Definition and Constraints
Given the `root` of a binary tree, return its maximum depth.
The maximum depth is the number of nodes along the longest path from the root node down to the farthest leaf node.

* Constraints:
  * 0 <= The number of nodes in the tree <= 10^4
  * -100 <= Node.val <= 100

### Examples
* **Example 1:**
  * Input: `root = [3,9,20,null,null,15,7]`
  * Output: `3`
* **Example 2:**
  * Input: `root = [1,null,2]`
  * Output: `2`

### Optimized Approach (Recursive DFS)
Because depth is naturally calculated from the bottom up, we use a post-order Depth First Search (DFS). 
Our base case checks if the current node is `None`; if it is, we have fallen off the bottom of the tree, so we return `0` as the depth. 
For a valid node, we recursively call the function on both its `left` child and its `right` child, pausing the current node on the call stack. Once both children return their maximum depths, the current node takes the `max()` of those two values, adds `1` to account for its own existence in the path, and returns that final integer up to its parent.
* Time Complexity: $O(n)$ — We must visit every single node in the tree exactly once to calculate its depth.
* Space Complexity: $O(n)$ — In the worst-case scenario (a completely unbalanced tree), the recursive call stack will grow to size $n$. In a perfectly balanced tree, the space complexity is bounded by the height of the tree, $O(\log n)$.

```python
from typing import Optional

# Definition for a binary tree node.
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

class Solution:
    
    # --- OPTIMIZED APPROACH (RECURSIVE DFS) ---
    def maxDepth(self, root: Optional[TreeNode]) -> int:
        # BASE CASE: If we hit empty space, the depth is 0
        if not root:
            return 0
            
        # 1. DELEGATE: Ask the left and right sub-trees for their depths
        # The current node is paused on the Call Stack while these run
        left_depth = self.maxDepth(root.left)
        right_depth = self.maxDepth(root.right)
        
        # 2. CALCULATE: Take the biggest depth of the two branches, 
        # add 1 for the current node we are standing on, and return it upwards.
        return 1 + max(left_depth, right_depth)
```

### Problem 003: Diameter of Binary Tree (LeetCode 543)

### Problem Definition and Constraints
Given the `root` of a binary tree, return the length of the diameter of the tree.
The diameter is the length of the longest path between any two nodes. This path may or may not pass through the root. The length of a path is defined as the number of edges between them.

* Constraints:
  * 1 <= number of nodes in the tree <= 100
  * -100 <= Node.val <= 100

### Examples
* **Example 1:**
  * Input: `root = [1,2,3,4,5]`
  * Output: `3`
  * Explanation: 3 is the length of the path [4,2,1,3] or [5,2,1,3].
* **Example 2:**
  * Input: `root = [1,2,3]`
  * Output: `2`

### Optimized Approach (Recursive DFS with Global State)
We use a bottom-up Depth First Search (DFS). The core logic is identical to finding the maximum depth of a tree. However, we introduce a class-level variable, `self.max_diameter`, to keep track of the largest diameter found at any point during the traversal.
At each node, we recursively calculate the depth of the left subtree and the right subtree. The diameter passing *through* the current node is simply `left_depth + right_depth`. We update `self.max_diameter` if this sum is larger than our current record. Finally, the node returns its own depth (`1 + max(left_depth, right_depth)`) to its parent so the parent can calculate its own metrics.
* Time Complexity: O(n) — We visit every single node exactly once.
* Space Complexity: O(n) — The recursive call stack will grow to size n in the worst-case (unbalanced tree) or O(log n) in the best-case (balanced tree).

```python
from typing import Optional

# Definition for a binary tree node.
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

class Solution:
    def diameterOfBinaryTree(self, root: Optional[TreeNode]) -> int:
        # This is our global "whiteboard" to track the biggest diameter we've seen
        self.max_diameter = 0
        
        # We define our recursive DFS helper function inside the main function
        def dfs(node):
            # BASE CASE: An empty space has a depth of 0
            if not node:
                return 0
                
            # DELEGATE: Get the max depths of the left and right branches
            left_depth = dfs(node.left)
            right_depth = dfs(node.right)
            
            # SECRET JOB: The diameter passing through THIS specific node is left + right.
            # Update our global whiteboard if this is a new record.
            self.max_diameter = max(self.max_diameter, left_depth + right_depth)
            
            # PUBLIC JOB: Return the depth of this node back up to its parent
            return 1 + max(left_depth, right_depth)
            
        # Kick off the recursion starting at the root
        dfs(root)
        
        # Once the entire tree is processed, our whiteboard holds the final answer
        return self.max_diameter
```

### Problem 004: Balanced Binary Tree (LeetCode 110)

### Problem Definition and Constraints
Given a binary tree, return `true` if it is height-balanced and `false` otherwise.
A height-balanced binary tree is defined as a binary tree in which the left and right subtrees of every node differ in height by no more than 1.

* Constraints:
  * The number of nodes in the tree is in the range [0, 1000].
  * -1000 <= Node.val <= 1000

### Examples
* **Example 1:**
  * Input: `root = [3,9,20,null,null,15,7]`
  * Output: `true`
* **Example 2:**
  * Input: `root = [1,2,2,3,3,null,null,4,4]`
  * Output: `false`

### Optimized Approach (Bottom-Up DFS)
We use a bottom-up Depth First Search (DFS) that returns two values from every node: a boolean indicating if the subtree is balanced, and an integer representing the maximum height of the subtree `[is_balanced, height]`.
Our base case for a `None` node returns `[True, 0]`. 
For a valid node, we recursively get the status of the left and right subtrees. The current node is balanced if and only if: the left subtree is balanced, the right subtree is balanced, and the absolute difference between their heights is $\le 1$. We return this new boolean alongside the current node's calculated height (`1 + max(left_height, right_height)`) back up the call stack. 
* Time Complexity: $O(n)$ — We visit every node exactly once, determining balance and height in a single pass.
* Space Complexity: $O(n)$ — The recursive call stack will grow to size $n$ in the worst-case (unbalanced tree) or $O(\log n)$ in the best-case (balanced tree).

```python
from typing import Optional

# Definition for a binary tree node.
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

class Solution:
    def isBalanced(self, root: Optional[TreeNode]) -> bool:
        
        # Helper function that returns [is_balanced, height]
        def dfs(node):
            # BASE CASE: An empty space is perfectly balanced with a height of 0
            if not node:
                return [True, 0]
                
            # DELEGATE: Get the reports from the left and right branches
            left_report = dfs(node.left)
            right_report = dfs(node.right)
            
            # CHECK BALANCE: 
            # 1. Did the left branch say it was balanced?
            # 2. Did the right branch say it was balanced?
            # 3. Is the difference between their heights <= 1?
            is_balanced = (
                left_report[0] and 
                right_report[0] and 
                abs(left_report[1] - right_report[1]) <= 1
            )
            
            # CALCULATE HEIGHT: Same logic as the Max Depth problem
            current_height = 1 + max(left_report[1], right_report[1])
            
            # RETURN: Pass the updated report back up to the parent
            return [is_balanced, current_height]
            
        # Kick off the recursion and extract just the boolean (index 0) from the final root report
        return dfs(root)[0]
```

### Problem 005: Same Binary Tree (LeetCode 100)

### Problem Definition and Constraints
Given the roots of two binary trees `p` and `q`, return `true` if the trees are equivalent, otherwise return `false`.
Two binary trees are considered equivalent if they share the exact same structure and the nodes have the exact same values.

* Constraints:
  * 0 <= The number of nodes in both trees <= 100
  * -100 <= Node.val <= 100

### Examples
* **Example 1:**
  * Input: `p = [1,2,3], q = [1,2,3]`
  * Output: `true`
* **Example 2:**
  * Input: `p = [4,7], q = [4,null,7]`
  * Output: `false`
  * Explanation: The structure is different. One has a left child, the other has a right child.

### Optimized Approach (Simultaneous Recursive DFS)
We use a simultaneous Depth First Search (DFS) to traverse both trees at the exact same time. 
At every step of the recursion, we check our base cases. If both nodes are `None`, the paths match (`True`). If only one node is `None`, or if the values of the two nodes differ, the trees are not identical (`False`). 
If the current nodes match, we recursively call the function on both the left children (`p.left, q.left`) and the right children (`p.right, q.right`). The current sub-tree is only identical if both the left recursive call `AND` the right recursive call return `True`.
* Time Complexity: $O(n)$ — Where $n$ is the number of nodes in the smaller tree. We visit each matching node exactly once. The moment a mismatch is found, the recursion stops.
* Space Complexity: $O(n)$ — The recursive call stack will grow to size $n$ in the worst-case (unbalanced tree) or $O(\log n)$ in the best-case (balanced tree).

```python
from typing import Optional

# Definition for a binary tree node.
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

class Solution:
    def isSameTree(self, p: Optional[TreeNode], q: Optional[TreeNode]) -> bool:
        
        # BASE CASE 1: Both nodes are empty (We reached the bottom together perfectly)
        if not p and not q:
            return True
            
        # BASE CASE 2: One node is empty, but the other has a real value (Structure mismatch)
        if not p or not q:
            return False
            
        # BASE CASE 3: Both are real nodes, but their values don't match (Value mismatch)
        if p.val != q.val:
            return False
            
        # DELEGATE: The current nodes match! Now go check both left wings and both right wings.
        # We use 'and' because the whole tree is only the same if BOTH halves are the same.
        return self.isSameTree(p.left, q.left) and self.isSameTree(p.right, q.right)
```

### Problem 006: Subtree of Another Tree (LeetCode 572)

### Problem Definition and Constraints
Given the roots of two binary trees `root` and `subRoot`, return `true` if there is a subtree of `root` with the same structure and node values of `subRoot` and `false` otherwise.
A subtree of a binary tree `tree` is a tree that consists of a node in `tree` and all of this node's descendants. The tree `tree` could also be considered as a subtree of itself.

* Constraints:
  * 1 <= The number of nodes in both trees <= 1000.
  * -10^4 <= root.val, subRoot.val <= 10^4

### Examples
* **Example 1:**
  * Input: `root = [3,4,5,1,2]`, `subRoot = [4,1,2]`
  * Output: `true`
  * Explanation: The subtree rooted at node 4 perfectly matches subRoot.
* **Example 2:**
  * Input: `root = [3,4,5,1,2,null,null,null,null,0]`, `subRoot = [4,1,2]`
  * Output: `false`
  * Explanation: Even though a portion matches, node 2 in `root` has a left child `0`, making the structure different from `subRoot`.

### Optimized Approach (Recursive DFS with Helper Function)
This problem requires two nested recursive Depth First Search (DFS) traversals. 
First, we need a helper function `isSameTree(p, q)` which checks if two trees are structurally and strictly identical (the exact same function from Problem 005).
Next, our main function traverses every node in the main `root` tree. At each node, it uses `isSameTree` to check if the current node is the starting point of a perfect match with `subRoot`. If it is not a match, the main function recursively calls itself to check the `root.left` and `root.right` branches. 
* Time Complexity: $O(m \times n)$ — Where $m$ is the number of nodes in `root` and $n$ is the number of nodes in `subRoot`. In the worst-case scenario, for every single node in `root`, we might have to traverse all of `subRoot` to check for a match.
* Space Complexity: $O(m + n)$ — Because we are using two recursive functions, the maximum depth of the call stack will be the height of `root` plus the height of `subRoot`.


```python
from typing import Optional

# Definition for a binary tree node.
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

class Solution:
    def isSubtree(self, root: Optional[TreeNode], subRoot: Optional[TreeNode]) -> bool:
        
        # MAIN FUNCTION BASE CASES:
        # If the subRoot is completely empty, it is technically a subtree of everything
        if not subRoot: 
            return True
            
        # If the main tree is empty, but the subRoot is not, we can't find a match
        if not root:
            return False
            
        # 1. Check if the CURRENT node starts a perfect match
        if self.isSameTree(root, subRoot):
            return True
            
        # 2. If it's not a match here, delegate the search down to the left and right children.
        # We use 'or' because the subtree only needs to exist on ONE of the sides.
        return self.isSubtree(root.left, subRoot) or self.isSubtree(root.right, subRoot)
        
        
    # HELPER FUNCTION: The exact logic from the "Same Binary Tree" problem
    def isSameTree(self, p: Optional[TreeNode], q: Optional[TreeNode]) -> bool:
        # Both are empty -> Match
        if not p and not q:
            return True
            
        # Only one is empty, or the values differ -> Mismatch
        if not p or not q or p.val != q.val:
            return False
            
        # Check both left and right wings
        return self.isSameTree(p.left, q.left) and self.isSameTree(p.right, q.right)
```

### Problem 007: Lowest Common Ancestor of a Binary Search Tree (LeetCode 235)

### Problem Definition and Constraints
Given a Binary Search Tree (BST), find the **Lowest Common Ancestor (LCA)** of two given nodes, `p` and `q`.
The LCA is the lowest node in the tree that has both `p` and `q` as descendants (a node can be a descendant of itself).

* **The BST Property:** For any node, all values in its left subtree are smaller, and all values in its right subtree are larger.
* Constraints:
  * 2 <= The number of nodes in the tree <= 100.
  * `p` and `q` will always exist in the tree.
  * `p` != `q`

### Examples
* **Example 1:**
  * Input: `root = [6,2,8,0,4,7,9,null,null,3,5]`, `p = 2`, `q = 8`
  * Output: `6`
  * Explanation: `2` is on the left of `6`, and `8` is on the right of `6`. The root `6` is the point where their paths split, making it the LCA.
* **Example 2:**
  * Input: `root = [6,2,8,0,4,7,9,null,null,3,5]`, `p = 2`, `q = 4`
  * Output: `2`
  * Explanation: The LCA of nodes `2` and `4` is `2`, since a node can be a descendant of itself.

### Optimized Approach (The "Split Point" Logic)
Because this is a BST, we don't need to do a full DFS traversal. We just start at the root and look for the **Split Point**. 
1. If both `p` and `q` are **greater** than the current node, they must both be hiding down the right branch. We dive right.
2. If both `p` and `q` are **smaller** than the current node, they must both be hiding down the left branch. We dive left.
3. **The Split:** If one is greater and one is smaller, the paths to `p` and `q` are officially splitting in opposite directions. The exact node where they split **must** be the Lowest Common Ancestor!
4. **The Self-Ancestor:** If the current node is exactly equal to `p` or `q`, we have physically bumped into one of them. Since the other node must be further down this exact path, the node we bumped into is the LCA.

* Time Complexity: $O(h)$ — Where $h$ is the height of the tree. In the worst-case, we dive straight down to a leaf node.
* Space Complexity: $O(1)$ Iterative, or $O(h)$ Recursive. Because we only ever dive down *one* specific path, we never have to pause and check the other side. This means we can write this iteratively with a simple `while` loop, completely eliminating the recursive call stack!

```python
# Definition for a binary tree node.
class TreeNode:
    def __init__(self, x):
        self.val = x
        self.left = None
        self.right = None

class Solution:
    
    # --- OPTIMIZED APPROACH (ITERATIVE O(1) SPACE) ---
    def lowestCommonAncestor(self, root: 'TreeNode', p: 'TreeNode', q: 'TreeNode') -> 'TreeNode':
        
        # We use a pointer 'curr' to walk down the tree
        curr = root
        
        while curr:
            # 1. Are they both GREATER than the current node?
            # If so, the LCA must be in the right subtree.
            if p.val > curr.val and q.val > curr.val:
                curr = curr.right
                
            # 2. Are they both SMALLER than the current node?
            # If so, the LCA must be in the left subtree.
            elif p.val < curr.val and q.val < curr.val:
                curr = curr.left
                
            # 3. THE SPLIT POINT!
            # If one is greater and one is smaller, OR if curr is exactly p or q...
            # We have found the Lowest Common Ancestor.
            else:
                return curr

    # --- ALTERNATIVE APPROACH (RECURSIVE O(H) SPACE) ---
    def lowestCommonAncestor_recursive(self, root: 'TreeNode', p: 'TreeNode', q: 'TreeNode') -> 'TreeNode':
        if p.val > root.val and q.val > root.val:
            return self.lowestCommonAncestor_recursive(root.right, p, q)
            
        elif p.val < root.val and q.val < root.val:
            return self.lowestCommonAncestor_recursive(root.left, p, q)
            
        else:
            return root
```

### Problem 008: Binary Tree Level Order Traversal (LeetCode 102)

### Problem Definition and Constraints
Given the root of a binary tree, return a nested list representing its level order traversal. 
Each sub-list should contain the values of the nodes at that specific level, reading from left to right.

* Constraints:
  * 0 <= The number of nodes in the tree <= 1000.
  * -1000 <= Node.val <= 1000

### Examples
* **Example 1:**
  * Input: `root = [3,9,20,null,null,15,7]`
  * Output: `[[3], [9,20], [15,7]]`
  * Explanation: Level 0 has 3. Level 1 has 9 and 20. Level 2 has 15 and 7.
* **Example 2:**
  * Input: `root = []`
  * Output: `[]`

### Optimized Approach (Breadth-First Search / BFS)
To sweep across the tree level by level, we use a **Queue** (specifically Python's `collections.deque`). 
The algorithm works in waves:
1. **Initialize:** Put the `root` node into the queue.
2. **The Snapshot:** While the queue has nodes, take a snapshot of its current length (`level_size`). This number tells us exactly how many nodes belong to the current level we are looking at.
3. **Process the Level:** Loop exactly `level_size` times. For each loop:
   * Pop the front node out of the queue.
   * Add its value to a temporary `current_level` list.
   * If it has a left child, push it to the back of the queue.
   * If it has a right child, push it to the back of the queue.
4. **Save and Repeat:** Once that loop finishes, the `current_level` list is complete. We append it to our final result, and the queue is now perfectly prepped with only the children of the next level.

* Time Complexity: $O(n)$ — We process every node in the tree exactly once.
* Space Complexity: $O(n)$ — In the worst-case scenario (a completely balanced tree), the bottom level holds roughly half of all nodes in the entire tree ($n/2$). Therefore, our queue will need to hold up to $O(n)$ nodes at once.

```python
import collections
from typing import Optional, List

# Definition for a binary tree node.
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

class Solution:
    
    # --- OPTIMIZED APPROACH (BREADTH-FIRST SEARCH) ---
    def levelOrder(self, root: Optional[TreeNode]) -> List[List[int]]:
        # Edge case: if the tree is completely empty, return an empty list
        if not root:
            return []
            
        result = []
        
        # We use deque (double-ended queue) because popping from the front is O(1) fast
        queue = collections.deque([root])
        
        while queue:
            # 1. Take a snapshot of how many nodes are on THIS specific level
            level_size = len(queue)
            current_level = []
            
            # 2. Process exactly that many nodes
            for _ in range(level_size):
                
                # Pop the guy at the front of the line
                node = queue.popleft()
                
                # Write his value down for this level's report
                current_level.append(node.val)
                
                # If he has kids, send them to the back of the line for the NEXT level
                if node.left:
                    queue.append(node.left)
                if node.right:
                    queue.append(node.right)
                    
            # 3. We finished processing all nodes for this level. 
            # Add the completed level to our final result.
            result.append(current_level)
            
        return result
```

### Problem 009: Binary Tree Right Side View (LeetCode 199)

### Problem Definition and Constraints
Given the `root` of a binary tree, imagine yourself standing on the right side of it. Return the values of the nodes you can see, ordered from top to bottom.

* Constraints:
  * 0 <= The number of nodes in the tree <= 100.
  * -100 <= Node.val <= 100

### Examples
* **Example 1:**
  * Input: `root = [1,2,3,null,5,null,4]`
  * Output: `[1,3,4]`
  * Explanation: Level 0 ends with 1. Level 1 ends with 3. Level 2 ends with 4. 
* **Example 2:**
  * Input: `root = [1,2,3,4,null,null,null,5]`
  * Output: `[1,3,4,5]`
  * Explanation: Even though 5 is on the left side of the tree's structure, because there are no right-side nodes blocking it at level 3, you can still see it from the right side!

### Optimized Approach (Breadth-First Search)
We use the exact same Breadth-First Search (BFS) template as "Level Order Traversal". 
1. We use a queue to process the tree one level at a time.
2. We take our snapshot (`level_size`) to lock in the number of nodes on the current level.
3. We run our `for` loop to process that exact number of nodes. 
4. **The Only Difference:** As we process the nodes in the `for` loop, we check if the current node is the *very last one* in the loop. If it is, we append its value to our final result!

* Time Complexity: $O(n)$ — We process every node in the tree exactly once.
* Space Complexity: $O(n)$ — Our queue will need to hold up to $O(n)$ nodes at once (the maximum width of the tree).

```python
import collections
from typing import Optional, List

# Definition for a binary tree node.
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

class Solution:
    def rightSideView(self, root: Optional[TreeNode]) -> List[int]:
        # 1. Edge Case: Empty tree
        if not root:
            return []
            
        result = []
        queue = collections.deque([root])
        
        # 2. The Wave Manager
        while queue:
            level_size = len(queue)
            
            # 3. Process the current level
            for i in range(level_size):
                node = queue.popleft()
                
                # THE MAGIC FILTER: Is this the very last person in this wave?
                # Since 'i' goes from 0 to level_size - 1, we check if i is at the end.
                if i == level_size - 1:
                    result.append(node.val)
                    
                # Send the kids to the back of the line for the next wave
                if node.left:
                    queue.append(node.left)
                if node.right:
                    queue.append(node.right)
                    
        return result
```

### Problem 010: Count Good Nodes in Binary Tree (LeetCode 1448)

### Problem Definition and Constraints
Given the root of a binary tree, you need to find the total number of "good" nodes.
A node is considered **good** if there are no nodes with a strictly greater value on the path from the root down to that node. 
Think of it like hiking a mountain: if a node is as tall as (or taller than) the highest peak you've seen *so far* on this specific trail, it's a good node.

**Constraints:**
* 1 <= number of nodes in the tree <= 100,000
* -100 <= Node.val <= 100

**Examples:**
* **Example 1:**
  * **Input:** `root = [2,1,1,3,null,1,5]`
  * **Output:** `3`
* **Example 2:**
  * **Input:** `root = [1,2,-1,3,4]`
  * **Output:** `4`

### Core Logic: Top-Down DFS (Passing State Down)
Instead of checking every single node back up to the root (which would be $O(n^2)$ time), we use a top-down Depth First Search. 

As we travel down the tree, we simply carry a "backpack" containing the highest number we have seen so far on this specific path (`max_val`).
1. **The Check:** When we land on a node, we compare its value to our backpack's `max_val`. If it is $\ge$ the backpack value, it is a good node! We count it as `1`. If it's smaller, it's hidden behind a taller mountain, so we count it as `0`.
2. **The Update:** Before we step down to the children, we update our backpack. The new highest number seen is `max(max_val, node.val)`. 
3. **The Delegation:** We pass this updated backpack down to the left and right children and ask them how many good nodes they found.
4. **The Return:** We add up our current node's score (1 or 0) plus whatever the left and right branches found, and return that total sum back up to the parent.

* **Time Complexity:** $O(n)$ — We visit every single node in the tree exactly once.
* **Space Complexity:** $O(n)$ — In the worst-case scenario (a completely unbalanced tree), the recursive call stack will grow to size $n$. In a balanced tree, it will be $O(\log n)$.

```python
from typing import Optional

# Definition for a binary tree node.
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

class Solution:
    def goodNodes(self, root: TreeNode) -> int:
        
        # Helper function that passes the 'max_val' state DOWN the tree
        def dfs(node, max_val):
            # BASE CASE: An empty space has 0 good nodes
            if not node:
                return 0
                
            # 1. CHECK: Is this node a "good" node?
            # It is good if it's >= the highest value seen on this path so far.
            if node.val >= max_val:
                current_is_good = 1
            else:
                current_is_good = 0
                
            # 2. UPDATE: What is the new highest value for the rest of this path?
            new_max = max(max_val, node.val)
            
            # 3. RECURSE: Ask the left and right branches to count their good nodes,
            # making sure to hand them the updated 'new_max' backpack.
            left_count = dfs(node.left, new_max)
            right_count = dfs(node.right, new_max)
            
            # 4. RETURN: The total good nodes from this branch
            return current_is_good + left_count + right_count
            
        # Kick off the recursion starting at the root. 
        # The root is always a good node, so the initial max_val is just root.val.
        return dfs(root, root.val)


        """
        =========================================================
        SIMULATION NOTES: root = [3, 1, 4, 3, null, 1, 5]
        =========================================================
        
        Start at Root (3):
          dfs(Node(3), max_val = 3)
          3 >= 3 -> Good (1)
          new_max = max(3, 3) = 3
          
          Dive Left to Node(1):
            dfs(Node(1), max_val = 3)
            1 is NOT >= 3 -> Bad (0)
            new_max = max(3, 1) = 3
            
            Dive Left to Node(3):
              dfs(Node(3), max_val = 3)
              3 >= 3 -> Good (1)
              (Returns 1 up the chain)
              
          Dive Right to Node(4):
            dfs(Node(4), max_val = 3)
            4 >= 3 -> Good (1)
            new_max = max(3, 4) = 4  <-- The highest peak is now 4!
            
            Dive Right to Node(5):
              dfs(Node(5), max_val = 4)
              5 >= 4 -> Good (1)
              (Returns 1 up the chain)
              
        Total Sum = 1 (Root) + 1 (Left branch) + 2 (Right branch) = 4 Good Nodes.
        """
```

### Problem 011: Valid Binary Search Tree (LeetCode 98)

### Problem Definition and Constraints
Given the root of a binary tree, you must determine if it is a valid Binary Search Tree (BST).
A valid BST has three strict rules:
1. Every single node in the **left subtree** must be strictly *less than* the current node.
2. Every single node in the **right subtree** must be strictly *greater than* the current node.
3. Both left and right subtrees must also be valid BSTs.

**Constraints:**
* 1 <= The number of nodes in the tree <= 10,000.
* -1,000,000,000 <= Node.val <= 1,000,000,000

**Examples:**
* **Example 1:**
  * **Input:** `root = [2,1,3]`
  * **Output:** `true`
* **Example 2:**
  * **Input:** `root = [5,1,4,null,null,3,6]`
  * **Output:** `false` 
  * **Explanation:** Even though the immediate children of `4` (`3` and `6`) form a valid mini-BST, the node `3` is in the right subtree of the root `5`. Therefore, it violates the rule because `3` is not greater than `5`.

### Core Logic: Top-Down DFS (The "Valid Range" Backpack)
The biggest trap in this problem is only checking a node against its immediate parent. As seen in Example 2, a node deep in the tree still has to obey the rules set by its grand-parents!

To solve this, we use a top-down Depth First Search and carry a "Valid Range" backpack containing a `min_bound` and a `max_bound`.
1. **The Root:** The root node can be any number, so its valid range is `[-infinity, infinity]`.
2. **Going Left (Update Ceiling):** When we step into a left child, that child (and everything below it) must be smaller than the current node. So, the current node's value becomes the new `max_bound`.
3. **Going Right (Update Floor):** When we step into a right child, that child (and everything below it) must be larger than the current node. So, the current node's value becomes the new `min_bound`.
4. **The Check:** The moment any node's value steps outside its valid `(min_bound, max_bound)` range, we immediately return `False`.

* **Time Complexity:** $O(n)$ — We visit every node in the tree at most once. Mismatches short-circuit the recursion early.
* **Space Complexity:** $O(n)$ — The recursive call stack will grow to $O(n)$ in the worst-case (a straight line tree) or $O(\log n)$ in the best-case (a balanced tree).

```python
from typing import Optional

# Definition for a binary tree node.
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

class Solution:
    def isValidBST(self, root: Optional[TreeNode]) -> bool:
        
        # Helper function that passes the valid boundaries down the tree
        def dfs(node, min_bound, max_bound):
            
            # BASE CASE: An empty space doesn't violate any rules, so it is valid
            if not node:
                return True
                
            # THE CHECK: Does this node break the rules set by its ancestors?
            # It must be strictly greater than min_bound and strictly less than max_bound.
            if not (min_bound < node.val < max_bound):
                return False
                
            # DELEGATE: Check both the left and right subtrees.
            # When going LEFT, the current node becomes the absolute Ceiling (max_bound).
            # When going RIGHT, the current node becomes the absolute Floor (min_bound).
            left_is_valid = dfs(node.left, min_bound, node.val)
            right_is_valid = dfs(node.right, node.val, max_bound)
            
            # RETURN: The whole tree is only valid if both sides report back as valid.
            return left_is_valid and right_is_valid
            
        # Kick off the recursion. 
        # The root has no ancestors, so it can be any number from -infinity to infinity.
        return dfs(root, float('-inf'), float('inf'))


        """
        =========================================================
        SIMULATION NOTES: root = [5, 1, 6, null, null, 3, 7]
        =========================================================
        
        Start at Root (5):
          dfs(Node(5), min = -inf, max = inf)
          -inf < 5 < inf -> Valid!
          
          Dive Left to Node(1):
            dfs(Node(1), min = -inf, max = 5)  <-- 5 is the new ceiling
            -inf < 1 < 5 -> Valid!
            (Its children are null, so it returns True)
            
          Dive Right to Node(6):
            dfs(Node(6), min = 5, max = inf)   <-- 5 is the new floor
            5 < 6 < inf -> Valid!
            
            Dive Left to Node(3):
              dfs(Node(3), min = 5, max = 6)   <-- Floor is 5, Ceiling is 6
              Wait... is 5 < 3 < 6 ? NO! 
              3 is not greater than 5. (Rule violation!)
              Returns False.
              
        The False bubbles all the way to the top. Final answer: False.
        """
```

### Problem 012: Kth Smallest Integer in BST (LeetCode 230)

### Problem Definition and Constraints
Given the root of a Binary Search Tree (BST) and an integer `k`, return the `k`th smallest value (1-indexed) in the tree.

**Constraints:**
* 1 <= k <= The number of nodes in the tree <= 10,000.
* 0 <= Node.val <= 10,000

**Examples:**
* **Example 1:**
  * **Input:** `root = [2,1,3], k = 1`
  * **Output:** `1`
* **Example 2:**
  * **Input:** `root = [4,3,5,2,null], k = 4`
  * **Output:** `5`

### Core Logic: In-Order Traversal (The Sorting Cheat Code)
The most important property of a Binary Search Tree is that if you read it in a specific order—**Left, Node, Right**—you will naturally read the values in perfectly sorted ascending order. This is called an **In-Order Traversal**.

Instead of pulling all the numbers out and paying the heavy time cost to sort them, we just walk the tree In-Order and count our steps.
1. **The Setup:** We use two global "whiteboard" variables attached to `self`: one to count how many nodes we've processed (`self.count`), and one to hold our final answer (`self.result`).
2. **Dive Left:** We recursively dive down the left branches as far as possible. This guarantees we hit the absolute smallest number in the entire tree first.
3. **Process the Node:** When we step onto a node, we increment our `self.count`. We then check: *Is `count == k`?* If it is, this node is exactly the $k$ th smallest! We write its value to `self.result` and stop.
4. **Dive Right:** If we haven't hit $k$ yet, we dive into the right subtree to look at the next set of slightly larger numbers.

* **Time Complexity:** $O(n)$ — In the worst-case scenario (where $k$ is the very last node in an unbalanced tree), we visit every node. In reality, it runs in $O(h + k)$ because we stop searching the moment we find the target.
* **Space Complexity:** $O(n)$ — The maximum size of the recursive call stack is bounded by the height of the tree.

```python
from typing import Optional

# Definition for a binary tree node.
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

class Solution:
    def kthSmallest(self, root: Optional[TreeNode], k: int) -> int:
        
        # Global "whiteboards" to track our progress across all recursive calls
        self.count = 0
        self.result = None
        
        # Helper function for In-Order Traversal (Left, Node, Right)
        def inorder(node):
            
            # BASE CASE: Stop if the node is empty, OR if we already found our answer.
            # (Checking self.result ensures we don't waste time processing the rest of the tree)
            if not node or self.result is not None:
                return
                
            # 1. LEFT: Dive down to the smaller numbers first
            inorder(node.left)
            
            # 2. NODE: Process the current number
            self.count += 1
            if self.count == k:
                self.result = node.val
                return  # We found the target! 
                
            # 3. RIGHT: Move on to the larger numbers
            inorder(node.right)
            
        # Kick off the recursion starting at the root
        inorder(root)
        
        # Once the traversal stops, our whiteboard holds the final answer
        return self.result


        """
        =========================================================
        SIMULATION NOTES: root = [4, 2, 5, 1, 3], k = 3
        =========================================================
        
        In-Order means we always try to go LEFT before we process the NODE.
        
        Dive down left: 4 -> 2 -> 1
        At Node(1): 
          Left is null.
          Process NODE: count = 1. (Not 3).
          Right is null.
          
        Back up to Node(2):
          Left is done.
          Process NODE: count = 2. (Not 3).
          Dive Right to Node(3).
          
        At Node(3):
          Left is null.
          Process NODE: count = 3. 
          count == k! We write '3' to self.result and stop.
          
        The recursive calls instantly return because self.result is no longer None.
        Final answer: 3.
        """
```

### Problem 013: Construct Binary Tree from Preorder and Inorder Traversal (LeetCode 105)

### Problem Definition and Constraints
You are given two integer arrays `preorder` and `inorder`.
* `preorder` is the preorder traversal of a binary tree (Root $\rightarrow$ Left $\rightarrow$ Right).
* `inorder` is the inorder traversal of the same tree (Left $\rightarrow$ Root $\rightarrow$ Right).
Both arrays are of the same size and consist of unique values.
Rebuild the binary tree from these traversals and return its root.

**Examples:**
* **Example 1:**
  * **Input:** `preorder = [1,2,3,4]`, `inorder = [2,1,3,4]`
  * **Output:** `[1,2,3,null,null,null,4]`
* **Example 2:**
  * **Input:** `preorder = [1]`, `inorder = [1]`
  * **Output:** `[1]`

* Constraints:
  * 1 <= inorder.length <= 2000
  * inorder.length == preorder.length
  * -1000 <= preorder[i], inorder[i] <= 1000
  * Both arrays consist of **unique** values.

### Core Logic: The "Anchor and Split"
To build a tree, you need to know what the Root is, what belongs in the Left Subtree, and what belongs in the Right Subtree. 

By definition of the traversals:
1. **The Anchor (Preorder):** The very first element in a preorder array is **always** the root of the tree. 
2. **The Split (Inorder):** If you find that root value inside the inorder array, everything strictly to its left belongs to the left subtree, and everything strictly to its right belongs to the right subtree.

Because `preorder` natively traverses Root $\rightarrow$ Left $\rightarrow$ Right, we can process it sequentially. For every number we read in `preorder`, we create a node, find its index in the `inorder` array, and then recursively pass the left and right boundaries to build its children. 

### Approach: Hash Map + Top-Down DFS
A brute-force method would use `inorder.index(root_val)` to find the split point, taking $O(n)$ time per node and ruining the complexity. 

Instead, we map it out first:
1. **The Blueprint Registry:** Iterate through the `inorder` array once and store the values and their indices in a Hash Map (`val -> index`). This gives us $O(1)$ lookups to find the split points.
2. **The Global Pointer:** Maintain a `pre_idx` variable that tracks which value in the `preorder` array we are currently turning into a node.
3. **The Recursive Builder:** Pass a `left` and `right` boundary (representing the current valid range in the `inorder` array) into a DFS helper.
   * If `left > right`, the range is invalid (empty subtree). Return `None`.
   * Grab the value at `preorder[pre_idx]`, increment `pre_idx`, and create the `root` node.
   * Look up this value in the Hash Map to find the `mid` split point.
   * Recursively build `root.left` using the range `[left, mid - 1]`.
   * Recursively build `root.right` using the range `[mid + 1, right]`.
   * Return the `root`.

* **Time Complexity:** $O(n)$ — Building the Hash Map takes $O(n)$. We create each of the $n$ nodes exactly once, and Hash Map lookups are $O(1)$.
* **Space Complexity:** $O(n)$ — The Hash Map stores $n$ elements, and the recursive call stack can grow up to $O(n)$ in an unbalanced tree.

```python
from typing import List, Optional

# Definition for a binary tree node.
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

class Solution:
    def buildTree(self, preorder: List[int], inorder: List[int]) -> Optional[TreeNode]:
        
        # 1. THE BLUEPRINT REGISTRY: Hash Map for O(1) index lookups
        inorder_map = {val: idx for idx, val in enumerate(inorder)}
        
        # 2. THE GLOBAL POINTER: Tracks which node we are building from preorder
        self.pre_idx = 0
        
        # 3. THE RECURSIVE BUILDER
        def build(left, right):
            # BASE CASE: If the boundaries cross, there are no nodes left for this subtree
            if left > right:
                return None
                
            # Create the current root node using the preorder array
            root_val = preorder[self.pre_idx]
            root = TreeNode(root_val)
            
            # Move the pointer forward for the next recursive call
            self.pre_idx += 1
            
            # Find where this root lives in the inorder array to split the branches
            mid = inorder_map[root_val]
            
            # DELEGATE: Build the left branch using elements strictly left of 'mid'
            root.left = build(left, mid - 1)
            
            # DELEGATE: Build the right branch using elements strictly right of 'mid'
            # Note: Because preorder processes Left entirely before Right, the self.pre_idx 
            # will naturally be in the exact correct spot when it finishes building root.left!
            root.right = build(mid + 1, right)
            
            # Return the fully constructed sub-tree upward
            return root
            
        # Kick off the recursion using the entire inorder array boundaries
        return build(0, len(inorder) - 1)


        """
        =========================================================
        SIMULATION NOTES: preorder = [3,9,20,15,7], inorder = [9,3,15,20,7]
        =========================================================
        inorder_map = {9: 0, 3: 1, 15: 2, 20: 3, 7: 4}
        pre_idx = 0
        
        build(0, 4):
          root_val = preorder[0] = 3. 
          pre_idx becomes 1.
          mid = inorder_map[3] = 1.
          
          Dive Left -> build(0, mid - 1) -> build(0, 0):
            root_val = preorder[1] = 9. 
            pre_idx becomes 2.
            mid = inorder_map[9] = 0.
            
            Dive Left -> build(0, -1) -> Returns None
            Dive Right -> build(1, 0) -> Returns None
            Returns Node(9).
            
          Dive Right -> build(mid + 1, 4) -> build(2, 4):
            root_val = preorder[2] = 20. 
            pre_idx becomes 3.
            mid = inorder_map[20] = 3.
            
            Dive Left -> build(2, 2):
              root_val = preorder[3] = 15. Returns Node(15).
              
            Dive Right -> build(4, 4):
              root_val = preorder[4] = 7. Returns Node(7).
              
            Returns Node(20).
            
        Final tree root is Node(3).
        """
```

### Problem 014: Binary Tree Maximum Path Sum (LeetCode 124) [HARD]

### Problem Definition and Constraints
Given the root of a non-empty binary tree, return the maximum path sum of any non-empty path.
A path is a sequence of connected nodes. It can start and end at *any* node in the tree. The path sum is simply the sum of the node values along that path.

**Examples:**
* **Example 1:**
  * **Input:** `root = [1,2,3]`
  * **Output:** `6`
  * **Explanation:** The optimal path is `2 -> 1 -> 3` with a sum of 2 + 1 + 3 = 6.
* **Example 2:**
  * **Input:** `root = [-15,10,20,null,null,15,5,-5]`
  * **Output:** `40`
  * **Explanation:** The optimal path is `15 -> 20 -> 5` with a sum of 15 + 20 + 5 = 40. Notice that it completely ignores the negative root.

* Constraints:
  * 1 <= The number of nodes in the tree <= 30,000.
  * -1000 <= Node.val <= 1000 (Negative numbers are allowed and common!)

### Core Logic: The "Inverted V" vs. The "Straight Line"
This is the ultimate boss of Bottom-Up DFS problems. It combines the logic of "Diameter of Binary Tree" with the "Toxic Baggage" concept from Greedy algorithms.

Because the path can start and end anywhere, every single node in the tree has to make a choice. It acts as the "highest point" (the peak) of a potential path. 
At any given node, we calculate two different things:
1. **The Split (The "Inverted V"):** What is the maximum path sum if this current node is the absolute peak of the path? That would be `left_branch + current_node + right_branch`. We calculate this and check if it beats our global high score.
2. **The Extension (The "Straight Line"):** The current node's parent doesn't care about the Inverted V. A path can't fork in three directions. The parent asks: "What is the single highest straight-line path you can offer me?" The current node must choose its best leg and return `current_node + max(left_branch, right_branch)`.

**The Toxic Baggage Rule:**
Because nodes can be negative, a branch might have a negative total sum. If a left or right branch returns a negative number, it acts as toxic baggage—it will only drag our sum down. The Greedy choice? We simply ignore it. If a branch is negative, we pretend it doesn't exist and treat its sum as `0`.

### Approach: Bottom-Up DFS with Global State
1. Initialize a global whiteboard variable `self.max_sum = float('-inf')`.
2. Create a recursive `dfs(node)` helper function.
3. **Base Case:** If `node` is `None`, return `0`.
4. **Delegate & Filter:** Call `dfs` on the left and right children. Wrap the recursive call in a `max(0, ...)` to instantly eliminate negative toxic branches.
5. **The Secret Job (Update Global Max):** Calculate the "Inverted V" sum passing through this node: `node.val + left_max + right_max`. If it's bigger than `self.max_sum`, update it.
6. **The Public Job (Return to Parent):** Return the "Straight Line" sum to the parent: `node.val + max(left_max, right_max)`.

* **Time Complexity:** $O(n)$ — We visit every single node in the tree exactly once.
* **Space Complexity:** $O(h)$ — The recursive call stack grows to the height of the tree. In the worst case (a linked list), this is $O(n)$; in a perfectly balanced tree, it is $O(\log n)$.

```python
from typing import Optional

# Definition for a binary tree node.
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

class Solution:
    def maxPathSum(self, root: Optional[TreeNode]) -> int:
        # Global whiteboard to track the absolute highest sum we've ever seen
        # We initialize it to negative infinity to handle trees with entirely negative nodes
        self.max_sum = float('-inf')
        
        def dfs(node):
            # BASE CASE: An empty space contributes 0 to the sum
            if not node:
                return 0
                
            # 1. DELEGATE & FILTER (Toxic Baggage Check)
            # Ask the left and right branches for their best straight-line paths.
            # If a branch returns a negative number, we use max(0, ...) to just ignore it.
            left_max = max(0, dfs(node.left))
            right_max = max(0, dfs(node.right))
            
            # 2. SECRET JOB: Calculate the "Inverted V" path through THIS node
            # This path goes up the left branch, through the node, and down the right branch.
            current_split_sum = node.val + left_max + right_max
            
            # Update our global record if this inverted V beats the high score
            self.max_sum = max(self.max_sum, current_split_sum)
            
            # 3. PUBLIC JOB: Return the best "Straight Line" path to the parent
            # A valid path moving upward can only choose ONE branch to connect to.
            return node.val + max(left_max, right_max)
            
        # Kick off the recursion starting at the root
        dfs(root)
        
        # The recursion finishes, and our whiteboard holds the true maximum sum
        return self.max_sum


        """
        =========================================================
        SIMULATION NOTES: root = [-15, 10, 20, null, null, 15, 5]
        =========================================================
        
        Start at Root (-15):
          Dive Left to Node(10):
            Left child is null (returns 0). Right child is null (returns 0).
            Secret Job: current_split = 10 + 0 + 0 = 10. (max_sum = 10)
            Public Job: Returns 10 + max(0, 0) = 10 up to Root.
            
          Dive Right to Node(20):
            Dive Left to Node(15):
              Left/Right are null (0).
              Secret Job: current_split = 15. (max_sum = max(10, 15) = 15)
              Public Job: Returns 15 up to Node(20).
              
            Dive Right to Node(5):
              Left/Right are null (0).
              Secret Job: current_split = 5. (max_sum stays 15)
              Public Job: Returns 5 up to Node(20).
              
            Back at Node(20):
              left_max = 15, right_max = 5.
              Secret Job: current_split = 20 + 15 + 5 = 40! (max_sum updates to 40)
              Public Job: Returns 20 + max(15, 5) = 35 up to Root.
              
          Back at Root (-15):
            left_max = 10, right_max = 35.
            Secret Job: current_split = -15 + 10 + 35 = 30. (max_sum stays 40).
            
        Final Answer: 40.
        """
```

### Problem 015: Serialize and Deserialize Binary Tree (LeetCode 297) [HARD]

### Problem Definition and Constraints
Implement an algorithm to serialize and deserialize a binary tree.
* **Serialization** is the process of converting an in-memory structure into a sequence of bits (or a string) so that it can be stored or sent across a network.
* **Deserialization** is the process of reconstructing that exact same tree from the string.
There is no restriction on how your algorithm works, as long as a tree can be serialized to a string, and that string can be deserialized back into the original tree structure.

**Examples:**
* **Example 1:**
  * **Input:** `root = [1,2,3,null,null,4,5]`
  * **Output:** `[1,2,3,null,null,4,5]`
* **Example 2:**
  * **Input:** `root = []`
  * **Output:** `[]`

* Constraints:
  * 0 <= The number of nodes in the tree <= 10,000.
  * -1000 <= Node.val <= 1000

### Core Logic: The "Preorder Blueprint" and the "Null Placeholder"
To perfectly reconstruct a binary tree, you need to know exactly when a branch ends. If you just write down the numbers, a string like `"1,2,3"` is highly ambiguous—it could be a tree that branches left and right, or a straight line of nodes leaning to the left.

**The Fix (Null Placeholders):** We must explicitly record the empty spaces (`None`). If we mark every `None` with a special character like `"N"`, the structure becomes strictly mathematically locked. 

**The Preorder Traversal:**
We use a Depth First Search (DFS) in **Preorder** (Root $\rightarrow$ Left $\rightarrow$ Right). 
*   Why Preorder? Because when we reconstruct the tree, we need to know what the root of the current subtree is *before* we can attach children to it. Preorder guarantees the very first item we read is the root.

### Approach: DFS String Joining & Splitting
1. **Serialization:** 
   * Create an empty array to hold the string components.
   * Run a Preorder DFS. If the node is `None`, append `"N"`. If it's a real node, append its string value (`str(node.val)`), then recursively call DFS on its left and right children.
   * Join the array with commas (e.g., `"1,2,N,N,3,4,N,N,5,N,N"`) and return it. (Joining an array is $O(n)$ and much faster in Python than repeatedly concatenating immutable strings).
2. **Deserialization:**
   * Split the incoming string by commas to get an array of values.
   * Initialize a global pointer `self.i = 0` to track which value we are currently reading.
   * Run a Preorder DFS to build the tree. Read the value at `self.i` and increment the pointer.
   * If the value is `"N"`, return `None`.
   * If it's a number, create a `TreeNode` with that integer. Then, recursively set `node.left = dfs()` and `node.right = dfs()`. 
   * Return the fully built `node`.

* **Time Complexity:** $O(n)$ — Both serialization and deserialization visit every single node (and every `null` placeholder) exactly once.
* **Space Complexity:** $O(n)$ — We create a string/array of size proportional to the number of nodes, and the recursive call stack scales with the height of the tree.

```python
# Definition for a binary tree node.
class TreeNode(object):
    def __init__(self, x):
        self.val = x
        self.left = None
        self.right = None

class Codec:

    def serialize(self, root):
        """Encodes a tree to a single string."""
        res = []
        
        def dfs(node):
            # BASE CASE: Explicitly mark the dead ends
            if not node:
                res.append("N")
                return
                
            # PREORDER: Process Root, then Left, then Right
            res.append(str(node.val))
            dfs(node.left)
            dfs(node.right)
            
        dfs(root)
        
        # Join into a single comma-separated string for O(n) efficiency
        return ",".join(res)
        

    def deserialize(self, data):
        """Decodes your encoded data to tree."""
        # Split the string back into an array of values
        vals = data.split(",")
        self.i = 0
        
        def dfs():
            # If we reach the end of the array or hit a placeholder, return None
            if vals[self.i] == "N":
                self.i += 1
                return None
                
            # PREORDER: Create the Root first
            node = TreeNode(int(vals[self.i]))
            self.i += 1
            
            # Since the string was built in Preorder, the next sequential values 
            # naturally belong to the left branch, followed by the right branch!
            node.left = dfs()
            node.right = dfs()
            
            return node
            
        return dfs()


        """
        =========================================================
        SIMULATION NOTES: root = [1, 2, 3, null, null, 4, 5]
        =========================================================
        
        --- SERIALIZATION ---
        dfs(Node 1): appends "1"
          dfs(Node 2): appends "2"
            dfs(null): appends "N"
            dfs(null): appends "N"
          dfs(Node 3): appends "3"
            dfs(Node 4): appends "4"
              dfs(null): appends "N"
              dfs(null): appends "N"
            dfs(Node 5): appends "5"
              dfs(null): appends "N"
              dfs(null): appends "N"
              
        Result String: "1,2,N,N,3,4,N,N,5,N,N"
        
        --- DESERIALIZATION ---
        vals = ["1", "2", "N", "N", "3", "4", "N", "N", "5", "N", "N"]
        i = 0
        
        dfs() at i=0 ("1"):
          node = TreeNode(1). i becomes 1.
          
          Build Left -> dfs() at i=1 ("2"):
            node = TreeNode(2). i becomes 2.
            Build Left -> dfs() at i=2 ("N"): returns None. i=3.
            Build Right -> dfs() at i=3 ("N"): returns None. i=4.
            Returns Node(2).
            
          Build Right -> dfs() at i=4 ("3"):
            node = TreeNode(3). i becomes 5.
            Build Left -> dfs() at i=5 ("4") -> processes 4 and its two Ns.
            Build Right -> dfs() at i=8 ("5") -> processes 5 and its two Ns.
            Returns Node(3).
            
        Returns Node(1) fully reconstructed!
        """
```
